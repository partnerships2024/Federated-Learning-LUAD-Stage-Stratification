## Step 1A — Create Output Folders and Locate Source-File Candidates

This part creates only the approved output folders and searches for possible original LUAD source files.

Is part mein sirf file paths locate hongi. Abhi headers inspect, DNA reconstruct, clinical mapping ya model training nahi hogi.

```python
from pathlib import Path
import re


# ============================================================
# STEP 1A — PROJECT SETUP AND SOURCE-FILE LOCATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\LAPTOP\Downloads\luad_reserch_code"
)

OUTPUT_DIR = PROJECT_DIR / "final_luad_results"

COHORTS = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
]

COHORT_ALIASES = {
    "TCGA_Atlas_2018": ["tcga_atlas_2018"],
    "LUNG_MSK_2017": ["lung_msk_2017"],
    "MSK_NPJPO_2021": ["msk_npjpo_2021"],
    "MSKCC_2020": ["mskcc_2020"],
    "MSKCC_2023": ["mskcc_2023"],
    "NCI_2022": ["nci_2022"],
    "ONCOSG_2020": ["oncosg_2020"],
}


def normalise_name(value):
    value = str(value).lower()
    value = re.sub(r"[^a-z0-9]+", "_", value)
    return value.strip("_")


print("=" * 80)
print("STEP 1A — PROJECT SETUP AND FILE LOCATION")
print("=" * 80)

print(f"\nPROJECT_DIR: {PROJECT_DIR}")
print(f"Project directory exists: {PROJECT_DIR.exists()}")

if not PROJECT_DIR.exists():
    raise FileNotFoundError(
        f"Project directory not found:\n{PROJECT_DIR}"
    )


# Create only the approved folders.
for folder_name in ["data", "models", "figures", "results"]:
    folder_path = OUTPUT_DIR / folder_name
    folder_path.mkdir(parents=True, exist_ok=True)

    print(
        f"Folder ready: {folder_path} "
        f"| Exists: {folder_path.exists()}"
    )


# Supported source-like extensions.
supported_extensions = (
    ".txt",
    ".tsv",
    ".csv",
    ".maf",
    ".fasta",
    ".fa",
    ".fna",
    ".txt.gz",
    ".tsv.gz",
    ".csv.gz",
    ".maf.gz",
    ".fasta.gz",
    ".fa.gz",
    ".fna.gz",
)


# Do not search the new output directory.
all_source_files = []

for path in PROJECT_DIR.rglob("*"):

    if not path.is_file():
        continue

    if OUTPUT_DIR in path.parents:
        continue

    if not path.name.lower().endswith(supported_extensions):
        continue

    all_source_files.append(path.resolve())


all_source_files = sorted(
    set(all_source_files),
    key=lambda item: str(item).lower(),
)

print(f"\nSource-like files found: {len(all_source_files)}")


# Locate cohort files.
cohort_candidates = {
    cohort: {
        "mutation": [],
        "clinical_patient": [],
        "clinical_sample": [],
    }
    for cohort in COHORTS
}


for path in all_source_files:

    path_text = normalise_name(path)
    filename = normalise_name(path.name)

    identified_cohort = None

    for cohort, aliases in COHORT_ALIASES.items():
        if any(
            normalise_name(alias) in path_text
            for alias in aliases
        ):
            identified_cohort = cohort
            break

    if identified_cohort is None:
        continue

    if "data_mutations" in filename:
        cohort_candidates[identified_cohort]["mutation"].append(path)

    elif "data_clinical_patient" in filename:
        cohort_candidates[identified_cohort]["clinical_patient"].append(path)

    elif "data_clinical_sample" in filename:
        cohort_candidates[identified_cohort]["clinical_sample"].append(path)


print("\n" + "=" * 80)
print("COHORT FILE CANDIDATES")
print("=" * 80)

for cohort in COHORTS:

    print(f"\nCOHORT: {cohort}")

    for file_type in [
        "mutation",
        "clinical_patient",
        "clinical_sample",
    ]:
        paths = cohort_candidates[cohort][file_type]

        print(f"  {file_type}: {len(paths)} candidate(s)")

        if not paths:
            print("    MISSING")
        else:
            for path in paths:
                print(f"    {path}")


# Locate Ensembl candidates.
ensembl_candidates = []

for path in all_source_files:
    path_text = normalise_name(path)

    if (
        "ensembl" in path_text
        and "cds" in path_text
    ):
        ensembl_candidates.append(path)


# Locate IntOGen candidates.
intogen_candidates = []

expected_intogen_names = {
    "compendium_cancer_genes.tsv",
    "unfiltered_drivers.tsv",
    "drivers.tsv",
}

for path in all_source_files:

    if path.name.lower() in expected_intogen_names:
        intogen_candidates.append(path)


print("\n" + "=" * 80)
print("ENSEMBL CDS CANDIDATES")
print("=" * 80)

if ensembl_candidates:
    for path in ensembl_candidates:
        print(path)
else:
    print("No Ensembl CDS candidate found.")


print("\n" + "=" * 80)
print("INTOGEN CANDIDATES")
print("=" * 80)

if intogen_candidates:
    for path in intogen_candidates:
        print(path)
else:
    print("No IntOGen candidate found.")


# Keep results available for the next small part.
STEP_1A_RESULTS = {
    "cohort_candidates": cohort_candidates,a
    "ensembl_candidates": ensembl_candidates,
    "intogen_candidates": intogen_candidates,
}


print("\n" + "=" * 80)
print("STEP 1A FINISHED — OUTPUT MUST BE REVIEWED")
print("=" * 80)

print("No file contents were loaded.")
print("No 101-bp dataset was used.")
print("No DNA reconstruction was performed.")
print("No clinical mapping was created.")
print("No model was trained.")
```

## Step 1B — Missing and Duplicate Candidate Audit

This part checks whether any cohort source file is missing or has multiple candidates.

Is part mein files load nahi hongi. Sirf discovered paths ka audit hoga.

```python
# ============================================================
# STEP 1B — MISSING AND DUPLICATE CANDIDATE AUDIT
# ============================================================

if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError(
        "STEP_1A_RESULTS not found. Please run Step 1A first."
    )

cohort_candidates = STEP_1A_RESULTS["cohort_candidates"]
ensembl_candidates = STEP_1A_RESULTS["ensembl_candidates"]
intogen_candidates = STEP_1A_RESULTS["intogen_candidates"]

print("=" * 70)
print("STEP 1B — MISSING AND DUPLICATE AUDIT")
print("=" * 70)

missing_items = []
duplicate_items = []

for cohort, file_groups in cohort_candidates.items():

    for file_type, paths in file_groups.items():

        if len(paths) == 0:
            missing_items.append((cohort, file_type))

        elif len(paths) > 1:
            duplicate_items.append((cohort, file_type, paths))


print("\nMISSING COHORT FILES")

if not missing_items:
    print("None")
else:
    for cohort, file_type in missing_items:
        print(f"- {cohort}: {file_type}")


print("\nMULTIPLE COHORT CANDIDATES")

if not duplicate_items:
    print("None")
else:
    for cohort, file_type, paths in duplicate_items:
        print(f"\n{cohort} — {file_type}")

        for path in paths:
            print(f"  {path}")


print("\nREFERENCE CANDIDATES")

print(f"Ensembl CDS candidates: {len(ensembl_candidates)}")
for path in ensembl_candidates:
    print(f"  {path}")

print(f"\nIntOGen candidates: {len(intogen_candidates)}")
for path in intogen_candidates:
    print(f"  {path}")


STEP_1B_RESULTS = {
    "missing_items": missing_items,
    "duplicate_items": duplicate_items,
    "ensembl_candidate_count": len(ensembl_candidates),
    "intogen_candidate_count": len(intogen_candidates),
}


print("\n" + "=" * 70)
print("STEP 1B FINISHED")
print("=" * 70)

print(f"Missing cohort files: {len(missing_items)}")
print(f"Multiple cohort candidates: {len(duplicate_items)}")
print(f"Ensembl candidates requiring later verification: {len(ensembl_candidates)}")
print(f"IntOGen candidates found: {len(intogen_candidates)}")

print("\nNo file contents were loaded.")
```

## Step 1C-1 — Inspect Headers for TCGA Atlas 2018 and LUNG MSK 2017

This part checks the delimiters, column names and one sample row from the mutation, clinical-patient and clinical-sample files of the first two cohorts.

Is part mein complete files memory mein load nahi hongi. Sirf headers aur one preview row inspect hogi.

```python
# ============================================================
# STEP 1C-1 — INSPECT FIRST TWO COHORTS
# ============================================================

import csv


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 1A first.")


# Confirm the selected Ensembl CDS file.
SELECTED_ENSEMBL_CDS_PATH = (
    PROJECT_DIR / "LUAD_Ensembl_CDS_sequences_96genes.fasta.txt"
)

print("=" * 75)
print("STEP 1C-1 — HEADER INSPECTION")
print("=" * 75)

print(f"\nSelected Ensembl CDS file:\n{SELECTED_ENSEMBL_CDS_PATH}")
print(f"Exists: {SELECTED_ENSEMBL_CDS_PATH.exists()}")

if not SELECTED_ENSEMBL_CDS_PATH.exists():
    raise FileNotFoundError(
        f"Selected Ensembl CDS file not found:\n"
        f"{SELECTED_ENSEMBL_CDS_PATH}"
    )


def inspect_small_header(file_path):
    """
    Read only the first non-comment header and one data row.
    """

    print("\n" + "-" * 75)
    print(f"FILE: {file_path.name}")
    print(f"PATH: {file_path}")
    print(f"Exists: {file_path.exists()}")

    if not file_path.exists():
        print("ERROR: File is missing.")
        return

    try:
        with file_path.open(
            "r",
            encoding="utf-8-sig",
            errors="replace"
        ) as file_handle:

            noncomment_lines = []

            for line in file_handle:
                clean_line = line.rstrip("\r\n")

                if not clean_line.strip():
                    continue

                if clean_line.lstrip().startswith("#"):
                    continue

                noncomment_lines.append(clean_line)

                # Only header and one data row are required.
                if len(noncomment_lines) == 2:
                    break

        if not noncomment_lines:
            print("ERROR: No header found.")
            return

        header_line = noncomment_lines[0]

        delimiter_counts = {
            "TAB": header_line.count("\t"),
            "COMMA": header_line.count(","),
            "SEMICOLON": header_line.count(";"),
        }

        delimiter_name = max(
            delimiter_counts,
            key=delimiter_counts.get
        )

        delimiter_map = {
            "TAB": "\t",
            "COMMA": ",",
            "SEMICOLON": ";",
        }

        if delimiter_counts[delimiter_name] == 0:
            print("Delimiter: Could not determine")
            print(f"Header preview: {header_line[:300]}")
            return

        delimiter = delimiter_map[delimiter_name]

        columns = next(
            csv.reader(
                [header_line],
                delimiter=delimiter
            )
        )

        print(f"Delimiter: {delimiter_name}")
        print(f"Column count: {len(columns)}")
        print("Columns:")

        for index, column in enumerate(columns, start=1):
            print(f"  {index:03d}. {column.strip()}")

        if len(noncomment_lines) > 1:
            preview_row = next(
                csv.reader(
                    [noncomment_lines[1]],
                    delimiter=delimiter
                )
            )

            print(f"Preview row field count: {len(preview_row)}")
            print(
                "Preview row: "
                f"{noncomment_lines[1][:250]}"
            )
        else:
            print("No data row available for preview.")

    except Exception as error:
        print(
            f"Inspection failed: "
            f"{type(error).__name__}: {error}"
        )


COHORTS_TO_INSPECT = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
]


for cohort in COHORTS_TO_INSPECT:

    print("\n" + "=" * 75)
    print(f"COHORT: {cohort}")
    print("=" * 75)

    for file_type in [
        "mutation",
        "clinical_patient",
        "clinical_sample",
    ]:

        candidates = STEP_1A_RESULTS[
            "cohort_candidates"
        ][cohort][file_type]

        if len(candidates) != 1:
            print(
                f"\n{file_type}: expected 1 candidate, "
                f"found {len(candidates)}"
            )
            continue

        inspect_small_header(candidates[0])


STEP_1C1_RESULTS = {
    "inspected_cohorts": COHORTS_TO_INSPECT,
    "selected_ensembl_cds": SELECTED_ENSEMBL_CDS_PATH,
}


print("\n" + "=" * 75)
print("STEP 1C-1 FINISHED")
print("=" * 75)

print("Inspected cohorts: TCGA_Atlas_2018, LUNG_MSK_2017")
print("Only headers and one row were read.")
print("No clinical mapping or DNA reconstruction was performed.")
```

### Step 1C-2 — Inspect MSK NPJPO 2021 Source Headers

This part inspects only the mutation, clinical-patient and clinical-sample files for `MSK_NPJPO_2021`.

Sirf headers aur one preview row read hogi. Koi clinical mapping, DNA reconstruction ya data processing nahi hogi.

```python
# ============================================================
# STEP 1C-2 — INSPECT MSK_NPJPO_2021 HEADERS
# ============================================================

from pathlib import Path
import csv


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 1A first.")


COHORT_TO_INSPECT = "MSK_NPJPO_2021"

candidate_files = STEP_1A_RESULTS[
    "cohort_candidates"
][COHORT_TO_INSPECT]


def inspect_small_header(file_path):
    print("\n" + "-" * 75)
    print(f"FILE: {file_path.name}")
    print(f"PATH: {file_path}")
    print(f"Exists: {file_path.exists()}")

    if not file_path.exists():
        print("ERROR: File does not exist.")
        return

    try:
        noncomment_lines = []

        with file_path.open(
            "r",
            encoding="utf-8-sig",
            errors="replace",
        ) as file_handle:

            for line in file_handle:

                if not line.strip():
                    continue

                if line.lstrip().startswith("#"):
                    continue

                noncomment_lines.append(
                    line.rstrip("\r\n")
                )

                if len(noncomment_lines) == 2:
                    break

        if not noncomment_lines:
            print("ERROR: No readable header found.")
            return

        header_line = noncomment_lines[0]

        delimiter = (
            "\t"
            if header_line.count("\t") > header_line.count(",")
            else ","
        )

        columns = next(
            csv.reader(
                [header_line],
                delimiter=delimiter,
            )
        )

        print(
            "Delimiter:",
            "TAB" if delimiter == "\t" else "COMMA",
        )

        print(f"Column count: {len(columns)}")
        print("Columns:")
        print(columns)

        if len(noncomment_lines) > 1:
            preview_row = next(
                csv.reader(
                    [noncomment_lines[1]],
                    delimiter=delimiter,
                )
            )

            print(
                f"Preview row field count: {len(preview_row)}"
            )

            print(
                "Preview row:",
                noncomment_lines[1][:400],
            )
        else:
            print("No data-row preview available.")

    except Exception as error:
        print(
            f"ERROR: {type(error).__name__}: {error}"
        )


print("=" * 75)
print(f"STEP 1C-2 — {COHORT_TO_INSPECT}")
print("=" * 75)

for file_type in [
    "mutation",
    "clinical_patient",
    "clinical_sample",
]:
    paths = candidate_files[file_type]

    if len(paths) != 1:
        print(
            f"\n{file_type}: expected 1 candidate, "
            f"found {len(paths)}"
        )

        for path in paths:
            print(path)

        continue

    inspect_small_header(paths[0])


print("\n" + "=" * 75)
print("STEP 1C-2 FINISHED")
print("=" * 75)

print("Only MSK_NPJPO_2021 was inspected.")
print("No clinical mapping was created.")
print("No DNA reconstruction was performed.")
```

### Step 1D-1 — Inspect the Selected 96-Gene Ensembl CDS FASTA

This part inspects only the first few records of the selected Ensembl CDS source:

`LUAD_Ensembl_CDS_sequences_96genes.fasta.txt`

Is step mein sirf FASTA header format, transcript identifiers, gene symbols and sequence previews check hongi. Koi DNA window reconstruct nahi hogi.

```python
# ============================================================
# STEP 1D-1 — INSPECT SELECTED ENSEMBL CDS FASTA
# ============================================================

from pathlib import Path
import re


ENSEMBL_CDS_FILE = PROJECT_DIR / (
    "LUAD_Ensembl_CDS_sequences_96genes.fasta.txt"
)

print("=" * 75)
print("STEP 1D-1 — SELECTED ENSEMBL CDS INSPECTION")
print("=" * 75)

print(f"\nFile: {ENSEMBL_CDS_FILE}")
print(f"Exists: {ENSEMBL_CDS_FILE.exists()}")

if not ENSEMBL_CDS_FILE.exists():
    raise FileNotFoundError(
        f"Selected Ensembl CDS file not found:\n{ENSEMBL_CDS_FILE}"
    )


preview_records = []
current_header = None
current_sequence_parts = []

try:
    with ENSEMBL_CDS_FILE.open(
        "r",
        encoding="utf-8-sig",
        errors="replace",
    ) as file_handle:

        for raw_line in file_handle:
            line = raw_line.strip()

            if not line:
                continue

            if line.startswith(">"):

                if current_header is not None:
                    preview_records.append(
                        (
                            current_header,
                            "".join(current_sequence_parts),
                        )
                    )

                    if len(preview_records) == 3:
                        break

                current_header = line
                current_sequence_parts = []

            else:
                current_sequence_parts.append(line)

        if (
            len(preview_records) < 3
            and current_header is not None
        ):
            preview_records.append(
                (
                    current_header,
                    "".join(current_sequence_parts),
                )
            )

except Exception as error:
    raise RuntimeError(
        f"FASTA inspection failed: {type(error).__name__}: {error}"
    )


print(f"\nPreview records read: {len(preview_records)}")

for record_number, (header, sequence) in enumerate(
    preview_records,
    start=1,
):
    header_parts = header[1:].split("|")

    gene_id = header_parts[0] if len(header_parts) > 0 else None
    transcript_id = header_parts[1] if len(header_parts) > 1 else None
    gene_symbol = header_parts[2] if len(header_parts) > 2 else None

    invalid_characters = sorted(
        set(sequence.upper()) - set("ACGTN")
    )

    print("\n" + "-" * 75)
    print(f"Record {record_number}")
    print(f"Full header: {header}")
    print(f"Header field count: {len(header_parts)}")
    print(f"Ensembl gene ID: {gene_id}")
    print(f"Ensembl transcript ID: {transcript_id}")
    print(f"Gene symbol: {gene_symbol}")
    print(f"CDS length: {len(sequence)}")
    print(f"Sequence preview: {sequence[:100]}")
    print(f"Invalid DNA characters: {invalid_characters}")


header_pattern = re.compile(
    r"^>ENSG\d+(?:\.\d+)?\|ENST\d+(?:\.\d+)?\|[^|]+$"
)

header_pattern_results = [
    bool(header_pattern.match(header))
    for header, sequence in preview_records
]

print("\n" + "=" * 75)
print("STEP 1D-1 SUMMARY")
print("=" * 75)

print(
    "All preview headers match expected "
    "ENSG|ENST|GENE format:",
    all(header_pattern_results),
)

print(
    "All preview sequences contain only A/C/G/T/N:",
    all(
        len(set(sequence.upper()) - set("ACGTN")) == 0
        for header, sequence in preview_records
    ),
)

STEP_1D_ENSEMBL_FILE = ENSEMBL_CDS_FILE

print("\nNo 301-bp window was created.")
print("No mutation was applied.")
print("No previous 101-bp data was loaded.")
```

### Step 1D-2 — Inspect Compendium Cancer Genes Source File

This part inspects only the header and three preview rows of:

`Compendium_Cancer_Genes.tsv`

Is part mein sirf delimiter, columns aur small preview check hoga. Driver-gene panel abhi create nahi hoga.

```python
# ============================================================
# STEP 1D-2 — INSPECT COMPENDIUM CANCER GENES
# ============================================================

from pathlib import Path
import csv


COMPENDIUM_FILE = (
    PROJECT_DIR / "Compendium_Cancer_Genes.tsv"
)

print("=" * 75)
print("STEP 1D-2 — COMPENDIUM CANCER GENES INSPECTION")
print("=" * 75)

print(f"\nFile: {COMPENDIUM_FILE}")
print(f"Exists: {COMPENDIUM_FILE.exists()}")

if not COMPENDIUM_FILE.exists():
    raise FileNotFoundError(
        f"File not found:\n{COMPENDIUM_FILE}"
    )


try:
    preview_lines = []

    with COMPENDIUM_FILE.open(
        "r",
        encoding="utf-8-sig",
        errors="replace",
    ) as file_handle:

        for line in file_handle:
            if not line.strip():
                continue

            preview_lines.append(
                line.rstrip("\r\n")
            )

            # Header + 3 rows only
            if len(preview_lines) == 4:
                break

    if not preview_lines:
        raise ValueError("The file is empty.")

    header_line = preview_lines[0]

    delimiter = (
        "\t"
        if header_line.count("\t") >= header_line.count(",")
        else ","
    )

    columns = next(
        csv.reader(
            [header_line],
            delimiter=delimiter,
        )
    )

    print(
        "\nDetected delimiter:",
        "TAB" if delimiter == "\t" else "COMMA",
    )

    print(f"Column count: {len(columns)}")

    print("\nColumns:")
    for number, column in enumerate(columns, start=1):
        print(f"{number:02d}. {column}")

    print("\nPreview rows:")

    for row_number, raw_row in enumerate(
        preview_lines[1:],
        start=1,
    ):
        parsed_row = next(
            csv.reader(
                [raw_row],
                delimiter=delimiter,
            )
        )

        print(
            f"\nRow {row_number} field count: "
            f"{len(parsed_row)}"
        )
        print(raw_row[:500])

except Exception as error:
    raise RuntimeError(
        f"Inspection failed: "
        f"{type(error).__name__}: {error}"
    )


print("\n" + "=" * 75)
print("STEP 1D-2 FINISHED")
print("=" * 75)

print("Only the header and three rows were inspected.")
print("No driver-gene panel was created.")
print("No patient gene indicators were created.")
```

```python
# ============================================================
# STEP 1D-3 — INSPECT UNFILTERED INTOGEN DRIVERS
# ============================================================

from pathlib import Path
import csv


UNFILTERED_DRIVERS_FILE = (
    PROJECT_DIR / "Unfiltered_drivers.tsv"
)

print("=" * 75)
print("STEP 1D-3 — UNFILTERED DRIVERS INSPECTION")
print("=" * 75)

print(f"\nFile: {UNFILTERED_DRIVERS_FILE}")
print(f"Exists: {UNFILTERED_DRIVERS_FILE.exists()}")

if not UNFILTERED_DRIVERS_FILE.exists():
    raise FileNotFoundError(
        f"File not found:\n{UNFILTERED_DRIVERS_FILE}"
    )


try:
    preview_lines = []

    with UNFILTERED_DRIVERS_FILE.open(
        "r",
        encoding="utf-8-sig",
        errors="replace",
    ) as file_handle:

        for line in file_handle:

            if not line.strip():
                continue

            preview_lines.append(
                line.rstrip("\r\n")
            )

            # Header + 3 rows only
            if len(preview_lines) == 4:
                break

    if not preview_lines:
        raise ValueError("The file is empty.")

    header_line = preview_lines[0]

    delimiter = (
        "\t"
        if header_line.count("\t") >= header_line.count(",")
        else ","
    )

    columns = next(
        csv.reader(
            [header_line],
            delimiter=delimiter,
        )
    )

    print(
        "\nDetected delimiter:",
        "TAB" if delimiter == "\t" else "COMMA",
    )

    print(f"Column count: {len(columns)}")

    print("\nColumns:")
    for number, column in enumerate(columns, start=1):
        print(f"{number:02d}. {column}")

    print("\nPreview rows:")

    for row_number, raw_row in enumerate(
        preview_lines[1:],
        start=1,
    ):
        parsed_row = next(
            csv.reader(
                [raw_row],
                delimiter=delimiter,
            )
        )

        print(
            f"\nRow {row_number} field count: "
            f"{len(parsed_row)}"
        )

        print(raw_row[:500])

except Exception as error:
    raise RuntimeError(
        f"Inspection failed: "
        f"{type(error).__name__}: {error}"
    )


print("\n" + "=" * 75)
print("STEP 1D-3 FINISHED")
print("=" * 75)

print("Only the header and three rows were inspected.")
print("No driver-gene panel was created.")
print("No patient gene indicators were created.")
```

## Step 2A — Discover LUAD Driver-Gene Candidates

This step checks both original IntOGen files for LUAD-related records and compares their gene symbols with the selected 96-gene Ensembl CDS FASTA.

Is part mein:

- LUAD-related cancer-type and cohort values identify hongi
- unique candidate gene counts print honge
- selected FASTA ke gene symbols count honge
- IntOGen aur FASTA gene overlap check hoga

The final 96-gene panel will not be saved yet.

```python
# ============================================================
# STEP 2A — DISCOVER LUAD DRIVER-GENE CANDIDATES
# ============================================================

from pathlib import Path
import pandas as pd


COMPENDIUM_FILE = PROJECT_DIR / "Compendium_Cancer_Genes.tsv"
UNFILTERED_FILE = PROJECT_DIR / "Unfiltered_drivers.tsv"
SELECTED_FASTA_FILE = (
    PROJECT_DIR / "LUAD_Ensembl_CDS_sequences_96genes.fasta.txt"
)


print("=" * 75)
print("STEP 2A — LUAD DRIVER-GENE CANDIDATE DISCOVERY")
print("=" * 75)


# ------------------------------------------------------------
# 1. Confirm files
# ------------------------------------------------------------

required_files = [
    COMPENDIUM_FILE,
    UNFILTERED_FILE,
    SELECTED_FASTA_FILE,
]

for file_path in required_files:
    print(f"\nFile: {file_path.name}")
    print(f"Exists: {file_path.exists()}")

    if not file_path.exists():
        raise FileNotFoundError(
            f"Required file was not found:\n{file_path}"
        )


# ------------------------------------------------------------
# 2. Load only the two IntOGen tabular source files
# ------------------------------------------------------------

try:
    compendium_df = pd.read_csv(
        COMPENDIUM_FILE,
        sep="\t",
        dtype=str,
        low_memory=False,
    )

    unfiltered_df = pd.read_csv(
        UNFILTERED_FILE,
        sep="\t",
        dtype=str,
        low_memory=False,
    )

except Exception as error:
    raise RuntimeError(
        f"IntOGen loading failed: "
        f"{type(error).__name__}: {error}"
    )


print("\n" + "-" * 75)
print("INTOGEN FILE SHAPES")
print("-" * 75)

print(f"Compendium shape: {compendium_df.shape}")
print(f"Unfiltered shape: {unfiltered_df.shape}")


# ------------------------------------------------------------
# 3. Validate required columns
# ------------------------------------------------------------

required_columns = {
    "Compendium": {
        "SYMBOL",
        "TRANSCRIPT",
        "COHORT",
        "CANCER_TYPE",
    },
    "Unfiltered": {
        "SYMBOL",
        "TRANSCRIPT",
        "COHORT",
        "CANCER_TYPE",
    },
}

for dataset_name, required in required_columns.items():

    dataframe = (
        compendium_df
        if dataset_name == "Compendium"
        else unfiltered_df
    )

    missing_columns = sorted(
        required - set(dataframe.columns)
    )

    print(
        f"\n{dataset_name} missing required columns: "
        f"{missing_columns}"
    )

    if missing_columns:
        raise ValueError(
            f"{dataset_name} is missing required columns: "
            f"{missing_columns}"
        )


# ------------------------------------------------------------
# 4. Find LUAD-related rows without assuming one exact label
# ------------------------------------------------------------

def find_luad_related_rows(dataframe):
    cancer_type_text = (
        dataframe["CANCER_TYPE"]
        .fillna("")
        .astype(str)
        .str.upper()
    )

    cohort_text = (
        dataframe["COHORT"]
        .fillna("")
        .astype(str)
        .str.upper()
    )

    luad_mask = (
        cancer_type_text.str.contains(
            r"LUAD|LUNG.*ADENO|ADENO.*LUNG",
            regex=True,
        )
        |
        cohort_text.str.contains(
            r"LUAD|LUNG.*ADENO|ADENO.*LUNG",
            regex=True,
        )
    )

    return dataframe.loc[luad_mask].copy()


compendium_luad = find_luad_related_rows(
    compendium_df
)

unfiltered_luad = find_luad_related_rows(
    unfiltered_df
)


print("\n" + "-" * 75)
print("LUAD-RELATED INTOGEN RECORDS")
print("-" * 75)

print(
    f"Compendium LUAD-related rows: "
    f"{len(compendium_luad):,}"
)

print(
    f"Unfiltered LUAD-related rows: "
    f"{len(unfiltered_luad):,}"
)


print("\nCompendium matching CANCER_TYPE values:")
print(
    sorted(
        compendium_luad["CANCER_TYPE"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)


print("\nCompendium matching COHORT values:")
print(
    sorted(
        compendium_luad["COHORT"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)


print("\nUnfiltered matching CANCER_TYPE values:")
print(
    sorted(
        unfiltered_luad["CANCER_TYPE"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)


print("\nUnfiltered matching COHORT values:")
print(
    sorted(
        unfiltered_luad["COHORT"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )
)


# ------------------------------------------------------------
# 5. Standardise candidate gene symbols
# ------------------------------------------------------------

def standardise_gene_symbols(series):
    symbols = (
        series
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
    )

    symbols = symbols[
        ~symbols.isin(
            ["", "NA", "NAN", "NONE", "NULL", "."]
        )
    ]

    return sorted(set(symbols))


compendium_luad_genes = standardise_gene_symbols(
    compendium_luad["SYMBOL"]
)

unfiltered_luad_genes = standardise_gene_symbols(
    unfiltered_luad["SYMBOL"]
)


print("\n" + "-" * 75)
print("LUAD CANDIDATE GENE COUNTS")
print("-" * 75)

print(
    f"Compendium unique LUAD genes: "
    f"{len(compendium_luad_genes)}"
)

print(
    f"Unfiltered unique LUAD genes: "
    f"{len(unfiltered_luad_genes)}"
)


# ------------------------------------------------------------
# 6. Parse gene symbols from selected 96-gene FASTA
# ------------------------------------------------------------

fasta_gene_symbols = []
invalid_fasta_headers = []

try:
    with SELECTED_FASTA_FILE.open(
        "r",
        encoding="utf-8-sig",
        errors="replace",
    ) as fasta_handle:

        for raw_line in fasta_handle:
            line = raw_line.strip()

            if not line.startswith(">"):
                continue

            parts = line[1:].split("|")

            if len(parts) < 3:
                invalid_fasta_headers.append(line)
                continue

            gene_symbol = parts[2].strip().upper()

            if gene_symbol:
                fasta_gene_symbols.append(gene_symbol)

except Exception as error:
    raise RuntimeError(
        f"FASTA parsing failed: "
        f"{type(error).__name__}: {error}"
    )


fasta_unique_genes = sorted(
    set(fasta_gene_symbols)
)


print("\n" + "-" * 75)
print("SELECTED FASTA GENE AUDIT")
print("-" * 75)

print(
    f"Total FASTA headers parsed: "
    f"{len(fasta_gene_symbols):,}"
)

print(
    f"Unique FASTA gene symbols: "
    f"{len(fasta_unique_genes)}"
)

print(
    f"Duplicate gene-header occurrences: "
    f"{len(fasta_gene_symbols) - len(fasta_unique_genes)}"
)

print(
    f"Invalid FASTA headers: "
    f"{len(invalid_fasta_headers)}"
)


# ------------------------------------------------------------
# 7. Compare FASTA genes with IntOGen LUAD candidates
# ------------------------------------------------------------

compendium_gene_set = set(
    compendium_luad_genes
)

unfiltered_gene_set = set(
    unfiltered_luad_genes
)

fasta_gene_set = set(
    fasta_unique_genes
)

combined_intogen_gene_set = (
    compendium_gene_set |
    unfiltered_gene_set
)


fasta_in_compendium = sorted(
    fasta_gene_set & compendium_gene_set
)

fasta_in_unfiltered = sorted(
    fasta_gene_set & unfiltered_gene_set
)

fasta_in_any_intogen = sorted(
    fasta_gene_set & combined_intogen_gene_set
)

fasta_not_in_luad_intogen = sorted(
    fasta_gene_set - combined_intogen_gene_set
)

intogen_not_in_fasta = sorted(
    combined_intogen_gene_set - fasta_gene_set
)


print("\n" + "-" * 75)
print("FASTA AND INTOGEN COMPARISON")
print("-" * 75)

print(
    f"FASTA genes found in Compendium LUAD rows: "
    f"{len(fasta_in_compendium)}"
)

print(
    f"FASTA genes found in Unfiltered LUAD rows: "
    f"{len(fasta_in_unfiltered)}"
)

print(
    f"FASTA genes found in either IntOGen source: "
    f"{len(fasta_in_any_intogen)}"
)

print(
    f"FASTA genes absent from LUAD-related IntOGen rows: "
    f"{len(fasta_not_in_luad_intogen)}"
)

print(
    "Names absent from LUAD-related IntOGen rows:"
)
print(fasta_not_in_luad_intogen)

print(
    f"\nAdditional LUAD IntOGen genes not in selected FASTA: "
    f"{len(intogen_not_in_fasta)}"
)

print(
    "First 30 additional IntOGen genes:"
)
print(intogen_not_in_fasta[:30])


# Keep results for the next small part.
STEP_2A_RESULTS = {
    "compendium_luad": compendium_luad,
    "unfiltered_luad": unfiltered_luad,
    "compendium_luad_genes": compendium_luad_genes,
    "unfiltered_luad_genes": unfiltered_luad_genes,
    "fasta_unique_genes": fasta_unique_genes,
    "fasta_not_in_luad_intogen": fasta_not_in_luad_intogen,
    "intogen_not_in_fasta": intogen_not_in_fasta,
}


print("\n" + "=" * 75)
print("STEP 2A FINISHED — OUTPUT MUST BE REVIEWED")
print("=" * 75)

print("No final driver-gene panel was saved.")
print("No patient gene indicators were created.")
print("No mutation file was filtered.")
```

### Step 2B — Verify and Save the Fixed 96-Gene LUAD Panel

This step creates the fixed 96-gene research panel using the unique gene symbols from the selected Ensembl FASTA and verifies them against the original IntOGen LUAD records.

Is step mein gene symbols standardise, duplicates remove aur missing symbols report honge. Patient gene indicators abhi create nahi honge.

```python
# ============================================================
# STEP 2B — VERIFY AND SAVE THE 96-GENE LUAD PANEL
# ============================================================

import pandas as pd
from collections import defaultdict


if "STEP_2A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 2A first.")


OUTPUT_GENE_PANEL = (
    OUTPUT_DIR / "data" / "luad_96_driver_genes.csv"
)

SELECTED_FASTA_FILE = (
    PROJECT_DIR / "LUAD_Ensembl_CDS_sequences_96genes.fasta.txt"
)


# ------------------------------------------------------------
# 1. Parse gene and transcript information from FASTA
# ------------------------------------------------------------

gene_transcripts = defaultdict(set)
missing_gene_headers = []
invalid_headers = []

with SELECTED_FASTA_FILE.open(
    "r",
    encoding="utf-8-sig",
    errors="replace",
) as fasta_handle:

    for raw_line in fasta_handle:

        line = raw_line.strip()

        if not line.startswith(">"):
            continue

        parts = line[1:].split("|")

        if len(parts) != 3:
            invalid_headers.append(line)
            continue

        gene_id = parts[0].strip()
        transcript_id = parts[1].strip()
        gene_symbol = parts[2].strip().upper()

        if not gene_symbol:
            missing_gene_headers.append(line)
            continue

        gene_transcripts[gene_symbol].add(transcript_id)


# ------------------------------------------------------------
# 2. Retrieve IntOGen LUAD gene sets from Step 2A
# ------------------------------------------------------------

compendium_gene_set = set(
    STEP_2A_RESULTS["compendium_luad_genes"]
)

unfiltered_gene_set = set(
    STEP_2A_RESULTS["unfiltered_luad_genes"]
)


# ------------------------------------------------------------
# 3. Build the verified panel
# ------------------------------------------------------------

panel_rows = []

for gene_symbol in sorted(gene_transcripts):

    panel_rows.append(
        {
            "GENE_SYMBOL": gene_symbol,
            "ENSEMBL_TRANSCRIPT_COUNT": len(
                gene_transcripts[gene_symbol]
            ),
            "IN_COMPENDIUM_LUAD": int(
                gene_symbol in compendium_gene_set
            ),
            "IN_UNFILTERED_LUAD": int(
                gene_symbol in unfiltered_gene_set
            ),
            "PANEL_SOURCE": (
                "Fixed_96_gene_panel_verified_with_IntOGen"
            ),
        }
    )


gene_panel_df = pd.DataFrame(panel_rows)


# ------------------------------------------------------------
# 4. Validation checks
# ------------------------------------------------------------

duplicate_symbols = gene_panel_df[
    gene_panel_df["GENE_SYMBOL"].duplicated(keep=False)
]["GENE_SYMBOL"].tolist()

missing_symbols = gene_panel_df[
    gene_panel_df["GENE_SYMBOL"].isna()
    | gene_panel_df["GENE_SYMBOL"].eq("")
]["GENE_SYMBOL"].tolist()

not_in_unfiltered = gene_panel_df.loc[
    gene_panel_df["IN_UNFILTERED_LUAD"] == 0,
    "GENE_SYMBOL",
].tolist()

not_in_compendium = gene_panel_df.loc[
    gene_panel_df["IN_COMPENDIUM_LUAD"] == 0,
    "GENE_SYMBOL",
].tolist()


print("=" * 75)
print("STEP 2B — 96-GENE PANEL VALIDATION")
print("=" * 75)

print(f"\nFinal unique gene count: {len(gene_panel_df)}")
print(f"Duplicate gene symbols: {len(duplicate_symbols)}")
print(f"Missing gene symbols: {len(missing_symbols)}")
print(f"Invalid FASTA headers: {len(invalid_headers)}")
print(f"Missing-symbol FASTA headers: {len(missing_gene_headers)}")

print(
    "Genes verified in Unfiltered IntOGen LUAD records:",
    int(gene_panel_df["IN_UNFILTERED_LUAD"].sum()),
)

print(
    "Genes verified in Compendium LUAD records:",
    int(gene_panel_df["IN_COMPENDIUM_LUAD"].sum()),
)

print(
    f"\nGenes absent from Unfiltered LUAD records: "
    f"{not_in_unfiltered}"
)

print(
    f"\nGenes absent from Compendium LUAD records "
    f"({len(not_in_compendium)}):"
)
print(not_in_compendium)


if len(gene_panel_df) != 96:
    raise ValueError(
        f"Expected 96 unique genes, but found "
        f"{len(gene_panel_df)}."
    )

if duplicate_symbols:
    raise ValueError(
        f"Duplicate gene symbols found: {duplicate_symbols}"
    )

if missing_symbols:
    raise ValueError(
        "Missing gene symbols were found."
    )

if not_in_unfiltered:
    raise ValueError(
        "Some selected genes were not verified in the "
        "Unfiltered IntOGen LUAD records."
    )


# ------------------------------------------------------------
# 5. Save final panel
# ------------------------------------------------------------

gene_panel_df.to_csv(
    OUTPUT_GENE_PANEL,
    index=False,
)


print("\n" + "=" * 75)
print("STEP 2B FINISHED")
print("=" * 75)

print(f"Saved file: {OUTPUT_GENE_PANEL}")
print(f"Saved file exists: {OUTPUT_GENE_PANEL.exists()}")
print(f"Saved rows: {len(gene_panel_df)}")
print(f"Saved columns: {gene_panel_df.shape[1]}")

print("\nFirst 10 genes:")
print(gene_panel_df.head(10).to_string(index=False))

print("\nNo patient gene indicators were created.")
print("No mutation files were filtered.")
print("No DNA reconstruction was performed.")
```

## Step 3A — Discover Clinical Columns for TCGA Atlas 2018 and LUNG MSK 2017

This step reads only the clinical file headers for the first two cohorts.

It searches for actual columns related to:

- patient and sample identifiers
- age
- sex
- smoking
- pack-years
- passive smoking
- clinical or pathological stage

Is step mein clinical values harmonise nahi hongi aur koi final dataset create nahi hoga.

```python
# ============================================================
# STEP 3A — CLINICAL COLUMN DISCOVERY
# TCGA_ATLAS_2018 AND LUNG_MSK_2017
# ============================================================

import pandas as pd
import re


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 1A first.")


COHORTS_TO_INSPECT = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
]


SEARCH_PATTERNS = {
    "patient_identifier": [
        r"^PATIENT_ID$",
        r"PATIENT.*ID",
    ],
    "sample_identifier": [
        r"^SAMPLE_ID$",
        r"SAMPLE.*ID",
    ],
    "age": [
        r"(^|_)AGE($|_)",
        r"AGE_AT",
        r"DAYS_TO_BIRTH",
    ],
    "sex": [
        r"^SEX$",
        r"GENDER",
    ],
    "smoking_history_or_status": [
        r"SMOK",
        r"TOBACCO",
        r"CIGARETTE",
    ],
    "pack_years": [
        r"PACK.*YEAR",
        r"YEAR.*PACK",
    ],
    "passive_smoking": [
        r"PASSIVE.*SMOK",
        r"SECOND.*HAND",
    ],
    "stage": [
        r"STAGE",
    ],
}


def read_header_only(file_path):
    """
    Read only the tabular header.
    cBioPortal metadata lines beginning with # are ignored.
    """
    return pd.read_csv(
        file_path,
        sep="\t",
        comment="#",
        nrows=0,
        dtype=str,
    )


def find_matching_columns(columns, patterns):
    """
    Return actual column names matching one or more patterns.
    """
    matches = []

    for column in columns:
        column_upper = str(column).strip().upper()

        if any(
            re.search(pattern, column_upper)
            for pattern in patterns
        ):
            matches.append(column)

    return matches


CLINICAL_COLUMN_DISCOVERY = {}

print("=" * 78)
print("STEP 3A — CLINICAL COLUMN DISCOVERY")
print("=" * 78)


for cohort in COHORTS_TO_INSPECT:

    print("\n" + "=" * 78)
    print(f"COHORT: {cohort}")
    print("=" * 78)

    CLINICAL_COLUMN_DISCOVERY[cohort] = {}

    for file_type in [
        "clinical_patient",
        "clinical_sample",
    ]:

        candidate_paths = STEP_1A_RESULTS[
            "cohort_candidates"
        ][cohort][file_type]

        print(f"\nFILE TYPE: {file_type}")

        if len(candidate_paths) != 1:
            print(
                f"ERROR: Expected 1 file, found "
                f"{len(candidate_paths)}"
            )

            for path in candidate_paths:
                print(path)

            continue

        file_path = candidate_paths[0]

        print(f"Path: {file_path}")
        print(f"Exists: {file_path.exists()}")

        try:
            header_df = read_header_only(file_path)
            actual_columns = header_df.columns.tolist()

        except Exception as error:
            print(
                f"ERROR reading header: "
                f"{type(error).__name__}: {error}"
            )
            continue

        print(f"Total columns: {len(actual_columns)}")

        matched_groups = {}

        for variable_group, patterns in SEARCH_PATTERNS.items():

            matched_columns = find_matching_columns(
                actual_columns,
                patterns,
            )

            matched_groups[variable_group] = matched_columns

            print(
                f"  {variable_group}: "
                f"{matched_columns if matched_columns else 'None'}"
            )

        CLINICAL_COLUMN_DISCOVERY[
            cohort
        ][file_type] = {
            "file_path": str(file_path),
            "all_columns": actual_columns,
            "matched_columns": matched_groups,
        }


print("\n" + "=" * 78)
print("STEP 3A FINISHED")
print("=" * 78)

print("Cohorts inspected:", COHORTS_TO_INSPECT)
print("Only clinical headers were read.")
print("No age, sex, smoking or stage values were harmonised.")
print("No clinical dataset was created.")
```

### Step 3A Correction — Fix False Age-Column Matches

The previous age search incorrectly identified `STAGE_AT_DIAGNOSIS` as an age column because the word `STAGE` contains the letters `AGE`.

This correction uses stricter age-column rules and repeats the discovery for the same two cohorts.

Is cell mein sirf column discovery correct hogi. Clinical values harmonise nahi hongi.

```python
# ============================================================
# STEP 3A CORRECTION — SAFE CLINICAL COLUMN MATCHING
# ============================================================

import pandas as pd
import re


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 1A first.")


COHORTS_TO_RECHECK = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
]


SAFE_SEARCH_PATTERNS = {
    "patient_identifier": [
        r"^PATIENT_ID$",
        r"^OTHER_PATIENT_ID$",
    ],
    "sample_identifier": [
        r"^SAMPLE_ID$",
    ],
    "age": [
        r"^AGE$",
        r"^AGE_AT_",
        r"^AGE_",
        r"_AGE$",
        r"^DAYS_TO_BIRTH$",
    ],
    "sex": [
        r"^SEX$",
        r"^GENDER$",
    ],
    "smoking_history_or_status": [
        r"SMOKING",
        r"TOBACCO",
        r"CIGARETTE",
    ],
    "pack_years": [
        r"PACK_YEARS",
        r"PACK.*YEAR",
        r"YEAR.*PACK",
    ],
    "passive_smoking": [
        r"PASSIVE.*SMOK",
        r"SECOND.*HAND",
    ],
    "stage": [
        r"STAGE",
    ],
}


def read_clinical_header(file_path):
    return pd.read_csv(
        file_path,
        sep="\t",
        comment="#",
        nrows=0,
        dtype=str,
    ).columns.tolist()


def safe_column_matches(columns, patterns):
    matches = []

    for column in columns:
        cleaned_column = str(column).strip().upper()

        if any(
            re.search(pattern, cleaned_column)
            for pattern in patterns
        ):
            matches.append(column)

    return matches


CLINICAL_COLUMN_DISCOVERY_CORRECTED = {}

print("=" * 78)
print("STEP 3A CORRECTION — SAFE COLUMN DISCOVERY")
print("=" * 78)


for cohort in COHORTS_TO_RECHECK:

    print("\n" + "=" * 78)
    print(f"COHORT: {cohort}")
    print("=" * 78)

    CLINICAL_COLUMN_DISCOVERY_CORRECTED[cohort] = {}

    for file_type in [
        "clinical_patient",
        "clinical_sample",
    ]:

        candidate_paths = STEP_1A_RESULTS[
            "cohort_candidates"
        ][cohort][file_type]

        print(f"\nFILE TYPE: {file_type}")

        if len(candidate_paths) != 1:
            print(
                f"ERROR: Expected 1 candidate, "
                f"found {len(candidate_paths)}"
            )
            continue

        file_path = candidate_paths[0]

        try:
            columns = read_clinical_header(file_path)

        except Exception as error:
            print(
                f"ERROR: {type(error).__name__}: {error}"
            )
            continue

        discovered = {}

        for variable_name, patterns in SAFE_SEARCH_PATTERNS.items():

            matches = safe_column_matches(
                columns,
                patterns,
            )

            discovered[variable_name] = matches

            print(
                f"  {variable_name}: "
                f"{matches if matches else 'None'}"
            )

        CLINICAL_COLUMN_DISCOVERY_CORRECTED[
            cohort
        ][file_type] = {
            "file_path": str(file_path),
            "matched_columns": discovered,
        }


# Explicit validation for the previous false match.
false_age_matches = []

for cohort, file_groups in (
    CLINICAL_COLUMN_DISCOVERY_CORRECTED.items()
):
    for file_type, details in file_groups.items():

        age_columns = details[
            "matched_columns"
        ]["age"]

        for column in age_columns:
            if "STAGE" in column.upper():
                false_age_matches.append(
                    (cohort, file_type, column)
                )


print("\n" + "=" * 78)
print("CORRECTION VALIDATION")
print("=" * 78)

print(
    "Stage columns incorrectly detected as age:",
    false_age_matches,
)

if false_age_matches:
    raise ValueError(
        "False age-column matches still exist."
    )

print("\nStep 3A correction finished.")
print("No clinical values were harmonised.")
print("No clinical dataset was created.")
```

### Step 3B-1 Correction — Separate Smoking Status and Pack-Years Columns

This correction prevents pack-year columns from being classified as smoking-history or smoking-status columns.

Is cell mein sirf column-discovery rule correct hoga. Clinical values harmonise nahi hongi.

```python
# ============================================================
# STEP 3B-1 CORRECTION — SEPARATE SMOKING AND PACK-YEAR FIELDS
# ============================================================

import re


def discover_clinical_columns_safely(columns):
    discovered = {
        "patient_identifier": [],
        "sample_identifier": [],
        "age": [],
        "sex": [],
        "smoking_history_or_status": [],
        "pack_years": [],
        "passive_smoking": [],
        "stage": [],
    }

    for column in columns:
        name = str(column).strip().upper()

        if name in {"PATIENT_ID", "OTHER_PATIENT_ID"}:
            discovered["patient_identifier"].append(column)

        if name == "SAMPLE_ID":
            discovered["sample_identifier"].append(column)

        if (
            name == "AGE"
            or name.startswith("AGE_AT_")
            or name.startswith("AGE_")
            or name.endswith("_AGE")
            or name == "DAYS_TO_BIRTH"
        ):
            discovered["age"].append(column)

        if name in {"SEX", "GENDER"}:
            discovered["sex"].append(column)

        is_pack_years = bool(
            re.search(r"PACK.*YEAR|YEAR.*PACK", name)
        )

        is_passive_smoking = bool(
            re.search(r"PASSIVE.*SMOK|SECOND.*HAND", name)
        )

        if is_pack_years:
            discovered["pack_years"].append(column)

        if is_passive_smoking:
            discovered["passive_smoking"].append(column)

        if (
            re.search(r"SMOKING|TOBACCO|CIGARETTE", name)
            and not is_pack_years
            and not is_passive_smoking
        ):
            discovered["smoking_history_or_status"].append(column)

        if "STAGE" in name:
            discovered["stage"].append(column)

    return discovered


COHORT_TO_RECHECK = "MSK_NPJPO_2021"

print("=" * 78)
print("STEP 3B-1 CORRECTION — MSK_NPJPO_2021")
print("=" * 78)

for file_type in ["clinical_patient", "clinical_sample"]:

    file_path = STEP_1A_RESULTS[
        "cohort_candidates"
    ][COHORT_TO_RECHECK][file_type][0]

    columns = read_clinical_header(file_path)

    discovered = discover_clinical_columns_safely(columns)

    print(f"\nFILE TYPE: {file_type}")

    for variable_name, matches in discovered.items():
        print(
            f"  {variable_name}: "
            f"{matches if matches else 'None'}"
        )

    CLINICAL_COLUMN_DISCOVERY_CORRECTED[
        COHORT_TO_RECHECK
    ][file_type] = {
        "file_path": str(file_path),
        "matched_columns": discovered,
    }


patient_matches = CLINICAL_COLUMN_DISCOVERY_CORRECTED[
    COHORT_TO_RECHECK
]["clinical_patient"]["matched_columns"]

incorrect_overlap = sorted(
    set(patient_matches["smoking_history_or_status"])
    & set(patient_matches["pack_years"])
)

print("\n" + "=" * 78)
print("CORRECTION VALIDATION")
print("=" * 78)

print(
    "Overlap between smoking-status and pack-years:",
    incorrect_overlap,
)

if incorrect_overlap:
    raise ValueError(
        "Smoking-status and pack-year fields still overlap."
    )

print("\nStep 3B-1 correction finished.")
print("No clinical values were harmonised.")
```

### Step 3C — Consolidate and Save the Clinical Column Discovery Audit

This step combines the discovered clinical columns for all seven cohorts into one audit table.

Is step mein sirf discovered column names save hongi. Age, sex, smoking aur stage values abhi harmonise nahi hongi.

```python
# ============================================================
# STEP 3C — CONSOLIDATE CLINICAL COLUMN DISCOVERY
# ============================================================

import pandas as pd


if "CLINICAL_COLUMN_DISCOVERY_CORRECTED" not in globals():
    raise RuntimeError(
        "Clinical column discovery results were not found."
    )


EXPECTED_COHORTS = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
]

OUTPUT_AUDIT_FILE = (
    OUTPUT_DIR
    / "results"
    / "luad_clinical_column_discovery.csv"
)


print("=" * 78)
print("STEP 3C — CLINICAL COLUMN DISCOVERY AUDIT")
print("=" * 78)


missing_cohorts = [
    cohort
    for cohort in EXPECTED_COHORTS
    if cohort not in CLINICAL_COLUMN_DISCOVERY_CORRECTED
]

print(f"\nMissing cohort discovery results: {missing_cohorts}")

if missing_cohorts:
    raise ValueError(
        "Clinical discovery is incomplete for: "
        f"{missing_cohorts}"
    )


audit_rows = []

for cohort in EXPECTED_COHORTS:

    for file_type in [
        "clinical_patient",
        "clinical_sample",
    ]:

        details = CLINICAL_COLUMN_DISCOVERY_CORRECTED[
            cohort
        ].get(file_type)

        if not details:
            raise ValueError(
                f"Missing {file_type} discovery for {cohort}"
            )

        matched = details["matched_columns"]

        audit_rows.append(
            {
                "COHORT": cohort,
                "FILE_TYPE": file_type,
                "SOURCE_FILE": details["file_path"],
                "PATIENT_ID_COLUMNS": " | ".join(
                    matched.get("patient_identifier", [])
                ),
                "SAMPLE_ID_COLUMNS": " | ".join(
                    matched.get("sample_identifier", [])
                ),
                "AGE_COLUMNS": " | ".join(
                    matched.get("age", [])
                ),
                "SEX_COLUMNS": " | ".join(
                    matched.get("sex", [])
                ),
                "SMOKING_COLUMNS": " | ".join(
                    matched.get(
                        "smoking_history_or_status",
                        [],
                    )
                ),
                "PACK_YEARS_COLUMNS": " | ".join(
                    matched.get("pack_years", [])
                ),
                "PASSIVE_SMOKING_COLUMNS": " | ".join(
                    matched.get("passive_smoking", [])
                ),
                "STAGE_COLUMNS": " | ".join(
                    matched.get("stage", [])
                ),
            }
        )


clinical_column_audit_df = pd.DataFrame(audit_rows)


# Save discovery audit.
clinical_column_audit_df.to_csv(
    OUTPUT_AUDIT_FILE,
    index=False,
)


print(f"\nAudit rows: {len(clinical_column_audit_df)}")
print(f"Expected rows: {len(EXPECTED_COHORTS) * 2}")
print(f"Saved file: {OUTPUT_AUDIT_FILE}")
print(f"Saved file exists: {OUTPUT_AUDIT_FILE.exists()}")


print("\nClinical discovery summary:")

summary_columns = [
    "COHORT",
    "FILE_TYPE",
    "AGE_COLUMNS",
    "SEX_COLUMNS",
    "SMOKING_COLUMNS",
    "PACK_YEARS_COLUMNS",
    "PASSIVE_SMOKING_COLUMNS",
    "STAGE_COLUMNS",
]

print(
    clinical_column_audit_df[
        summary_columns
    ].to_string(index=False)
)


print("\n" + "=" * 78)
print("STEP 3C FINISHED")
print("=" * 78)

print("All seven cohorts are included.")
print("No clinical values were harmonised.")
print("No target labels were created.")
print("No patient-level dataset was created.")
```

## Step 4A — Audit Raw Age Values for TCGA Atlas 2018 and LUNG MSK 2017

This step examines the original age columns for the first two cohorts.

It will:

- preserve raw age values
- convert possible numeric values temporarily for auditing
- inspect missing and non-numeric values
- check minimum, maximum and suspicious values
- compare TCGA `AGE` with `DAYS_TO_BIRTH`

Is part mein age imputation, scaling ya final `age_for_model` create nahi hoga.

```python
# ============================================================
# STEP 4A — RAW AGE AUDIT
# TCGA_ATLAS_2018 AND LUNG_MSK_2017
# ============================================================

import pandas as pd
import numpy as np


AGE_AUDIT_COHORTS = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "age_columns": ["AGE", "DAYS_TO_BIRTH"],
    },
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "age_columns": ["AGE"],
    },
}


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "NOT AVAILABLE",
    "UNKNOWN",
    ".",
}


def clean_numeric_age(series):
    """
    Convert valid raw values to numeric only for auditing.
    Original values remain unchanged.
    """
    cleaned = series.astype("string").str.strip()

    cleaned = cleaned.mask(
        cleaned.str.upper().isin(MISSING_TOKENS)
    )

    return pd.to_numeric(cleaned, errors="coerce")


AGE_RAW_AUDIT_RESULTS = {}

print("=" * 78)
print("STEP 4A — RAW AGE AUDIT")
print("=" * 78)


for cohort, settings in AGE_AUDIT_COHORTS.items():

    file_path = settings["file"]
    requested_columns = ["PATIENT_ID"] + settings["age_columns"]

    print("\n" + "=" * 78)
    print(f"COHORT: {cohort}")
    print("=" * 78)

    print(f"File: {file_path}")
    print(f"Exists: {file_path.exists()}")

    if not file_path.exists():
        print("ERROR: Clinical-patient file is missing.")
        continue

    try:
        age_df = pd.read_csv(
            file_path,
            sep="\t",
            comment="#",
            usecols=requested_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        print(
            f"ERROR loading age columns: "
            f"{type(error).__name__}: {error}"
        )
        continue

    print(f"Patient rows loaded: {len(age_df):,}")

    cohort_results = {
        "file_path": str(file_path),
        "row_count": len(age_df),
        "columns": {},
    }

    for age_column in settings["age_columns"]:

        raw_series = age_df[age_column]
        numeric_series = clean_numeric_age(raw_series)

        raw_nonmissing = (
            raw_series.astype("string").str.strip()
        )

        raw_nonmissing = raw_nonmissing[
            ~raw_nonmissing.str.upper().isin(MISSING_TOKENS)
        ].dropna()

        nonnumeric_values = sorted(
            set(
                raw_nonmissing[
                    pd.to_numeric(
                        raw_nonmissing,
                        errors="coerce",
                    ).isna()
                ].tolist()
            )
        )

        print("\n" + "-" * 78)
        print(f"AGE COLUMN: {age_column}")
        print(f"Non-missing numeric values: {numeric_series.notna().sum():,}")
        print(f"Missing/unusable values: {numeric_series.isna().sum():,}")
        print(f"Non-numeric raw values: {nonnumeric_values[:20]}")

        if numeric_series.notna().any():
            print(f"Numeric minimum: {numeric_series.min()}")
            print(f"Numeric maximum: {numeric_series.max()}")
            print(f"Numeric median: {numeric_series.median()}")

        if age_column == "DAYS_TO_BIRTH":
            suspicious_mask = (
                numeric_series.notna()
                & (
                    (numeric_series > 0)
                    | (numeric_series.abs() > 120 * 365.25)
                )
            )

            estimated_years = numeric_series.abs() / 365.25

            print(
                "Likely unit: days "
                "(conversion not applied permanently yet)"
            )
            print(
                f"Suspicious day values: "
                f"{int(suspicious_mask.sum())}"
            )

            if estimated_years.notna().any():
                print(
                    f"Estimated age range in years: "
                    f"{estimated_years.min():.2f} to "
                    f"{estimated_years.max():.2f}"
                )

        else:
            suspicious_mask = (
                numeric_series.notna()
                & (
                    (numeric_series < 0)
                    | (numeric_series > 120)
                )
            )

            print("Likely unit: years")
            print(
                f"Suspicious year values: "
                f"{int(suspicious_mask.sum())}"
            )

        suspicious_examples = age_df.loc[
            suspicious_mask,
            ["PATIENT_ID", age_column],
        ].head(10)

        if len(suspicious_examples) > 0:
            print("Suspicious examples:")
            print(suspicious_examples.to_string(index=False))

        cohort_results["columns"][age_column] = {
            "numeric_values": numeric_series,
            "missing_count": int(numeric_series.isna().sum()),
            "suspicious_count": int(suspicious_mask.sum()),
            "nonnumeric_values": nonnumeric_values,
        }

    # TCGA internal comparison only.
    if cohort == "TCGA_Atlas_2018":

        age_years = clean_numeric_age(age_df["AGE"])
        birth_days = clean_numeric_age(age_df["DAYS_TO_BIRTH"])
        birth_days_as_years = birth_days.abs() / 365.25

        comparison_mask = (
            age_years.notna()
            & birth_days_as_years.notna()
        )

        absolute_difference = (
            age_years - birth_days_as_years
        ).abs()

        large_disagreement_mask = (
            comparison_mask
            & (absolute_difference > 1.0)
        )

        print("\n" + "-" * 78)
        print("TCGA AGE vs DAYS_TO_BIRTH COMPARISON")
        print(
            f"Rows with both values: "
            f"{int(comparison_mask.sum()):,}"
        )
        print(
            f"Rows differing by more than 1 year: "
            f"{int(large_disagreement_mask.sum()):,}"
        )

        if comparison_mask.any():
            print(
                f"Median absolute difference: "
                f"{absolute_difference[comparison_mask].median():.3f} years"
            )

        disagreement_examples = age_df.loc[
            large_disagreement_mask,
            ["PATIENT_ID", "AGE", "DAYS_TO_BIRTH"],
        ].copy()

        disagreement_examples["DAYS_TO_BIRTH_AS_YEARS"] = (
            birth_days_as_years[large_disagreement_mask].round(2)
        )

        if len(disagreement_examples) > 0:
            print("First disagreement examples:")
            print(
                disagreement_examples.head(10).to_string(
                    index=False
                )
            )

        cohort_results["tcga_comparison"] = {
            "both_available_count": int(comparison_mask.sum()),
            "over_one_year_difference_count": int(
                large_disagreement_mask.sum()
            ),
        }

    AGE_RAW_AUDIT_RESULTS[cohort] = cohort_results


print("\n" + "=" * 78)
print("STEP 4A FINISHED — OUTPUT MUST BE REVIEWED")
print("=" * 78)

print("No age value was imputed.")
print("No age scaling was performed.")
print("No final age column was created.")
print("No clinical dataset was saved.")
```

### Step 4A Correction — Inspect Suspicious Young and Extreme Ages

The previous audit treated only ages below 0 or above 120 as suspicious.

This correction separately identifies:

- impossible ages: below 0 or above 120
- suspicious ages: below 18 or above 100

Suspicious values will only be reported. They will not be silently corrected, removed or imputed.

```python
# ============================================================
# STEP 4A CORRECTION — SUSPICIOUS AGE INSPECTION
# ============================================================

import pandas as pd


COHORTS_TO_RECHECK = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "age_column": "AGE",
    },
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "age_column": "AGE",
    },
}


print("=" * 78)
print("STEP 4A CORRECTION — SUSPICIOUS AGE INSPECTION")
print("=" * 78)


for cohort, settings in COHORTS_TO_RECHECK.items():

    file_path = settings["file"]
    age_column = settings["age_column"]

    print("\n" + "=" * 78)
    print(f"COHORT: {cohort}")
    print("=" * 78)

    try:
        age_df = pd.read_csv(
            file_path,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", age_column],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        print(
            f"ERROR: {type(error).__name__}: {error}"
        )
        continue

    age_numeric = pd.to_numeric(
        age_df[age_column].astype("string").str.strip(),
        errors="coerce",
    )

    impossible_mask = (
        age_numeric.notna()
        & (
            (age_numeric < 0)
            | (age_numeric > 120)
        )
    )

    suspicious_mask = (
        age_numeric.notna()
        & (
            (age_numeric < 18)
            | (age_numeric > 100)
        )
        & ~impossible_mask
    )

    print(f"Total patient rows: {len(age_df):,}")
    print(f"Missing/unusable ages: {age_numeric.isna().sum():,}")
    print(f"Impossible ages: {int(impossible_mask.sum())}")
    print(f"Suspicious ages: {int(suspicious_mask.sum())}")

    if suspicious_mask.any():
        print("\nSuspicious age records:")
        print(
            age_df.loc[
                suspicious_mask,
                ["PATIENT_ID", age_column],
            ].to_string(index=False)
        )

    if impossible_mask.any():
        print("\nImpossible age records:")
        print(
            age_df.loc[
                impossible_mask,
                ["PATIENT_ID", age_column],
            ].to_string(index=False)
        )


print("\n" + "=" * 78)
print("STEP 4A CORRECTION FINISHED")
print("=" * 78)

print("No suspicious age was corrected or removed.")
print("No age value was imputed.")
print("No final age_for_model column was created.")
```

```python
# ============================================================
# STEP 4B-1 — RAW AGE AUDIT: MSK_NPJPO_2021
# ============================================================

import pandas as pd


AGE_FILE = (
    PROJECT_DIR
    / "msk_npjpo_2021_data_clinical_patient.txt"
)

AGE_COLUMN = "AGE_AT_RESECTION"


print("=" * 78)
print("STEP 4B-1 — MSK_NPJPO_2021 AGE AUDIT")
print("=" * 78)

print(f"\nFile: {AGE_FILE}")
print(f"Exists: {AGE_FILE.exists()}")

if not AGE_FILE.exists():
    raise FileNotFoundError(
        f"Clinical-patient file not found:\n{AGE_FILE}"
    )


try:
    age_df = pd.read_csv(
        AGE_FILE,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", AGE_COLUMN],
        dtype=str,
        low_memory=False,
    )

except Exception as error:
    raise RuntimeError(
        f"Age-file loading failed: "
        f"{type(error).__name__}: {error}"
    )


raw_age = age_df[AGE_COLUMN].astype("string").str.strip()

missing_tokens = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    ".",
}

raw_age = raw_age.mask(
    raw_age.str.upper().isin(missing_tokens)
)

age_numeric = pd.to_numeric(
    raw_age,
    errors="coerce",
)


impossible_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 0)
        | (age_numeric > 120)
    )
)

suspicious_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 18)
        | (age_numeric > 100)
    )
    & ~impossible_mask
)


non_numeric_mask = (
    raw_age.notna()
    & age_numeric.isna()
)

non_numeric_values = sorted(
    raw_age[non_numeric_mask]
    .dropna()
    .unique()
    .tolist()
)


print(f"\nTotal patient rows: {len(age_df):,}")
print(f"Numeric age values: {age_numeric.notna().sum():,}")
print(f"Missing/unusable ages: {age_numeric.isna().sum():,}")
print(f"Non-numeric raw values: {non_numeric_values}")

if age_numeric.notna().any():
    print(f"Minimum age: {age_numeric.min()}")
    print(f"Maximum age: {age_numeric.max()}")
    print(f"Median age: {age_numeric.median()}")

print(f"Impossible ages: {int(impossible_mask.sum())}")
print(f"Suspicious ages: {int(suspicious_mask.sum())}")


if suspicious_mask.any():
    print("\nSuspicious age records:")
    print(
        age_df.loc[
            suspicious_mask,
            ["PATIENT_ID", AGE_COLUMN],
        ].to_string(index=False)
    )


if impossible_mask.any():
    print("\nImpossible age records:")
    print(
        age_df.loc[
            impossible_mask,
            ["PATIENT_ID", AGE_COLUMN],
        ].to_string(index=False)
    )


STEP_4B1_AGE_AUDIT = {
    "cohort": "MSK_NPJPO_2021",
    "source_file": str(AGE_FILE),
    "source_column": AGE_COLUMN,
    "total_rows": len(age_df),
    "numeric_count": int(age_numeric.notna().sum()),
    "missing_count": int(age_numeric.isna().sum()),
    "impossible_count": int(impossible_mask.sum()),
    "suspicious_count": int(suspicious_mask.sum()),
}


print("\n" + "=" * 78)
print("STEP 4B-1 FINISHED")
print("=" * 78)

print("No age value was corrected or removed.")
print("No age value was imputed.")
print("No age_for_model column was created.")
```

### Step 4B-3 — Audit Sample-Level Age Values for MSKCC 2023

`MSKCC_2023` stores age in the clinical-sample file as `AGE_AT_DOS_BX`.

This step will inspect:

- numeric and missing age coverage
- minimum, maximum and median
- suspicious and impossible values
- patients with multiple samples
- patients whose samples contain different age values

No representative sample will be selected and no age value will be harmonised, corrected or imputed yet.

```python
# ============================================================
# STEP 4B-3 — RAW AGE AUDIT: MSKCC_2023
# ============================================================

import pandas as pd


AGE_FILE = (
    PROJECT_DIR
    / "mskcc_2023_data_clinical_sample.txt"
)

AGE_COLUMN = "AGE_AT_DOS_BX"


print("=" * 78)
print("STEP 4B-3 — MSKCC_2023 SAMPLE-LEVEL AGE AUDIT")
print("=" * 78)

print(f"\nFile: {AGE_FILE}")
print(f"Exists: {AGE_FILE.exists()}")

if not AGE_FILE.exists():
    raise FileNotFoundError(
        f"Clinical-sample file not found:\n{AGE_FILE}"
    )


try:
    age_df = pd.read_csv(
        AGE_FILE,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", "SAMPLE_ID", AGE_COLUMN],
        dtype=str,
        low_memory=False,
    )

except Exception as error:
    raise RuntimeError(
        f"Age-file loading failed: "
        f"{type(error).__name__}: {error}"
    )


missing_tokens = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    ".",
}


raw_age = age_df[AGE_COLUMN].astype("string").str.strip()

raw_age = raw_age.mask(
    raw_age.str.upper().isin(missing_tokens)
)

age_numeric = pd.to_numeric(
    raw_age,
    errors="coerce",
)


non_numeric_mask = (
    raw_age.notna()
    & age_numeric.isna()
)

non_numeric_values = sorted(
    raw_age[non_numeric_mask]
    .dropna()
    .unique()
    .tolist()
)


impossible_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 0)
        | (age_numeric > 120)
    )
)

suspicious_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 18)
        | (age_numeric > 100)
    )
    & ~impossible_mask
)


age_df["AGE_NUMERIC_AUDIT"] = age_numeric


print(f"\nTotal sample rows: {len(age_df):,}")
print(
    f"Unique patients: "
    f"{age_df['PATIENT_ID'].nunique(dropna=True):,}"
)
print(
    f"Unique samples: "
    f"{age_df['SAMPLE_ID'].nunique(dropna=True):,}"
)

print(f"Numeric age values: {age_numeric.notna().sum():,}")
print(f"Missing/unusable ages: {age_numeric.isna().sum():,}")
print(f"Non-numeric raw values: {non_numeric_values}")

if age_numeric.notna().any():
    print(f"Minimum age: {age_numeric.min():.2f}")
    print(f"Maximum age: {age_numeric.max():.2f}")
    print(f"Median age: {age_numeric.median():.2f}")

print(f"Impossible age rows: {int(impossible_mask.sum())}")
print(f"Suspicious age rows: {int(suspicious_mask.sum())}")


if suspicious_mask.any():
    print("\nSuspicious age records:")
    print(
        age_df.loc[
            suspicious_mask,
            ["PATIENT_ID", "SAMPLE_ID", AGE_COLUMN],
        ].head(20).to_string(index=False)
    )


if impossible_mask.any():
    print("\nImpossible age records:")
    print(
        age_df.loc[
            impossible_mask,
            ["PATIENT_ID", "SAMPLE_ID", AGE_COLUMN],
        ].head(20).to_string(index=False)
    )


# ------------------------------------------------------------
# Multiple-sample and within-patient age consistency audit
# ------------------------------------------------------------

samples_per_patient = (
    age_df.groupby("PATIENT_ID", dropna=False)["SAMPLE_ID"]
    .nunique(dropna=True)
)

patients_with_multiple_samples = samples_per_patient[
    samples_per_patient > 1
]


unique_ages_per_patient = (
    age_df.dropna(subset=["AGE_NUMERIC_AUDIT"])
    .groupby("PATIENT_ID")["AGE_NUMERIC_AUDIT"]
    .nunique()
)

patients_with_different_ages = unique_ages_per_patient[
    unique_ages_per_patient > 1
]


print("\n" + "-" * 78)
print("MULTIPLE-SAMPLE AGE AUDIT")
print("-" * 78)

print(
    f"Patients with more than one sample: "
    f"{len(patients_with_multiple_samples):,}"
)

print(
    f"Patients with more than one numeric age value: "
    f"{len(patients_with_different_ages):,}"
)


if len(patients_with_different_ages) > 0:

    example_patient_ids = (
        patients_with_different_ages
        .index
        .astype(str)
        .tolist()[:10]
    )

    print("\nExamples with different sample-level ages:")

    print(
        age_df.loc[
            age_df["PATIENT_ID"].isin(example_patient_ids),
            [
                "PATIENT_ID",
                "SAMPLE_ID",
                AGE_COLUMN,
                "AGE_NUMERIC_AUDIT",
            ],
        ]
        .sort_values(["PATIENT_ID", "SAMPLE_ID"])
        .to_string(index=False)
    )


STEP_4B3_AGE_AUDIT = {
    "cohort": "MSKCC_2023",
    "source_file": str(AGE_FILE),
    "source_column": AGE_COLUMN,
    "total_sample_rows": len(age_df),
    "unique_patients": int(
        age_df["PATIENT_ID"].nunique(dropna=True)
    ),
    "numeric_count": int(age_numeric.notna().sum()),
    "missing_count": int(age_numeric.isna().sum()),
    "impossible_count": int(impossible_mask.sum()),
    "suspicious_count": int(suspicious_mask.sum()),
    "patients_with_multiple_samples": int(
        len(patients_with_multiple_samples)
    ),
    "patients_with_different_ages": int(
        len(patients_with_different_ages)
    ),
}


print("\n" + "=" * 78)
print("STEP 4B-3 FINISHED")
print("=" * 78)

print("No representative sample was selected.")
print("No age value was corrected or imputed.")
print("No final age_for_model column was created.")
```

### Step 4B-4 — Audit Raw Age Values for NCI 2022

This step audits the original `AGE` values for `NCI_2022`.

It checks numeric coverage, missing values, age range, impossible ages and suspicious ages.

No age value will be corrected, removed, imputed or scaled.

```python
# ============================================================
# STEP 4B-4 — RAW AGE AUDIT: NCI_2022
# ============================================================

import pandas as pd


AGE_FILE = (
    PROJECT_DIR
    / "nci_2022_data_clinical_patient.txt"
)

AGE_COLUMN = "AGE"


print("=" * 78)
print("STEP 4B-4 — NCI_2022 AGE AUDIT")
print("=" * 78)

print(f"\nFile: {AGE_FILE}")
print(f"Exists: {AGE_FILE.exists()}")

if not AGE_FILE.exists():
    raise FileNotFoundError(
        f"Clinical-patient file not found:\n{AGE_FILE}"
    )


try:
    age_df = pd.read_csv(
        AGE_FILE,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", AGE_COLUMN],
        dtype=str,
        low_memory=False,
    )

except Exception as error:
    raise RuntimeError(
        f"Age-file loading failed: "
        f"{type(error).__name__}: {error}"
    )


missing_tokens = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    ".",
}


raw_age = age_df[AGE_COLUMN].astype("string").str.strip()

raw_age = raw_age.mask(
    raw_age.str.upper().isin(missing_tokens)
)

age_numeric = pd.to_numeric(
    raw_age,
    errors="coerce",
)


non_numeric_mask = (
    raw_age.notna()
    & age_numeric.isna()
)

non_numeric_values = sorted(
    raw_age[non_numeric_mask]
    .dropna()
    .unique()
    .tolist()
)


impossible_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 0)
        | (age_numeric > 120)
    )
)

suspicious_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 18)
        | (age_numeric > 100)
    )
    & ~impossible_mask
)


print(f"\nTotal patient rows: {len(age_df):,}")
print(f"Numeric age values: {age_numeric.notna().sum():,}")
print(f"Missing/unusable ages: {age_numeric.isna().sum():,}")
print(f"Non-numeric raw values: {non_numeric_values}")

if age_numeric.notna().any():
    print(f"Minimum age: {age_numeric.min()}")
    print(f"Maximum age: {age_numeric.max()}")
    print(f"Median age: {age_numeric.median()}")

print(f"Impossible ages: {int(impossible_mask.sum())}")
print(f"Suspicious ages: {int(suspicious_mask.sum())}")


if suspicious_mask.any():
    print("\nSuspicious age records:")
    print(
        age_df.loc[
            suspicious_mask,
            ["PATIENT_ID", AGE_COLUMN],
        ].to_string(index=False)
    )


if impossible_mask.any():
    print("\nImpossible age records:")
    print(
        age_df.loc[
            impossible_mask,
            ["PATIENT_ID", AGE_COLUMN],
        ].to_string(index=False)
    )


STEP_4B4_AGE_AUDIT = {
    "cohort": "NCI_2022",
    "source_file": str(AGE_FILE),
    "source_column": AGE_COLUMN,
    "total_rows": len(age_df),
    "numeric_count": int(age_numeric.notna().sum()),
    "missing_count": int(age_numeric.isna().sum()),
    "impossible_count": int(impossible_mask.sum()),
    "suspicious_count": int(suspicious_mask.sum()),
}


print("\n" + "=" * 78)
print("STEP 4B-4 FINISHED")
print("=" * 78)

print("No age value was corrected or removed.")
print("No age value was imputed.")
print("No final age_for_model column was created.")
```

### Step 4C-1 — Harmonise Patient-Level Age for Six Cohorts

This step creates standard age fields for the six cohorts where age is stored in the clinical-patient file.

For TCGA:

- `AGE` will have first priority
- `DAYS_TO_BIRTH` will be used only when `AGE` is missing
- days will be converted to years only after validation

Ages below 18 or above 100 will be flagged as suspicious and excluded from `age_for_model`, but their original values will remain preserved.

`MSKCC_2023` is not included here because its age is sample-level and requires separate handling.

No age imputation or scaling will be performed.

```python
# ============================================================
# STEP 4C-1 — HARMONISE PATIENT-LEVEL AGE FOR SIX COHORTS
# ============================================================

import pandas as pd
import numpy as np


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError("Please run Step 1A first.")


AGE_CONFIG = {
    "TCGA_Atlas_2018": {
        "primary_column": "AGE",
        "fallback_days_column": "DAYS_TO_BIRTH",
    },
    "LUNG_MSK_2017": {
        "primary_column": "AGE",
        "fallback_days_column": None,
    },
    "MSK_NPJPO_2021": {
        "primary_column": "AGE_AT_RESECTION",
        "fallback_days_column": None,
    },
    "MSKCC_2020": {
        "primary_column": "AGE_AT_SURGERY",
        "fallback_days_column": None,
    },
    "NCI_2022": {
        "primary_column": "AGE",
        "fallback_days_column": None,
    },
    "ONCOSG_2020": {
        "primary_column": "AGE",
        "fallback_days_column": None,
    },
}


OUTPUT_AGE_AUDIT = (
    OUTPUT_DIR
    / "results"
    / "luad_age_harmonisation_patient_level_audit.csv"
)


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


def clean_raw_series(series):
    cleaned = series.astype("string").str.strip()

    return cleaned.mask(
        cleaned.str.upper().isin(MISSING_TOKENS)
    )


harmonised_age_frames = []


print("=" * 78)
print("STEP 4C-1 — PATIENT-LEVEL AGE HARMONISATION")
print("=" * 78)


for cohort, config in AGE_CONFIG.items():

    candidate_paths = STEP_1A_RESULTS[
        "cohort_candidates"
    ][cohort]["clinical_patient"]

    if len(candidate_paths) != 1:
        raise ValueError(
            f"{cohort}: expected one clinical-patient file, "
            f"found {len(candidate_paths)}."
        )

    source_file = candidate_paths[0]
    primary_column = config["primary_column"]
    fallback_column = config["fallback_days_column"]

    required_columns = ["PATIENT_ID", primary_column]

    if fallback_column:
        required_columns.append(fallback_column)

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} age loading failed: "
            f"{type(error).__name__}: {error}"
        )

    result = pd.DataFrame({
        "COHORT": cohort,
        "PATIENT_ID": source_df["PATIENT_ID"].astype("string").str.strip(),
    })

    primary_raw = clean_raw_series(
        source_df[primary_column]
    )

    primary_numeric = pd.to_numeric(
        primary_raw,
        errors="coerce",
    )

    result["age_original_value"] = primary_raw
    result["age_numeric"] = primary_numeric
    result["age_source_column"] = primary_column


    # --------------------------------------------------------
    # TCGA fallback: DAYS_TO_BIRTH
    # --------------------------------------------------------

    if fallback_column:

        fallback_raw = clean_raw_series(
            source_df[fallback_column]
        )

        fallback_numeric = pd.to_numeric(
            fallback_raw,
            errors="coerce",
        )

        available_fallback = fallback_numeric.dropna()

        invalid_day_values = available_fallback[
            (available_fallback >= 0)
            | (available_fallback.abs() < (10 * 365.25))
            | (available_fallback.abs() > (120 * 365.25))
        ]

        print(
            f"\n{cohort} invalid fallback day values: "
            f"{len(invalid_day_values)}"
        )

        if len(invalid_day_values) > 0:
            print(
                "Examples:",
                invalid_day_values.head(10).tolist(),
            )

            raise ValueError(
                "DAYS_TO_BIRTH contains values that cannot "
                "be safely converted."
            )

        use_fallback_mask = (
            primary_numeric.isna()
            & fallback_numeric.notna()
        )

        converted_years = (
            fallback_numeric.abs() / 365.25
        )

        result.loc[
            use_fallback_mask,
            "age_original_value",
        ] = fallback_raw[use_fallback_mask]

        result.loc[
            use_fallback_mask,
            "age_numeric",
        ] = converted_years[use_fallback_mask]

        result.loc[
            use_fallback_mask,
            "age_source_column",
        ] = fallback_column


    # --------------------------------------------------------
    # Flags and modelling field
    # --------------------------------------------------------

    numeric_age = pd.to_numeric(
        result["age_numeric"],
        errors="coerce",
    )

    result["age_missing_flag"] = (
        numeric_age.isna()
    ).astype(int)

    result["age_outlier_flag"] = (
        numeric_age.notna()
        & (
            (numeric_age < 18)
            | (numeric_age > 100)
        )
    ).astype(int)

    impossible_mask = (
        numeric_age.notna()
        & (
            (numeric_age < 0)
            | (numeric_age > 120)
        )
    )

    result["age_unusable_flag"] = (
        numeric_age.isna()
        | impossible_mask
        | result["age_outlier_flag"].eq(1)
    ).astype(int)

    result["age_for_model"] = numeric_age.where(
        result["age_unusable_flag"].eq(0),
        np.nan,
    )

    result["age_source_file"] = str(source_file)

    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "age_original_value",
            "age_numeric",
            "age_for_model",
            "age_missing_flag",
            "age_outlier_flag",
            "age_unusable_flag",
            "age_source_column",
            "age_source_file",
        ]
    ]

    duplicate_patient_count = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Rows: {len(result):,}")
    print(f"Duplicate patient rows: {duplicate_patient_count}")
    print(
        f"Numeric age coverage: "
        f"{result['age_numeric'].notna().sum():,}"
    )
    print(
        f"Usable age coverage: "
        f"{result['age_for_model'].notna().sum():,}"
    )
    print(
        f"Missing ages: "
        f"{result['age_missing_flag'].sum():,}"
    )
    print(
        f"Suspicious/outlier ages: "
        f"{result['age_outlier_flag'].sum():,}"
    )
    print(
        f"Unusable ages: "
        f"{result['age_unusable_flag'].sum():,}"
    )

    if result["age_numeric"].notna().any():
        print(
            f"Age range: "
            f"{result['age_numeric'].min():.2f} to "
            f"{result['age_numeric'].max():.2f}"
        )

    harmonised_age_frames.append(result)


patient_level_age_df = pd.concat(
    harmonised_age_frames,
    ignore_index=True,
)


# ------------------------------------------------------------
# Final validation and save
# ------------------------------------------------------------

duplicate_keys = patient_level_age_df.duplicated(
    subset=["COHORT", "PATIENT_ID"],
    keep=False,
)

print("\n" + "=" * 78)
print("COMBINED AGE AUDIT")
print("=" * 78)

print(f"Combined rows: {len(patient_level_age_df):,}")
print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_keys.sum())}"
)
print(
    f"Total numeric age coverage: "
    f"{patient_level_age_df['age_numeric'].notna().sum():,}"
)
print(
    f"Total usable age coverage: "
    f"{patient_level_age_df['age_for_model'].notna().sum():,}"
)
print(
    f"Total missing ages: "
    f"{patient_level_age_df['age_missing_flag'].sum():,}"
)
print(
    f"Total suspicious ages: "
    f"{patient_level_age_df['age_outlier_flag'].sum():,}"
)


patient_level_age_df.to_csv(
    OUTPUT_AGE_AUDIT,
    index=False,
)


print("\n" + "=" * 78)
print("STEP 4C-1 FINISHED")
print("=" * 78)

print(f"Saved audit: {OUTPUT_AGE_AUDIT}")
print(f"Saved file exists: {OUTPUT_AGE_AUDIT.exists()}")

print("\nMSKCC_2023 age is deferred to the next step.")
print("No age imputation was performed.")
print("No age scaling was performed.")
print("No final clinical dataset was created.")
```

### Step 4C-2 — Harmonise Sample-Level Age for MSKCC 2023

`MSKCC_2023` stores age in the clinical-sample file as `AGE_AT_DOS_BX`.

This step creates standard age fields for every sample while preserving:

- `PATIENT_ID`
- `SAMPLE_ID`
- original age value
- numeric age
- modelling-age field
- missing, outlier and unusable flags
- source column and source file

Multiple samples per patient will remain separate. Representative-sample selection will happen later during clinical record merging.

No age imputation or scaling will be performed.

```python
# ============================================================
# STEP 4C-2 — HARMONISE SAMPLE-LEVEL AGE FOR MSKCC_2023
# ============================================================

import pandas as pd
import numpy as np


SOURCE_FILE = (
    PROJECT_DIR
    / "mskcc_2023_data_clinical_sample.txt"
)

AGE_SOURCE_COLUMN = "AGE_AT_DOS_BX"

OUTPUT_AUDIT_FILE = (
    OUTPUT_DIR
    / "results"
    / "luad_age_harmonisation_mskcc_2023_sample_level_audit.csv"
)


print("=" * 78)
print("STEP 4C-2 — MSKCC_2023 SAMPLE-LEVEL AGE HARMONISATION")
print("=" * 78)

print(f"\nSource file: {SOURCE_FILE}")
print(f"Exists: {SOURCE_FILE.exists()}")

if not SOURCE_FILE.exists():
    raise FileNotFoundError(
        f"Source file was not found:\n{SOURCE_FILE}"
    )


# ------------------------------------------------------------
# 1. Load only required columns
# ------------------------------------------------------------

required_columns = [
    "PATIENT_ID",
    "SAMPLE_ID",
    AGE_SOURCE_COLUMN,
]

try:
    source_df = pd.read_csv(
        SOURCE_FILE,
        sep="\t",
        comment="#",
        usecols=required_columns,
        dtype=str,
        low_memory=False,
    )

except Exception as error:
    raise RuntimeError(
        f"Age loading failed: "
        f"{type(error).__name__}: {error}"
    )


# ------------------------------------------------------------
# 2. Standardise identifiers and raw age
# ------------------------------------------------------------

missing_tokens = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


result = pd.DataFrame()

result["COHORT"] = "MSKCC_2023"

result["PATIENT_ID"] = (
    source_df["PATIENT_ID"]
    .astype("string")
    .str.strip()
)

result["SAMPLE_ID"] = (
    source_df["SAMPLE_ID"]
    .astype("string")
    .str.strip()
)

raw_age = (
    source_df[AGE_SOURCE_COLUMN]
    .astype("string")
    .str.strip()
)

raw_age = raw_age.mask(
    raw_age.str.upper().isin(missing_tokens)
)


# ------------------------------------------------------------
# 3. Convert age to numeric years
# ------------------------------------------------------------

age_numeric = pd.to_numeric(
    raw_age,
    errors="coerce",
)

non_numeric_mask = (
    raw_age.notna()
    & age_numeric.isna()
)

non_numeric_values = sorted(
    raw_age.loc[non_numeric_mask]
    .dropna()
    .unique()
    .tolist()
)


# AGE_AT_DOS_BX values are already expressed in years.
impossible_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 0)
        | (age_numeric > 120)
    )
)

outlier_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 18)
        | (age_numeric > 100)
    )
    & ~impossible_mask
)


result["age_original_value"] = raw_age
result["age_numeric"] = age_numeric

result["age_missing_flag"] = (
    age_numeric.isna()
).astype(int)

result["age_outlier_flag"] = (
    outlier_mask
).astype(int)

result["age_unusable_flag"] = (
    age_numeric.isna()
    | impossible_mask
    | outlier_mask
).astype(int)

result["age_for_model"] = age_numeric.where(
    result["age_unusable_flag"].eq(0),
    np.nan,
)

result["age_source_column"] = AGE_SOURCE_COLUMN
result["age_source_file"] = str(SOURCE_FILE)


result = result[
    [
        "COHORT",
        "SAMPLE_ID",
        "PATIENT_ID",
        "age_original_value",
        "age_numeric",
        "age_for_model",
        "age_missing_flag",
        "age_outlier_flag",
        "age_unusable_flag",
        "age_source_column",
        "age_source_file",
    ]
]


# ------------------------------------------------------------
# 4. Identifier and multiple-sample audit
# ------------------------------------------------------------

duplicate_sample_keys = result.duplicated(
    subset=["COHORT", "SAMPLE_ID"],
    keep=False,
)

missing_patient_ids = (
    result["PATIENT_ID"].isna()
    | result["PATIENT_ID"].eq("")
)

missing_sample_ids = (
    result["SAMPLE_ID"].isna()
    | result["SAMPLE_ID"].eq("")
)


samples_per_patient = (
    result.groupby(
        ["COHORT", "PATIENT_ID"],
        dropna=False,
    )["SAMPLE_ID"]
    .nunique(dropna=True)
)

patients_with_multiple_samples = samples_per_patient[
    samples_per_patient > 1
]


numeric_ages_per_patient = (
    result.dropna(subset=["age_numeric"])
    .groupby(
        ["COHORT", "PATIENT_ID"]
    )["age_numeric"]
    .nunique()
)

patients_with_multiple_ages = numeric_ages_per_patient[
    numeric_ages_per_patient > 1
]


# ------------------------------------------------------------
# 5. Print audit
# ------------------------------------------------------------

print(f"\nTotal sample rows: {len(result):,}")

print(
    f"Unique patients: "
    f"{result['PATIENT_ID'].nunique(dropna=True):,}"
)

print(
    f"Unique samples: "
    f"{result['SAMPLE_ID'].nunique(dropna=True):,}"
)

print(
    f"Duplicate sample-key rows: "
    f"{int(duplicate_sample_keys.sum())}"
)

print(
    f"Missing patient IDs: "
    f"{int(missing_patient_ids.sum())}"
)

print(
    f"Missing sample IDs: "
    f"{int(missing_sample_ids.sum())}"
)

print(
    f"Numeric age coverage: "
    f"{result['age_numeric'].notna().sum():,}"
)

print(
    f"Usable age coverage: "
    f"{result['age_for_model'].notna().sum():,}"
)

print(
    f"Missing ages: "
    f"{result['age_missing_flag'].sum():,}"
)

print(
    f"Suspicious/outlier ages: "
    f"{result['age_outlier_flag'].sum():,}"
)

print(
    f"Unusable ages: "
    f"{result['age_unusable_flag'].sum():,}"
)

print(f"Non-numeric raw values: {non_numeric_values}")

if result["age_numeric"].notna().any():
    print(
        f"Numeric age range: "
        f"{result['age_numeric'].min():.2f} to "
        f"{result['age_numeric'].max():.2f}"
    )

print(
    f"Patients with multiple samples: "
    f"{len(patients_with_multiple_samples):,}"
)

print(
    f"Patients with multiple numeric age values: "
    f"{len(patients_with_multiple_ages):,}"
)


# ------------------------------------------------------------
# 6. Validation
# ------------------------------------------------------------

if duplicate_sample_keys.any():
    print("\nDuplicate sample-key examples:")
    print(
        result.loc[
            duplicate_sample_keys,
            ["COHORT", "SAMPLE_ID", "PATIENT_ID"],
        ].head(20).to_string(index=False)
    )

if impossible_mask.any():
    print("\nImpossible age examples:")
    print(
        result.loc[
            impossible_mask,
            [
                "PATIENT_ID",
                "SAMPLE_ID",
                "age_original_value",
                "age_numeric",
            ],
        ].head(20).to_string(index=False)
    )


# ------------------------------------------------------------
# 7. Save sample-level age audit
# ------------------------------------------------------------

result.to_csv(
    OUTPUT_AUDIT_FILE,
    index=False,
)

MSKCC_2023_SAMPLE_LEVEL_AGE_DF = result


print("\n" + "=" * 78)
print("STEP 4C-2 FINISHED")
print("=" * 78)

print(f"Saved audit: {OUTPUT_AUDIT_FILE}")
print(f"Saved file exists: {OUTPUT_AUDIT_FILE.exists()}")
print(f"Saved rows: {len(result):,}")

print("\nNo representative sample was selected.")
print("No age imputation was performed.")
print("No age scaling was performed.")
print("No final clinical dataset was created.")
```

### Step 4C-2 Correction — Fix MSKCC 2023 Cohort Assignment and Age Consistency Audit

The previous cell may have created an empty `COHORT` column before adding sample rows. This affected the patient-level grouping audit.

This correction recreates the MSKCC 2023 sample-level age table with a valid cohort value on every row and repeats the multiple-age audit.

No representative sample will be selected.

```python
# ============================================================
# STEP 4C-2 CORRECTION — MSKCC_2023 AGE TABLE
# ============================================================

import pandas as pd
import numpy as np


SOURCE_FILE = (
    PROJECT_DIR
    / "mskcc_2023_data_clinical_sample.txt"
)

OUTPUT_AUDIT_FILE = (
    OUTPUT_DIR
    / "results"
    / "luad_age_harmonisation_mskcc_2023_sample_level_audit.csv"
)

AGE_COLUMN = "AGE_AT_DOS_BX"

MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


print("=" * 78)
print("STEP 4C-2 CORRECTION — MSKCC_2023 AGE AUDIT")
print("=" * 78)


source_df = pd.read_csv(
    SOURCE_FILE,
    sep="\t",
    comment="#",
    usecols=["PATIENT_ID", "SAMPLE_ID", AGE_COLUMN],
    dtype=str,
    low_memory=False,
)


# Start with the source index so every row receives a cohort value.
result = pd.DataFrame(index=source_df.index)

result["COHORT"] = "MSKCC_2023"

result["PATIENT_ID"] = (
    source_df["PATIENT_ID"]
    .astype("string")
    .str.strip()
)

result["SAMPLE_ID"] = (
    source_df["SAMPLE_ID"]
    .astype("string")
    .str.strip()
)

raw_age = (
    source_df[AGE_COLUMN]
    .astype("string")
    .str.strip()
)

raw_age = raw_age.mask(
    raw_age.str.upper().isin(MISSING_TOKENS)
)

age_numeric = pd.to_numeric(
    raw_age,
    errors="coerce",
)


impossible_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 0)
        | (age_numeric > 120)
    )
)

outlier_mask = (
    age_numeric.notna()
    & (
        (age_numeric < 18)
        | (age_numeric > 100)
    )
    & ~impossible_mask
)


result["age_original_value"] = raw_age
result["age_numeric"] = age_numeric

result["age_missing_flag"] = (
    age_numeric.isna()
).astype(int)

result["age_outlier_flag"] = (
    outlier_mask
).astype(int)

result["age_unusable_flag"] = (
    age_numeric.isna()
    | impossible_mask
    | outlier_mask
).astype(int)

result["age_for_model"] = age_numeric.where(
    result["age_unusable_flag"].eq(0),
    np.nan,
)

result["age_source_column"] = AGE_COLUMN
result["age_source_file"] = str(SOURCE_FILE)


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

missing_cohort_values = int(
    result["COHORT"].isna().sum()
)

duplicate_sample_rows = int(
    result.duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    ).sum()
)

samples_per_patient = (
    result.groupby(
        ["COHORT", "PATIENT_ID"],
        dropna=False,
    )["SAMPLE_ID"]
    .nunique()
)

patients_with_multiple_samples = samples_per_patient[
    samples_per_patient > 1
]


ages_per_patient = (
    result.dropna(subset=["age_numeric"])
    .groupby(
        ["COHORT", "PATIENT_ID"],
        dropna=False,
    )["age_numeric"]
    .nunique()
)

patients_with_multiple_ages = ages_per_patient[
    ages_per_patient > 1
]


print(f"\nTotal rows: {len(result):,}")
print(f"Missing COHORT values: {missing_cohort_values}")
print(f"Unique patients: {result['PATIENT_ID'].nunique():,}")
print(f"Unique samples: {result['SAMPLE_ID'].nunique():,}")
print(f"Duplicate sample rows: {duplicate_sample_rows}")

print(
    f"Patients with multiple samples: "
    f"{len(patients_with_multiple_samples):,}"
)

print(
    f"Patients with multiple numeric ages: "
    f"{len(patients_with_multiple_ages):,}"
)

print(
    f"Numeric age coverage: "
    f"{result['age_numeric'].notna().sum():,}"
)

print(
    f"Missing ages: "
    f"{result['age_missing_flag'].sum():,}"
)


if missing_cohort_values != 0:
    raise ValueError(
        "COHORT still contains missing values."
    )

if duplicate_sample_rows != 0:
    raise ValueError(
        "Duplicate cohort-sample keys were found."
    )


result.to_csv(
    OUTPUT_AUDIT_FILE,
    index=False,
)

MSKCC_2023_SAMPLE_LEVEL_AGE_DF = result


print("\n" + "=" * 78)
print("STEP 4C-2 CORRECTION FINISHED")
print("=" * 78)

print(f"Corrected file saved: {OUTPUT_AUDIT_FILE}")
print(f"Saved file exists: {OUTPUT_AUDIT_FILE.exists()}")

print("\nNo representative sample was selected.")
print("No age imputation or scaling was performed.")
```

### Step 4D — Consolidate the Age Harmonisation Audit

This step creates one age-coverage summary for all seven cohorts.

For the first six cohorts, age is already patient-level.

For `MSKCC_2023`, coverage will be calculated at both sample level and unique-patient level without selecting a representative sample.

Is step mein:

- total and usable age coverage report hogi
- missing and suspicious age counts report honge
- cohort-wise summary save hogi
- no age imputation or scaling will be performed

```python
# ============================================================
# STEP 4D — CONSOLIDATE AGE HARMONISATION AUDIT
# ============================================================

import pandas as pd


if "patient_level_age_df" not in globals():
    raise RuntimeError(
        "patient_level_age_df was not found. "
        "Please run Step 4C-1 first."
    )

if "MSKCC_2023_SAMPLE_LEVEL_AGE_DF" not in globals():
    raise RuntimeError(
        "MSKCC_2023_SAMPLE_LEVEL_AGE_DF was not found. "
        "Please run the corrected Step 4C-2 cell first."
    )


OUTPUT_AGE_SUMMARY = (
    OUTPUT_DIR
    / "results"
    / "luad_age_harmonisation_summary.csv"
)


print("=" * 78)
print("STEP 4D — AGE HARMONISATION SUMMARY")
print("=" * 78)


summary_rows = []


# ------------------------------------------------------------
# 1. Six cohorts with patient-level age
# ------------------------------------------------------------

for cohort, cohort_df in patient_level_age_df.groupby(
    "COHORT",
    sort=False,
):

    total_patients = cohort_df["PATIENT_ID"].nunique(
        dropna=True
    )

    numeric_patients = cohort_df.loc[
        cohort_df["age_numeric"].notna(),
        "PATIENT_ID",
    ].nunique()

    usable_patients = cohort_df.loc[
        cohort_df["age_for_model"].notna(),
        "PATIENT_ID",
    ].nunique()

    missing_patients = cohort_df.loc[
        cohort_df["age_missing_flag"].eq(1),
        "PATIENT_ID",
    ].nunique()

    outlier_patients = cohort_df.loc[
        cohort_df["age_outlier_flag"].eq(1),
        "PATIENT_ID",
    ].nunique()

    unusable_patients = cohort_df.loc[
        cohort_df["age_unusable_flag"].eq(1),
        "PATIENT_ID",
    ].nunique()

    usable_values = pd.to_numeric(
        cohort_df["age_for_model"],
        errors="coerce",
    ).dropna()

    summary_rows.append(
        {
            "COHORT": cohort,
            "AGE_SOURCE_LEVEL": "patient",
            "TOTAL_SOURCE_ROWS": len(cohort_df),
            "TOTAL_UNIQUE_PATIENTS": total_patients,
            "PATIENTS_WITH_NUMERIC_AGE": numeric_patients,
            "PATIENTS_WITH_USABLE_AGE": usable_patients,
            "PATIENTS_WITH_MISSING_AGE": missing_patients,
            "PATIENTS_WITH_OUTLIER_AGE": outlier_patients,
            "PATIENTS_WITH_UNUSABLE_AGE": unusable_patients,
            "NUMERIC_AGE_COVERAGE_PERCENT": round(
                100 * numeric_patients / total_patients,
                2,
            ) if total_patients else 0.0,
            "USABLE_AGE_COVERAGE_PERCENT": round(
                100 * usable_patients / total_patients,
                2,
            ) if total_patients else 0.0,
            "USABLE_AGE_MIN": (
                float(usable_values.min())
                if not usable_values.empty
                else None
            ),
            "USABLE_AGE_MAX": (
                float(usable_values.max())
                if not usable_values.empty
                else None
            ),
            "USABLE_AGE_MEDIAN": (
                float(usable_values.median())
                if not usable_values.empty
                else None
            ),
            "PATIENTS_WITH_MULTIPLE_SAMPLES": 0,
            "PATIENTS_WITH_MULTIPLE_AGE_VALUES": 0,
            "REPRESENTATIVE_SAMPLE_SELECTED": 0,
        }
    )


# ------------------------------------------------------------
# 2. MSKCC 2023 sample-level age
# ------------------------------------------------------------

msk_df = MSKCC_2023_SAMPLE_LEVEL_AGE_DF.copy()

total_patients = msk_df["PATIENT_ID"].nunique(
    dropna=True
)

patients_with_numeric_age = (
    msk_df.loc[
        msk_df["age_numeric"].notna(),
        "PATIENT_ID",
    ]
    .nunique()
)

patients_with_usable_age = (
    msk_df.loc[
        msk_df["age_for_model"].notna(),
        "PATIENT_ID",
    ]
    .nunique()
)

patients_with_no_numeric_age = (
    total_patients - patients_with_numeric_age
)

patients_with_any_outlier = (
    msk_df.loc[
        msk_df["age_outlier_flag"].eq(1),
        "PATIENT_ID",
    ]
    .nunique()
)

patients_with_no_usable_age = (
    total_patients - patients_with_usable_age
)

samples_per_patient = (
    msk_df.groupby("PATIENT_ID")["SAMPLE_ID"]
    .nunique()
)

patients_with_multiple_samples = int(
    (samples_per_patient > 1).sum()
)

ages_per_patient = (
    msk_df.dropna(subset=["age_numeric"])
    .groupby("PATIENT_ID")["age_numeric"]
    .nunique()
)

patients_with_multiple_ages = int(
    (ages_per_patient > 1).sum()
)

usable_values = pd.to_numeric(
    msk_df["age_for_model"],
    errors="coerce",
).dropna()


summary_rows.append(
    {
        "COHORT": "MSKCC_2023",
        "AGE_SOURCE_LEVEL": "sample",
        "TOTAL_SOURCE_ROWS": len(msk_df),
        "TOTAL_UNIQUE_PATIENTS": total_patients,
        "PATIENTS_WITH_NUMERIC_AGE": patients_with_numeric_age,
        "PATIENTS_WITH_USABLE_AGE": patients_with_usable_age,
        "PATIENTS_WITH_MISSING_AGE": patients_with_no_numeric_age,
        "PATIENTS_WITH_OUTLIER_AGE": patients_with_any_outlier,
        "PATIENTS_WITH_UNUSABLE_AGE": patients_with_no_usable_age,
        "NUMERIC_AGE_COVERAGE_PERCENT": round(
            100 * patients_with_numeric_age / total_patients,
            2,
        ),
        "USABLE_AGE_COVERAGE_PERCENT": round(
            100 * patients_with_usable_age / total_patients,
            2,
        ),
        "USABLE_AGE_MIN": (
            float(usable_values.min())
            if not usable_values.empty
            else None
        ),
        "USABLE_AGE_MAX": (
            float(usable_values.max())
            if not usable_values.empty
            else None
        ),
        "USABLE_AGE_MEDIAN": (
            float(usable_values.median())
            if not usable_values.empty
            else None
        ),
        "PATIENTS_WITH_MULTIPLE_SAMPLES": (
            patients_with_multiple_samples
        ),
        "PATIENTS_WITH_MULTIPLE_AGE_VALUES": (
            patients_with_multiple_ages
        ),
        "REPRESENTATIVE_SAMPLE_SELECTED": 0,
    }
)


# ------------------------------------------------------------
# 3. Create and validate summary
# ------------------------------------------------------------

age_summary_df = pd.DataFrame(summary_rows)

expected_cohorts = {
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
}

found_cohorts = set(age_summary_df["COHORT"])

missing_cohorts = sorted(
    expected_cohorts - found_cohorts
)

duplicate_cohorts = age_summary_df.loc[
    age_summary_df["COHORT"].duplicated(keep=False),
    "COHORT",
].tolist()


print(f"\nSummary rows: {len(age_summary_df)}")
print(f"Missing cohorts: {missing_cohorts}")
print(f"Duplicate cohort rows: {duplicate_cohorts}")

if missing_cohorts:
    raise ValueError(
        f"Age summary is missing cohorts: {missing_cohorts}"
    )

if duplicate_cohorts:
    raise ValueError(
        f"Duplicate cohort summaries found: {duplicate_cohorts}"
    )


# ------------------------------------------------------------
# 4. Save and print
# ------------------------------------------------------------

age_summary_df.to_csv(
    OUTPUT_AGE_SUMMARY,
    index=False,
)


print("\nAge harmonisation summary:")

print(
    age_summary_df[
        [
            "COHORT",
            "AGE_SOURCE_LEVEL",
            "TOTAL_UNIQUE_PATIENTS",
            "PATIENTS_WITH_USABLE_AGE",
            "PATIENTS_WITH_MISSING_AGE",
            "PATIENTS_WITH_OUTLIER_AGE",
            "USABLE_AGE_COVERAGE_PERCENT",
            "PATIENTS_WITH_MULTIPLE_AGE_VALUES",
        ]
    ].to_string(index=False)
)


print("\n" + "=" * 78)
print("STEP 4D FINISHED")
print("=" * 78)

print(f"Saved summary: {OUTPUT_AGE_SUMMARY}")
print(f"Saved file exists: {OUTPUT_AGE_SUMMARY.exists()}")

print("\nNo representative sample was selected.")
print("No age imputation was performed.")
print("No age scaling was performed.")
print("No final clinical dataset was created.")
```

## Step 5A-1 — Audit Raw Sex Values for TCGA Atlas 2018 and LUNG MSK 2017

This step inspects the original sex values in the first two cohorts.

It will report:

- all unique raw sex values
- value counts
- missing or unknown values
- whether sex exists in patient and/or sample files

No sex encoding or final harmonisation will be performed yet.

```python
# ============================================================
# STEP 5A-1 — RAW SEX AUDIT
# TCGA_ATLAS_2018 AND LUNG_MSK_2017
# ============================================================

import pandas as pd


SEX_AUDIT_CONFIG = {
    "TCGA_Atlas_2018": {
        "patient_file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "patient_sex_column": "SEX",
        "sample_file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_sample.txt",
        "sample_sex_column": None,
    },
    "LUNG_MSK_2017": {
        "patient_file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "patient_sex_column": "SEX",
        "sample_file": PROJECT_DIR / "lung_msk_2017_data_clinical_sample.txt",
        "sample_sex_column": None,
    },
}


MISSING_SEX_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


STEP_5A1_SEX_AUDIT = {}


print("=" * 78)
print("STEP 5A-1 — RAW SEX VALUE AUDIT")
print("=" * 78)


for cohort, config in SEX_AUDIT_CONFIG.items():

    patient_file = config["patient_file"]
    patient_sex_column = config["patient_sex_column"]
    sample_sex_column = config["sample_sex_column"]

    print("\n" + "=" * 78)
    print(f"COHORT: {cohort}")
    print("=" * 78)

    print(f"Patient file: {patient_file}")
    print(f"Exists: {patient_file.exists()}")

    if not patient_file.exists():
        print("ERROR: Patient clinical file is missing.")
        continue

    try:
        patient_df = pd.read_csv(
            patient_file,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", patient_sex_column],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        print(
            f"ERROR loading sex values: "
            f"{type(error).__name__}: {error}"
        )
        continue

    raw_sex = (
        patient_df[patient_sex_column]
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        raw_sex.isna()
        | raw_sex.str.upper().isin(MISSING_SEX_TOKENS)
    )

    known_values = raw_sex.mask(missing_mask)

    value_counts = (
        known_values
        .value_counts(dropna=False)
    )

    unique_raw_values = sorted(
        known_values
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    print(f"\nPatient rows: {len(patient_df):,}")
    print(f"Sex source column: {patient_sex_column}")
    print(f"Unique known raw values: {unique_raw_values}")
    print(f"Missing/unknown values: {int(missing_mask.sum())}")

    print("\nRaw value counts:")
    print(value_counts.to_string())

    print(
        "\nSample-level sex column:",
        sample_sex_column if sample_sex_column else "None",
    )

    STEP_5A1_SEX_AUDIT[cohort] = {
        "patient_rows": len(patient_df),
        "patient_source_column": patient_sex_column,
        "sample_source_column": sample_sex_column,
        "unique_raw_values": unique_raw_values,
        "missing_count": int(missing_mask.sum()),
    }


print("\n" + "=" * 78)
print("STEP 5A-1 FINISHED")
print("=" * 78)

print("No sex values were harmonised.")
print("No sex encoding was created.")
print("No final clinical dataset was created.")
```

### Step 5A-2 — Audit Raw Sex Values for MSK NPJPO 2021

This step inspects the original `SEX` values in the `MSK_NPJPO_2021` clinical-patient file.

It will report unique values, counts and missing values.

Abhi sex harmonisation, encoding ya clinical merge nahi hogi.

## Step 5B — Harmonise Sex for All Seven LUAD Cohorts

This step standardises sex directly from each cohort's original clinical-patient file.

Mapping:

- Female / F → Female, encoded as 0
- Male / M → Male, encoded as 1
- missing or unclear values → Unknown, encoded as missing

All seven sample-level clinical files lack a sex column, so the patient-level `SEX` field is the deterministic source.

Is step mein harmonised patient-level sex table `final_luad_results/data` mein save hogi. Clinical merge aur model training abhi nahi hogi.

```python
# ============================================================
# STEP 5B — HARMONISE SEX FOR ALL SEVEN COHORTS
# ============================================================

import pandas as pd


SEX_CONFIG = {
    "TCGA_Atlas_2018": (
        PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt"
    ),
    "LUNG_MSK_2017": (
        PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt"
    ),
    "MSK_NPJPO_2021": (
        PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt"
    ),
    "MSKCC_2020": (
        PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt"
    ),
    "MSKCC_2023": (
        PROJECT_DIR / "mskcc_2023_data_clinical_patient.txt"
    ),
    "NCI_2022": (
        PROJECT_DIR / "nci_2022_data_clinical_patient.txt"
    ),
    "ONCOSG_2020": (
        PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt"
    ),
}


OUTPUT_SEX_DATA = (
    OUTPUT_DIR
    / "data"
    / "luad_sex_harmonisation_patient_level.csv"
)

OUTPUT_SEX_SUMMARY = (
    OUTPUT_DIR
    / "results"
    / "luad_sex_harmonisation_summary.csv"
)


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


def standardise_sex(raw_value):
    """
    Standardise one original sex value.
    """
    if pd.isna(raw_value):
        return "Unknown"

    value = str(raw_value).strip().upper()

    if value in MISSING_TOKENS:
        return "Unknown"

    if value in {"F", "FEMALE"}:
        return "Female"

    if value in {"M", "MALE"}:
        return "Male"

    return "Unknown"


print("=" * 78)
print("STEP 5B — SEX HARMONISATION")
print("=" * 78)


harmonised_frames = []
summary_rows = []
unrecognised_values = {}


for cohort, source_file in SEX_CONFIG.items():

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source file exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} source file was not found:\n{source_file}"
        )

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", "SEX"],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} sex loading failed: "
            f"{type(error).__name__}: {error}"
        )

    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    result["sex_original_value"] = (
        source_df["SEX"]
        .astype("string")
        .str.strip()
    )

    result["SEX"] = result[
        "sex_original_value"
    ].apply(standardise_sex)

    result["sex_encoded"] = result["SEX"].map(
        {
            "Female": 0,
            "Male": 1,
        }
    ).astype("Int64")

    result["sex_missing_flag"] = (
        result["SEX"].eq("Unknown")
    ).astype(int)

    result["sex_source_column"] = "SEX"
    result["sex_source_file"] = str(source_file)

    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "sex_original_value",
            "SEX",
            "sex_encoded",
            "sex_missing_flag",
            "sex_source_column",
            "sex_source_file",
        ]
    ]


    # --------------------------------------------------------
    # Audit raw and unrecognised values
    # --------------------------------------------------------

    raw_upper = (
        result["sex_original_value"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()
    )

    recognised_values = {
        "",
        "F",
        "FEMALE",
        "M",
        "MALE",
    } | MISSING_TOKENS

    unrecognised = sorted(
        set(
            raw_upper[
                ~raw_upper.isin(recognised_values)
            ].tolist()
        )
    )

    unrecognised_values[cohort] = unrecognised

    duplicate_patient_rows = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    female_count = int(
        result["SEX"].eq("Female").sum()
    )

    male_count = int(
        result["SEX"].eq("Male").sum()
    )

    unknown_count = int(
        result["SEX"].eq("Unknown").sum()
    )

    known_count = female_count + male_count

    coverage_percent = (
        round(100 * known_count / len(result), 2)
        if len(result) > 0
        else 0.0
    )

    print(f"Patient rows: {len(result):,}")
    print(f"Female: {female_count:,}")
    print(f"Male: {male_count:,}")
    print(f"Unknown: {unknown_count:,}")
    print(f"Known-sex coverage: {coverage_percent:.2f}%")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")
    print(f"Unrecognised raw values: {unrecognised}")

    summary_rows.append(
        {
            "COHORT": cohort,
            "TOTAL_PATIENTS": len(result),
            "FEMALE_COUNT": female_count,
            "MALE_COUNT": male_count,
            "UNKNOWN_COUNT": unknown_count,
            "KNOWN_SEX_COVERAGE_PERCENT": coverage_percent,
            "DUPLICATE_PATIENT_ROWS": duplicate_patient_rows,
            "UNRECOGNISED_RAW_VALUES": " | ".join(unrecognised),
            "SEX_PRIORITY_RULE": "Patient-level SEX",
        }
    )

    harmonised_frames.append(result)


# ------------------------------------------------------------
# Combine all cohorts
# ------------------------------------------------------------

patient_level_sex_df = pd.concat(
    harmonised_frames,
    ignore_index=True,
)

sex_summary_df = pd.DataFrame(summary_rows)


# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

duplicate_keys = patient_level_sex_df.duplicated(
    subset=["COHORT", "PATIENT_ID"],
    keep=False,
)

invalid_standard_values = sorted(
    set(patient_level_sex_df["SEX"])
    - {"Female", "Male", "Unknown"}
)

invalid_encoded_values = sorted(
    patient_level_sex_df["sex_encoded"]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

missing_cohort_values = int(
    patient_level_sex_df["COHORT"].isna().sum()
)

missing_patient_values = int(
    patient_level_sex_df["PATIENT_ID"].isna().sum()
)


print("\n" + "=" * 78)
print("COMBINED SEX VALIDATION")
print("=" * 78)

print(f"Combined patient rows: {len(patient_level_sex_df):,}")
print(f"Duplicate cohort-patient rows: {int(duplicate_keys.sum())}")
print(f"Missing cohort values: {missing_cohort_values}")
print(f"Missing patient IDs: {missing_patient_values}")
print(f"Invalid standard sex values: {invalid_standard_values}")
print(f"Observed encoded values: {invalid_encoded_values}")

print(
    f"Overall known-sex coverage: "
    f"{patient_level_sex_df['sex_encoded'].notna().mean() * 100:.2f}%"
)


if duplicate_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were found."
    )

if invalid_standard_values:
    raise ValueError(
        f"Invalid standard sex values found: "
        f"{invalid_standard_values}"
    )

if not set(invalid_encoded_values).issubset({0, 1}):
    raise ValueError(
        f"Invalid sex encodings found: "
        f"{invalid_encoded_values}"
    )


# ------------------------------------------------------------
# Save actual harmonised data and audit summary
# ------------------------------------------------------------

patient_level_sex_df.to_csv(
    OUTPUT_SEX_DATA,
    index=False,
)

sex_summary_df.to_csv(
    OUTPUT_SEX_SUMMARY,
    index=False,
)


print("\n" + "=" * 78)
print("STEP 5B FINISHED")
print("=" * 78)

print(f"Harmonised sex data: {OUTPUT_SEX_DATA}")
print(f"Data file exists: {OUTPUT_SEX_DATA.exists()}")

print(f"\nSex audit summary: {OUTPUT_SEX_SUMMARY}")
print(f"Summary file exists: {OUTPUT_SEX_SUMMARY.exists()}")

print("\nNo unknown patient was assigned Female or Male.")
print("No patient/sample clinical merge was performed.")
print("No final clinical dataset was created.")
```

### Step 6A-4 — Audit Raw Smoking Values for MSKCC 2023

This step inspects every original value from `CIGARETTE_HX` in the `MSKCC_2023` clinical-patient file.

### Step 6B-1 — Audit Smoking Pack-Years Across Available Cohorts

Smoking pack-years are available in:

- `MSK_NPJPO_2021`
- `MSKCC_2020`
- `ONCOSG_2020`

This step checks numeric coverage, missing values, non-numeric entries, ranges, negative values and unusually high values.

No pack-year value will be corrected, removed or imputed.

```python
# ============================================================
# STEP 6B-1 — SMOKING PACK-YEARS AUDIT
# ============================================================

import pandas as pd


PACK_YEAR_CONFIG = {
    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "column": "SMOKING_PACK_YEARS",
    },
    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt",
        "column": "SMOKING_PACK_YEARS",
    },
    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "column": "SMOKING_PACK_YEARS",
    },
}


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


STEP_6B1_PACK_YEAR_AUDIT = {}


print("=" * 78)
print("STEP 6B-1 — SMOKING PACK-YEARS AUDIT")
print("=" * 78)


for cohort, config in PACK_YEAR_CONFIG.items():

    source_file = config["file"]
    pack_column = config["column"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"File exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} file was not found:\n{source_file}"
        )

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", pack_column],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} pack-year loading failed: "
            f"{type(error).__name__}: {error}"
        )

    raw_pack_years = (
        source_df[pack_column]
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        raw_pack_years.isna()
        | raw_pack_years.str.upper().isin(MISSING_TOKENS)
    )

    cleaned_raw = raw_pack_years.mask(missing_mask)

    numeric_pack_years = pd.to_numeric(
        cleaned_raw,
        errors="coerce",
    )

    non_numeric_mask = (
        cleaned_raw.notna()
        & numeric_pack_years.isna()
    )

    non_numeric_values = sorted(
        cleaned_raw.loc[non_numeric_mask]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    negative_count = int(
        (numeric_pack_years < 0).sum()
    )

    above_200_count = int(
        (numeric_pack_years > 200).sum()
    )

    zero_count = int(
        numeric_pack_years.eq(0).sum()
    )

    duplicate_patient_rows = int(
        source_df.duplicated(
            subset=["PATIENT_ID"],
            keep=False,
        ).sum()
    )

    print(f"Patient rows: {len(source_df):,}")
    print(
        f"Numeric pack-year values: "
        f"{numeric_pack_years.notna().sum():,}"
    )
    print(
        f"Missing pack-year values: "
        f"{int(missing_mask.sum()):,}"
    )
    print(
        f"Non-numeric raw values: "
        f"{non_numeric_values}"
    )
    print(f"Zero values: {zero_count:,}")
    print(f"Negative values: {negative_count:,}")
    print(f"Values above 200: {above_200_count:,}")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")

    if numeric_pack_years.notna().any():
        print(
            f"Range: "
            f"{numeric_pack_years.min():.2f} to "
            f"{numeric_pack_years.max():.2f}"
        )
        print(
            f"Median: "
            f"{numeric_pack_years.median():.2f}"
        )

    STEP_6B1_PACK_YEAR_AUDIT[cohort] = {
        "source_file": str(source_file),
        "source_column": pack_column,
        "total_rows": len(source_df),
        "numeric_count": int(numeric_pack_years.notna().sum()),
        "missing_count": int(missing_mask.sum()),
        "non_numeric_values": non_numeric_values,
        "zero_count": zero_count,
        "negative_count": negative_count,
        "above_200_count": above_200_count,
        "duplicate_patient_rows": duplicate_patient_rows,
    }


print("\n" + "=" * 78)
print("STEP 6B-1 FINISHED")
print("=" * 78)

print("No pack-year value was corrected or removed.")
print("No pack-year value was imputed.")
print("No final smoking dataset was created.")
```

### Step 6B-2 — Audit Smoking Status and Pack-Year Consistency

This step checks whether smoking-status categories agree with available pack-year values.

It will identify:

- never smokers with positive pack-years
- ever smokers with zero pack-years
- known smoking status with missing pack-years
- missing smoking status with numeric pack-years
- pack-year values above 200

The original values will remain unchanged. No correction, exclusion or imputation will be performed.

```python
# ============================================================
# STEP 6B-2 — SMOKING STATUS AND PACK-YEAR CONSISTENCY AUDIT
# ============================================================

import pandas as pd


CONSISTENCY_CONFIG = {
    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "status_column": "SMOKING_STATUS",
        "pack_column": "SMOKING_PACK_YEARS",
        "never_values": {"NEVER SMOKER"},
        "ever_values": {"EVER SMOKER"},
    },
    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt",
        "status_column": "SMOKING_HISTORY",
        "pack_column": "SMOKING_PACK_YEARS",
        "never_values": {"NEVER SMOKER"},
        "ever_values": {"EVER SMOKER"},
    },
    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "status_column": "SMOKING_STATUS",
        "pack_column": "SMOKING_PACK_YEARS",
        "never_values": {"NO"},
        "ever_values": {"YES"},
    },
}


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


STEP_6B2_SMOKING_CONSISTENCY_AUDIT = {}


print("=" * 78)
print("STEP 6B-2 — SMOKING STATUS AND PACK-YEAR CONSISTENCY")
print("=" * 78)


for cohort, config in CONSISTENCY_CONFIG.items():

    source_file = config["file"]
    status_column = config["status_column"]
    pack_column = config["pack_column"]

    source_df = pd.read_csv(
        source_file,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", status_column, pack_column],
        dtype=str,
        low_memory=False,
    )

    audit_df = pd.DataFrame(index=source_df.index)

    audit_df["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    raw_status = (
        source_df[status_column]
        .astype("string")
        .str.strip()
    )

    status_upper = raw_status.str.upper()

    status_missing = (
        raw_status.isna()
        | status_upper.isin(MISSING_TOKENS)
    )

    raw_pack = (
        source_df[pack_column]
        .astype("string")
        .str.strip()
    )

    pack_missing = (
        raw_pack.isna()
        | raw_pack.str.upper().isin(MISSING_TOKENS)
    )

    pack_numeric = pd.to_numeric(
        raw_pack.mask(pack_missing),
        errors="coerce",
    )

    never_mask = status_upper.isin(
        config["never_values"]
    )

    ever_mask = status_upper.isin(
        config["ever_values"]
    )

    never_with_positive = (
        never_mask
        & pack_numeric.gt(0)
    )

    ever_with_zero = (
        ever_mask
        & pack_numeric.eq(0)
    )

    known_status_missing_pack = (
        (never_mask | ever_mask)
        & pack_numeric.isna()
    )

    missing_status_numeric_pack = (
        status_missing
        & pack_numeric.notna()
    )

    extreme_above_200 = (
        pack_numeric.gt(200)
    )

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Patient rows: {len(audit_df):,}")

    print(
        f"Never smokers with positive pack-years: "
        f"{int(never_with_positive.sum())}"
    )

    print(
        f"Ever smokers with zero pack-years: "
        f"{int(ever_with_zero.sum())}"
    )

    print(
        f"Known status with missing pack-years: "
        f"{int(known_status_missing_pack.sum())}"
    )

    print(
        f"Missing status with numeric pack-years: "
        f"{int(missing_status_numeric_pack.sum())}"
    )

    print(
        f"Pack-year values above 200: "
        f"{int(extreme_above_200.sum())}"
    )


    # Print only exceptional rows
    exception_mask = (
        never_with_positive
        | ever_with_zero
        | missing_status_numeric_pack
        | extreme_above_200
    )

    if exception_mask.any():

        exceptions = pd.DataFrame({
            "PATIENT_ID": audit_df.loc[
                exception_mask,
                "PATIENT_ID"
            ],
            "smoking_status_raw": raw_status.loc[
                exception_mask
            ],
            "pack_years_raw": raw_pack.loc[
                exception_mask
            ],
            "pack_years_numeric": pack_numeric.loc[
                exception_mask
            ],
            "never_with_positive_pack": (
                never_with_positive.loc[exception_mask]
                .astype(int)
            ),
            "ever_with_zero_pack": (
                ever_with_zero.loc[exception_mask]
                .astype(int)
            ),
            "missing_status_with_numeric_pack": (
                missing_status_numeric_pack.loc[exception_mask]
                .astype(int)
            ),
            "pack_years_above_200": (
                extreme_above_200.loc[exception_mask]
                .astype(int)
            ),
        })

        print("\nExceptional rows:")
        print(
            exceptions.head(15).to_string(index=False)
        )

    STEP_6B2_SMOKING_CONSISTENCY_AUDIT[cohort] = {
        "total_rows": len(audit_df),
        "never_with_positive_pack": int(
            never_with_positive.sum()
        ),
        "ever_with_zero_pack": int(
            ever_with_zero.sum()
        ),
        "known_status_missing_pack": int(
            known_status_missing_pack.sum()
        ),
        "missing_status_numeric_pack": int(
            missing_status_numeric_pack.sum()
        ),
        "pack_years_above_200": int(
            extreme_above_200.sum()
        ),
    }


print("\n" + "=" * 78)
print("STEP 6B-2 FINISHED")
print("=" * 78)

print("No smoking or pack-year value was changed.")
print("No extreme value was automatically removed.")
print("No missing value was imputed.")
```

## Step 6C-1 — Harmonise Active Smoking Status Across All Seven Cohorts

This step standardises active smoking status as:

- `Never` → encoded as 0
- `Ever` → encoded as 1
- `Unknown` → encoded as missing

Missing smoking information will not be interpreted as `Never`.

For cohorts without active-smoking data, smoking status will remain `Unknown`.

Where a reported `Never` smoking status conflicts with a positive pack-year value, the original values will be preserved, an inconsistency flag will be created, and the standard smoking status will be set to `Unknown`.

Passive smoking and pack-year modelling fields remain separate.

```python
# ============================================================
# STEP 6C-1 — HARMONISE ACTIVE SMOKING STATUS
# CORRECTED FULL READY-TO-RUN VERSION
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# 1. Check required notebook variables
# ------------------------------------------------------------

if "PROJECT_DIR" not in globals():
    raise RuntimeError(
        "PROJECT_DIR was not found. Please run the initial setup cell first."
    )

if "OUTPUT_DIR" not in globals():
    raise RuntimeError(
        "OUTPUT_DIR was not found. Please run the initial setup cell first."
    )


# ------------------------------------------------------------
# 2. Cohort-specific smoking configuration
# ------------------------------------------------------------

SMOKING_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "status_column": None,
        "pack_column": None,
    },

    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "status_column": "SMOKING_HISTORY",
        "pack_column": None,
    },

    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "status_column": "SMOKING_STATUS",
        "pack_column": "SMOKING_PACK_YEARS",
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt",
        "status_column": "SMOKING_HISTORY",
        "pack_column": "SMOKING_PACK_YEARS",
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_patient.txt",
        "status_column": "CIGARETTE_HX",
        "pack_column": None,
    },

    "NCI_2022": {
        "file": PROJECT_DIR / "nci_2022_data_clinical_patient.txt",
        "status_column": None,
        "pack_column": None,
    },

    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "status_column": "SMOKING_STATUS",
        "pack_column": "SMOKING_PACK_YEARS",
    },
}


# ------------------------------------------------------------
# 3. Output paths
# ------------------------------------------------------------

DATA_OUTPUT_DIR = OUTPUT_DIR / "data"
RESULTS_OUTPUT_DIR = OUTPUT_DIR / "results"

DATA_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


OUTPUT_SMOKING_DATA = (
    DATA_OUTPUT_DIR
    / "luad_active_smoking_harmonisation_patient_level.csv"
)

OUTPUT_SMOKING_SUMMARY = (
    RESULTS_OUTPUT_DIR
    / "luad_active_smoking_harmonisation_summary.csv"
)


# ------------------------------------------------------------
# 4. Standardisation rules
# ------------------------------------------------------------

MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


NEVER_VALUES = {
    "NEVER",
    "NEVER SMOKER",
    "NO",
}


EVER_VALUES = {
    "EVER",
    "EVER SMOKER",
    "YES",
    "CURRENT",
    "CURRENT SMOKER",
    "CURRENT HEAVY",
    "CURRENT LIGHT",
    "FORMER",
    "FORMER SMOKER",
    "FORMER HEAVY",
    "FORMER LIGHT",
}


def clean_raw_values(series):
    """
    Strip whitespace and convert known missing tokens to pandas NA.
    """

    cleaned = (
        series
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        cleaned.isna()
        | cleaned.str.upper().isin(MISSING_TOKENS)
    )

    return cleaned.mask(missing_mask)


def map_smoking_status(raw_value):
    """
    Convert an original cohort-specific smoking value to:
    Never, Ever or Unknown.
    """

    if pd.isna(raw_value):
        return "Unknown"

    value = str(raw_value).strip().upper()

    if value in NEVER_VALUES:
        return "Never"

    if value in EVER_VALUES:
        return "Ever"

    return "Unknown"


# ------------------------------------------------------------
# 5. Harmonise each cohort
# ------------------------------------------------------------

print("=" * 78)
print("STEP 6C-1 — ACTIVE SMOKING HARMONISATION")
print("=" * 78)


harmonised_frames = []
summary_rows = []


for cohort, config in SMOKING_CONFIG.items():

    source_file = config["file"]
    status_column = config["status_column"]
    pack_column = config["pack_column"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source file: {source_file}")
    print(f"File exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-patient file was not found:\n"
            f"{source_file}"
        )


    # --------------------------------------------------------
    # Load only required source columns
    # --------------------------------------------------------

    required_columns = ["PATIENT_ID"]

    if status_column is not None:
        required_columns.append(status_column)

    if pack_column is not None:
        required_columns.append(pack_column)

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} smoking data loading failed: "
            f"{type(error).__name__}: {error}"
        )


    # --------------------------------------------------------
    # Create output table with source index
    # --------------------------------------------------------

    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Preserve original active-smoking value
    # --------------------------------------------------------

    if status_column is not None:

        original_status = clean_raw_values(
            source_df[status_column]
        )

    else:

        original_status = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="string",
        )


    result["smoking_original_value"] = original_status

    result["smoking_status_standard"] = (
        original_status
        .apply(map_smoking_status)
        .astype("string")
    )


    # Whether this cohort has an active-smoking source column
    result["smoking_source_available_flag"] = (
        1 if status_column is not None else 0
    )


    # Missing original smoking value
    result["smoking_missing_flag"] = (
        original_status.isna()
    ).astype("int8")


    # --------------------------------------------------------
    # Check smoking status against pack-years
    # --------------------------------------------------------

    inconsistency_mask = pd.Series(
        False,
        index=source_df.index,
        dtype=bool,
    )


    if pack_column is not None:

        raw_pack_years = clean_raw_values(
            source_df[pack_column]
        )

        numeric_pack_years = pd.to_numeric(
            raw_pack_years,
            errors="coerce",
        )

        # Contradiction only exists when:
        # standard smoking status is Never
        # AND a known numeric pack-year value is above zero.
        never_mask = (
            result["smoking_status_standard"]
            .eq("Never")
            .fillna(False)
        )

        positive_pack_mask = (
            numeric_pack_years
            .gt(0)
            .fillna(False)
        )

        inconsistency_mask = (
            never_mask & positive_pack_mask
        ).astype(bool)


    # Safe integer conversion: no pd.NA remains here.
    result["smoking_inconsistency_flag"] = (
        inconsistency_mask
        .fillna(False)
        .astype("int8")
    )


    # Contradictory records are not used as Never or Ever.
    result.loc[
        inconsistency_mask,
        "smoking_status_standard",
    ] = "Unknown"


    # --------------------------------------------------------
    # Final active-smoking model fields
    # --------------------------------------------------------

    result["smoking_unusable_flag"] = (
        result["smoking_status_standard"]
        .eq("Unknown")
        .fillna(True)
        .astype("int8")
    )


    result["smoking_binary"] = (
        result["smoking_status_standard"]
        .map({
            "Never": 0,
            "Ever": 1,
        })
        .astype("Int64")
    )


    result["smoking_source_column"] = (
        status_column
        if status_column is not None
        else pd.NA
    )

    result["smoking_source_file"] = str(source_file)


    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "smoking_original_value",
            "smoking_status_standard",
            "smoking_binary",
            "smoking_source_available_flag",
            "smoking_missing_flag",
            "smoking_inconsistency_flag",
            "smoking_unusable_flag",
            "smoking_source_column",
            "smoking_source_file",
        ]
    ]


    # --------------------------------------------------------
    # Cohort validation and summary
    # --------------------------------------------------------

    duplicate_patient_rows = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    missing_patient_ids = int(
        result["PATIENT_ID"].isna().sum()
        + result["PATIENT_ID"].eq("").sum()
    )

    never_count = int(
        result["smoking_status_standard"]
        .eq("Never")
        .sum()
    )

    ever_count = int(
        result["smoking_status_standard"]
        .eq("Ever")
        .sum()
    )

    unknown_count = int(
        result["smoking_status_standard"]
        .eq("Unknown")
        .sum()
    )

    inconsistency_count = int(
        result["smoking_inconsistency_flag"]
        .sum()
    )

    usable_count = never_count + ever_count

    usable_percent = (
        round(
            100 * usable_count / len(result),
            2,
        )
        if len(result) > 0
        else 0.0
    )


    print(f"Patient rows: {len(result):,}")
    print(f"Never: {never_count:,}")
    print(f"Ever: {ever_count:,}")
    print(f"Unknown: {unknown_count:,}")
    print(f"Inconsistent records: {inconsistency_count:,}")
    print(f"Usable smoking coverage: {usable_percent:.2f}%")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")
    print(f"Missing patient IDs: {missing_patient_ids}")


    summary_rows.append(
        {
            "COHORT": cohort,
            "TOTAL_PATIENTS": len(result),
            "SMOKING_SOURCE_AVAILABLE": (
                1 if status_column is not None else 0
            ),
            "SMOKING_SOURCE_COLUMN": (
                status_column
                if status_column is not None
                else "Not available"
            ),
            "NEVER_COUNT": never_count,
            "EVER_COUNT": ever_count,
            "UNKNOWN_COUNT": unknown_count,
            "INCONSISTENCY_COUNT": inconsistency_count,
            "USABLE_SMOKING_COUNT": usable_count,
            "USABLE_SMOKING_PERCENT": usable_percent,
            "DUPLICATE_PATIENT_ROWS": duplicate_patient_rows,
            "MISSING_PATIENT_IDS": missing_patient_ids,
        }
    )

    harmonised_frames.append(result)


# ------------------------------------------------------------
# 6. Combine all seven cohorts
# ------------------------------------------------------------

patient_level_smoking_df = pd.concat(
    harmonised_frames,
    ignore_index=True,
)

smoking_summary_df = pd.DataFrame(
    summary_rows
)


# ------------------------------------------------------------
# 7. Combined validation
# ------------------------------------------------------------

duplicate_keys = patient_level_smoking_df.duplicated(
    subset=["COHORT", "PATIENT_ID"],
    keep=False,
)

missing_cohort_values = int(
    patient_level_smoking_df["COHORT"].isna().sum()
)

missing_patient_values = int(
    patient_level_smoking_df["PATIENT_ID"].isna().sum()
    + patient_level_smoking_df["PATIENT_ID"].eq("").sum()
)

invalid_status_values = sorted(
    set(
        patient_level_smoking_df[
            "smoking_status_standard"
        ]
        .dropna()
        .astype(str)
    )
    - {"Never", "Ever", "Unknown"}
)

observed_binary_values = sorted(
    patient_level_smoking_df[
        "smoking_binary"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

expected_cohorts = set(
    SMOKING_CONFIG.keys()
)

found_cohorts = set(
    smoking_summary_df["COHORT"]
)

missing_cohorts = sorted(
    expected_cohorts - found_cohorts
)


print("\n" + "=" * 78)
print("COMBINED ACTIVE-SMOKING VALIDATION")
print("=" * 78)

print(
    f"Combined patient rows: "
    f"{len(patient_level_smoking_df):,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_keys.sum())}"
)

print(f"Missing cohort values: {missing_cohort_values}")
print(f"Missing patient IDs: {missing_patient_values}")
print(f"Missing cohort summaries: {missing_cohorts}")
print(f"Invalid standard values: {invalid_status_values}")
print(f"Observed binary values: {observed_binary_values}")

print(
    f"Total inconsistency flags: "
    f"{int(patient_level_smoking_df['smoking_inconsistency_flag'].sum()):,}"
)

print(
    f"Overall usable active-smoking coverage: "
    f"{patient_level_smoking_df['smoking_binary'].notna().mean() * 100:.2f}%"
)


if duplicate_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID records were found."
    )

if missing_cohort_values != 0:
    raise ValueError(
        "Missing COHORT values were found."
    )

if missing_patient_values != 0:
    raise ValueError(
        "Missing PATIENT_ID values were found."
    )

if missing_cohorts:
    raise ValueError(
        f"Smoking summary is missing cohorts: {missing_cohorts}"
    )

if invalid_status_values:
    raise ValueError(
        f"Invalid smoking-status values were found: "
        f"{invalid_status_values}"
    )

if not set(observed_binary_values).issubset({0, 1}):
    raise ValueError(
        f"Invalid smoking binary values were found: "
        f"{observed_binary_values}"
    )


# ------------------------------------------------------------
# 8. Display inconsistent records
# ------------------------------------------------------------

inconsistent_records = patient_level_smoking_df.loc[
    patient_level_smoking_df[
        "smoking_inconsistency_flag"
    ].eq(1),
    [
        "COHORT",
        "PATIENT_ID",
        "smoking_original_value",
        "smoking_status_standard",
        "smoking_binary",
        "smoking_inconsistency_flag",
    ],
]


if not inconsistent_records.empty:

    print("\nInconsistent smoking records:")

    print(
        inconsistent_records.to_string(
            index=False
        )
    )

else:

    print("\nNo smoking-status contradictions were found.")


# ------------------------------------------------------------
# 9. Save harmonised data and summary
# ------------------------------------------------------------

patient_level_smoking_df.to_csv(
    OUTPUT_SMOKING_DATA,
    index=False,
)

smoking_summary_df.to_csv(
    OUTPUT_SMOKING_SUMMARY,
    index=False,
)


# Keep tables available for later notebook steps.
PATIENT_LEVEL_ACTIVE_SMOKING_DF = (
    patient_level_smoking_df.copy()
)

ACTIVE_SMOKING_SUMMARY_DF = (
    smoking_summary_df.copy()
)


print("\n" + "=" * 78)
print("STEP 6C-1 FINISHED")
print("=" * 78)

print(f"Harmonised smoking data: {OUTPUT_SMOKING_DATA}")
print(f"Data file exists: {OUTPUT_SMOKING_DATA.exists()}")

print(f"\nSmoking summary: {OUTPUT_SMOKING_SUMMARY}")
print(f"Summary file exists: {OUTPUT_SMOKING_SUMMARY.exists()}")

print("\nMissing smoking values were not mapped to Never.")
print("Contradictory smoking records were mapped to Unknown.")
print("Passive smoking was not merged with active smoking.")
print("No pack-year value was changed or imputed.")
print("No final clinical dataset was created.")
```

## Step 6C-2 — Harmonise Passive Smoking Across All Seven Cohorts

Passive-smoking information is available only in `NCI_2022` through the `PASSIVE_SMOKING` column.

Standard mapping:

- `No` → encoded as 0
- `Yes` → encoded as 1
- missing or unavailable → `Unknown`, encoded as missing

For cohorts without a passive-smoking source column, passive smoking will remain `Unknown`.

Passive smoking will not replace, modify or infer active smoking status.

```python
# ============================================================
# STEP 6C-2 — HARMONISE PASSIVE SMOKING
# ============================================================

import pandas as pd


if "PROJECT_DIR" not in globals():
    raise RuntimeError(
        "PROJECT_DIR was not found. Run the setup cell first."
    )

if "OUTPUT_DIR" not in globals():
    raise RuntimeError(
        "OUTPUT_DIR was not found. Run the setup cell first."
    )


# ------------------------------------------------------------
# 1. Cohort configuration
# ------------------------------------------------------------

PASSIVE_SMOKING_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "passive_column": None,
    },
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "passive_column": None,
    },
    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "passive_column": None,
    },
    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt",
        "passive_column": None,
    },
    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_patient.txt",
        "passive_column": None,
    },
    "NCI_2022": {
        "file": PROJECT_DIR / "nci_2022_data_clinical_patient.txt",
        "passive_column": "PASSIVE_SMOKING",
    },
    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "passive_column": None,
    },
}


OUTPUT_PASSIVE_DATA = (
    OUTPUT_DIR
    / "data"
    / "luad_passive_smoking_harmonisation_patient_level.csv"
)

OUTPUT_PASSIVE_SUMMARY = (
    OUTPUT_DIR
    / "results"
    / "luad_passive_smoking_harmonisation_summary.csv"
)


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


def clean_passive_value_series(series):
    """
    Strip whitespace and convert known missing tokens to pandas NA.
    """

    cleaned = (
        series
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        cleaned.isna()
        | cleaned.str.upper().isin(MISSING_TOKENS)
    )

    return cleaned.mask(missing_mask)


def map_passive_smoking(raw_value):
    """
    Map an original passive-smoking value to:
    No, Yes or Unknown.
    """

    if pd.isna(raw_value):
        return "Unknown"

    value = str(raw_value).strip().upper()

    if value in {"NO", "N"}:
        return "No"

    if value in {"YES", "Y"}:
        return "Yes"

    return "Unknown"


# ------------------------------------------------------------
# 2. Process all seven cohorts
# ------------------------------------------------------------

print("=" * 78)
print("STEP 6C-2 — PASSIVE SMOKING HARMONISATION")
print("=" * 78)


harmonised_frames = []
summary_rows = []


for cohort, config in PASSIVE_SMOKING_CONFIG.items():

    source_file = config["file"]
    passive_column = config["passive_column"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source file exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-patient file not found:\n"
            f"{source_file}"
        )

    required_columns = ["PATIENT_ID"]

    if passive_column is not None:
        required_columns.append(passive_column)

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} passive-smoking loading failed: "
            f"{type(error).__name__}: {error}"
        )


    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Preserve original passive-smoking value
    # --------------------------------------------------------

    if passive_column is not None:

        original_passive = clean_passive_value_series(
            source_df[passive_column]
        )

    else:

        original_passive = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="string",
        )


    result["passive_smoking_original_value"] = (
        original_passive
    )

    result["passive_smoking_standard"] = (
        original_passive
        .apply(map_passive_smoking)
        .astype("string")
    )


    result["passive_smoking_binary"] = (
        result["passive_smoking_standard"]
        .map({
            "No": 0,
            "Yes": 1,
        })
        .astype("Int64")
    )


    result["passive_smoking_source_available_flag"] = (
        1 if passive_column is not None else 0
    )

    result["passive_smoking_missing_flag"] = (
        original_passive.isna()
    ).astype("int8")

    result["passive_smoking_unusable_flag"] = (
        result["passive_smoking_standard"]
        .eq("Unknown")
        .fillna(True)
        .astype("int8")
    )

    result["passive_smoking_source_column"] = (
        passive_column
        if passive_column is not None
        else pd.NA
    )

    result["passive_smoking_source_file"] = str(
        source_file
    )


    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "passive_smoking_original_value",
            "passive_smoking_standard",
            "passive_smoking_binary",
            "passive_smoking_source_available_flag",
            "passive_smoking_missing_flag",
            "passive_smoking_unusable_flag",
            "passive_smoking_source_column",
            "passive_smoking_source_file",
        ]
    ]


    # --------------------------------------------------------
    # Cohort audit
    # --------------------------------------------------------

    duplicate_patient_rows = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    no_count = int(
        result["passive_smoking_standard"]
        .eq("No")
        .sum()
    )

    yes_count = int(
        result["passive_smoking_standard"]
        .eq("Yes")
        .sum()
    )

    unknown_count = int(
        result["passive_smoking_standard"]
        .eq("Unknown")
        .sum()
    )

    usable_count = no_count + yes_count

    usable_percent = (
        round(
            100 * usable_count / len(result),
            2,
        )
        if len(result) > 0
        else 0.0
    )


    print(f"Patient rows: {len(result):,}")
    print(f"No: {no_count:,}")
    print(f"Yes: {yes_count:,}")
    print(f"Unknown: {unknown_count:,}")
    print(f"Usable passive-smoking coverage: {usable_percent:.2f}%")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")


    summary_rows.append(
        {
            "COHORT": cohort,
            "TOTAL_PATIENTS": len(result),
            "PASSIVE_SOURCE_AVAILABLE": (
                1 if passive_column is not None else 0
            ),
            "PASSIVE_SOURCE_COLUMN": (
                passive_column
                if passive_column is not None
                else "Not available"
            ),
            "NO_COUNT": no_count,
            "YES_COUNT": yes_count,
            "UNKNOWN_COUNT": unknown_count,
            "USABLE_PASSIVE_COUNT": usable_count,
            "USABLE_PASSIVE_PERCENT": usable_percent,
            "DUPLICATE_PATIENT_ROWS": duplicate_patient_rows,
        }
    )

    harmonised_frames.append(result)


# ------------------------------------------------------------
# 3. Combine all cohorts
# ------------------------------------------------------------

patient_level_passive_smoking_df = pd.concat(
    harmonised_frames,
    ignore_index=True,
)

passive_smoking_summary_df = pd.DataFrame(
    summary_rows
)


# ------------------------------------------------------------
# 4. Combined validation
# ------------------------------------------------------------

duplicate_keys = (
    patient_level_passive_smoking_df
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

invalid_standard_values = sorted(
    set(
        patient_level_passive_smoking_df[
            "passive_smoking_standard"
        ]
        .dropna()
        .astype(str)
    )
    - {"No", "Yes", "Unknown"}
)

observed_binary_values = sorted(
    patient_level_passive_smoking_df[
        "passive_smoking_binary"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

missing_cohort_values = int(
    patient_level_passive_smoking_df[
        "COHORT"
    ].isna().sum()
)

missing_patient_ids = int(
    patient_level_passive_smoking_df[
        "PATIENT_ID"
    ].isna().sum()
    +
    patient_level_passive_smoking_df[
        "PATIENT_ID"
    ].eq("").sum()
)


print("\n" + "=" * 78)
print("COMBINED PASSIVE-SMOKING VALIDATION")
print("=" * 78)

print(
    f"Combined patient rows: "
    f"{len(patient_level_passive_smoking_df):,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_keys.sum())}"
)

print(f"Missing cohort values: {missing_cohort_values}")
print(f"Missing patient IDs: {missing_patient_ids}")
print(f"Invalid standard values: {invalid_standard_values}")
print(f"Observed binary values: {observed_binary_values}")

print(
    f"Overall usable passive-smoking coverage: "
    f"{patient_level_passive_smoking_df['passive_smoking_binary'].notna().mean() * 100:.2f}%"
)


if duplicate_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were found."
    )

if missing_cohort_values != 0:
    raise ValueError(
        "Missing COHORT values were found."
    )

if missing_patient_ids != 0:
    raise ValueError(
        "Missing PATIENT_ID values were found."
    )

if invalid_standard_values:
    raise ValueError(
        f"Invalid passive-smoking values found: "
        f"{invalid_standard_values}"
    )

if not set(observed_binary_values).issubset({0, 1}):
    raise ValueError(
        f"Invalid passive-smoking encodings found: "
        f"{observed_binary_values}"
    )


# ------------------------------------------------------------
# 5. Save data and summary
# ------------------------------------------------------------

patient_level_passive_smoking_df.to_csv(
    OUTPUT_PASSIVE_DATA,
    index=False,
)

passive_smoking_summary_df.to_csv(
    OUTPUT_PASSIVE_SUMMARY,
    index=False,
)


PATIENT_LEVEL_PASSIVE_SMOKING_DF = (
    patient_level_passive_smoking_df.copy()
)

PASSIVE_SMOKING_SUMMARY_DF = (
    passive_smoking_summary_df.copy()
)


print("\n" + "=" * 78)
print("STEP 6C-2 FINISHED")
print("=" * 78)

print(f"Passive-smoking data: {OUTPUT_PASSIVE_DATA}")
print(f"Data file exists: {OUTPUT_PASSIVE_DATA.exists()}")

print(f"\nSummary: {OUTPUT_PASSIVE_SUMMARY}")
print(f"Summary file exists: {OUTPUT_PASSIVE_SUMMARY.exists()}")

print("\nPassive smoking was kept separate from active smoking.")
print("Missing passive-smoking values were not mapped to No.")
print("No active-smoking value was changed.")
print("No final clinical dataset was created.")
```

## Step 6C-3 — Harmonise Smoking Pack-Years Across All Seven Cohorts

Smoking pack-years are available in:

- `MSK_NPJPO_2021`
- `MSKCC_2020`
- `ONCOSG_2020`

For all other cohorts, pack-years will remain missing.

The original value and numeric value will be preserved. Values above 200 will be flagged as unusual and excluded only from `pack_years_for_model`, without deleting or correcting the original record.

A positive pack-year value reported with `Never/No` smoking status will be flagged as inconsistent and excluded from the modelling field.

No imputation, scaling or model fitting will be performed.

```python
# ============================================================
# STEP 6C-3 — HARMONISE SMOKING PACK-YEARS
# ============================================================

import pandas as pd
import numpy as np


if "PROJECT_DIR" not in globals():
    raise RuntimeError(
        "PROJECT_DIR was not found. Run the setup cell first."
    )

if "OUTPUT_DIR" not in globals():
    raise RuntimeError(
        "OUTPUT_DIR was not found. Run the setup cell first."
    )


# ------------------------------------------------------------
# 1. Cohort configuration
# ------------------------------------------------------------

PACK_YEAR_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "pack_column": None,
        "status_column": None,
        "never_values": set(),
    },

    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_patient.txt",
        "pack_column": None,
        "status_column": "SMOKING_HISTORY",
        "never_values": {"NEVER"},
    },

    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "pack_column": "SMOKING_PACK_YEARS",
        "status_column": "SMOKING_STATUS",
        "never_values": {"NEVER SMOKER"},
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_patient.txt",
        "pack_column": "SMOKING_PACK_YEARS",
        "status_column": "SMOKING_HISTORY",
        "never_values": {"NEVER SMOKER"},
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_patient.txt",
        "pack_column": None,
        "status_column": "CIGARETTE_HX",
        "never_values": {"NEVER"},
    },

    "NCI_2022": {
        "file": PROJECT_DIR / "nci_2022_data_clinical_patient.txt",
        "pack_column": None,
        "status_column": None,
        "never_values": set(),
    },

    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "pack_column": "SMOKING_PACK_YEARS",
        "status_column": "SMOKING_STATUS",
        "never_values": {"NO"},
    },
}


OUTPUT_PACK_DATA = (
    OUTPUT_DIR
    / "data"
    / "luad_pack_years_harmonisation_patient_level.csv"
)

OUTPUT_PACK_SUMMARY = (
    OUTPUT_DIR
    / "results"
    / "luad_pack_years_harmonisation_summary.csv"
)


MISSING_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    ".",
}


def clean_source_series(series):
    cleaned = (
        series
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        cleaned.isna()
        | cleaned.str.upper().isin(MISSING_TOKENS)
    )

    return cleaned.mask(missing_mask)


# ------------------------------------------------------------
# 2. Process all seven cohorts
# ------------------------------------------------------------

print("=" * 78)
print("STEP 6C-3 — SMOKING PACK-YEARS HARMONISATION")
print("=" * 78)


harmonised_frames = []
summary_rows = []


for cohort, config in PACK_YEAR_CONFIG.items():

    source_file = config["file"]
    pack_column = config["pack_column"]
    status_column = config["status_column"]
    never_values = config["never_values"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source file exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-patient file not found:\n"
            f"{source_file}"
        )

    required_columns = ["PATIENT_ID"]

    if pack_column is not None:
        required_columns.append(pack_column)

    if status_column is not None:
        required_columns.append(status_column)

    required_columns = list(dict.fromkeys(required_columns))

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} pack-year loading failed: "
            f"{type(error).__name__}: {error}"
        )


    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Original and numeric pack-year values
    # --------------------------------------------------------

    if pack_column is not None:

        raw_pack_years = clean_source_series(
            source_df[pack_column]
        )

        numeric_pack_years = pd.to_numeric(
            raw_pack_years,
            errors="coerce",
        )

    else:

        raw_pack_years = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="string",
        )

        numeric_pack_years = pd.Series(
            np.nan,
            index=source_df.index,
            dtype="float64",
        )


    result["pack_years_original_value"] = raw_pack_years

    result["pack_years_numeric"] = (
        numeric_pack_years.astype("float64")
    )


    # --------------------------------------------------------
    # Missing and invalid numeric checks
    # --------------------------------------------------------

    non_numeric_mask = (
        raw_pack_years.notna()
        & numeric_pack_years.isna()
    ).fillna(False)

    negative_mask = (
        numeric_pack_years.lt(0)
        .fillna(False)
    )

    above_200_mask = (
        numeric_pack_years.gt(200)
        .fillna(False)
    )


    # --------------------------------------------------------
    # Check Never smoking status versus positive pack-years
    # --------------------------------------------------------

    inconsistency_mask = pd.Series(
        False,
        index=source_df.index,
        dtype=bool,
    )

    if (
        pack_column is not None
        and status_column is not None
    ):

        raw_status = clean_source_series(
            source_df[status_column]
        )

        status_upper = (
            raw_status
            .str.upper()
        )

        reported_never_mask = (
            status_upper
            .isin(never_values)
            .fillna(False)
        )

        positive_pack_mask = (
            numeric_pack_years
            .gt(0)
            .fillna(False)
        )

        inconsistency_mask = (
            reported_never_mask
            & positive_pack_mask
        ).astype(bool)


    # --------------------------------------------------------
    # Flags
    # --------------------------------------------------------

    result["pack_years_source_available_flag"] = (
        1 if pack_column is not None else 0
    )

    result["pack_years_missing_flag"] = (
        numeric_pack_years.isna()
    ).astype("int8")

    result["pack_years_non_numeric_flag"] = (
        non_numeric_mask.astype("int8")
    )

    result["pack_years_negative_flag"] = (
        negative_mask.astype("int8")
    )

    result["pack_years_outlier_flag"] = (
        above_200_mask.astype("int8")
    )

    result["pack_years_inconsistency_flag"] = (
        inconsistency_mask.astype("int8")
    )


    unusable_mask = (
        numeric_pack_years.isna()
        | negative_mask
        | above_200_mask
        | inconsistency_mask
    ).fillna(True)


    result["pack_years_unusable_flag"] = (
        unusable_mask.astype("int8")
    )


    result["pack_years_for_model"] = (
        numeric_pack_years.where(
            ~unusable_mask,
            np.nan,
        )
    )


    result["pack_years_source_column"] = (
        pack_column
        if pack_column is not None
        else pd.NA
    )

    result["pack_years_source_file"] = str(
        source_file
    )


    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "pack_years_original_value",
            "pack_years_numeric",
            "pack_years_for_model",
            "pack_years_source_available_flag",
            "pack_years_missing_flag",
            "pack_years_non_numeric_flag",
            "pack_years_negative_flag",
            "pack_years_outlier_flag",
            "pack_years_inconsistency_flag",
            "pack_years_unusable_flag",
            "pack_years_source_column",
            "pack_years_source_file",
        ]
    ]


    # --------------------------------------------------------
    # Cohort audit
    # --------------------------------------------------------

    duplicate_patient_rows = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    numeric_count = int(
        result["pack_years_numeric"]
        .notna()
        .sum()
    )

    usable_count = int(
        result["pack_years_for_model"]
        .notna()
        .sum()
    )

    missing_count = int(
        result["pack_years_missing_flag"]
        .sum()
    )

    outlier_count = int(
        result["pack_years_outlier_flag"]
        .sum()
    )

    inconsistency_count = int(
        result["pack_years_inconsistency_flag"]
        .sum()
    )


    print(f"Patient rows: {len(result):,}")
    print(f"Numeric pack-years: {numeric_count:,}")
    print(f"Usable pack-years: {usable_count:,}")
    print(f"Missing pack-years: {missing_count:,}")
    print(f"Values above 200: {outlier_count:,}")
    print(f"Inconsistent records: {inconsistency_count:,}")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")


    summary_rows.append(
        {
            "COHORT": cohort,
            "TOTAL_PATIENTS": len(result),
            "PACK_YEAR_SOURCE_AVAILABLE": (
                1 if pack_column is not None else 0
            ),
            "PACK_YEAR_SOURCE_COLUMN": (
                pack_column
                if pack_column is not None
                else "Not available"
            ),
            "NUMERIC_PACK_YEAR_COUNT": numeric_count,
            "USABLE_PACK_YEAR_COUNT": usable_count,
            "MISSING_PACK_YEAR_COUNT": missing_count,
            "OUTLIER_ABOVE_200_COUNT": outlier_count,
            "INCONSISTENCY_COUNT": inconsistency_count,
            "DUPLICATE_PATIENT_ROWS": duplicate_patient_rows,
        }
    )

    harmonised_frames.append(result)


# ------------------------------------------------------------
# 3. Combine all cohorts
# ------------------------------------------------------------

patient_level_pack_years_df = pd.concat(
    harmonised_frames,
    ignore_index=True,
)

pack_years_summary_df = pd.DataFrame(
    summary_rows
)


# ------------------------------------------------------------
# 4. Combined validation
# ------------------------------------------------------------

duplicate_keys = (
    patient_level_pack_years_df
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

missing_cohort_values = int(
    patient_level_pack_years_df[
        "COHORT"
    ].isna().sum()
)

missing_patient_ids = int(
    patient_level_pack_years_df[
        "PATIENT_ID"
    ].isna().sum()
    +
    patient_level_pack_years_df[
        "PATIENT_ID"
    ].eq("").sum()
)

negative_model_values = int(
    patient_level_pack_years_df[
        "pack_years_for_model"
    ].lt(0).sum()
)

above_200_model_values = int(
    patient_level_pack_years_df[
        "pack_years_for_model"
    ].gt(200).sum()
)


print("\n" + "=" * 78)
print("COMBINED PACK-YEAR VALIDATION")
print("=" * 78)

print(
    f"Combined patient rows: "
    f"{len(patient_level_pack_years_df):,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_keys.sum())}"
)

print(f"Missing cohort values: {missing_cohort_values}")
print(f"Missing patient IDs: {missing_patient_ids}")

print(
    f"Total usable pack-year values: "
    f"{patient_level_pack_years_df['pack_years_for_model'].notna().sum():,}"
)

print(
    f"Total outlier flags: "
    f"{patient_level_pack_years_df['pack_years_outlier_flag'].sum():,}"
)

print(
    f"Total inconsistency flags: "
    f"{patient_level_pack_years_df['pack_years_inconsistency_flag'].sum():,}"
)

print(
    f"Negative values in modelling field: "
    f"{negative_model_values}"
)

print(
    f"Values above 200 in modelling field: "
    f"{above_200_model_values}"
)


if duplicate_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were found."
    )

if missing_cohort_values != 0:
    raise ValueError(
        "Missing COHORT values were found."
    )

if missing_patient_ids != 0:
    raise ValueError(
        "Missing PATIENT_ID values were found."
    )

if negative_model_values != 0:
    raise ValueError(
        "Negative pack-year values entered the modelling field."
    )

if above_200_model_values != 0:
    raise ValueError(
        "Pack-year values above 200 entered the modelling field."
    )


# ------------------------------------------------------------
# 5. Display exceptional records
# ------------------------------------------------------------

exception_mask = (
    patient_level_pack_years_df[
        "pack_years_outlier_flag"
    ].eq(1)
    |
    patient_level_pack_years_df[
        "pack_years_inconsistency_flag"
    ].eq(1)
)


exception_records = (
    patient_level_pack_years_df
    .loc[
        exception_mask,
        [
            "COHORT",
            "PATIENT_ID",
            "pack_years_original_value",
            "pack_years_numeric",
            "pack_years_for_model",
            "pack_years_outlier_flag",
            "pack_years_inconsistency_flag",
        ],
    ]
)


if not exception_records.empty:

    print("\nExceptional pack-year records:")

    print(
        exception_records.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 6. Save outputs
# ------------------------------------------------------------

patient_level_pack_years_df.to_csv(
    OUTPUT_PACK_DATA,
    index=False,
)

pack_years_summary_df.to_csv(
    OUTPUT_PACK_SUMMARY,
    index=False,
)


PATIENT_LEVEL_PACK_YEARS_DF = (
    patient_level_pack_years_df.copy()
)

PACK_YEARS_SUMMARY_DF = (
    pack_years_summary_df.copy()
)


print("\n" + "=" * 78)
print("STEP 6C-3 FINISHED")
print("=" * 78)

print(f"Pack-year data: {OUTPUT_PACK_DATA}")
print(f"Data file exists: {OUTPUT_PACK_DATA.exists()}")

print(f"\nSummary: {OUTPUT_PACK_SUMMARY}")
print(f"Summary file exists: {OUTPUT_PACK_SUMMARY.exists()}")

print("\nOriginal pack-year values were preserved.")
print("No pack-year value was imputed or scaled.")
print("No model fitting was performed.")
print("No train/validation/test split was created yet.")
```

## Step 7A-1 — Audit Patient-Level Stage Values

This step inspects original stage values from cohorts where stage is stored in the clinical-patient file:

- `TCGA_Atlas_2018` — `AJCC_PATHOLOGIC_TUMOR_STAGE`
- `MSK_NPJPO_2021` — `CLINICAL_STAGE`
- `NCI_2022` — `TUMOR_STAGE`
- `ONCOSG_2020` — `STAGE`

It will report unique raw values, counts, missing values and duplicate patient identifiers.

Stage fields are label sources only. They will not be used as model input features.

No stage harmonisation or encoding will be performed yet.

```python
# ============================================================
# STEP 7A-1 — PATIENT-LEVEL RAW STAGE AUDIT
# ============================================================

import pandas as pd


STAGE_AUDIT_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "stage_column": "AJCC_PATHOLOGIC_TUMOR_STAGE",
    },
    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "stage_column": "CLINICAL_STAGE",
    },
    "NCI_2022": {
        "file": PROJECT_DIR / "nci_2022_data_clinical_patient.txt",
        "stage_column": "TUMOR_STAGE",
    },
    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "stage_column": "STAGE",
    },
}


MISSING_STAGE_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    "NOT REPORTED",
    ".",
}


STEP_7A1_STAGE_AUDIT = {}


print("=" * 78)
print("STEP 7A-1 — PATIENT-LEVEL RAW STAGE AUDIT")
print("=" * 78)


for cohort, config in STAGE_AUDIT_CONFIG.items():

    source_file = config["file"]
    stage_column = config["stage_column"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Stage column: {stage_column}")
    print(f"File exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-patient file not found:\n"
            f"{source_file}"
        )

    try:
        stage_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", stage_column],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} stage loading failed: "
            f"{type(error).__name__}: {error}"
        )


    patient_ids = (
        stage_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    raw_stage = (
        stage_df[stage_column]
        .astype("string")
        .str.strip()
    )

    missing_mask = (
        raw_stage.isna()
        | raw_stage.str.upper().isin(
            MISSING_STAGE_TOKENS
        )
    )

    clean_stage = raw_stage.mask(
        missing_mask
    )

    value_counts = (
        clean_stage
        .value_counts(dropna=False)
    )

    unique_values = sorted(
        clean_stage
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    duplicate_patient_rows = int(
        pd.DataFrame({
            "PATIENT_ID": patient_ids
        }).duplicated(
            subset=["PATIENT_ID"],
            keep=False,
        ).sum()
    )

    missing_patient_ids = int(
        patient_ids.isna().sum()
        + patient_ids.eq("").sum()
    )


    print(f"Patient rows: {len(stage_df):,}")
    print(
        f"Unique patient IDs: "
        f"{patient_ids.nunique(dropna=True):,}"
    )
    print(
        f"Known stage values: "
        f"{clean_stage.notna().sum():,}"
    )
    print(
        f"Missing/unknown stage values: "
        f"{int(missing_mask.sum()):,}"
    )
    print(
        f"Duplicate patient rows: "
        f"{duplicate_patient_rows}"
    )
    print(
        f"Missing patient IDs: "
        f"{missing_patient_ids}"
    )

    print("\nRaw stage value counts:")
    print(value_counts.to_string())

    STEP_7A1_STAGE_AUDIT[cohort] = {
        "source_file": str(source_file),
        "source_column": stage_column,
        "total_rows": len(stage_df),
        "known_stage_count": int(
            clean_stage.notna().sum()
        ),
        "missing_stage_count": int(
            missing_mask.sum()
        ),
        "unique_raw_values": unique_values,
        "duplicate_patient_rows": duplicate_patient_rows,
        "missing_patient_ids": missing_patient_ids,
    }


print("\n" + "=" * 78)
print("STEP 7A-1 FINISHED")
print("=" * 78)

print("No stage value was harmonised.")
print("No stage label was encoded.")
print("No stage field was added as a model feature.")
print("No patient was selected or excluded.")
```

### Step 7A-2 — Audit Sample-Level Stage Values

This step inspects stage values stored in clinical-sample files:

- `LUNG_MSK_2017` — `STAGE_AT_DIAGNOSIS`
- `MSKCC_2020` — `PATHOLOGIC_STAGE`
- `MSKCC_2023` — `CSTAGE` and `PSTAGE`

Multiple samples per patient will remain separate.

No representative sample will be selected and no file will be saved. Stage fields remain label sources only and will never be used as model input features.

```python
# ============================================================
# STEP 7A-2 — SAMPLE-LEVEL RAW STAGE AUDIT
# No output file is created
# ============================================================

import pandas as pd


SAMPLE_STAGE_CONFIG = {
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_sample.txt",
        "stage_columns": ["STAGE_AT_DIAGNOSIS"],
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_sample.txt",
        "stage_columns": ["PATHOLOGIC_STAGE"],
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_sample.txt",
        "stage_columns": ["CSTAGE", "PSTAGE"],
    },
}


MISSING_STAGE_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    "NOT REPORTED",
    ".",
}


STEP_7A2_SAMPLE_STAGE_AUDIT = {}


print("=" * 78)
print("STEP 7A-2 — SAMPLE-LEVEL RAW STAGE AUDIT")
print("=" * 78)


for cohort, config in SAMPLE_STAGE_CONFIG.items():

    source_file = config["file"]
    stage_columns = config["stage_columns"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"File: {source_file}")
    print(f"Exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-sample file not found:\n"
            f"{source_file}"
        )

    required_columns = [
        "PATIENT_ID",
        "SAMPLE_ID",
        *stage_columns,
    ]

    try:
        sample_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} stage loading failed: "
            f"{type(error).__name__}: {error}"
        )


    patient_ids = (
        sample_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    sample_ids = (
        sample_df["SAMPLE_ID"]
        .astype("string")
        .str.strip()
    )


    duplicate_sample_rows = int(
        pd.DataFrame({
            "COHORT": cohort,
            "SAMPLE_ID": sample_ids,
        }).duplicated(
            subset=["COHORT", "SAMPLE_ID"],
            keep=False,
        ).sum()
    )


    samples_per_patient = (
        pd.DataFrame({
            "PATIENT_ID": patient_ids,
            "SAMPLE_ID": sample_ids,
        })
        .groupby("PATIENT_ID", dropna=False)["SAMPLE_ID"]
        .nunique()
    )

    patients_with_multiple_samples = int(
        (samples_per_patient > 1).sum()
    )


    print(f"Sample rows: {len(sample_df):,}")
    print(
        f"Unique patients: "
        f"{patient_ids.nunique(dropna=True):,}"
    )
    print(
        f"Unique samples: "
        f"{sample_ids.nunique(dropna=True):,}"
    )
    print(
        f"Patients with multiple samples: "
        f"{patients_with_multiple_samples:,}"
    )
    print(
        f"Duplicate sample-ID rows: "
        f"{duplicate_sample_rows}"
    )


    cohort_audit = {
        "total_sample_rows": len(sample_df),
        "unique_patients": int(
            patient_ids.nunique(dropna=True)
        ),
        "unique_samples": int(
            sample_ids.nunique(dropna=True)
        ),
        "patients_with_multiple_samples": (
            patients_with_multiple_samples
        ),
        "duplicate_sample_rows": duplicate_sample_rows,
        "stage_columns": {},
    }


    for stage_column in stage_columns:

        raw_stage = (
            sample_df[stage_column]
            .astype("string")
            .str.strip()
        )

        missing_mask = (
            raw_stage.isna()
            | raw_stage.str.upper().isin(
                MISSING_STAGE_TOKENS
            )
        )

        clean_stage = raw_stage.mask(
            missing_mask
        )

        value_counts = clean_stage.value_counts(
            dropna=False
        )

        unique_values = sorted(
            clean_stage
            .dropna()
            .astype(str)
            .unique()
            .tolist()
        )


        print("\n" + f"Stage column: {stage_column}")

        print(
            f"Known values: "
            f"{clean_stage.notna().sum():,}"
        )

        print(
            f"Missing/unknown values: "
            f"{int(missing_mask.sum()):,}"
        )

        print("Raw value counts:")
        print(value_counts.to_string())


        cohort_audit["stage_columns"][stage_column] = {
            "known_count": int(
                clean_stage.notna().sum()
            ),
            "missing_count": int(
                missing_mask.sum()
            ),
            "unique_raw_values": unique_values,
        }


    STEP_7A2_SAMPLE_STAGE_AUDIT[cohort] = (
        cohort_audit
    )


print("\n" + "=" * 78)
print("STEP 7A-2 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("No stage value was harmonised or encoded.")
print("Stage fields remain label-only fields.")
```

## Step 7B-1 — Define and Test Stage Normalisation Rules

This step creates a reusable stage-normalisation function for all observed patient-level and sample-level stage values.

Rules:

- Stage IA/IB or 1 → Stage I
- Stage IIA/IIB or 2 → Stage II
- Stage IIIA/IIIB or 3 → Stage III
- Stage IV or 4 → Stage IV
- Stage I–II → Early
- Stage III–IV → Advanced
- missing or unrecognised values → Unknown

When one raw field contains more than one stage, all stages will be detected, the highest documented stage will be used, and a multiple-stage flag will be preserved.

This is only a mapping audit. No representative sample will be selected and no output file will be created.

Stage labels will never be included in model input features.

```python
# ============================================================
# STEP 7B-1 — DEFINE AND TEST STAGE NORMALISATION RULES
# No output file is created
# ============================================================

import re
import pandas as pd


MISSING_STAGE_TOKENS = {
    "",
    "NA",
    "N/A",
    "NAN",
    "NONE",
    "NULL",
    "UNKNOWN",
    "NOT AVAILABLE",
    "NOT REPORTED",
    ".",
}


STAGE_RANK = {
    "I": 1,
    "II": 2,
    "III": 3,
    "IV": 4,
}


STAGE_LABEL_4CLASS = {
    "I": "Stage I",
    "II": "Stage II",
    "III": "Stage III",
    "IV": "Stage IV",
}


STAGE_ENCODE_4CLASS = {
    "I": 0,
    "II": 1,
    "III": 2,
    "IV": 3,
}


def normalise_stage_value(raw_value):
    """
    Convert one raw stage value into harmonised label fields.

    Stage-derived fields are labels only, not model features.
    """

    default_result = {
        "stage_major": "Unknown",
        "stage_label_4class": "Unknown",
        "label_stage_4class": pd.NA,
        "stage_label_binary": "Unknown",
        "label_stage_binary": pd.NA,
        "stage_multiple_values_flag": 0,
        "stage_mapping_status": "missing",
    }

    if pd.isna(raw_value):
        return default_result

    original_text = str(raw_value).strip()

    if original_text.upper() in MISSING_STAGE_TOKENS:
        return default_result

    text = original_text.upper().strip()

    # Remove the word STAGE while preserving Roman/numeric stage.
    text_without_stage = re.sub(
        r"\bSTAGE\b",
        " ",
        text,
    ).strip()

    detected_major_stages = []


    # --------------------------------------------------------
    # Detect numeric stage values such as 1, 2, 3 or 4
    # --------------------------------------------------------

    numeric_matches = re.findall(
        r"(?<![0-9])([1-4])(?:[ABC])?(?![0-9])",
        text_without_stage,
    )

    numeric_to_roman = {
        "1": "I",
        "2": "II",
        "3": "III",
        "4": "IV",
    }

    detected_major_stages.extend(
        numeric_to_roman[value]
        for value in numeric_matches
    )


    # --------------------------------------------------------
    # Detect Roman stages and substages:
    # I, IA, IB, II, IIA, IIIA, IIIB, IV
    # --------------------------------------------------------

    roman_matches = re.findall(
        r"(?<![A-Z0-9])"
        r"(IV|III|II|I)"
        r"(?:[ABC])?"
        r"(?![A-Z0-9])",
        text_without_stage,
    )

    detected_major_stages.extend(
        roman_matches
    )


    # Remove duplicate detections while preserving order.
    unique_detected_stages = list(
        dict.fromkeys(detected_major_stages)
    )


    if not unique_detected_stages:

        return {
            **default_result,
            "stage_mapping_status": "unrecognised",
        }


    # Highest documented stage is used when multiple stages occur.
    selected_major_stage = max(
        unique_detected_stages,
        key=lambda stage: STAGE_RANK[stage],
    )

    multiple_stage_flag = int(
        len(unique_detected_stages) > 1
    )

    binary_label = (
        "Early"
        if selected_major_stage in {"I", "II"}
        else "Advanced"
    )

    binary_encoded = (
        0 if binary_label == "Early" else 1
    )


    return {
        "stage_major": selected_major_stage,
        "stage_label_4class": (
            STAGE_LABEL_4CLASS[selected_major_stage]
        ),
        "label_stage_4class": (
            STAGE_ENCODE_4CLASS[selected_major_stage]
        ),
        "stage_label_binary": binary_label,
        "label_stage_binary": binary_encoded,
        "stage_multiple_values_flag": multiple_stage_flag,
        "stage_mapping_status": (
            "mapped_multiple"
            if multiple_stage_flag
            else "mapped"
        ),
    }


# ------------------------------------------------------------
# Source configuration
# ------------------------------------------------------------

STAGE_SOURCE_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": PROJECT_DIR / "tcga_atlas_2018_data_clinical_patient.txt",
        "source_level": "patient",
        "id_columns": ["PATIENT_ID"],
        "stage_columns": ["AJCC_PATHOLOGIC_TUMOR_STAGE"],
    },

    "MSK_NPJPO_2021": {
        "file": PROJECT_DIR / "msk_npjpo_2021_data_clinical_patient.txt",
        "source_level": "patient",
        "id_columns": ["PATIENT_ID"],
        "stage_columns": ["CLINICAL_STAGE"],
    },

    "NCI_2022": {
        "file": PROJECT_DIR / "nci_2022_data_clinical_patient.txt",
        "source_level": "patient",
        "id_columns": ["PATIENT_ID"],
        "stage_columns": ["TUMOR_STAGE"],
    },

    "ONCOSG_2020": {
        "file": PROJECT_DIR / "oncosg_2020_data_clinical_patient.txt",
        "source_level": "patient",
        "id_columns": ["PATIENT_ID"],
        "stage_columns": ["STAGE"],
    },

    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_sample.txt",
        "source_level": "sample",
        "id_columns": ["PATIENT_ID", "SAMPLE_ID"],
        "stage_columns": ["STAGE_AT_DIAGNOSIS"],
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_sample.txt",
        "source_level": "sample",
        "id_columns": ["PATIENT_ID", "SAMPLE_ID"],
        "stage_columns": ["PATHOLOGIC_STAGE"],
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_sample.txt",
        "source_level": "sample",
        "id_columns": ["PATIENT_ID", "SAMPLE_ID"],
        "stage_columns": ["CSTAGE", "PSTAGE"],
    },
}


print("=" * 78)
print("STEP 7B-1 — STAGE NORMALISATION MAPPING AUDIT")
print("=" * 78)


STEP_7B1_STAGE_MAPPING_AUDIT = {}
unrecognised_records = []


for cohort, config in STAGE_SOURCE_CONFIG.items():

    source_file = config["file"]
    source_level = config["source_level"]
    id_columns = config["id_columns"]
    stage_columns = config["stage_columns"]

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} source file not found:\n{source_file}"
        )

    required_columns = list(
        dict.fromkeys(
            id_columns + stage_columns
        )
    )

    source_df = pd.read_csv(
        source_file,
        sep="\t",
        comment="#",
        usecols=required_columns,
        dtype=str,
        low_memory=False,
    )

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source level: {source_level}")
    print(f"Rows: {len(source_df):,}")


    cohort_results = {}


    for stage_column in stage_columns:

        mapping_df = (
            source_df[stage_column]
            .apply(normalise_stage_value)
            .apply(pd.Series)
        )

        raw_stage = (
            source_df[stage_column]
            .astype("string")
            .str.strip()
        )

        audit_df = pd.DataFrame({
            "raw_stage": raw_stage,
            "stage_major": mapping_df["stage_major"],
            "stage_label_4class": (
                mapping_df["stage_label_4class"]
            ),
            "stage_label_binary": (
                mapping_df["stage_label_binary"]
            ),
            "multiple_stage_flag": (
                mapping_df[
                    "stage_multiple_values_flag"
                ]
            ),
            "mapping_status": (
                mapping_df["stage_mapping_status"]
            ),
        })


        mapped_count = int(
            audit_df["mapping_status"]
            .isin(["mapped", "mapped_multiple"])
            .sum()
        )

        missing_count = int(
            audit_df["mapping_status"]
            .eq("missing")
            .sum()
        )

        unrecognised_count = int(
            audit_df["mapping_status"]
            .eq("unrecognised")
            .sum()
        )

        multiple_stage_count = int(
            audit_df["multiple_stage_flag"]
            .sum()
        )


        print(f"\nStage column: {stage_column}")
        print(f"Mapped rows: {mapped_count:,}")
        print(f"Missing rows: {missing_count:,}")
        print(f"Unrecognised rows: {unrecognised_count:,}")
        print(
            f"Rows containing multiple major stages: "
            f"{multiple_stage_count:,}"
        )


        # Show each unique raw-to-standard mapping once.
        unique_mapping_table = (
            audit_df
            .drop_duplicates()
            .sort_values(
                by=[
                    "mapping_status",
                    "raw_stage",
                ],
                na_position="last",
            )
        )

        print("\nUnique mapping results:")

        print(
            unique_mapping_table.to_string(
                index=False
            )
        )


        if unrecognised_count > 0:

            unresolved_indices = audit_df.index[
                audit_df["mapping_status"]
                .eq("unrecognised")
            ]

            for row_index in unresolved_indices:

                record = {
                    "COHORT": cohort,
                    "SOURCE_LEVEL": source_level,
                    "STAGE_COLUMN": stage_column,
                    "RAW_STAGE": raw_stage.loc[row_index],
                }

                for id_column in id_columns:
                    record[id_column] = (
                        source_df.loc[
                            row_index,
                            id_column,
                        ]
                    )

                unrecognised_records.append(record)


        cohort_results[stage_column] = {
            "total_rows": len(audit_df),
            "mapped_count": mapped_count,
            "missing_count": missing_count,
            "unrecognised_count": unrecognised_count,
            "multiple_stage_count": multiple_stage_count,
        }


    STEP_7B1_STAGE_MAPPING_AUDIT[cohort] = (
        cohort_results
    )


# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

print("\n" + "=" * 78)
print("COMBINED STAGE MAPPING VALIDATION")
print("=" * 78)

print(
    f"Total unrecognised stage rows: "
    f"{len(unrecognised_records):,}"
)


if unrecognised_records:

    unrecognised_stage_df = pd.DataFrame(
        unrecognised_records
    )

    print("\nUnrecognised stage records:")

    print(
        unrecognised_stage_df
        .head(30)
        .to_string(index=False)
    )

else:

    unrecognised_stage_df = pd.DataFrame()

    print("All non-missing stage values were recognised.")


print("\n" + "=" * 78)
print("STEP 7B-1 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("CSTAGE and PSTAGE were kept separate.")
print("Stage-derived fields remain labels only.")
print("No model feature table was created.")
```

### Step 7B-2 — Audit Stage Consistency Across Samples and Stage Sources

This step checks:

- whether different samples from the same patient report different stages
- whether `CSTAGE` and `PSTAGE` disagree within the same `MSKCC_2023` sample
- how many patients have no usable stage information

This is an audit only.

Representative-sample selection will not use stage, binary labels or four-class labels. No output file will be created.

```python
# ============================================================
# STEP 7B-2 — STAGE CONSISTENCY AUDIT
# No output file is created
# ============================================================

import pandas as pd


if "normalise_stage_value" not in globals():
    raise RuntimeError(
        "normalise_stage_value was not found. "
        "Please run Step 7B-1 first."
    )


SAMPLE_STAGE_CONFIG = {
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_sample.txt",
        "stage_columns": ["STAGE_AT_DIAGNOSIS"],
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_sample.txt",
        "stage_columns": ["PATHOLOGIC_STAGE"],
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_sample.txt",
        "stage_columns": ["CSTAGE", "PSTAGE"],
    },
}


def get_stage_major(raw_value):
    """
    Return only the harmonised major stage:
    I, II, III, IV or Unknown.
    """
    return normalise_stage_value(
        raw_value
    )["stage_major"]


print("=" * 78)
print("STEP 7B-2 — STAGE CONSISTENCY AUDIT")
print("=" * 78)


STEP_7B2_STAGE_CONSISTENCY_AUDIT = {}


for cohort, config in SAMPLE_STAGE_CONFIG.items():

    source_file = config["file"]
    stage_columns = config["stage_columns"]

    required_columns = [
        "PATIENT_ID",
        "SAMPLE_ID",
        *stage_columns,
    ]

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=required_columns,
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} loading failed: "
            f"{type(error).__name__}: {error}"
        )


    source_df["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    source_df["SAMPLE_ID"] = (
        source_df["SAMPLE_ID"]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Harmonise each stage source in memory
    # --------------------------------------------------------

    major_columns = []

    for stage_column in stage_columns:

        major_column = f"{stage_column}_major"

        source_df[major_column] = (
            source_df[stage_column]
            .apply(get_stage_major)
            .astype("string")
        )

        major_columns.append(major_column)


    # --------------------------------------------------------
    # Patient-level consistency across all available samples
    # and all stage sources
    # --------------------------------------------------------

    long_stage_df = source_df.melt(
        id_vars=["PATIENT_ID", "SAMPLE_ID"],
        value_vars=major_columns,
        var_name="STAGE_SOURCE",
        value_name="STAGE_MAJOR",
    )

    known_stage_df = long_stage_df.loc[
        long_stage_df["STAGE_MAJOR"].ne("Unknown")
    ].copy()


    unique_stage_count_per_patient = (
        known_stage_df
        .groupby("PATIENT_ID")["STAGE_MAJOR"]
        .nunique()
    )


    patients_with_stage = int(
        unique_stage_count_per_patient.shape[0]
    )

    total_patients = int(
        source_df["PATIENT_ID"]
        .nunique(dropna=True)
    )

    patients_without_stage = (
        total_patients - patients_with_stage
    )

    patients_with_conflicting_stages = (
        unique_stage_count_per_patient.loc[
            unique_stage_count_per_patient > 1
        ]
    )


    # --------------------------------------------------------
    # Samples per patient
    # --------------------------------------------------------

    samples_per_patient = (
        source_df
        .groupby("PATIENT_ID")["SAMPLE_ID"]
        .nunique()
    )

    patients_with_multiple_samples = int(
        samples_per_patient.gt(1).sum()
    )


    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Sample rows: {len(source_df):,}")
    print(f"Unique patients: {total_patients:,}")
    print(
        f"Patients with multiple samples: "
        f"{patients_with_multiple_samples:,}"
    )
    print(
        f"Patients with at least one known stage: "
        f"{patients_with_stage:,}"
    )
    print(
        f"Patients without a known stage: "
        f"{patients_without_stage:,}"
    )
    print(
        f"Patients with conflicting known stages: "
        f"{len(patients_with_conflicting_stages):,}"
    )


    cohort_audit = {
        "sample_rows": len(source_df),
        "unique_patients": total_patients,
        "patients_with_multiple_samples": (
            patients_with_multiple_samples
        ),
        "patients_with_known_stage": patients_with_stage,
        "patients_without_known_stage": patients_without_stage,
        "patients_with_conflicting_stages": int(
            len(patients_with_conflicting_stages)
        ),
    }


    # --------------------------------------------------------
    # Show a few cross-sample conflict examples
    # --------------------------------------------------------

    if len(patients_with_conflicting_stages) > 0:

        conflict_patient_ids = (
            patients_with_conflicting_stages
            .index
            .tolist()
        )

        conflict_examples = source_df.loc[
            source_df["PATIENT_ID"].isin(
                conflict_patient_ids[:10]
            ),
            [
                "PATIENT_ID",
                "SAMPLE_ID",
                *stage_columns,
                *major_columns,
            ],
        ].sort_values(
            ["PATIENT_ID", "SAMPLE_ID"]
        )

        print("\nCross-sample/source conflict examples:")

        print(
            conflict_examples.to_string(
                index=False
            )
        )


    # --------------------------------------------------------
    # MSKCC_2023: compare CSTAGE and PSTAGE within each sample
    # --------------------------------------------------------

    if cohort == "MSKCC_2023":

        cstage_known = (
            source_df["CSTAGE_major"]
            .ne("Unknown")
        )

        pstage_known = (
            source_df["PSTAGE_major"]
            .ne("Unknown")
        )

        both_known = (
            cstage_known & pstage_known
        )

        same_when_both_known = (
            both_known
            & source_df["CSTAGE_major"].eq(
                source_df["PSTAGE_major"]
            )
        )

        different_when_both_known = (
            both_known
            & source_df["CSTAGE_major"].ne(
                source_df["PSTAGE_major"]
            )
        )

        only_cstage_known = (
            cstage_known & ~pstage_known
        )

        only_pstage_known = (
            ~cstage_known & pstage_known
        )

        neither_known = (
            ~cstage_known & ~pstage_known
        )


        print("\nMSKCC_2023 within-sample CSTAGE/PSTAGE audit:")

        print(
            f"Both known and same: "
            f"{int(same_when_both_known.sum()):,}"
        )

        print(
            f"Both known but different: "
            f"{int(different_when_both_known.sum()):,}"
        )

        print(
            f"Only CSTAGE known: "
            f"{int(only_cstage_known.sum()):,}"
        )

        print(
            f"Only PSTAGE known: "
            f"{int(only_pstage_known.sum()):,}"
        )

        print(
            f"Neither known: "
            f"{int(neither_known.sum()):,}"
        )


        if different_when_both_known.any():

            within_sample_examples = source_df.loc[
                different_when_both_known,
                [
                    "PATIENT_ID",
                    "SAMPLE_ID",
                    "CSTAGE",
                    "PSTAGE",
                    "CSTAGE_major",
                    "PSTAGE_major",
                ],
            ].head(10)

            print(
                "\nCSTAGE/PSTAGE disagreement examples:"
            )

            print(
                within_sample_examples.to_string(
                    index=False
                )
            )


        cohort_audit.update({
            "samples_both_stage_sources_known_same": int(
                same_when_both_known.sum()
            ),
            "samples_both_stage_sources_known_different": int(
                different_when_both_known.sum()
            ),
            "samples_only_cstage_known": int(
                only_cstage_known.sum()
            ),
            "samples_only_pstage_known": int(
                only_pstage_known.sum()
            ),
            "samples_neither_stage_known": int(
                neither_known.sum()
            ),
        })


    STEP_7B2_STAGE_CONSISTENCY_AUDIT[cohort] = (
        cohort_audit
    )


print("\n" + "=" * 78)
print("STEP 7B-2 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("Stage information was not used as a selection criterion.")
print("No train, validation or test split was created.")
print("Stage-derived fields remain label-only fields.")
```

### Step 7B-3 — Resolve Stage Source Per Sample and Audit True Cross-Sample Conflicts

For `MSKCC_2023`, pathological stage (`PSTAGE`) has priority when available. Clinical stage (`CSTAGE`) is used only when pathological stage is missing.

Any disagreement between known `CSTAGE` and `PSTAGE` values will remain explicitly flagged.

After resolving one label per sample, this step checks whether different samples belonging to the same patient still have different resolved stages.

No representative sample will be selected, no stage-based sample selection will occur, and no file will be created.

```python
# ============================================================
# STEP 7B-3 — RESOLVE STAGE SOURCE PER SAMPLE
# AND AUDIT TRUE CROSS-SAMPLE CONFLICTS
# No output file is created
# ============================================================

import pandas as pd


if "normalise_stage_value" not in globals():
    raise RuntimeError(
        "normalise_stage_value was not found. "
        "Please run Step 7B-1 first."
    )


SAMPLE_STAGE_RESOLUTION_CONFIG = {
    "LUNG_MSK_2017": {
        "file": PROJECT_DIR / "lung_msk_2017_data_clinical_sample.txt",
        "stage_type": "single",
        "single_column": "STAGE_AT_DIAGNOSIS",
    },

    "MSKCC_2020": {
        "file": PROJECT_DIR / "mskcc_2020_data_clinical_sample.txt",
        "stage_type": "single",
        "single_column": "PATHOLOGIC_STAGE",
    },

    "MSKCC_2023": {
        "file": PROJECT_DIR / "mskcc_2023_data_clinical_sample.txt",
        "stage_type": "clinical_pathologic",
        "clinical_column": "CSTAGE",
        "pathologic_column": "PSTAGE",
    },
}


def map_stage_series(stage_series):
    """
    Apply the existing stage-normalisation function
    and return its harmonised fields as a DataFrame.
    """

    return (
        stage_series
        .apply(normalise_stage_value)
        .apply(pd.Series)
    )


print("=" * 78)
print("STEP 7B-3 — SAMPLE STAGE SOURCE RESOLUTION")
print("=" * 78)


resolved_frames = []
audit_results = {}


for cohort, config in SAMPLE_STAGE_RESOLUTION_CONFIG.items():

    source_file = config["file"]
    stage_type = config["stage_type"]

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-sample file not found:\n"
            f"{source_file}"
        )


    # --------------------------------------------------------
    # Load required columns
    # --------------------------------------------------------

    if stage_type == "single":

        stage_column = config["single_column"]

        required_columns = [
            "PATIENT_ID",
            "SAMPLE_ID",
            stage_column,
        ]

    else:

        clinical_column = config["clinical_column"]
        pathologic_column = config["pathologic_column"]

        required_columns = [
            "PATIENT_ID",
            "SAMPLE_ID",
            clinical_column,
            pathologic_column,
        ]


    source_df = pd.read_csv(
        source_file,
        sep="\t",
        comment="#",
        usecols=required_columns,
        dtype=str,
        low_memory=False,
    )


    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    result["SAMPLE_ID"] = (
        source_df["SAMPLE_ID"]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Single-stage-source cohorts
    # --------------------------------------------------------

    if stage_type == "single":

        mapped_stage = map_stage_series(
            source_df[stage_column]
        )

        result["stage_original_primary"] = (
            source_df[stage_column]
            .astype("string")
            .str.strip()
        )

        result["stage_original_secondary"] = pd.NA

        result["stage_primary_source_column"] = (
            stage_column
        )

        result["stage_resolved_source_column"] = (
            stage_column
        )

        result["stage_source_disagreement_flag"] = 0

        result["stage_major_resolved"] = (
            mapped_stage["stage_major"]
            .astype("string")
        )

        result["stage_label_4class"] = (
            mapped_stage["stage_label_4class"]
            .astype("string")
        )

        result["label_stage_4class"] = (
            mapped_stage["label_stage_4class"]
            .astype("Int64")
        )

        result["stage_label_binary"] = (
            mapped_stage["stage_label_binary"]
            .astype("string")
        )

        result["label_stage_binary"] = (
            mapped_stage["label_stage_binary"]
            .astype("Int64")
        )

        result["stage_multiple_values_flag"] = (
            mapped_stage["stage_multiple_values_flag"]
            .astype("int8")
        )

        result["stage_mapping_status"] = (
            mapped_stage["stage_mapping_status"]
            .astype("string")
        )


    # --------------------------------------------------------
    # MSKCC_2023:
    # PSTAGE priority, CSTAGE fallback
    # --------------------------------------------------------

    else:

        clinical_mapped = map_stage_series(
            source_df[clinical_column]
        )

        pathologic_mapped = map_stage_series(
            source_df[pathologic_column]
        )


        clinical_known = (
            clinical_mapped["stage_major"]
            .ne("Unknown")
            .fillna(False)
        )

        pathologic_known = (
            pathologic_mapped["stage_major"]
            .ne("Unknown")
            .fillna(False)
        )


        both_known = (
            clinical_known & pathologic_known
        )

        disagreement_mask = (
            both_known
            & clinical_mapped["stage_major"].ne(
                pathologic_mapped["stage_major"]
            )
        ).fillna(False)


        # Pathological stage is primary.
        # Clinical stage is used only when PSTAGE is unavailable.
        use_pathologic = pathologic_known

        use_clinical_fallback = (
            ~pathologic_known
            & clinical_known
        )


        result["stage_original_primary"] = (
            source_df[pathologic_column]
            .astype("string")
            .str.strip()
        )

        result["stage_original_secondary"] = (
            source_df[clinical_column]
            .astype("string")
            .str.strip()
        )

        result["stage_primary_source_column"] = (
            pathologic_column
        )

        result["stage_resolved_source_column"] = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="string",
        )

        result.loc[
            use_pathologic,
            "stage_resolved_source_column",
        ] = pathologic_column

        result.loc[
            use_clinical_fallback,
            "stage_resolved_source_column",
        ] = clinical_column


        result["stage_source_disagreement_flag"] = (
            disagreement_mask
            .astype("int8")
        )


        # Begin with missing/unknown defaults.
        result["stage_major_resolved"] = "Unknown"
        result["stage_label_4class"] = "Unknown"

        result["label_stage_4class"] = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="Int64",
        )

        result["stage_label_binary"] = "Unknown"

        result["label_stage_binary"] = pd.Series(
            pd.NA,
            index=source_df.index,
            dtype="Int64",
        )

        result["stage_multiple_values_flag"] = 0
        result["stage_mapping_status"] = "missing"


        # Use PSTAGE when available.
        pathologic_fields = [
            "stage_major",
            "stage_label_4class",
            "label_stage_4class",
            "stage_label_binary",
            "label_stage_binary",
            "stage_multiple_values_flag",
            "stage_mapping_status",
        ]

        output_fields = [
            "stage_major_resolved",
            "stage_label_4class",
            "label_stage_4class",
            "stage_label_binary",
            "label_stage_binary",
            "stage_multiple_values_flag",
            "stage_mapping_status",
        ]


        for source_field, output_field in zip(
            pathologic_fields,
            output_fields,
        ):

            result.loc[
                use_pathologic,
                output_field,
            ] = pathologic_mapped.loc[
                use_pathologic,
                source_field,
            ].values


        # Use CSTAGE only when PSTAGE is unavailable.
        for source_field, output_field in zip(
            pathologic_fields,
            output_fields,
        ):

            result.loc[
                use_clinical_fallback,
                output_field,
            ] = clinical_mapped.loc[
                use_clinical_fallback,
                source_field,
            ].values


        result["label_stage_4class"] = (
            result["label_stage_4class"]
            .astype("Int64")
        )

        result["label_stage_binary"] = (
            result["label_stage_binary"]
            .astype("Int64")
        )

        result["stage_multiple_values_flag"] = (
            result["stage_multiple_values_flag"]
            .astype("int8")
        )

        result["stage_source_disagreement_flag"] = (
            result["stage_source_disagreement_flag"]
            .astype("int8")
        )


    result["stage_source_file"] = str(
        source_file
    )


    # --------------------------------------------------------
    # True cross-sample patient audit
    # Uses one resolved stage per sample.
    # --------------------------------------------------------

    known_stage_rows = result.loc[
        result["stage_major_resolved"].ne("Unknown")
    ].copy()


    resolved_stage_count_per_patient = (
        known_stage_rows
        .groupby("PATIENT_ID")[
            "stage_major_resolved"
        ]
        .nunique()
    )


    true_cross_sample_conflict_patients = (
        resolved_stage_count_per_patient.loc[
            resolved_stage_count_per_patient > 1
        ]
    )


    samples_per_patient = (
        result
        .groupby("PATIENT_ID")["SAMPLE_ID"]
        .nunique()
    )


    patients_with_multiple_samples = int(
        samples_per_patient.gt(1).sum()
    )

    total_patients = int(
        result["PATIENT_ID"]
        .nunique(dropna=True)
    )

    patients_with_known_resolved_stage = int(
        known_stage_rows["PATIENT_ID"]
        .nunique(dropna=True)
    )

    patients_without_known_resolved_stage = (
        total_patients
        - patients_with_known_resolved_stage
    )

    patients_with_source_disagreement = int(
        result.loc[
            result["stage_source_disagreement_flag"].eq(1),
            "PATIENT_ID",
        ].nunique()
    )


    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Sample rows: {len(result):,}")
    print(f"Unique patients: {total_patients:,}")

    print(
        f"Patients with multiple samples: "
        f"{patients_with_multiple_samples:,}"
    )

    print(
        f"Patients with known resolved stage: "
        f"{patients_with_known_resolved_stage:,}"
    )

    print(
        f"Patients without known resolved stage: "
        f"{patients_without_known_resolved_stage:,}"
    )

    print(
        f"Patients with CSTAGE/PSTAGE disagreement: "
        f"{patients_with_source_disagreement:,}"
    )

    print(
        f"Patients with true cross-sample resolved-stage conflict: "
        f"{len(true_cross_sample_conflict_patients):,}"
    )


    # --------------------------------------------------------
    # Show true cross-sample conflicts only
    # --------------------------------------------------------

    if len(true_cross_sample_conflict_patients) > 0:

        conflict_patient_ids = (
            true_cross_sample_conflict_patients
            .index
            .tolist()
        )

        conflict_examples = result.loc[
            result["PATIENT_ID"].isin(
                conflict_patient_ids[:10]
            ),
            [
                "PATIENT_ID",
                "SAMPLE_ID",
                "stage_original_primary",
                "stage_original_secondary",
                "stage_resolved_source_column",
                "stage_major_resolved",
                "stage_source_disagreement_flag",
            ],
        ].sort_values(
            ["PATIENT_ID", "SAMPLE_ID"]
        )

        print("\nTrue cross-sample conflict examples:")

        print(
            conflict_examples.to_string(
                index=False
            )
        )


    audit_results[cohort] = {
        "sample_rows": len(result),
        "unique_patients": total_patients,
        "patients_with_multiple_samples": (
            patients_with_multiple_samples
        ),
        "patients_with_known_resolved_stage": (
            patients_with_known_resolved_stage
        ),
        "patients_without_known_resolved_stage": (
            patients_without_known_resolved_stage
        ),
        "patients_with_source_disagreement": (
            patients_with_source_disagreement
        ),
        "patients_with_true_cross_sample_conflict": int(
            len(true_cross_sample_conflict_patients)
        ),
    }


    resolved_frames.append(result)


# ------------------------------------------------------------
# Combine in memory for later clinical merge
# ------------------------------------------------------------

RESOLVED_SAMPLE_STAGE_DF = pd.concat(
    resolved_frames,
    ignore_index=True,
)

STEP_7B3_STAGE_RESOLUTION_AUDIT = (
    audit_results
)


# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

duplicate_sample_keys = (
    RESOLVED_SAMPLE_STAGE_DF
    .duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)


invalid_stage_values = sorted(
    set(
        RESOLVED_SAMPLE_STAGE_DF[
            "stage_major_resolved"
        ]
        .dropna()
        .astype(str)
    )
    - {"I", "II", "III", "IV", "Unknown"}
)


print("\n" + "=" * 78)
print("COMBINED RESOLVED-STAGE VALIDATION")
print("=" * 78)

print(
    f"Combined sample rows: "
    f"{len(RESOLVED_SAMPLE_STAGE_DF):,}"
)

print(
    f"Duplicate cohort-sample rows: "
    f"{int(duplicate_sample_keys.sum())}"
)

print(
    f"Invalid resolved-stage values: "
    f"{invalid_stage_values}"
)


if duplicate_sample_keys.any():
    raise ValueError(
        "Duplicate COHORT + SAMPLE_ID rows were found."
    )

if invalid_stage_values:
    raise ValueError(
        f"Invalid resolved-stage values found: "
        f"{invalid_stage_values}"
    )


print("\n" + "=" * 78)
print("STEP 7B-3 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("PSTAGE had priority; CSTAGE was fallback only.")
print("Stage disagreement flags were preserved.")
print("Stage information was not used as a model feature.")
print("No train, validation or test split was created.")
```

### Step 7B-4 — Harmonise Patient-Level Stage Labels

This step harmonises stage labels for cohorts where stage is stored directly in the clinical-patient file:

- `TCGA_Atlas_2018`
- `MSK_NPJPO_2021`
- `NCI_2022`
- `ONCOSG_2020`

The original stage value, source column and source file will remain preserved.

Stage-derived fields are target labels only. They will not be included among model input features.

No output file, representative-sample selection, dataset split or model training will be performed.

```python
# ============================================================
# STEP 7B-4 — HARMONISE PATIENT-LEVEL STAGE LABELS
# No output file is created
# ============================================================

import pandas as pd


if "normalise_stage_value" not in globals():
    raise RuntimeError(
        "normalise_stage_value was not found. "
        "Please run Step 7B-1 first."
    )


PATIENT_STAGE_CONFIG = {
    "TCGA_Atlas_2018": {
        "file": (
            PROJECT_DIR
            / "tcga_atlas_2018_data_clinical_patient.txt"
        ),
        "stage_column": "AJCC_PATHOLOGIC_TUMOR_STAGE",
    },

    "MSK_NPJPO_2021": {
        "file": (
            PROJECT_DIR
            / "msk_npjpo_2021_data_clinical_patient.txt"
        ),
        "stage_column": "CLINICAL_STAGE",
    },

    "NCI_2022": {
        "file": (
            PROJECT_DIR
            / "nci_2022_data_clinical_patient.txt"
        ),
        "stage_column": "TUMOR_STAGE",
    },

    "ONCOSG_2020": {
        "file": (
            PROJECT_DIR
            / "oncosg_2020_data_clinical_patient.txt"
        ),
        "stage_column": "STAGE",
    },
}


print("=" * 78)
print("STEP 7B-4 — PATIENT-LEVEL STAGE HARMONISATION")
print("=" * 78)


harmonised_stage_frames = []
stage_summary_rows = []


for cohort, config in PATIENT_STAGE_CONFIG.items():

    source_file = config["file"]
    stage_column = config["stage_column"]

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Source column: {stage_column}")
    print(f"File exists: {source_file.exists()}")

    if not source_file.exists():
        raise FileNotFoundError(
            f"{cohort} clinical-patient file not found:\n"
            f"{source_file}"
        )


    # --------------------------------------------------------
    # Load identifiers and original stage only
    # --------------------------------------------------------

    try:
        source_df = pd.read_csv(
            source_file,
            sep="\t",
            comment="#",
            usecols=["PATIENT_ID", stage_column],
            dtype=str,
            low_memory=False,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} stage loading failed: "
            f"{type(error).__name__}: {error}"
        )


    result = pd.DataFrame(index=source_df.index)

    result["COHORT"] = cohort

    result["PATIENT_ID"] = (
        source_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    result["SAMPLE_ID"] = pd.Series(
        pd.NA,
        index=source_df.index,
        dtype="string",
    )

    result["stage_original_value"] = (
        source_df[stage_column]
        .astype("string")
        .str.strip()
    )


    # --------------------------------------------------------
    # Apply previously audited stage rules
    # --------------------------------------------------------

    mapped_stage = (
        source_df[stage_column]
        .apply(normalise_stage_value)
        .apply(pd.Series)
    )


    result["stage_major_resolved"] = (
        mapped_stage["stage_major"]
        .astype("string")
    )

    result["stage_label_4class"] = (
        mapped_stage["stage_label_4class"]
        .astype("string")
    )

    result["label_stage_4class"] = (
        mapped_stage["label_stage_4class"]
        .astype("Int64")
    )

    result["stage_label_binary"] = (
        mapped_stage["stage_label_binary"]
        .astype("string")
    )

    result["label_stage_binary"] = (
        mapped_stage["label_stage_binary"]
        .astype("Int64")
    )

    result["stage_multiple_values_flag"] = (
        mapped_stage["stage_multiple_values_flag"]
        .fillna(0)
        .astype("int8")
    )

    result["stage_mapping_status"] = (
        mapped_stage["stage_mapping_status"]
        .astype("string")
    )


    # Patient-level cohorts have one audited source column.
    result["stage_source_disagreement_flag"] = 0

    result["stage_resolved_source_column"] = (
        stage_column
    )

    result["stage_source_level"] = "patient"

    result["stage_source_file"] = str(
        source_file
    )


    result = result[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "stage_original_value",
            "stage_major_resolved",
            "stage_label_4class",
            "label_stage_4class",
            "stage_label_binary",
            "label_stage_binary",
            "stage_multiple_values_flag",
            "stage_source_disagreement_flag",
            "stage_mapping_status",
            "stage_resolved_source_column",
            "stage_source_level",
            "stage_source_file",
        ]
    ]


    # --------------------------------------------------------
    # Cohort validation
    # --------------------------------------------------------

    duplicate_patient_rows = int(
        result.duplicated(
            subset=["COHORT", "PATIENT_ID"],
            keep=False,
        ).sum()
    )

    missing_patient_ids = int(
        result["PATIENT_ID"].isna().sum()
        +
        result["PATIENT_ID"].eq("").sum()
    )

    known_stage_count = int(
        result["label_stage_4class"]
        .notna()
        .sum()
    )

    unknown_stage_count = int(
        result["label_stage_4class"]
        .isna()
        .sum()
    )

    stage_i_count = int(
        result["stage_major_resolved"]
        .eq("I")
        .sum()
    )

    stage_ii_count = int(
        result["stage_major_resolved"]
        .eq("II")
        .sum()
    )

    stage_iii_count = int(
        result["stage_major_resolved"]
        .eq("III")
        .sum()
    )

    stage_iv_count = int(
        result["stage_major_resolved"]
        .eq("IV")
        .sum()
    )

    early_count = int(
        result["stage_label_binary"]
        .eq("Early")
        .sum()
    )

    advanced_count = int(
        result["stage_label_binary"]
        .eq("Advanced")
        .sum()
    )


    print(f"Patient rows: {len(result):,}")
    print(f"Known stage labels: {known_stage_count:,}")
    print(f"Unknown stage labels: {unknown_stage_count:,}")
    print(f"Stage I: {stage_i_count:,}")
    print(f"Stage II: {stage_ii_count:,}")
    print(f"Stage III: {stage_iii_count:,}")
    print(f"Stage IV: {stage_iv_count:,}")
    print(f"Early: {early_count:,}")
    print(f"Advanced: {advanced_count:,}")
    print(f"Duplicate patient rows: {duplicate_patient_rows}")
    print(f"Missing patient IDs: {missing_patient_ids}")


    stage_summary_rows.append(
        {
            "COHORT": cohort,
            "TOTAL_PATIENTS": len(result),
            "KNOWN_STAGE_COUNT": known_stage_count,
            "UNKNOWN_STAGE_COUNT": unknown_stage_count,
            "STAGE_I_COUNT": stage_i_count,
            "STAGE_II_COUNT": stage_ii_count,
            "STAGE_III_COUNT": stage_iii_count,
            "STAGE_IV_COUNT": stage_iv_count,
            "EARLY_COUNT": early_count,
            "ADVANCED_COUNT": advanced_count,
            "DUPLICATE_PATIENT_ROWS": duplicate_patient_rows,
        }
    )

    harmonised_stage_frames.append(result)


# ------------------------------------------------------------
# Combine patient-level stage cohorts in memory
# ------------------------------------------------------------

HARMONISED_PATIENT_STAGE_DF = pd.concat(
    harmonised_stage_frames,
    ignore_index=True,
)

PATIENT_STAGE_SUMMARY_DF = pd.DataFrame(
    stage_summary_rows
)


# ------------------------------------------------------------
# Combined validation
# ------------------------------------------------------------

duplicate_keys = (
    HARMONISED_PATIENT_STAGE_DF
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

invalid_stage_values = sorted(
    set(
        HARMONISED_PATIENT_STAGE_DF[
            "stage_major_resolved"
        ]
        .dropna()
        .astype(str)
    )
    - {"I", "II", "III", "IV", "Unknown"}
)

invalid_four_class_values = sorted(
    HARMONISED_PATIENT_STAGE_DF[
        "label_stage_4class"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

invalid_binary_values = sorted(
    HARMONISED_PATIENT_STAGE_DF[
        "label_stage_binary"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)


print("\n" + "=" * 78)
print("COMBINED PATIENT-STAGE VALIDATION")
print("=" * 78)

print(
    f"Combined patient rows: "
    f"{len(HARMONISED_PATIENT_STAGE_DF):,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_keys.sum())}"
)

print(
    f"Invalid resolved stages: "
    f"{invalid_stage_values}"
)

print(
    f"Observed four-class labels: "
    f"{invalid_four_class_values}"
)

print(
    f"Observed binary labels: "
    f"{invalid_binary_values}"
)


if duplicate_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were found."
    )

if invalid_stage_values:
    raise ValueError(
        f"Invalid resolved stage values found: "
        f"{invalid_stage_values}"
    )

if not set(invalid_four_class_values).issubset(
    {0, 1, 2, 3}
):
    raise ValueError(
        "Unexpected four-class stage encoding found."
    )

if not set(invalid_binary_values).issubset(
    {0, 1}
):
    raise ValueError(
        "Unexpected binary stage encoding found."
    )


print("\n" + "=" * 78)
print("STEP 7B-4 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("Stage fields remain target labels only.")
print("No model input table was created.")
print("No dataset split or model training was performed.")
```

## Step 7C-1 — Create One Combined Patient-Level Stage Table

This step combines harmonised stage labels from all seven cohorts.

For cohorts with patient-level stage, the existing patient record will be used.

For cohorts with sample-level stage:

- all samples belonging to the patient will be checked
- one patient-level stage will be created only when the resolved stages agree
- patients with no known stage will remain `Unknown`
- conflicting patient stages will trigger an error
- no representative sample will be selected

Stage labels are target columns only and will not be model input features.


```python
# ============================================================
# STEP 7C-1 — COMBINE STAGE LABELS AT PATIENT LEVEL
# No output file is created
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# 1. Required previous objects
# ------------------------------------------------------------

if "HARMONISED_PATIENT_STAGE_DF" not in globals():
    raise RuntimeError(
        "HARMONISED_PATIENT_STAGE_DF was not found. "
        "Please run Step 7B-4 first."
    )

if "RESOLVED_SAMPLE_STAGE_DF" not in globals():
    raise RuntimeError(
        "RESOLVED_SAMPLE_STAGE_DF was not found. "
        "Please run Step 7B-3 first."
    )

if "normalise_stage_value" not in globals():
    raise RuntimeError(
        "normalise_stage_value was not found. "
        "Please run Step 7B-1 first."
    )


EXPECTED_COHORTS = {
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
}


def join_unique_values(series):
    """
    Join unique non-missing values without changing their content.
    """

    values = []

    for value in series:

        if pd.isna(value):
            continue

        text = str(value).strip()

        if not text:
            continue

        if text.upper() in {
            "NA",
            "N/A",
            "NAN",
            "NONE",
            "NULL",
        }:
            continue

        if text not in values:
            values.append(text)

    return " | ".join(values) if values else pd.NA


print("=" * 78)
print("STEP 7C-1 — COMBINED PATIENT-LEVEL STAGE TABLE")
print("=" * 78)


# ------------------------------------------------------------
# 2. Prepare cohorts with patient-level stage
# ------------------------------------------------------------

patient_source_df = (
    HARMONISED_PATIENT_STAGE_DF.copy()
)

patient_stage_rows = pd.DataFrame({
    "COHORT": patient_source_df["COHORT"],
    "PATIENT_ID": patient_source_df["PATIENT_ID"],

    "stage_original_value_summary": (
        patient_source_df["stage_original_value"]
    ),

    "stage_major_resolved": (
        patient_source_df["stage_major_resolved"]
    ),

    "stage_label_4class": (
        patient_source_df["stage_label_4class"]
    ),

    "label_stage_4class": (
        patient_source_df["label_stage_4class"]
    ),

    "stage_label_binary": (
        patient_source_df["stage_label_binary"]
    ),

    "label_stage_binary": (
        patient_source_df["label_stage_binary"]
    ),

    "stage_source_level": "patient",

    "stage_source_column": (
        patient_source_df[
            "stage_resolved_source_column"
        ]
    ),

    "stage_source_file": (
        patient_source_df["stage_source_file"]
    ),

    "stage_source_record_count": 1,

    "stage_known_record_count": (
        patient_source_df[
            "stage_major_resolved"
        ]
        .ne("Unknown")
        .astype("int64")
    ),

    "stage_source_disagreement_flag": (
        patient_source_df[
            "stage_source_disagreement_flag"
        ]
        .fillna(0)
        .astype("int8")
    ),

    "stage_multiple_values_flag": (
        patient_source_df[
            "stage_multiple_values_flag"
        ]
        .fillna(0)
        .astype("int8")
    ),

    "stage_patient_conflict_flag": 0,
})


# ------------------------------------------------------------
# 3. Collapse sample-level stage to one row per patient
# ------------------------------------------------------------

sample_source_df = (
    RESOLVED_SAMPLE_STAGE_DF.copy()
)

sample_patient_rows = []
conflicting_patient_records = []


for (cohort, patient_id), patient_group in (
    sample_source_df.groupby(
        ["COHORT", "PATIENT_ID"],
        sort=False,
        dropna=False,
    )
):

    known_group = patient_group.loc[
        patient_group[
            "stage_major_resolved"
        ].ne("Unknown")
    ].copy()

    unique_known_stages = (
        known_group["stage_major_resolved"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )


    # --------------------------------------------------------
    # A patient must not receive conflicting stage labels.
    # --------------------------------------------------------

    patient_conflict_flag = int(
        len(unique_known_stages) > 1
    )

    if patient_conflict_flag == 1:

        conflicting_patient_records.append({
            "COHORT": cohort,
            "PATIENT_ID": patient_id,
            "RESOLVED_STAGES": " | ".join(
                sorted(unique_known_stages)
            ),
            "SAMPLE_IDS": join_unique_values(
                patient_group["SAMPLE_ID"]
            ),
        })

        continue


    # --------------------------------------------------------
    # Resolve one patient-level stage
    # --------------------------------------------------------

    if len(unique_known_stages) == 1:

        selected_major_stage = (
            unique_known_stages[0]
        )

        mapped_stage = normalise_stage_value(
            selected_major_stage
        )

    else:

        selected_major_stage = "Unknown"

        mapped_stage = normalise_stage_value(
            pd.NA
        )


    sample_patient_rows.append({
        "COHORT": cohort,
        "PATIENT_ID": patient_id,

        "stage_original_value_summary": (
            join_unique_values(
                pd.concat(
                    [
                        patient_group[
                            "stage_original_primary"
                        ],
                        patient_group[
                            "stage_original_secondary"
                        ],
                    ],
                    ignore_index=True,
                )
            )
        ),

        "stage_major_resolved": (
            mapped_stage["stage_major"]
        ),

        "stage_label_4class": (
            mapped_stage["stage_label_4class"]
        ),

        "label_stage_4class": (
            mapped_stage["label_stage_4class"]
        ),

        "stage_label_binary": (
            mapped_stage["stage_label_binary"]
        ),

        "label_stage_binary": (
            mapped_stage["label_stage_binary"]
        ),

        "stage_source_level": "sample_consensus",

        "stage_source_column": (
            join_unique_values(
                known_group[
                    "stage_resolved_source_column"
                ]
            )
            if not known_group.empty
            else pd.NA
        ),

        "stage_source_file": (
            join_unique_values(
                patient_group[
                    "stage_source_file"
                ]
            )
        ),

        "stage_source_record_count": (
            int(len(patient_group))
        ),

        "stage_known_record_count": (
            int(len(known_group))
        ),

        "stage_source_disagreement_flag": int(
            patient_group[
                "stage_source_disagreement_flag"
            ]
            .fillna(0)
            .astype(int)
            .max()
        ),

        "stage_multiple_values_flag": int(
            patient_group[
                "stage_multiple_values_flag"
            ]
            .fillna(0)
            .astype(int)
            .max()
        ),

        "stage_patient_conflict_flag": (
            patient_conflict_flag
        ),
    })


# ------------------------------------------------------------
# 4. Stop if true patient-stage conflicts exist
# ------------------------------------------------------------

if conflicting_patient_records:

    conflicting_patient_stage_df = pd.DataFrame(
        conflicting_patient_records
    )

    print("\nTrue patient-level stage conflicts found:")

    print(
        conflicting_patient_stage_df
        .head(20)
        .to_string(index=False)
    )

    raise ValueError(
        f"{len(conflicting_patient_stage_df)} patients "
        "contain conflicting resolved stage labels."
    )

else:

    conflicting_patient_stage_df = pd.DataFrame()


sample_patient_stage_df = pd.DataFrame(
    sample_patient_rows
)


# ------------------------------------------------------------
# 5. Combine all seven cohorts
# ------------------------------------------------------------

COMBINED_PATIENT_STAGE_DF = pd.concat(
    [
        patient_stage_rows,
        sample_patient_stage_df,
    ],
    ignore_index=True,
)


COMBINED_PATIENT_STAGE_DF[
    "label_stage_4class"
] = (
    COMBINED_PATIENT_STAGE_DF[
        "label_stage_4class"
    ]
    .astype("Int64")
)


COMBINED_PATIENT_STAGE_DF[
    "label_stage_binary"
] = (
    COMBINED_PATIENT_STAGE_DF[
        "label_stage_binary"
    ]
    .astype("Int64")
)


# ------------------------------------------------------------
# 6. Validation
# ------------------------------------------------------------

duplicate_patient_keys = (
    COMBINED_PATIENT_STAGE_DF
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

found_cohorts = set(
    COMBINED_PATIENT_STAGE_DF["COHORT"]
)

missing_cohorts = sorted(
    EXPECTED_COHORTS - found_cohorts
)

unexpected_cohorts = sorted(
    found_cohorts - EXPECTED_COHORTS
)

missing_patient_ids = int(
    COMBINED_PATIENT_STAGE_DF[
        "PATIENT_ID"
    ].isna().sum()
    +
    COMBINED_PATIENT_STAGE_DF[
        "PATIENT_ID"
    ].eq("").sum()
)

invalid_stage_values = sorted(
    set(
        COMBINED_PATIENT_STAGE_DF[
            "stage_major_resolved"
        ]
        .dropna()
        .astype(str)
    )
    - {"I", "II", "III", "IV", "Unknown"}
)

four_class_values = sorted(
    COMBINED_PATIENT_STAGE_DF[
        "label_stage_4class"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

binary_values = sorted(
    COMBINED_PATIENT_STAGE_DF[
        "label_stage_binary"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)


print("\n" + "=" * 78)
print("COMBINED STAGE VALIDATION")
print("=" * 78)

print(
    f"Combined patient rows: "
    f"{len(COMBINED_PATIENT_STAGE_DF):,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{int(duplicate_patient_keys.sum())}"
)

print(f"Missing patient IDs: {missing_patient_ids}")
print(f"Missing cohorts: {missing_cohorts}")
print(f"Unexpected cohorts: {unexpected_cohorts}")
print(f"Invalid stage values: {invalid_stage_values}")
print(f"Observed four-class labels: {four_class_values}")
print(f"Observed binary labels: {binary_values}")

print(
    f"Known stage labels: "
    f"{COMBINED_PATIENT_STAGE_DF['label_stage_4class'].notna().sum():,}"
)

print(
    f"Unknown stage labels: "
    f"{COMBINED_PATIENT_STAGE_DF['label_stage_4class'].isna().sum():,}"
)

print(
    f"Patient conflict flags: "
    f"{COMBINED_PATIENT_STAGE_DF['stage_patient_conflict_flag'].sum():,}"
)

print(
    f"Stage-source disagreement flags: "
    f"{COMBINED_PATIENT_STAGE_DF['stage_source_disagreement_flag'].sum():,}"
)


if duplicate_patient_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were found."
    )

if missing_patient_ids != 0:
    raise ValueError(
        "Missing PATIENT_ID values were found."
    )

if missing_cohorts:
    raise ValueError(
        f"Missing cohorts: {missing_cohorts}"
    )

if unexpected_cohorts:
    raise ValueError(
        f"Unexpected cohorts: {unexpected_cohorts}"
    )

if invalid_stage_values:
    raise ValueError(
        f"Invalid stage values found: "
        f"{invalid_stage_values}"
    )

if not set(four_class_values).issubset(
    {0, 1, 2, 3}
):
    raise ValueError(
        "Unexpected four-class stage labels found."
    )

if not set(binary_values).issubset(
    {0, 1}
):
    raise ValueError(
        "Unexpected binary stage labels found."
    )


# ------------------------------------------------------------
# 7. Concise cohort summary
# ------------------------------------------------------------

cohort_stage_summary = (
    COMBINED_PATIENT_STAGE_DF
    .groupby("COHORT", sort=False)
    .agg(
        total_patients=("PATIENT_ID", "size"),
        known_stage=(
            "label_stage_4class",
            lambda values: int(values.notna().sum())
        ),
        unknown_stage=(
            "label_stage_4class",
            lambda values: int(values.isna().sum())
        ),
        stage_source_disagreements=(
            "stage_source_disagreement_flag",
            "sum",
        ),
    )
    .reset_index()
)


print("\nCohort stage summary:")

print(
    cohort_stage_summary.to_string(
        index=False
    )
)


print("\n" + "=" * 78)
print("STEP 7C-1 FINISHED")
print("=" * 78)

print("No output file was created.")
print("No representative sample was selected.")
print("No dataset split or model training was performed.")
print("Stage-derived columns remain target labels only.")
```

## Step 8A-1 — Audit 96-Gene Mutation Sample Matching

This step checks whether mutation sample identifiers can be matched exactly to the original clinical-sample files.

Only mutations belonging to the selected 96-gene LUAD panel are considered.

The audit reports:

- selected-gene mutation rows
- unique mutation sample IDs
- exact clinical-sample matches
- unmatched mutation sample IDs
- patients with at least one matched selected-gene mutation

This step does not select a representative sample. Stage labels are not used during matching.

```python
# ============================================================
# STEP 8A-1 — AUDIT 96-GENE MUTATION SAMPLE MATCHING
# ============================================================

import pandas as pd


if "STEP_1A_RESULTS" not in globals():
    raise RuntimeError(
        "STEP_1A_RESULTS was not found. Run Step 1A first."
    )

if "OUTPUT_DIR" not in globals():
    raise RuntimeError(
        "OUTPUT_DIR was not found. Run the setup cell first."
    )


PANEL_FILE = (
    OUTPUT_DIR
    / "data"
    / "luad_96_driver_genes.csv"
)

if not PANEL_FILE.exists():
    raise FileNotFoundError(
        f"96-gene panel file not found:\n{PANEL_FILE}"
    )


# ------------------------------------------------------------
# 1. Load the validated 96-gene panel
# ------------------------------------------------------------

panel_df = pd.read_csv(
    PANEL_FILE,
    dtype=str,
)

if "GENE_SYMBOL" not in panel_df.columns:
    raise ValueError(
        "GENE_SYMBOL column is missing from the 96-gene panel."
    )

selected_genes = set(
    panel_df["GENE_SYMBOL"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.upper()
)

if len(selected_genes) != 96:
    raise ValueError(
        f"Expected 96 genes, found {len(selected_genes)}."
    )


EXPECTED_COHORTS = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
]


GENE_COLUMN_CANDIDATES = [
    "Hugo_Symbol",
    "HUGO_SYMBOL",
    "GENE_SYMBOL",
    "Gene",
]

MUTATION_SAMPLE_COLUMN_CANDIDATES = [
    "Tumor_Sample_Barcode",
    "TUMOR_SAMPLE_BARCODE",
    "SAMPLE_ID",
    "Sample_ID",
]


def get_single_source_path(cohort, source_type):
    """
    Return the single discovered source path for one cohort.
    """

    candidates = STEP_1A_RESULTS[
        "cohort_candidates"
    ][cohort][source_type]

    if len(candidates) != 1:
        raise ValueError(
            f"{cohort}: expected one {source_type} file, "
            f"found {len(candidates)}."
        )

    return candidates[0]


def find_column(columns, candidates, field_name, cohort):
    """
    Find the first exact matching column from a candidate list.
    """

    for candidate in candidates:
        if candidate in columns:
            return candidate

    raise ValueError(
        f"{cohort}: could not identify {field_name}. "
        f"Available columns include: {list(columns)[:20]}"
    )


print("=" * 78)
print("STEP 8A-1 — 96-GENE MUTATION SAMPLE MATCHING AUDIT")
print("=" * 78)

print(f"\nValidated panel genes: {len(selected_genes)}")


audit_rows = []
STEP_8A1_MUTATION_SAMPLE_AUDIT = {}


for cohort in EXPECTED_COHORTS:

    mutation_file = get_single_source_path(
        cohort,
        "mutation",
    )

    clinical_sample_file = get_single_source_path(
        cohort,
        "clinical_sample",
    )


    # --------------------------------------------------------
    # 2. Discover mutation columns
    # --------------------------------------------------------

    mutation_header = pd.read_csv(
        mutation_file,
        sep="\t",
        comment="#",
        nrows=0,
        low_memory=False,
    )

    gene_column = find_column(
        mutation_header.columns,
        GENE_COLUMN_CANDIDATES,
        "mutation gene column",
        cohort,
    )

    mutation_sample_column = find_column(
        mutation_header.columns,
        MUTATION_SAMPLE_COLUMN_CANDIDATES,
        "mutation sample column",
        cohort,
    )


    # --------------------------------------------------------
    # 3. Load clinical sample mapping
    # --------------------------------------------------------

    clinical_sample_df = pd.read_csv(
        clinical_sample_file,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", "SAMPLE_ID"],
        dtype=str,
        low_memory=False,
    )

    clinical_sample_df["PATIENT_ID"] = (
        clinical_sample_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    clinical_sample_df["SAMPLE_ID"] = (
        clinical_sample_df["SAMPLE_ID"]
        .astype("string")
        .str.strip()
    )

    clinical_sample_df = (
        clinical_sample_df
        .dropna(subset=["PATIENT_ID", "SAMPLE_ID"])
        .drop_duplicates(
            subset=["PATIENT_ID", "SAMPLE_ID"]
        )
    )


    # --------------------------------------------------------
    # 4. Load only gene and sample fields from mutation data
    # --------------------------------------------------------

    mutation_df = pd.read_csv(
        mutation_file,
        sep="\t",
        comment="#",
        usecols=[gene_column, mutation_sample_column],
        dtype=str,
        low_memory=False,
        on_bad_lines="skip",
    )

    mutation_df["GENE_SYMBOL_STANDARD"] = (
        mutation_df[gene_column]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    mutation_df["MUTATION_SAMPLE_ID"] = (
        mutation_df[mutation_sample_column]
        .astype("string")
        .str.strip()
    )


    panel_mutation_df = mutation_df.loc[
        mutation_df["GENE_SYMBOL_STANDARD"]
        .isin(selected_genes)
    ].copy()

    panel_mutation_df = panel_mutation_df.dropna(
        subset=["MUTATION_SAMPLE_ID"]
    )


    # --------------------------------------------------------
    # 5. Exact sample matching
    # --------------------------------------------------------

    mutation_sample_ids = set(
        panel_mutation_df["MUTATION_SAMPLE_ID"]
        .dropna()
        .astype(str)
    )

    clinical_sample_ids = set(
        clinical_sample_df["SAMPLE_ID"]
        .dropna()
        .astype(str)
    )

    matched_sample_ids = (
        mutation_sample_ids
        & clinical_sample_ids
    )

    unmatched_mutation_sample_ids = sorted(
        mutation_sample_ids
        - clinical_sample_ids
    )

    clinical_samples_without_panel_mutation = (
        clinical_sample_ids
        - mutation_sample_ids
    )


    matched_mapping_df = clinical_sample_df.loc[
        clinical_sample_df["SAMPLE_ID"]
        .isin(matched_sample_ids)
    ].copy()

    matched_patient_count = int(
        matched_mapping_df["PATIENT_ID"]
        .nunique(dropna=True)
    )


    # --------------------------------------------------------
    # 6. Print concise cohort audit
    # --------------------------------------------------------

    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")

    print(
        f"Clinical patients: "
        f"{clinical_sample_df['PATIENT_ID'].nunique():,}"
    )

    print(
        f"Clinical samples: "
        f"{clinical_sample_df['SAMPLE_ID'].nunique():,}"
    )

    print(
        f"96-gene mutation rows: "
        f"{len(panel_mutation_df):,}"
    )

    print(
        f"Unique mutation sample IDs: "
        f"{len(mutation_sample_ids):,}"
    )

    print(
        f"Exact matched sample IDs: "
        f"{len(matched_sample_ids):,}"
    )

    print(
        f"Unmatched mutation sample IDs: "
        f"{len(unmatched_mutation_sample_ids):,}"
    )

    print(
        f"Matched patients with panel mutations: "
        f"{matched_patient_count:,}"
    )

    if unmatched_mutation_sample_ids:
        print(
            "Unmatched examples:",
            unmatched_mutation_sample_ids[:5],
        )


    audit_record = {
        "COHORT": cohort,
        "GENE_COLUMN": gene_column,
        "MUTATION_SAMPLE_COLUMN": mutation_sample_column,
        "CLINICAL_PATIENTS": int(
            clinical_sample_df["PATIENT_ID"].nunique()
        ),
        "CLINICAL_SAMPLES": int(
            clinical_sample_df["SAMPLE_ID"].nunique()
        ),
        "PANEL_MUTATION_ROWS": len(panel_mutation_df),
        "UNIQUE_MUTATION_SAMPLE_IDS": len(
            mutation_sample_ids
        ),
        "EXACT_MATCHED_SAMPLE_IDS": len(
            matched_sample_ids
        ),
        "UNMATCHED_MUTATION_SAMPLE_IDS": len(
            unmatched_mutation_sample_ids
        ),
        "CLINICAL_SAMPLES_WITHOUT_PANEL_MUTATION": len(
            clinical_samples_without_panel_mutation
        ),
        "MATCHED_PATIENTS_WITH_PANEL_MUTATIONS": (
            matched_patient_count
        ),
    }

    audit_rows.append(audit_record)

    STEP_8A1_MUTATION_SAMPLE_AUDIT[cohort] = {
        **audit_record,
        "matched_sample_ids": matched_sample_ids,
        "unmatched_mutation_sample_ids": (
            unmatched_mutation_sample_ids
        ),
    }


# ------------------------------------------------------------
# 7. Combined summary
# ------------------------------------------------------------

mutation_sample_audit_df = pd.DataFrame(
    audit_rows
)


print("\n" + "=" * 78)
print("COMBINED MUTATION SAMPLE AUDIT")
print("=" * 78)

print(
    mutation_sample_audit_df[
        [
            "COHORT",
            "PANEL_MUTATION_ROWS",
            "UNIQUE_MUTATION_SAMPLE_IDS",
            "EXACT_MATCHED_SAMPLE_IDS",
            "UNMATCHED_MUTATION_SAMPLE_IDS",
            "MATCHED_PATIENTS_WITH_PANEL_MUTATIONS",
        ]
    ].to_string(index=False)
)


print("\n" + "=" * 78)
print("STEP 8A-1 FINISHED")
print("=" * 78)

print("Stage labels were not used for mutation-sample matching.")
print("Representative samples have not been selected yet.")
```

## Step 8A-2 — Build Mutation-Positive Sample Candidates Per Patient

This step creates the exact-matched mutation-positive sample candidates for representative-sample selection.

For every cohort and sample, it calculates:

- number of selected 96-gene mutation rows
- number of unique selected genes
- linked patient identifier

It then reports patients who have more than one mutation-positive sample.

Representative-sample selection will use only identifiers and a deterministic rule. Stage and clinical labels are not used.

```python
# ============================================================
# STEP 8A-2 — BUILD MUTATION-POSITIVE SAMPLE CANDIDATES
# ============================================================

import pandas as pd


required_objects = [
    "STEP_1A_RESULTS",
    "selected_genes",
    "get_single_source_path",
    "find_column",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required previous objects are missing: "
        + ", ".join(missing_objects)
        + ". Please run Step 8A-1 first."
    )


EXPECTED_COHORTS = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
]


GENE_COLUMN_CANDIDATES = [
    "Hugo_Symbol",
    "HUGO_SYMBOL",
    "GENE_SYMBOL",
    "Gene",
]

MUTATION_SAMPLE_COLUMN_CANDIDATES = [
    "Tumor_Sample_Barcode",
    "TUMOR_SAMPLE_BARCODE",
    "SAMPLE_ID",
    "Sample_ID",
]


print("=" * 78)
print("STEP 8A-2 — MUTATION-POSITIVE SAMPLE CANDIDATES")
print("=" * 78)


candidate_frames = []
summary_rows = []


for cohort in EXPECTED_COHORTS:

    mutation_file = get_single_source_path(
        cohort,
        "mutation",
    )

    clinical_sample_file = get_single_source_path(
        cohort,
        "clinical_sample",
    )


    # --------------------------------------------------------
    # Discover required mutation columns
    # --------------------------------------------------------

    mutation_header = pd.read_csv(
        mutation_file,
        sep="\t",
        comment="#",
        nrows=0,
        low_memory=False,
    )

    gene_column = find_column(
        mutation_header.columns,
        GENE_COLUMN_CANDIDATES,
        "mutation gene column",
        cohort,
    )

    mutation_sample_column = find_column(
        mutation_header.columns,
        MUTATION_SAMPLE_COLUMN_CANDIDATES,
        "mutation sample column",
        cohort,
    )


    # --------------------------------------------------------
    # Load selected-gene mutation rows
    # --------------------------------------------------------

    mutation_df = pd.read_csv(
        mutation_file,
        sep="\t",
        comment="#",
        usecols=[
            gene_column,
            mutation_sample_column,
        ],
        dtype=str,
        low_memory=False,
        on_bad_lines="skip",
    )

    mutation_df["GENE_SYMBOL"] = (
        mutation_df[gene_column]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    mutation_df["SAMPLE_ID"] = (
        mutation_df[mutation_sample_column]
        .astype("string")
        .str.strip()
    )

    mutation_df = mutation_df.loc[
        mutation_df["GENE_SYMBOL"].isin(
            selected_genes
        )
        & mutation_df["SAMPLE_ID"].notna()
        & mutation_df["SAMPLE_ID"].ne("")
    ].copy()


    # --------------------------------------------------------
    # Count panel mutations and unique genes per sample
    # --------------------------------------------------------

    sample_mutation_counts = (
        mutation_df
        .groupby("SAMPLE_ID", as_index=False)
        .agg(
            PANEL_MUTATION_ROW_COUNT=(
                "GENE_SYMBOL",
                "size",
            ),
            UNIQUE_PANEL_GENE_COUNT=(
                "GENE_SYMBOL",
                "nunique",
            ),
        )
    )


    # --------------------------------------------------------
    # Load exact clinical sample-to-patient mapping
    # --------------------------------------------------------

    clinical_df = pd.read_csv(
        clinical_sample_file,
        sep="\t",
        comment="#",
        usecols=["PATIENT_ID", "SAMPLE_ID"],
        dtype=str,
        low_memory=False,
    )

    clinical_df["PATIENT_ID"] = (
        clinical_df["PATIENT_ID"]
        .astype("string")
        .str.strip()
    )

    clinical_df["SAMPLE_ID"] = (
        clinical_df["SAMPLE_ID"]
        .astype("string")
        .str.strip()
    )

    clinical_df = clinical_df.drop_duplicates(
        subset=["PATIENT_ID", "SAMPLE_ID"]
    )


    # --------------------------------------------------------
    # Keep exact-matched mutation-positive samples
    # --------------------------------------------------------

    candidates = clinical_df.merge(
        sample_mutation_counts,
        on="SAMPLE_ID",
        how="inner",
        validate="one_to_one",
    )

    candidates.insert(
        0,
        "COHORT",
        cohort,
    )


    candidate_count_per_patient = (
        candidates
        .groupby(
            ["COHORT", "PATIENT_ID"]
        )["SAMPLE_ID"]
        .nunique()
    )

    multi_candidate_patients = (
        candidate_count_per_patient[
            candidate_count_per_patient > 1
        ]
    )


    candidates["MUTATION_POSITIVE_SAMPLE_COUNT"] = (
        candidates
        .groupby(
            ["COHORT", "PATIENT_ID"]
        )["SAMPLE_ID"]
        .transform("nunique")
        .astype(int)
    )


    duplicate_sample_keys = int(
        candidates.duplicated(
            subset=["COHORT", "SAMPLE_ID"],
            keep=False,
        ).sum()
    )


    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")

    print(
        f"Mutation-positive candidate samples: "
        f"{len(candidates):,}"
    )

    print(
        f"Patients with candidate samples: "
        f"{candidates['PATIENT_ID'].nunique():,}"
    )

    print(
        f"Patients with multiple candidate samples: "
        f"{len(multi_candidate_patients):,}"
    )

    print(
        f"Maximum candidate samples for one patient: "
        f"{int(candidate_count_per_patient.max())}"
    )

    print(
        f"Duplicate cohort-sample keys: "
        f"{duplicate_sample_keys}"
    )


    if len(multi_candidate_patients) > 0:

        example_patient_ids = (
            multi_candidate_patients
            .index
            .get_level_values("PATIENT_ID")
            .tolist()[:5]
        )

        examples = candidates.loc[
            candidates["PATIENT_ID"].isin(
                example_patient_ids
            ),
            [
                "PATIENT_ID",
                "SAMPLE_ID",
                "PANEL_MUTATION_ROW_COUNT",
                "UNIQUE_PANEL_GENE_COUNT",
                "MUTATION_POSITIVE_SAMPLE_COUNT",
            ],
        ].sort_values(
            ["PATIENT_ID", "SAMPLE_ID"]
        )

        print("\nMultiple-candidate examples:")

        print(
            examples.to_string(
                index=False
            )
        )


    summary_rows.append(
        {
            "COHORT": cohort,
            "CANDIDATE_SAMPLES": len(candidates),
            "PATIENTS_WITH_CANDIDATES": int(
                candidates["PATIENT_ID"].nunique()
            ),
            "PATIENTS_WITH_MULTIPLE_CANDIDATES": int(
                len(multi_candidate_patients)
            ),
            "MAX_CANDIDATES_PER_PATIENT": int(
                candidate_count_per_patient.max()
            ),
            "DUPLICATE_SAMPLE_KEYS": duplicate_sample_keys,
        }
    )

    candidate_frames.append(candidates)


# ------------------------------------------------------------
# Combined candidate table
# ------------------------------------------------------------

MATCHED_PANEL_SAMPLE_CANDIDATES_DF = pd.concat(
    candidate_frames,
    ignore_index=True,
)

MUTATION_SAMPLE_CANDIDATE_SUMMARY_DF = pd.DataFrame(
    summary_rows
)


duplicate_sample_keys = (
    MATCHED_PANEL_SAMPLE_CANDIDATES_DF
    .duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)


print("\n" + "=" * 78)
print("COMBINED CANDIDATE VALIDATION")
print("=" * 78)

print(
    f"Total candidate samples: "
    f"{len(MATCHED_PANEL_SAMPLE_CANDIDATES_DF):,}"
)

print(
    f"Total patients with candidates: "
    f"{MATCHED_PANEL_SAMPLE_CANDIDATES_DF[['COHORT', 'PATIENT_ID']].drop_duplicates().shape[0]:,}"
)

print(
    f"Duplicate cohort-sample keys: "
    f"{int(duplicate_sample_keys.sum())}"
)

print("\nCandidate summary:")

print(
    MUTATION_SAMPLE_CANDIDATE_SUMMARY_DF
    .to_string(index=False)
)


if duplicate_sample_keys.any():
    raise ValueError(
        "Duplicate COHORT + SAMPLE_ID candidate rows were found."
    )


print("\n" + "=" * 78)
print("STEP 8A-2 FINISHED")
print("=" * 78)

print("Candidate table is available as:")
print("MATCHED_PANEL_SAMPLE_CANDIDATES_DF")

print("\nStage and labels were not used.")
```

## Step 8B — Deterministic Representative Sample Selection

For each patient with one or more mutation-positive samples, one representative sample is selected using identifiers only.

Deterministic priority:

1. Earliest tumour sample number from `T01`, `T02`, `T03`, etc.
2. Lexicographically smallest complete `SAMPLE_ID` as the tie-breaker

The selection does not use stage, target labels, mutation count, number of genes, age, sex, smoking or outcomes.

```python
# ============================================================
# REMAINING STEP 1 OF 15
# LEAKAGE-SAFE REPRESENTATIVE SAMPLE SELECTION
# ============================================================

import pandas as pd


if "MATCHED_PANEL_SAMPLE_CANDIDATES_DF" not in globals():
    raise RuntimeError(
        "MATCHED_PANEL_SAMPLE_CANDIDATES_DF was not found. "
        "Please run Step 8A-2 first."
    )


print("=" * 78)
print("REMAINING STEP 1 OF 15 — REPRESENTATIVE SAMPLE SELECTION")
print("=" * 78)


# ------------------------------------------------------------
# 1. Validate candidate table
# ------------------------------------------------------------

required_columns = {
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "PANEL_MUTATION_ROW_COUNT",
    "UNIQUE_PANEL_GENE_COUNT",
    "MUTATION_POSITIVE_SAMPLE_COUNT",
}

missing_columns = sorted(
    required_columns
    - set(MATCHED_PANEL_SAMPLE_CANDIDATES_DF.columns)
)

if missing_columns:
    raise ValueError(
        f"Candidate table is missing columns: {missing_columns}"
    )


candidates = (
    MATCHED_PANEL_SAMPLE_CANDIDATES_DF
    .copy()
)


for column in ["COHORT", "PATIENT_ID", "SAMPLE_ID"]:
    candidates[column] = (
        candidates[column]
        .astype("string")
        .str.strip()
    )


missing_identifier_rows = (
    candidates[
        ["COHORT", "PATIENT_ID", "SAMPLE_ID"]
    ]
    .isna()
    .any(axis=1)
)

if missing_identifier_rows.any():
    raise ValueError(
        f"{int(missing_identifier_rows.sum())} candidate rows "
        "contain missing identifiers."
    )


duplicate_sample_keys = candidates.duplicated(
    subset=["COHORT", "SAMPLE_ID"],
    keep=False,
)

if duplicate_sample_keys.any():
    raise ValueError(
        "Duplicate COHORT + SAMPLE_ID candidate rows were found."
    )


# ------------------------------------------------------------
# 2. Parse tumour sample order from identifiers
#
# Examples:
# T01 -> 1
# T02 -> 2
# T10 -> 10
#
# Samples without a T-number receive a large fallback value.
# ------------------------------------------------------------

tumour_order_text = (
    candidates["SAMPLE_ID"]
    .str.extract(
        r"(?:^|-)T(\d+)(?:-|$)",
        expand=False,
    )
)

candidates["IDENTIFIER_TUMOUR_ORDER"] = (
    pd.to_numeric(
        tumour_order_text,
        errors="coerce",
    )
    .fillna(999999)
    .astype(int)
)

candidates["IDENTIFIER_TUMOUR_ORDER_FOUND"] = (
    tumour_order_text.notna()
    .astype("int8")
)


# ------------------------------------------------------------
# 3. Sort using identifiers only
#
# Mutation counts and clinical fields are intentionally absent
# from the selection sort columns.
# ------------------------------------------------------------

SELECTION_SORT_COLUMNS = [
    "COHORT",
    "PATIENT_ID",
    "IDENTIFIER_TUMOUR_ORDER",
    "SAMPLE_ID",
]

candidates = candidates.sort_values(
    by=SELECTION_SORT_COLUMNS,
    ascending=[True, True, True, True],
    kind="mergesort",
).reset_index(drop=True)


candidates["REPRESENTATIVE_SAMPLE_RANK"] = (
    candidates
    .groupby(
        ["COHORT", "PATIENT_ID"],
        sort=False,
    )
    .cumcount()
    .add(1)
    .astype(int)
)


candidates["REPRESENTATIVE_SAMPLE_FLAG"] = (
    candidates["REPRESENTATIVE_SAMPLE_RANK"]
    .eq(1)
    .astype("int8")
)


candidates["REPRESENTATIVE_SELECTION_RULE"] = (
    "Earliest parsed T-number, then smallest SAMPLE_ID"
)


# ------------------------------------------------------------
# 4. Select one representative sample per patient
# ------------------------------------------------------------

REPRESENTATIVE_SAMPLE_DF = (
    candidates.loc[
        candidates["REPRESENTATIVE_SAMPLE_FLAG"].eq(1)
    ]
    .copy()
    .reset_index(drop=True)
)


REPRESENTATIVE_SAMPLE_RANKING_DF = (
    candidates.copy()
)


# ------------------------------------------------------------
# 5. Validation
# ------------------------------------------------------------

candidate_patient_count = (
    candidates[
        ["COHORT", "PATIENT_ID"]
    ]
    .drop_duplicates()
    .shape[0]
)

selected_patient_count = (
    REPRESENTATIVE_SAMPLE_DF[
        ["COHORT", "PATIENT_ID"]
    ]
    .drop_duplicates()
    .shape[0]
)

duplicate_selected_patients = (
    REPRESENTATIVE_SAMPLE_DF
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

selected_sample_duplicates = (
    REPRESENTATIVE_SAMPLE_DF
    .duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)

selected_not_in_candidates = (
    ~REPRESENTATIVE_SAMPLE_DF.set_index(
        ["COHORT", "SAMPLE_ID"]
    ).index.isin(
        candidates.set_index(
            ["COHORT", "SAMPLE_ID"]
        ).index
    )
)


print(f"\nCandidate sample rows: {len(candidates):,}")
print(f"Candidate patients: {candidate_patient_count:,}")
print(f"Selected representative samples: {len(REPRESENTATIVE_SAMPLE_DF):,}")
print(f"Selected representative patients: {selected_patient_count:,}")

print(
    f"Duplicate selected patient keys: "
    f"{int(duplicate_selected_patients.sum())}"
)

print(
    f"Duplicate selected sample keys: "
    f"{int(selected_sample_duplicates.sum())}"
)

print(
    f"Selected samples absent from candidates: "
    f"{int(selected_not_in_candidates.sum())}"
)

print(
    f"Samples with parsed T-number: "
    f"{int(candidates['IDENTIFIER_TUMOUR_ORDER_FOUND'].sum()):,}"
)

print(
    f"Samples without parsed T-number: "
    f"{int(candidates['IDENTIFIER_TUMOUR_ORDER_FOUND'].eq(0).sum()):,}"
)


if selected_patient_count != candidate_patient_count:
    raise ValueError(
        "The number of selected patients does not match "
        "the number of candidate patients."
    )

if duplicate_selected_patients.any():
    raise ValueError(
        "More than one representative sample was selected "
        "for a patient."
    )

if selected_sample_duplicates.any():
    raise ValueError(
        "A representative sample was duplicated."
    )

if selected_not_in_candidates.any():
    raise ValueError(
        "A selected sample was not present in the candidate table."
    )


# ------------------------------------------------------------
# 6. Cohort summary
# ------------------------------------------------------------

representative_summary_df = (
    candidates
    .groupby("COHORT", sort=False)
    .agg(
        candidate_samples=("SAMPLE_ID", "size"),
        candidate_patients=("PATIENT_ID", "nunique"),
        patients_with_multiple_candidates=(
            "MUTATION_POSITIVE_SAMPLE_COUNT",
            lambda values: int(
                pd.Series(values)
                .gt(1)
                .groupby(
                    candidates.loc[values.index, "PATIENT_ID"]
                )
                .max()
                .sum()
            )
        ),
        selected_samples=(
            "REPRESENTATIVE_SAMPLE_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print("\nRepresentative-sample summary:")

print(
    representative_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 7. Show examples from patients with multiple candidates
# ------------------------------------------------------------

multi_patient_keys = (
    candidates.loc[
        candidates[
            "MUTATION_POSITIVE_SAMPLE_COUNT"
        ].gt(1),
        ["COHORT", "PATIENT_ID"],
    ]
    .drop_duplicates()
    .head(5)
)


if not multi_patient_keys.empty:

    example_rows = candidates.merge(
        multi_patient_keys,
        on=["COHORT", "PATIENT_ID"],
        how="inner",
    )

    print("\nSelection examples:")

    print(
        example_rows[
            [
                "COHORT",
                "PATIENT_ID",
                "SAMPLE_ID",
                "IDENTIFIER_TUMOUR_ORDER",
                "PANEL_MUTATION_ROW_COUNT",
                "UNIQUE_PANEL_GENE_COUNT",
                "REPRESENTATIVE_SAMPLE_RANK",
                "REPRESENTATIVE_SAMPLE_FLAG",
            ]
        ]
        .sort_values(
            [
                "COHORT",
                "PATIENT_ID",
                "REPRESENTATIVE_SAMPLE_RANK",
            ]
        )
        .to_string(index=False)
    )

print("\n" + "=" * 78)
print("STEP 8B FINISHED")
print("=" * 78)


print("Representative sample table:")
print("REPRESENTATIVE_SAMPLE_DF")

print("\nSelection used identifiers only.")
print("Stage, labels and mutation counts were not ranking criteria.")
```

## Step 8C — Merge Representative Samples with Harmonised Clinical Data

This step creates one clinical record for every selected representative sample.

It combines:

- representative `COHORT`, `PATIENT_ID` and `SAMPLE_ID`
- age
- sex
- active smoking
- passive smoking
- smoking pack-years
- harmonised stage targets

For `MSKCC_2023`, age is taken from the selected sample's `AGE_AT_DOS_BX` record.

Missing clinical covariates remain missing and will not cause patient exclusion. Stage labels are retained only as target columns.

```python
# ============================================================
# STEP 8C — MERGE REPRESENTATIVE SAMPLES
# WITH HARMONISED CLINICAL DATA
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# 1. Check required previous objects
# ------------------------------------------------------------

required_objects = [
    "REPRESENTATIVE_SAMPLE_DF",
    "patient_level_age_df",
    "MSKCC_2023_SAMPLE_LEVEL_AGE_DF",
    "patient_level_sex_df",
    "PATIENT_LEVEL_ACTIVE_SMOKING_DF",
    "PATIENT_LEVEL_PASSIVE_SMOKING_DF",
    "PATIENT_LEVEL_PACK_YEARS_DF",
    "COMBINED_PATIENT_STAGE_DF",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required objects are missing: "
        + ", ".join(missing_objects)
    )


print("=" * 78)
print("STEP 8C — REPRESENTATIVE SAMPLE CLINICAL MERGE")
print("=" * 78)


# ------------------------------------------------------------
# 2. Prepare representative sample keys
# ------------------------------------------------------------

selected_samples = (
    REPRESENTATIVE_SAMPLE_DF[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "PANEL_MUTATION_ROW_COUNT",
            "UNIQUE_PANEL_GENE_COUNT",
            "MUTATION_POSITIVE_SAMPLE_COUNT",
            "IDENTIFIER_TUMOUR_ORDER",
            "IDENTIFIER_TUMOUR_ORDER_FOUND",
            "REPRESENTATIVE_SELECTION_RULE",
        ]
    ]
    .copy()
)


for column in ["COHORT", "PATIENT_ID", "SAMPLE_ID"]:
    selected_samples[column] = (
        selected_samples[column]
        .astype("string")
        .str.strip()
    )


duplicate_selected_patients = (
    selected_samples.duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

duplicate_selected_samples = (
    selected_samples.duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)

if duplicate_selected_patients.any():
    raise ValueError(
        "Duplicate representative patient keys were found."
    )

if duplicate_selected_samples.any():
    raise ValueError(
        "Duplicate representative sample keys were found."
    )


# ------------------------------------------------------------
# 3. Build one unified selected-sample age table
# ------------------------------------------------------------

age_columns = [
    "age_original_value",
    "age_numeric",
    "age_for_model",
    "age_missing_flag",
    "age_outlier_flag",
    "age_unusable_flag",
    "age_source_column",
    "age_source_file",
]


# Six cohorts with patient-level age
six_cohort_selected = selected_samples.loc[
    selected_samples["COHORT"].ne("MSKCC_2023"),
    ["COHORT", "PATIENT_ID", "SAMPLE_ID"],
].copy()


patient_age_source = (
    patient_level_age_df[
        [
            "COHORT",
            "PATIENT_ID",
            *age_columns,
        ]
    ]
    .copy()
)


selected_patient_age = six_cohort_selected.merge(
    patient_age_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


# MSKCC 2023 with sample-level age
msk_selected = selected_samples.loc[
    selected_samples["COHORT"].eq("MSKCC_2023"),
    ["COHORT", "PATIENT_ID", "SAMPLE_ID"],
].copy()


msk_age_source = (
    MSKCC_2023_SAMPLE_LEVEL_AGE_DF[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            *age_columns,
        ]
    ]
    .copy()
)


selected_msk_age = msk_selected.merge(
    msk_age_source,
    on=["COHORT", "PATIENT_ID", "SAMPLE_ID"],
    how="left",
    validate="one_to_one",
)


selected_age_df = pd.concat(
    [
        selected_patient_age,
        selected_msk_age,
    ],
    ignore_index=True,
)


duplicate_age_keys = selected_age_df.duplicated(
    subset=["COHORT", "PATIENT_ID"],
    keep=False,
)

if duplicate_age_keys.any():
    raise ValueError(
        "Duplicate patient keys were created in the selected age table."
    )


# ------------------------------------------------------------
# 4. Prepare sex table
# ------------------------------------------------------------

sex_columns = [
    "sex_original_value",
    "SEX",
    "sex_encoded",
    "sex_missing_flag",
    "sex_source_column",
    "sex_source_file",
]

sex_source = (
    patient_level_sex_df[
        [
            "COHORT",
            "PATIENT_ID",
            *sex_columns,
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 5. Prepare active-smoking table
# ------------------------------------------------------------

active_smoking_columns = [
    "smoking_original_value",
    "smoking_status_standard",
    "smoking_binary",
    "smoking_source_available_flag",
    "smoking_missing_flag",
    "smoking_inconsistency_flag",
    "smoking_unusable_flag",
    "smoking_source_column",
    "smoking_source_file",
]

active_smoking_source = (
    PATIENT_LEVEL_ACTIVE_SMOKING_DF[
        [
            "COHORT",
            "PATIENT_ID",
            *active_smoking_columns,
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 6. Prepare passive-smoking table
# ------------------------------------------------------------

passive_smoking_columns = [
    "passive_smoking_original_value",
    "passive_smoking_standard",
    "passive_smoking_binary",
    "passive_smoking_source_available_flag",
    "passive_smoking_missing_flag",
    "passive_smoking_unusable_flag",
    "passive_smoking_source_column",
    "passive_smoking_source_file",
]

passive_smoking_source = (
    PATIENT_LEVEL_PASSIVE_SMOKING_DF[
        [
            "COHORT",
            "PATIENT_ID",
            *passive_smoking_columns,
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 7. Prepare pack-year table
# ------------------------------------------------------------

pack_year_columns = [
    "pack_years_original_value",
    "pack_years_numeric",
    "pack_years_for_model",
    "pack_years_source_available_flag",
    "pack_years_missing_flag",
    "pack_years_non_numeric_flag",
    "pack_years_negative_flag",
    "pack_years_outlier_flag",
    "pack_years_inconsistency_flag",
    "pack_years_unusable_flag",
    "pack_years_source_column",
    "pack_years_source_file",
]

pack_year_source = (
    PATIENT_LEVEL_PACK_YEARS_DF[
        [
            "COHORT",
            "PATIENT_ID",
            *pack_year_columns,
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 8. Prepare patient-level stage target table
# ------------------------------------------------------------

stage_columns = [
    "stage_original_value_summary",
    "stage_major_resolved",
    "stage_label_4class",
    "label_stage_4class",
    "stage_label_binary",
    "label_stage_binary",
    "stage_source_level",
    "stage_source_column",
    "stage_source_file",
    "stage_source_record_count",
    "stage_known_record_count",
    "stage_source_disagreement_flag",
    "stage_multiple_values_flag",
    "stage_patient_conflict_flag",
]

stage_source = (
    COMBINED_PATIENT_STAGE_DF[
        [
            "COHORT",
            "PATIENT_ID",
            *stage_columns,
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 9. Merge all harmonised clinical components
# ------------------------------------------------------------

clinical_df = selected_samples.merge(
    selected_age_df,
    on=["COHORT", "PATIENT_ID", "SAMPLE_ID"],
    how="left",
    validate="one_to_one",
)


clinical_df = clinical_df.merge(
    sex_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


clinical_df = clinical_df.merge(
    active_smoking_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


clinical_df = clinical_df.merge(
    passive_smoking_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


clinical_df = clinical_df.merge(
    pack_year_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


clinical_df = clinical_df.merge(
    stage_source,
    on=["COHORT", "PATIENT_ID"],
    how="left",
    validate="one_to_one",
)


# ------------------------------------------------------------
# 10. Add source-record and target availability flags
# ------------------------------------------------------------

clinical_df["age_record_found_flag"] = (
    clinical_df["age_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["sex_record_found_flag"] = (
    clinical_df["sex_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["active_smoking_record_found_flag"] = (
    clinical_df["smoking_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["passive_smoking_record_found_flag"] = (
    clinical_df["passive_smoking_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["pack_year_record_found_flag"] = (
    clinical_df["pack_years_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["stage_record_found_flag"] = (
    clinical_df["stage_source_file"]
    .notna()
    .astype("int8")
)

clinical_df["stage_label_available_flag"] = (
    clinical_df["label_stage_4class"]
    .notna()
    .astype("int8")
)


# ------------------------------------------------------------
# 11. Ensure target encodings use nullable integers
# ------------------------------------------------------------

nullable_integer_columns = [
    "sex_encoded",
    "smoking_binary",
    "passive_smoking_binary",
    "label_stage_4class",
    "label_stage_binary",
]

for column in nullable_integer_columns:
    clinical_df[column] = (
        clinical_df[column]
        .astype("Int64")
    )


# ------------------------------------------------------------
# 12. Final validation
# ------------------------------------------------------------

SELECTED_SAMPLE_CLINICAL_BASE_DF = (
    clinical_df.copy()
)


duplicate_patient_keys = (
    SELECTED_SAMPLE_CLINICAL_BASE_DF
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

duplicate_sample_keys = (
    SELECTED_SAMPLE_CLINICAL_BASE_DF
    .duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)

missing_identifiers = (
    SELECTED_SAMPLE_CLINICAL_BASE_DF[
        ["COHORT", "PATIENT_ID", "SAMPLE_ID"]
    ]
    .isna()
    .any(axis=1)
)

expected_rows = len(
    REPRESENTATIVE_SAMPLE_DF
)

actual_rows = len(
    SELECTED_SAMPLE_CLINICAL_BASE_DF
)


print(f"\nExpected selected patients: {expected_rows:,}")
print(f"Merged clinical rows: {actual_rows:,}")

print(
    f"Duplicate cohort-patient keys: "
    f"{int(duplicate_patient_keys.sum())}"
)

print(
    f"Duplicate cohort-sample keys: "
    f"{int(duplicate_sample_keys.sum())}"
)

print(
    f"Rows with missing identifiers: "
    f"{int(missing_identifiers.sum())}"
)

print(
    f"Age source records found: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['age_record_found_flag'].sum():,}"
)

print(
    f"Sex source records found: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['sex_record_found_flag'].sum():,}"
)

print(
    f"Known sex values: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['sex_encoded'].notna().sum():,}"
)

print(
    f"Usable active-smoking values: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['smoking_binary'].notna().sum():,}"
)

print(
    f"Usable passive-smoking values: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['passive_smoking_binary'].notna().sum():,}"
)

print(
    f"Usable pack-year values: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['pack_years_for_model'].notna().sum():,}"
)

print(
    f"Known stage targets: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['stage_label_available_flag'].sum():,}"
)

print(
    f"Unknown stage targets: "
    f"{SELECTED_SAMPLE_CLINICAL_BASE_DF['stage_label_available_flag'].eq(0).sum():,}"
)


if actual_rows != expected_rows:
    raise ValueError(
        "Clinical merge changed the number of selected patients."
    )

if duplicate_patient_keys.any():
    raise ValueError(
        "Duplicate COHORT + PATIENT_ID rows were created."
    )

if duplicate_sample_keys.any():
    raise ValueError(
        "Duplicate COHORT + SAMPLE_ID rows were created."
    )

if missing_identifiers.any():
    raise ValueError(
        "Missing identifiers were found after the clinical merge."
    )


# ------------------------------------------------------------
# 13. Concise cohort-level coverage summary
# ------------------------------------------------------------

clinical_coverage_summary_df = (
    SELECTED_SAMPLE_CLINICAL_BASE_DF
    .groupby("COHORT", sort=False)
    .agg(
        selected_patients=("PATIENT_ID", "size"),
        age_available=(
            "age_for_model",
            lambda values: int(values.notna().sum())
        ),
        sex_available=(
            "sex_encoded",
            lambda values: int(values.notna().sum())
        ),
        active_smoking_available=(
            "smoking_binary",
            lambda values: int(values.notna().sum())
        ),
        pack_years_available=(
            "pack_years_for_model",
            lambda values: int(values.notna().sum())
        ),
        stage_available=(
            "label_stage_4class",
            lambda values: int(values.notna().sum())
        ),
    )
    .reset_index()
)


print("\nClinical coverage by cohort:")

print(
    clinical_coverage_summary_df.to_string(
        index=False
    )
)


print("\n" + "=" * 78)
print("STEP 8C FINISHED")
print("=" * 78)

print("Merged clinical table:")
print("SELECTED_SAMPLE_CLINICAL_BASE_DF")

print("\nRepresentative sample selection remains identifier-only.")
print("Unknown clinical values were not imputed.")
print("Stage columns remain target-only columns.")
```

## Step 9A — Build the Selected-Sample Mutation Base Table

This step reloads mutation annotations directly from the seven original mutation files and keeps only:

- the selected 96 LUAD genes
- the deterministic representative sample selected for each patient

For every retained mutation, it preserves the original:

- gene symbol
- sample and patient identifiers
- transcript ID
- HGVSc
- reference and tumour alleles
- variant classification and type
- genomic coordinates
- source mutation file and parsed source-row number

A unique mutation-row identifier is generated for later reconstruction and exclusion auditing.

Fresh 301-bp windows are not constructed in this step.

```python
# ============================================================
# STEP 9A — BUILD SELECTED-SAMPLE MUTATION BASE TABLE
# ============================================================

import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. Validate required previous objects
# ------------------------------------------------------------

required_objects = [
    "REPRESENTATIVE_SAMPLE_DF",
    "STEP_1A_RESULTS",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required previous objects are missing: "
        + ", ".join(missing_objects)
    )


# Load the validated 96-gene panel when needed.
if "selected_genes" not in globals():

    panel_file = (
        OUTPUT_DIR
        / "data"
        / "luad_96_driver_genes.csv"
    )

    if not panel_file.exists():
        raise FileNotFoundError(
            f"96-gene panel file was not found:\n{panel_file}"
        )

    panel_df = pd.read_csv(
        panel_file,
        dtype=str,
    )

    selected_genes = set(
        panel_df["GENE_SYMBOL"]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
    )


if len(selected_genes) != 96:
    raise ValueError(
        f"Expected 96 selected genes, found {len(selected_genes)}."
    )


EXPECTED_COHORTS = [
    "TCGA_Atlas_2018",
    "LUNG_MSK_2017",
    "MSK_NPJPO_2021",
    "MSKCC_2020",
    "MSKCC_2023",
    "NCI_2022",
    "ONCOSG_2020",
]


# ------------------------------------------------------------
# 2. Source-column candidates
# ------------------------------------------------------------

FIELD_CANDIDATES = {
    "gene": [
        "Hugo_Symbol",
        "HUGO_SYMBOL",
        "GENE_SYMBOL",
        "Gene",
    ],

    "sample": [
        "Tumor_Sample_Barcode",
        "TUMOR_SAMPLE_BARCODE",
        "SAMPLE_ID",
        "Sample_ID",
    ],

    "hgvsc": [
        "HGVSc",
        "HGVSC",
        "HGVSc_Short",
    ],

    "transcript": [
        "Transcript_ID",
        "TRANSCRIPT_ID",
        "Transcript",
        "Feature",
    ],

    "reference_allele": [
        "Reference_Allele",
        "REFERENCE_ALLELE",
        "Ref",
        "REF",
    ],

    "tumour_allele": [
        "Tumor_Seq_Allele2",
        "TUMOR_SEQ_ALLELE2",
        "Tumor_Allele",
        "ALT",
        "Alt",
    ],

    "variant_classification": [
        "Variant_Classification",
        "VARIANT_CLASSIFICATION",
    ],

    "variant_type": [
        "Variant_Type",
        "VARIANT_TYPE",
    ],

    "chromosome": [
        "Chromosome",
        "CHROMOSOME",
        "Chr",
    ],

    "start_position": [
        "Start_Position",
        "START_POSITION",
        "Start",
    ],

    "end_position": [
        "End_Position",
        "END_POSITION",
        "End",
    ],

    "hgvsp_short": [
        "HGVSp_Short",
        "HGVSP_SHORT",
        "Protein_Change",
    ],
}


def find_first_column(columns, candidates):
    """
    Return the first exact column match, otherwise None.
    """

    for candidate in candidates:
        if candidate in columns:
            return candidate

    return None


def get_mutation_file(cohort):
    """
    Return the single discovered original mutation file.
    """

    candidates = STEP_1A_RESULTS[
        "cohort_candidates"
    ][cohort]["mutation"]

    if len(candidates) != 1:
        raise ValueError(
            f"{cohort}: expected one mutation file, "
            f"found {len(candidates)}."
        )

    return candidates[0]


def get_optional_series(frame, column_name):
    """
    Return a cleaned string series or an all-missing series.
    """

    if column_name is None:
        return pd.Series(
            pd.NA,
            index=frame.index,
            dtype="string",
        )

    return (
        frame[column_name]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# 3. Prepare representative sample-to-patient mapping
# ------------------------------------------------------------

representative_keys = (
    REPRESENTATIVE_SAMPLE_DF[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ]
    ]
    .copy()
)


for column in [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
]:
    representative_keys[column] = (
        representative_keys[column]
        .astype("string")
        .str.strip()
    )


duplicate_patient_keys = (
    representative_keys
    .duplicated(
        subset=["COHORT", "PATIENT_ID"],
        keep=False,
    )
)

duplicate_sample_keys = (
    representative_keys
    .duplicated(
        subset=["COHORT", "SAMPLE_ID"],
        keep=False,
    )
)

if duplicate_patient_keys.any():
    raise ValueError(
        "Representative table contains duplicate patient keys."
    )

if duplicate_sample_keys.any():
    raise ValueError(
        "Representative table contains duplicate sample keys."
    )


# ------------------------------------------------------------
# 4. Reload and filter original mutation files
# ------------------------------------------------------------

print("=" * 78)
print("STEP 9A — SELECTED-SAMPLE MUTATION BASE TABLE")
print("=" * 78)


mutation_frames = []
source_audit_rows = []

CHUNK_SIZE = 100_000


for cohort in EXPECTED_COHORTS:

    mutation_file = get_mutation_file(
        cohort
    )

    cohort_selected = (
        representative_keys.loc[
            representative_keys["COHORT"].eq(cohort)
        ]
        .copy()
    )

    selected_sample_ids = set(
        cohort_selected["SAMPLE_ID"]
        .dropna()
        .astype(str)
    )

    sample_to_patient = dict(
        zip(
            cohort_selected["SAMPLE_ID"],
            cohort_selected["PATIENT_ID"],
        )
    )


    # --------------------------------------------------------
    # Inspect mutation header and resolve source columns
    # --------------------------------------------------------

    mutation_header = pd.read_csv(
        mutation_file,
        sep="\t",
        comment="#",
        nrows=0,
        low_memory=False,
    )

    source_columns = {
        field: find_first_column(
            mutation_header.columns,
            candidates,
        )
        for field, candidates in FIELD_CANDIDATES.items()
    }


    gene_column = source_columns["gene"]
    sample_column = source_columns["sample"]


    if gene_column is None:
        raise ValueError(
            f"{cohort}: mutation gene column was not found."
        )

    if sample_column is None:
        raise ValueError(
            f"{cohort}: mutation sample column was not found."
        )


    use_columns = list(
        dict.fromkeys(
            column
            for column in source_columns.values()
            if column is not None
        )
    )


    cohort_filtered_chunks = []

    total_parsed_rows = 0
    total_panel_rows = 0
    selected_panel_rows = 0


    # --------------------------------------------------------
    # Read source mutation file in memory-safe chunks
    # --------------------------------------------------------

    try:
        mutation_reader = pd.read_csv(
            mutation_file,
            sep="\t",
            comment="#",
            usecols=use_columns,
            dtype=str,
            low_memory=False,
            on_bad_lines="skip",
            chunksize=CHUNK_SIZE,
        )

    except Exception as error:
        raise RuntimeError(
            f"{cohort} mutation loading failed: "
            f"{type(error).__name__}: {error}"
        )


    for chunk in mutation_reader:

        total_parsed_rows += len(chunk)

        standard_gene = (
            chunk[gene_column]
            .astype("string")
            .str.strip()
            .str.upper()
        )

        standard_sample = (
            chunk[sample_column]
            .astype("string")
            .str.strip()
        )


        panel_mask = (
            standard_gene
            .isin(selected_genes)
            .fillna(False)
        )

        selected_sample_mask = (
            standard_sample
            .isin(selected_sample_ids)
            .fillna(False)
        )

        total_panel_rows += int(
            panel_mask.sum()
        )


        keep_mask = (
            panel_mask
            & selected_sample_mask
        )

        if not keep_mask.any():
            continue


        retained = chunk.loc[
            keep_mask
        ].copy()

        retained_gene = standard_gene.loc[
            keep_mask
        ]

        retained_sample = standard_sample.loc[
            keep_mask
        ]

        selected_panel_rows += len(
            retained
        )


        canonical = pd.DataFrame(
            index=retained.index
        )

        canonical["COHORT"] = cohort

        canonical["SAMPLE_ID"] = (
            retained_sample
        )

        canonical["PATIENT_ID"] = (
            retained_sample
            .map(sample_to_patient)
            .astype("string")
        )

        canonical["GENE_SYMBOL"] = (
            retained_gene
        )

        canonical["HGVSC_RAW"] = (
            get_optional_series(
                retained,
                source_columns["hgvsc"],
            )
        )

        canonical["TRANSCRIPT_ID_RAW"] = (
            get_optional_series(
                retained,
                source_columns["transcript"],
            )
        )

        canonical["REFERENCE_ALLELE_RAW"] = (
            get_optional_series(
                retained,
                source_columns["reference_allele"],
            )
        )

        canonical["TUMOUR_ALLELE_RAW"] = (
            get_optional_series(
                retained,
                source_columns["tumour_allele"],
            )
        )

        canonical["VARIANT_CLASSIFICATION_RAW"] = (
            get_optional_series(
                retained,
                source_columns[
                    "variant_classification"
                ],
            )
        )

        canonical["VARIANT_TYPE_RAW"] = (
            get_optional_series(
                retained,
                source_columns["variant_type"],
            )
        )

        canonical["CHROMOSOME_RAW"] = (
            get_optional_series(
                retained,
                source_columns["chromosome"],
            )
        )

        canonical["START_POSITION_RAW"] = (
            get_optional_series(
                retained,
                source_columns["start_position"],
            )
        )

        canonical["END_POSITION_RAW"] = (
            get_optional_series(
                retained,
                source_columns["end_position"],
            )
        )

        canonical["HGVSP_SHORT_RAW"] = (
            get_optional_series(
                retained,
                source_columns["hgvsp_short"],
            )
        )


        # Parsed data-row number within the mutation table.
        canonical["SOURCE_PARSED_ROW_NUMBER"] = (
            retained.index
            .to_series()
            .astype("int64")
            .add(1)
            .values
        )

        canonical["SOURCE_MUTATION_FILE"] = str(
            mutation_file
        )

        canonical["MUTATION_ROW_ID"] = (
            cohort
            + "__MUT__"
            + canonical[
                "SOURCE_PARSED_ROW_NUMBER"
            ]
            .astype(str)
            .str.zfill(9)
        )


        canonical = canonical[
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "SAMPLE_ID",
                "PATIENT_ID",
                "GENE_SYMBOL",
                "TRANSCRIPT_ID_RAW",
                "HGVSC_RAW",
                "REFERENCE_ALLELE_RAW",
                "TUMOUR_ALLELE_RAW",
                "VARIANT_CLASSIFICATION_RAW",
                "VARIANT_TYPE_RAW",
                "CHROMOSOME_RAW",
                "START_POSITION_RAW",
                "END_POSITION_RAW",
                "HGVSP_SHORT_RAW",
                "SOURCE_PARSED_ROW_NUMBER",
                "SOURCE_MUTATION_FILE",
            ]
        ]


        cohort_filtered_chunks.append(
            canonical
        )


    # --------------------------------------------------------
    # Combine the cohort's retained mutation rows
    # --------------------------------------------------------

    if cohort_filtered_chunks:

        cohort_mutations = pd.concat(
            cohort_filtered_chunks,
            ignore_index=True,
        )

    else:

        cohort_mutations = pd.DataFrame(
            columns=[
                "MUTATION_ROW_ID",
                "COHORT",
                "SAMPLE_ID",
                "PATIENT_ID",
                "GENE_SYMBOL",
                "TRANSCRIPT_ID_RAW",
                "HGVSC_RAW",
                "REFERENCE_ALLELE_RAW",
                "TUMOUR_ALLELE_RAW",
                "VARIANT_CLASSIFICATION_RAW",
                "VARIANT_TYPE_RAW",
                "CHROMOSOME_RAW",
                "START_POSITION_RAW",
                "END_POSITION_RAW",
                "HGVSP_SHORT_RAW",
                "SOURCE_PARSED_ROW_NUMBER",
                "SOURCE_MUTATION_FILE",
            ]
        )


    represented_samples = int(
        cohort_mutations["SAMPLE_ID"]
        .nunique(dropna=True)
    )

    represented_patients = int(
        cohort_mutations["PATIENT_ID"]
        .nunique(dropna=True)
    )

    expected_samples = len(
        selected_sample_ids
    )

    missing_selected_samples = sorted(
        selected_sample_ids
        - set(
            cohort_mutations["SAMPLE_ID"]
            .dropna()
            .astype(str)
        )
    )


    print("\n" + "-" * 78)
    print(f"COHORT: {cohort}")
    print(f"Parsed source rows: {total_parsed_rows:,}")
    print(f"96-gene rows in all samples: {total_panel_rows:,}")
    print(f"Selected-sample mutation rows: {len(cohort_mutations):,}")
    print(f"Expected representative samples: {expected_samples:,}")
    print(f"Represented samples: {represented_samples:,}")
    print(f"Represented patients: {represented_patients:,}")
    print(f"Missing representative samples: {len(missing_selected_samples):,}")

    print(
        f"HGVSc available: "
        f"{cohort_mutations['HGVSC_RAW'].notna().sum():,}"
    )

    print(
        f"Transcript ID available: "
        f"{cohort_mutations['TRANSCRIPT_ID_RAW'].notna().sum():,}"
    )

    print(
        f"Reference allele available: "
        f"{cohort_mutations['REFERENCE_ALLELE_RAW'].notna().sum():,}"
    )

    print(
        f"Tumour allele available: "
        f"{cohort_mutations['TUMOUR_ALLELE_RAW'].notna().sum():,}"
    )


    if missing_selected_samples:
        print(
            "Missing sample examples:",
            missing_selected_samples[:5],
        )


    source_audit_rows.append({
        "COHORT": cohort,
        "SOURCE_MUTATION_FILE": str(mutation_file),
        "GENE_SOURCE_COLUMN": gene_column,
        "SAMPLE_SOURCE_COLUMN": sample_column,
        "HGVSC_SOURCE_COLUMN": (
            source_columns["hgvsc"]
            if source_columns["hgvsc"] is not None
            else "Not available"
        ),
        "TRANSCRIPT_SOURCE_COLUMN": (
            source_columns["transcript"]
            if source_columns["transcript"] is not None
            else "Not available"
        ),
        "REFERENCE_ALLELE_SOURCE_COLUMN": (
            source_columns["reference_allele"]
            if source_columns["reference_allele"] is not None
            else "Not available"
        ),
        "TUMOUR_ALLELE_SOURCE_COLUMN": (
            source_columns["tumour_allele"]
            if source_columns["tumour_allele"] is not None
            else "Not available"
        ),
        "PARSED_SOURCE_ROWS": total_parsed_rows,
        "PANEL_ROWS_ALL_SAMPLES": total_panel_rows,
        "SELECTED_SAMPLE_MUTATION_ROWS": len(
            cohort_mutations
        ),
        "EXPECTED_SELECTED_SAMPLES": expected_samples,
        "REPRESENTED_SELECTED_SAMPLES": represented_samples,
        "REPRESENTED_PATIENTS": represented_patients,
        "MISSING_SELECTED_SAMPLES": len(
            missing_selected_samples
        ),
        "HGVSC_AVAILABLE_ROWS": int(
            cohort_mutations["HGVSC_RAW"]
            .notna()
            .sum()
        ),
        "TRANSCRIPT_AVAILABLE_ROWS": int(
            cohort_mutations["TRANSCRIPT_ID_RAW"]
            .notna()
            .sum()
        ),
        "REFERENCE_ALLELE_AVAILABLE_ROWS": int(
            cohort_mutations["REFERENCE_ALLELE_RAW"]
            .notna()
            .sum()
        ),
        "TUMOUR_ALLELE_AVAILABLE_ROWS": int(
            cohort_mutations["TUMOUR_ALLELE_RAW"]
            .notna()
            .sum()
        ),
    })


    mutation_frames.append(
        cohort_mutations
    )


# ------------------------------------------------------------
# 5. Combine all seven cohorts
# ------------------------------------------------------------

SELECTED_SAMPLE_MUTATION_BASE_DF = pd.concat(
    mutation_frames,
    ignore_index=True,
)

SELECTED_MUTATION_SOURCE_AUDIT_DF = pd.DataFrame(
    source_audit_rows
)


# ------------------------------------------------------------
# 6. Combined validation
# ------------------------------------------------------------

duplicate_mutation_ids = (
    SELECTED_SAMPLE_MUTATION_BASE_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)

missing_identifier_rows = (
    SELECTED_SAMPLE_MUTATION_BASE_DF[
        [
            "MUTATION_ROW_ID",
            "COHORT",
            "SAMPLE_ID",
            "PATIENT_ID",
            "GENE_SYMBOL",
        ]
    ]
    .isna()
    .any(axis=1)
)

unexpected_genes = sorted(
    set(
        SELECTED_SAMPLE_MUTATION_BASE_DF[
            "GENE_SYMBOL"
        ]
        .dropna()
        .astype(str)
    )
    - selected_genes
)

expected_representative_samples = set(
    zip(
        representative_keys["COHORT"],
        representative_keys["SAMPLE_ID"],
    )
)

represented_mutation_samples = set(
    zip(
        SELECTED_SAMPLE_MUTATION_BASE_DF[
            "COHORT"
        ],
        SELECTED_SAMPLE_MUTATION_BASE_DF[
            "SAMPLE_ID"
        ],
    )
)

missing_representative_sample_keys = (
    expected_representative_samples
    - represented_mutation_samples
)


print("\n" + "=" * 78)
print("COMBINED MUTATION BASE VALIDATION")
print("=" * 78)

print(
    f"Selected-sample mutation rows: "
    f"{len(SELECTED_SAMPLE_MUTATION_BASE_DF):,}"
)

print(
    f"Unique represented patients: "
    f"{SELECTED_SAMPLE_MUTATION_BASE_DF[['COHORT', 'PATIENT_ID']].drop_duplicates().shape[0]:,}"
)

print(
    f"Unique represented samples: "
    f"{SELECTED_SAMPLE_MUTATION_BASE_DF[['COHORT', 'SAMPLE_ID']].drop_duplicates().shape[0]:,}"
)

print(
    f"Unique represented genes: "
    f"{SELECTED_SAMPLE_MUTATION_BASE_DF['GENE_SYMBOL'].nunique():,}"
)

print(
    f"Duplicate mutation-row IDs: "
    f"{int(duplicate_mutation_ids.sum())}"
)

print(
    f"Rows with missing core identifiers: "
    f"{int(missing_identifier_rows.sum())}"
)

print(
    f"Unexpected genes outside panel: "
    f"{unexpected_genes}"
)

print(
    f"Representative samples without retained mutations: "
    f"{len(missing_representative_sample_keys):,}"
)


if duplicate_mutation_ids.any():
    raise ValueError(
        "Duplicate mutation-row identifiers were generated."
    )

if missing_identifier_rows.any():
    raise ValueError(
        "Mutation rows contain missing core identifiers."
    )

if unexpected_genes:
    raise ValueError(
        f"Genes outside the 96-gene panel were retained: "
        f"{unexpected_genes}"
    )

if missing_representative_sample_keys:
    raise ValueError(
        "Some representative samples have no retained "
        "96-gene mutation rows."
    )


# ------------------------------------------------------------
# 7. Patient mutation-count audit
# ------------------------------------------------------------

patient_mutation_counts = (
    SELECTED_SAMPLE_MUTATION_BASE_DF
    .groupby(
        ["COHORT", "PATIENT_ID"],
        as_index=False,
    )
    .agg(
        MUTATION_ROW_COUNT=(
            "MUTATION_ROW_ID",
            "size",
        ),
        UNIQUE_MUTATED_GENE_COUNT=(
            "GENE_SYMBOL",
            "nunique",
        ),
    )
)


print("\nPatient mutation-count range:")

print(
    f"Minimum mutation rows: "
    f"{patient_mutation_counts['MUTATION_ROW_COUNT'].min():,}"
)

print(
    f"Median mutation rows: "
    f"{patient_mutation_counts['MUTATION_ROW_COUNT'].median():.1f}"
)

print(
    f"Maximum mutation rows: "
    f"{patient_mutation_counts['MUTATION_ROW_COUNT'].max():,}"
)

print(
    f"Maximum unique mutated genes for one patient: "
    f"{patient_mutation_counts['UNIQUE_MUTATED_GENE_COUNT'].max():,}"
)


print("\n" + "=" * 78)
print("STEP 9A FINISHED")
print("=" * 78)

print("Mutation base table:")
print("SELECTED_SAMPLE_MUTATION_BASE_DF")

print("\nMutation source audit:")
print("SELECTED_MUTATION_SOURCE_AUDIT_DF")

print("\nStage and clinical target values were not used.")
```

### Step 9B-1 — Diagnose Ensembl CDS Sequence Characters

This step checks why some Ensembl CDS records failed the strict DNA validation.

It separately reports:

- empty CDS records
- every character outside `A`, `C`, `G`, `T`, and `N`
- number of records containing each unsupported character
- example genes, transcripts and sequence positions

Sequences are inspected without changing or removing any nucleotide.

```python
# ============================================================
# STEP 9B-1 — DIAGNOSE ENSEMBL CDS SEQUENCE CHARACTERS
# ============================================================

import pandas as pd
from collections import Counter, defaultdict


print("=" * 78)
print("STEP 9B-1 — ENSEMBL CDS CHARACTER DIAGNOSTIC")
print("=" * 78)


# ------------------------------------------------------------
# 1. Recover the parsed FASTA table when necessary
# ------------------------------------------------------------

if "ENSEMBL_CDS_REFERENCE_DF" not in globals():

    if "parse_fasta_records" not in globals():
        raise RuntimeError(
            "The parsed FASTA table and parser were not found. "
            "Please run the previous Step 9B cell up to FASTA parsing."
        )

    if "FASTA_FILE" not in globals():
        FASTA_FILE = (
            PROJECT_DIR
            / "LUAD_Ensembl_CDS_sequences_96genes.fasta.txt"
        )

    ENSEMBL_CDS_REFERENCE_DF = pd.DataFrame(
        parse_fasta_records(FASTA_FILE)
    )


diagnostic_df = ENSEMBL_CDS_REFERENCE_DF.copy()


required_columns = {
    "FASTA_RECORD_NUMBER",
    "GENE_SYMBOL",
    "ENSEMBL_TRANSCRIPT_ID_RAW",
    "CDS_SEQUENCE",
    "CDS_LENGTH",
}

missing_columns = sorted(
    required_columns - set(diagnostic_df.columns)
)

if missing_columns:
    raise ValueError(
        f"Parsed FASTA table is missing columns: {missing_columns}"
    )


# ------------------------------------------------------------
# 2. Recalculate length safely
# ------------------------------------------------------------

diagnostic_df["CDS_SEQUENCE"] = (
    diagnostic_df["CDS_SEQUENCE"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

diagnostic_df["CDS_LENGTH_RECHECKED"] = (
    diagnostic_df["CDS_SEQUENCE"]
    .str.len()
)

empty_mask = (
    diagnostic_df["CDS_LENGTH_RECHECKED"].eq(0)
)


# ------------------------------------------------------------
# 3. Detect every character outside strict A/C/G/T/N
# ------------------------------------------------------------

STRICT_DNA_CHARACTERS = set("ACGTN")

character_record_counts = Counter()
character_total_counts = Counter()
character_examples = defaultdict(list)

record_invalid_character_sets = []
record_invalid_character_positions = []


for row in diagnostic_df.itertuples(index=False):

    sequence = row.CDS_SEQUENCE

    invalid_positions = [
        (position + 1, character)
        for position, character in enumerate(sequence)
        if character not in STRICT_DNA_CHARACTERS
    ]

    invalid_characters = sorted(
        set(character for _, character in invalid_positions)
    )

    record_invalid_character_sets.append(
        "".join(invalid_characters)
    )

    record_invalid_character_positions.append(
        " | ".join(
            f"{character}@{position}"
            for position, character in invalid_positions[:20]
        )
    )

    for character in invalid_characters:

        character_record_counts[character] += 1

        if len(character_examples[character]) < 5:
            character_examples[character].append({
                "GENE_SYMBOL": row.GENE_SYMBOL,
                "TRANSCRIPT_ID": row.ENSEMBL_TRANSCRIPT_ID_RAW,
                "FASTA_RECORD_NUMBER": row.FASTA_RECORD_NUMBER,
            })

    for _, character in invalid_positions:
        character_total_counts[character] += 1


diagnostic_df["INVALID_CHARACTER_SET"] = (
    record_invalid_character_sets
)

diagnostic_df["INVALID_CHARACTER_POSITION_EXAMPLES"] = (
    record_invalid_character_positions
)

invalid_character_mask = (
    diagnostic_df["INVALID_CHARACTER_SET"].ne("")
)


# ------------------------------------------------------------
# 4. Create character summary
# ------------------------------------------------------------

all_invalid_characters = sorted(
    character_record_counts.keys()
)

character_summary_rows = []

for character in all_invalid_characters:

    character_summary_rows.append({
        "CHARACTER": repr(character),
        "RECORDS_CONTAINING_CHARACTER": (
            character_record_counts[character]
        ),
        "TOTAL_CHARACTER_OCCURRENCES": (
            character_total_counts[character]
        ),
    })


INVALID_FASTA_CHARACTER_SUMMARY_DF = pd.DataFrame(
    character_summary_rows
)

INVALID_FASTA_RECORDS_DF = diagnostic_df.loc[
    empty_mask | invalid_character_mask,
    [
        "FASTA_RECORD_NUMBER",
        "GENE_SYMBOL",
        "ENSEMBL_TRANSCRIPT_ID_RAW",
        "CDS_LENGTH_RECHECKED",
        "INVALID_CHARACTER_SET",
        "INVALID_CHARACTER_POSITION_EXAMPLES",
    ],
].copy()


# ------------------------------------------------------------
# 5. Print concise diagnostic
# ------------------------------------------------------------

print(f"\nTotal FASTA records: {len(diagnostic_df):,}")

print(
    f"Empty CDS records: "
    f"{int(empty_mask.sum()):,}"
)

print(
    f"Records containing characters outside A/C/G/T/N: "
    f"{int(invalid_character_mask.sum()):,}"
)

print(
    f"Total records failing strict validation: "
    f"{int((empty_mask | invalid_character_mask).sum()):,}"
)

print(
    f"Distinct unsupported characters: "
    f"{[repr(value) for value in all_invalid_characters]}"
)


if not INVALID_FASTA_CHARACTER_SUMMARY_DF.empty:

    print("\nUnsupported-character summary:")

    print(
        INVALID_FASTA_CHARACTER_SUMMARY_DF
        .to_string(index=False)
    )


if empty_mask.any():

    print("\nEmpty CDS examples:")

    print(
        diagnostic_df.loc[
            empty_mask,
            [
                "FASTA_RECORD_NUMBER",
                "GENE_SYMBOL",
                "ENSEMBL_TRANSCRIPT_ID_RAW",
            ],
        ]
        .head(10)
        .to_string(index=False)
    )


if invalid_character_mask.any():

    print("\nInvalid-character record examples:")

    print(
        INVALID_FASTA_RECORDS_DF.loc[
            INVALID_FASTA_RECORDS_DF[
                "INVALID_CHARACTER_SET"
            ].ne("")
        ]
        .head(15)
        .to_string(index=False)
    )


    print("\nExamples grouped by unsupported character:")

    for character in all_invalid_characters:

        print(f"\nCharacter {repr(character)}:")

        print(
            pd.DataFrame(
                character_examples[character]
            ).to_string(index=False)
        )


# Keep the diagnostic table available for the correction step.
ENSEMBL_CDS_CHARACTER_DIAGNOSTIC_DF = diagnostic_df


print("\n" + "=" * 78)
print("STEP 9B-1 FINISHED")
print("=" * 78)

print("Diagnostic tables:")
print("INVALID_FASTA_CHARACTER_SUMMARY_DF")
print("INVALID_FASTA_RECORDS_DF")
```

### Step 9B-2 — Remove Non-Sequence FASTA Placeholder Records

The Ensembl FASTA contains 579 transcript records whose sequence field is the literal placeholder `SEQUENCE UNAVAILABLE`, rather than a nucleotide CDS.

These records cannot support mutation reconstruction and will be excluded from the usable CDS reference table.

The placeholder records remain preserved in a separate in-memory exclusion audit. Valid CDS records must contain only `A`, `C`, `G`, `T`, or `N`, have a positive sequence length, and belong to the validated 96-gene panel.

No placeholder sequence will be converted to `N`, padded, resized, or reconstructed.

```python
# ============================================================
# STEP 9B-2 — REMOVE NON-SEQUENCE FASTA PLACEHOLDERS
# ============================================================

import pandas as pd
import re


if "ENSEMBL_CDS_REFERENCE_DF" not in globals():
    raise RuntimeError(
        "ENSEMBL_CDS_REFERENCE_DF was not found. "
        "Please run the FASTA parsing step first."
    )

if "selected_genes" not in globals():
    raise RuntimeError(
        "selected_genes was not found."
    )


print("=" * 78)
print("STEP 9B-2 — VALID ENSEMBL CDS REFERENCE TABLE")
print("=" * 78)


# ------------------------------------------------------------
# 1. Prepare sequences without changing biological content
# ------------------------------------------------------------

reference_df = ENSEMBL_CDS_REFERENCE_DF.copy()

reference_df["CDS_SEQUENCE"] = (
    reference_df["CDS_SEQUENCE"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

reference_df["CDS_LENGTH"] = (
    reference_df["CDS_SEQUENCE"]
    .str.len()
)


# ------------------------------------------------------------
# 2. Detect known non-sequence placeholders
# ------------------------------------------------------------

def normalise_placeholder_text(value):
    """
    Remove spaces and punctuation only for placeholder detection.
    The original CDS_SEQUENCE column remains unchanged.
    """

    return re.sub(
        r"[^A-Z]",
        "",
        str(value).upper(),
    )


reference_df["SEQUENCE_TEXT_NORMALISED"] = (
    reference_df["CDS_SEQUENCE"]
    .apply(normalise_placeholder_text)
)


known_placeholder_values = {
    "SEQUENCEUNAVAILABLE",
    "SEQUENCENOTAVAILABLE",
    "NOSEQUENCEAVAILABLE",
    "SEQUENCENOTFOUND",
}


reference_df["PLACEHOLDER_SEQUENCE_FLAG"] = (
    reference_df["SEQUENCE_TEXT_NORMALISED"]
    .isin(known_placeholder_values)
    .astype("int8")
)


# ------------------------------------------------------------
# 3. Strict nucleotide validation
# ------------------------------------------------------------

VALID_DNA_PATTERN = re.compile(
    r"^[ACGTN]+$"
)


reference_df["STRICT_DNA_VALID_FLAG"] = (
    reference_df["CDS_SEQUENCE"]
    .apply(
        lambda sequence: bool(
            VALID_DNA_PATTERN.fullmatch(sequence)
        )
    )
    .astype("int8")
)


reference_df["EMPTY_SEQUENCE_FLAG"] = (
    reference_df["CDS_LENGTH"]
    .eq(0)
    .astype("int8")
)


reference_df["GENE_IN_PANEL_FLAG"] = (
    reference_df["GENE_SYMBOL"]
    .isin(selected_genes)
    .astype("int8")
)


# ------------------------------------------------------------
# 4. Assign deterministic exclusion reasons
# ------------------------------------------------------------

def assign_fasta_exclusion_reason(row):

    if row["HEADER_VALID_FLAG"] != 1:
        return "INVALID_FASTA_HEADER"

    if row["EMPTY_SEQUENCE_FLAG"] == 1:
        return "EMPTY_CDS_SEQUENCE"

    if row["PLACEHOLDER_SEQUENCE_FLAG"] == 1:
        return "SEQUENCE_UNAVAILABLE_PLACEHOLDER"

    if row["STRICT_DNA_VALID_FLAG"] != 1:
        return "UNSUPPORTED_NON_DNA_CHARACTERS"

    if row["GENE_IN_PANEL_FLAG"] != 1:
        return "GENE_OUTSIDE_96_GENE_PANEL"

    return pd.NA


reference_df["FASTA_EXCLUSION_REASON"] = (
    reference_df.apply(
        assign_fasta_exclusion_reason,
        axis=1,
    )
)


# ------------------------------------------------------------
# 5. Separate usable CDS and excluded records
# ------------------------------------------------------------

ENSEMBL_CDS_EXCLUSION_AUDIT_DF = (
    reference_df.loc[
        reference_df[
            "FASTA_EXCLUSION_REASON"
        ].notna()
    ]
    .copy()
    .reset_index(drop=True)
)


ENSEMBL_VALID_CDS_REFERENCE_DF = (
    reference_df.loc[
        reference_df[
            "FASTA_EXCLUSION_REASON"
        ].isna()
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Confirm placeholder identity
# ------------------------------------------------------------

placeholder_records = reference_df.loc[
    reference_df[
        "PLACEHOLDER_SEQUENCE_FLAG"
    ].eq(1)
].copy()


unique_placeholder_sequences = sorted(
    placeholder_records[
        "CDS_SEQUENCE"
    ]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)


# ------------------------------------------------------------
# 7. Deduplicate exact repeated transcript records
# ------------------------------------------------------------

exact_duplicate_mask = (
    ENSEMBL_VALID_CDS_REFERENCE_DF
    .duplicated(
        subset=[
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
            "CDS_SEQUENCE",
        ],
        keep="first",
    )
)


ENSEMBL_EXACT_DUPLICATE_CDS_DF = (
    ENSEMBL_VALID_CDS_REFERENCE_DF.loc[
        exact_duplicate_mask
    ]
    .copy()
    .reset_index(drop=True)
)


ENSEMBL_VALID_CDS_REFERENCE_DF = (
    ENSEMBL_VALID_CDS_REFERENCE_DF.loc[
        ~exact_duplicate_mask
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. Detect conflicting sequences for the same transcript
# ------------------------------------------------------------

raw_transcript_audit = (
    ENSEMBL_VALID_CDS_REFERENCE_DF
    .groupby(
        [
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
        ],
        dropna=False,
    )
    .agg(
        RECORD_COUNT=(
            "FASTA_RECORD_NUMBER",
            "size",
        ),
        UNIQUE_SEQUENCE_COUNT=(
            "CDS_SEQUENCE",
            "nunique",
        ),
    )
    .reset_index()
)


RAW_TRANSCRIPT_CDS_CONFLICT_DF = (
    raw_transcript_audit.loc[
        raw_transcript_audit[
            "UNIQUE_SEQUENCE_COUNT"
        ].gt(1)
    ]
    .copy()
)


base_transcript_audit = (
    ENSEMBL_VALID_CDS_REFERENCE_DF
    .groupby(
        [
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_BASE",
        ],
        dropna=False,
    )
    .agg(
        RECORD_COUNT=(
            "FASTA_RECORD_NUMBER",
            "size",
        ),
        UNIQUE_RAW_TRANSCRIPT_COUNT=(
            "ENSEMBL_TRANSCRIPT_ID_RAW",
            "nunique",
        ),
        UNIQUE_SEQUENCE_COUNT=(
            "CDS_SEQUENCE",
            "nunique",
        ),
    )
    .reset_index()
)


BASE_TRANSCRIPT_CDS_CONFLICT_DF = (
    base_transcript_audit.loc[
        base_transcript_audit[
            "UNIQUE_SEQUENCE_COUNT"
        ].gt(1)
    ]
    .copy()
)


# ------------------------------------------------------------
# 9. Panel-gene coverage after placeholder removal
# ------------------------------------------------------------

valid_fasta_genes = set(
    ENSEMBL_VALID_CDS_REFERENCE_DF[
        "GENE_SYMBOL"
    ]
    .dropna()
    .astype(str)
)


missing_panel_genes_after_cleanup = sorted(
    selected_genes - valid_fasta_genes
)


valid_transcripts_per_gene = (
    ENSEMBL_VALID_CDS_REFERENCE_DF
    .groupby("GENE_SYMBOL")[
        "ENSEMBL_TRANSCRIPT_ID_RAW"
    ]
    .nunique()
)


genes_with_one_valid_transcript = int(
    valid_transcripts_per_gene.eq(1).sum()
)

genes_with_multiple_valid_transcripts = int(
    valid_transcripts_per_gene.gt(1).sum()
)


# ------------------------------------------------------------
# 10. Print validation results
# ------------------------------------------------------------

print(f"\nOriginal FASTA records: {len(reference_df):,}")

print(
    f"Placeholder records excluded: "
    f"{len(placeholder_records):,}"
)

print(
    f"Unique placeholder sequence values: "
    f"{unique_placeholder_sequences}"
)

print(
    f"Other excluded records: "
    f"{len(ENSEMBL_CDS_EXCLUSION_AUDIT_DF) - len(placeholder_records):,}"
)

print(
    f"Exact duplicate valid records removed: "
    f"{len(ENSEMBL_EXACT_DUPLICATE_CDS_DF):,}"
)

print(
    f"Usable CDS transcript records: "
    f"{len(ENSEMBL_VALID_CDS_REFERENCE_DF):,}"
)

print(
    f"Usable versioned transcript IDs: "
    f"{ENSEMBL_VALID_CDS_REFERENCE_DF['ENSEMBL_TRANSCRIPT_ID_RAW'].nunique():,}"
)

print(
    f"Usable version-free transcript IDs: "
    f"{ENSEMBL_VALID_CDS_REFERENCE_DF['ENSEMBL_TRANSCRIPT_ID_BASE'].nunique():,}"
)

print(
    f"Genes with at least one usable CDS: "
    f"{len(valid_fasta_genes):,}/96"
)

print(
    f"Genes missing all usable CDS records: "
    f"{missing_panel_genes_after_cleanup}"
)

print(
    f"Genes with one usable transcript: "
    f"{genes_with_one_valid_transcript:,}"
)

print(
    f"Genes with multiple usable transcripts: "
    f"{genes_with_multiple_valid_transcripts:,}"
)

print(
    f"Raw transcript IDs with conflicting CDS sequences: "
    f"{len(RAW_TRANSCRIPT_CDS_CONFLICT_DF):,}"
)

print(
    f"Version-free transcript IDs with conflicting CDS sequences: "
    f"{len(BASE_TRANSCRIPT_CDS_CONFLICT_DF):,}"
)


if not ENSEMBL_CDS_EXCLUSION_AUDIT_DF.empty:

    print("\nFASTA exclusion reasons:")

    print(
        ENSEMBL_CDS_EXCLUSION_AUDIT_DF[
            "FASTA_EXCLUSION_REASON"
        ]
        .value_counts(dropna=False)
        .to_string()
    )


valid_lengths = (
    ENSEMBL_VALID_CDS_REFERENCE_DF[
        "CDS_LENGTH"
    ]
    .astype(int)
)


print(
    f"\nUsable CDS length range: "
    f"{valid_lengths.min():,} to {valid_lengths.max():,}"
)

print(
    f"Median usable CDS length: "
    f"{valid_lengths.median():,.1f}"
)


# ------------------------------------------------------------
# 11. Hard validations
# ------------------------------------------------------------

if len(placeholder_records) != 579:
    raise ValueError(
        f"Expected 579 placeholder records, "
        f"found {len(placeholder_records)}."
    )

if not set(unique_placeholder_sequences).issubset(
    {
        "SEQUENCE UNAVAILABLE",
        "SEQUENCEUNAVAILABLE",
    }
):
    raise ValueError(
        "Unexpected placeholder sequence text was detected."
    )

if (
    ENSEMBL_VALID_CDS_REFERENCE_DF[
        "STRICT_DNA_VALID_FLAG"
    ].ne(1).any()
):
    raise ValueError(
        "A non-DNA sequence entered the usable CDS table."
    )

if missing_panel_genes_after_cleanup:
    raise ValueError(
        "Some panel genes have no usable CDS after placeholder "
        f"removal: {missing_panel_genes_after_cleanup}"
    )

if not RAW_TRANSCRIPT_CDS_CONFLICT_DF.empty:

    print("\nRaw transcript conflict examples:")

    print(
        RAW_TRANSCRIPT_CDS_CONFLICT_DF
        .head(10)
        .to_string(index=False)
    )

    raise ValueError(
        "The same versioned transcript ID has conflicting "
        "CDS sequences."
    )


# ------------------------------------------------------------
# 12. Keep cleaned reference table in memory
# ------------------------------------------------------------

ENSEMBL_CDS_REFERENCE_CLEAN_DF = (
    ENSEMBL_VALID_CDS_REFERENCE_DF.copy()
)


print("\n" + "=" * 78)
print("STEP 9B-2 FINISHED")
print("=" * 78)

print("Usable CDS reference table:")
print("ENSEMBL_CDS_REFERENCE_CLEAN_DF")

print("\nPlaceholder/exclusion audit:")
print("ENSEMBL_CDS_EXCLUSION_AUDIT_DF")

print("\nPlaceholder records were not converted into DNA.")
```

## Step 9C — Match Mutation Transcripts to Valid Ensembl CDS Records

This step normalises transcript identifiers from the original mutation annotations.

Transcript selection priority:

1. Ensembl transcript extracted from `TRANSCRIPT_ID_RAW`
2. Ensembl transcript extracted from `HGVSC_RAW` only when the transcript field does not provide one

Version suffixes such as `.1` or `.8` are removed only for matching.

Each mutation is then matched using both:

- gene symbol
- version-free Ensembl transcript ID

Transcript disagreements, missing transcript IDs, gene mismatches and transcripts absent from the valid CDS FASTA are preserved as explicit audit statuses.

Only successfully matched gene-transcript records receive a CDS reference sequence.

```python
# ============================================================
# STEP 9C — MATCH MUTATION TRANSCRIPTS TO VALID ENSEMBL CDS
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required previous objects
# ------------------------------------------------------------

required_objects = [
    "SELECTED_SAMPLE_MUTATION_BASE_DF",
    "ENSEMBL_CDS_REFERENCE_CLEAN_DF",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required previous objects are missing: "
        + ", ".join(missing_objects)
    )


print("=" * 78)
print("STEP 9C — MUTATION TRANSCRIPT MATCHING")
print("=" * 78)


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def clean_optional_text(series):
    """
    Strip whitespace and convert common missing tokens to NA.
    """

    cleaned = (
        series
        .astype("string")
        .str.strip()
    )

    missing_tokens = {
        "",
        "NA",
        "N/A",
        "NAN",
        "NONE",
        "NULL",
        "UNKNOWN",
        "NOT AVAILABLE",
        ".",
    }

    return cleaned.mask(
        cleaned.str.upper().isin(missing_tokens)
    )


def extract_ensembl_transcript(value):
    """
    Extract the first Ensembl transcript ID from text.

    Examples:
    ENST00000367770
    ENST00000367770.8
    ENST00000367770:c.35G>A
    """

    if pd.isna(value):
        return pd.NA

    text = str(value).strip().upper()

    match = re.search(
        r"(ENST[0-9]+(?:\.[0-9]+)?)",
        text,
    )

    if match is None:
        return pd.NA

    return match.group(1)


def remove_transcript_version(value):
    """
    Convert ENST000001.5 to ENST000001.
    """

    if pd.isna(value):
        return pd.NA

    return str(value).split(".")[0]


# ------------------------------------------------------------
# 3. Prepare mutation annotations
# ------------------------------------------------------------

mutation_df = (
    SELECTED_SAMPLE_MUTATION_BASE_DF
    .copy()
)


mutation_df["TRANSCRIPT_ID_RAW_CLEAN"] = (
    clean_optional_text(
        mutation_df["TRANSCRIPT_ID_RAW"]
    )
)

mutation_df["HGVSC_RAW_CLEAN"] = (
    clean_optional_text(
        mutation_df["HGVSC_RAW"]
    )
)


mutation_df["TRANSCRIPT_ID_FROM_SOURCE"] = (
    mutation_df["TRANSCRIPT_ID_RAW_CLEAN"]
    .apply(extract_ensembl_transcript)
    .astype("string")
)

mutation_df["TRANSCRIPT_ID_FROM_HGVSC"] = (
    mutation_df["HGVSC_RAW_CLEAN"]
    .apply(extract_ensembl_transcript)
    .astype("string")
)


mutation_df["TRANSCRIPT_BASE_FROM_SOURCE"] = (
    mutation_df["TRANSCRIPT_ID_FROM_SOURCE"]
    .apply(remove_transcript_version)
    .astype("string")
)

mutation_df["TRANSCRIPT_BASE_FROM_HGVSC"] = (
    mutation_df["TRANSCRIPT_ID_FROM_HGVSC"]
    .apply(remove_transcript_version)
    .astype("string")
)


# ------------------------------------------------------------
# 4. Audit source versus HGVSc transcript disagreement
# ------------------------------------------------------------

both_transcripts_available = (
    mutation_df["TRANSCRIPT_BASE_FROM_SOURCE"].notna()
    &
    mutation_df["TRANSCRIPT_BASE_FROM_HGVSC"].notna()
)

transcript_disagreement_mask = (
    both_transcripts_available
    &
    mutation_df["TRANSCRIPT_BASE_FROM_SOURCE"].ne(
        mutation_df["TRANSCRIPT_BASE_FROM_HGVSC"]
    )
).fillna(False)


mutation_df["TRANSCRIPT_SOURCE_HGVSC_DISAGREEMENT_FLAG"] = (
    transcript_disagreement_mask
    .astype("int8")
)


# ------------------------------------------------------------
# 5. Select transcript deterministically
# ------------------------------------------------------------

mutation_df["TRANSCRIPT_ID_SELECTED"] = (
    mutation_df["TRANSCRIPT_ID_FROM_SOURCE"]
    .combine_first(
        mutation_df["TRANSCRIPT_ID_FROM_HGVSC"]
    )
    .astype("string")
)


mutation_df["TRANSCRIPT_ID_BASE_SELECTED"] = (
    mutation_df["TRANSCRIPT_ID_SELECTED"]
    .apply(remove_transcript_version)
    .astype("string")
)


mutation_df["TRANSCRIPT_SELECTION_SOURCE"] = pd.Series(
    "MISSING",
    index=mutation_df.index,
    dtype="string",
)


source_transcript_available = (
    mutation_df["TRANSCRIPT_ID_FROM_SOURCE"]
    .notna()
)

hgvsc_fallback_used = (
    ~source_transcript_available
    &
    mutation_df["TRANSCRIPT_ID_FROM_HGVSC"]
    .notna()
)


mutation_df.loc[
    source_transcript_available,
    "TRANSCRIPT_SELECTION_SOURCE",
] = "TRANSCRIPT_ID_RAW"


mutation_df.loc[
    hgvsc_fallback_used,
    "TRANSCRIPT_SELECTION_SOURCE",
] = "HGVSC_RAW_FALLBACK"


# ------------------------------------------------------------
# 6. Prepare unique valid FASTA transcript lookup
# ------------------------------------------------------------

reference_columns = [
    "FASTA_RECORD_NUMBER",
    "FASTA_HEADER_RAW",
    "ENSEMBL_GENE_ID_RAW",
    "ENSEMBL_TRANSCRIPT_ID_RAW",
    "ENSEMBL_TRANSCRIPT_ID_BASE",
    "GENE_SYMBOL",
    "CDS_SEQUENCE",
    "CDS_LENGTH",
]


reference_lookup = (
    ENSEMBL_CDS_REFERENCE_CLEAN_DF[
        reference_columns
    ]
    .copy()
)


reference_lookup["GENE_SYMBOL"] = (
    reference_lookup["GENE_SYMBOL"]
    .astype("string")
    .str.strip()
    .str.upper()
)

reference_lookup["ENSEMBL_TRANSCRIPT_ID_BASE"] = (
    reference_lookup[
        "ENSEMBL_TRANSCRIPT_ID_BASE"
    ]
    .astype("string")
    .str.strip()
    .str.upper()
)


duplicate_reference_transcripts = (
    reference_lookup
    .duplicated(
        subset=["ENSEMBL_TRANSCRIPT_ID_BASE"],
        keep=False,
    )
)


print(
    f"\nDuplicate version-free transcript IDs in valid FASTA: "
    f"{int(duplicate_reference_transcripts.sum()):,}"
)


if duplicate_reference_transcripts.any():

    print("\nDuplicate transcript examples:")

    print(
        reference_lookup.loc[
            duplicate_reference_transcripts,
            [
                "GENE_SYMBOL",
                "ENSEMBL_TRANSCRIPT_ID_RAW",
                "ENSEMBL_TRANSCRIPT_ID_BASE",
                "CDS_LENGTH",
            ],
        ]
        .head(20)
        .to_string(index=False)
    )

    raise ValueError(
        "Valid FASTA transcript lookup is not unique."
    )


reference_lookup = reference_lookup.rename(
    columns={
        "GENE_SYMBOL": "REFERENCE_GENE_SYMBOL",
        "ENSEMBL_GENE_ID_RAW": "REFERENCE_ENSEMBL_GENE_ID",
        "ENSEMBL_TRANSCRIPT_ID_RAW": (
            "REFERENCE_TRANSCRIPT_ID_RAW"
        ),
        "ENSEMBL_TRANSCRIPT_ID_BASE": (
            "REFERENCE_TRANSCRIPT_ID_BASE"
        ),
        "CDS_SEQUENCE": "REFERENCE_CDS_SEQUENCE",
        "CDS_LENGTH": "REFERENCE_CDS_LENGTH",
        "FASTA_HEADER_RAW": "REFERENCE_FASTA_HEADER",
        "FASTA_RECORD_NUMBER": (
            "REFERENCE_FASTA_RECORD_NUMBER"
        ),
    }
)


# ------------------------------------------------------------
# 7. Match selected transcript to FASTA
# ------------------------------------------------------------

matched_df = mutation_df.merge(
    reference_lookup,
    left_on="TRANSCRIPT_ID_BASE_SELECTED",
    right_on="REFERENCE_TRANSCRIPT_ID_BASE",
    how="left",
    validate="many_to_one",
)


transcript_missing_mask = (
    matched_df["TRANSCRIPT_ID_BASE_SELECTED"]
    .isna()
)

transcript_not_in_fasta_mask = (
    matched_df["TRANSCRIPT_ID_BASE_SELECTED"]
    .notna()
    &
    matched_df["REFERENCE_TRANSCRIPT_ID_BASE"]
    .isna()
)

transcript_gene_mismatch_mask = (
    matched_df["REFERENCE_TRANSCRIPT_ID_BASE"]
    .notna()
    &
    matched_df["GENE_SYMBOL"].ne(
        matched_df["REFERENCE_GENE_SYMBOL"]
    )
).fillna(False)

successful_match_mask = (
    matched_df["REFERENCE_TRANSCRIPT_ID_BASE"]
    .notna()
    &
    ~transcript_gene_mismatch_mask
)


# ------------------------------------------------------------
# 8. Assign deterministic transcript-match status
# ------------------------------------------------------------

matched_df["TRANSCRIPT_MATCH_STATUS"] = (
    "MATCHED_GENE_TRANSCRIPT"
)

matched_df.loc[
    transcript_missing_mask,
    "TRANSCRIPT_MATCH_STATUS",
] = "MISSING_ENSEMBL_TRANSCRIPT"

matched_df.loc[
    transcript_not_in_fasta_mask,
    "TRANSCRIPT_MATCH_STATUS",
] = "TRANSCRIPT_NOT_IN_VALID_FASTA"

matched_df.loc[
    transcript_gene_mismatch_mask,
    "TRANSCRIPT_MATCH_STATUS",
] = "TRANSCRIPT_GENE_MISMATCH"


matched_df["TRANSCRIPT_MATCH_FLAG"] = (
    successful_match_mask
    .astype("int8")
)


# ------------------------------------------------------------
# 9. Prevent mismatched transcript sequences from being used
# ------------------------------------------------------------

reference_fields_to_clear = [
    "REFERENCE_FASTA_RECORD_NUMBER",
    "REFERENCE_FASTA_HEADER",
    "REFERENCE_ENSEMBL_GENE_ID",
    "REFERENCE_TRANSCRIPT_ID_RAW",
    "REFERENCE_TRANSCRIPT_ID_BASE",
    "REFERENCE_GENE_SYMBOL",
    "REFERENCE_CDS_SEQUENCE",
    "REFERENCE_CDS_LENGTH",
]


for column in reference_fields_to_clear:

    matched_df.loc[
        ~successful_match_mask,
        column,
    ] = pd.NA


# ------------------------------------------------------------
# 10. Final transcript-matched mutation table
# ------------------------------------------------------------

MUTATION_TRANSCRIPT_MATCH_DF = (
    matched_df.copy()
)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

duplicate_mutation_ids = (
    MUTATION_TRANSCRIPT_MATCH_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


matched_gene_mismatches = (
    MUTATION_TRANSCRIPT_MATCH_DF[
        "TRANSCRIPT_MATCH_FLAG"
    ].eq(1)
    &
    MUTATION_TRANSCRIPT_MATCH_DF[
        "GENE_SYMBOL"
    ].ne(
        MUTATION_TRANSCRIPT_MATCH_DF[
            "REFERENCE_GENE_SYMBOL"
        ]
    )
).fillna(False)


matched_missing_sequences = (
    MUTATION_TRANSCRIPT_MATCH_DF[
        "TRANSCRIPT_MATCH_FLAG"
    ].eq(1)
    &
    MUTATION_TRANSCRIPT_MATCH_DF[
        "REFERENCE_CDS_SEQUENCE"
    ].isna()
)


print(f"\nTotal mutation rows: {len(MUTATION_TRANSCRIPT_MATCH_DF):,}")

print(
    f"Transcript extracted from source field: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_ID_FROM_SOURCE'].notna().sum():,}"
)

print(
    f"Transcript extracted from HGVSc: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_ID_FROM_HGVSC'].notna().sum():,}"
)

print(
    f"HGVSc transcript fallback used: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_SELECTION_SOURCE'].eq('HGVSC_RAW_FALLBACK').sum():,}"
)

print(
    f"Source/HGVSc transcript disagreements: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_SOURCE_HGVSC_DISAGREEMENT_FLAG'].sum():,}"
)

print(
    f"Successful gene-transcript matches: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_MATCH_FLAG'].sum():,}"
)

print(
    f"Unmatched mutation rows: "
    f"{MUTATION_TRANSCRIPT_MATCH_DF['TRANSCRIPT_MATCH_FLAG'].eq(0).sum():,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{int(duplicate_mutation_ids.sum())}"
)

print(
    f"Matched rows with gene mismatch: "
    f"{int(matched_gene_mismatches.sum())}"
)

print(
    f"Matched rows without CDS sequence: "
    f"{int(matched_missing_sequences.sum())}"
)


print("\nTranscript match-status counts:")

print(
    MUTATION_TRANSCRIPT_MATCH_DF[
        "TRANSCRIPT_MATCH_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 12. Cohort-level transcript summary
# ------------------------------------------------------------

transcript_summary_df = (
    MUTATION_TRANSCRIPT_MATCH_DF
    .groupby("COHORT", sort=False)
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),
        source_transcript_available=(
            "TRANSCRIPT_ID_FROM_SOURCE",
            lambda values: int(values.notna().sum())
        ),
        hgvsc_fallback_used=(
            "TRANSCRIPT_SELECTION_SOURCE",
            lambda values: int(
                values.eq("HGVSC_RAW_FALLBACK").sum()
            )
        ),
        transcript_matched=(
            "TRANSCRIPT_MATCH_FLAG",
            "sum",
        ),
        transcript_unmatched=(
            "TRANSCRIPT_MATCH_FLAG",
            lambda values: int(values.eq(0).sum())
        ),
        source_hgvsc_disagreement=(
            "TRANSCRIPT_SOURCE_HGVSC_DISAGREEMENT_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print("\nTranscript matching by cohort:")

print(
    transcript_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 13. Show unresolved examples
# ------------------------------------------------------------

unresolved_examples = (
    MUTATION_TRANSCRIPT_MATCH_DF.loc[
        MUTATION_TRANSCRIPT_MATCH_DF[
            "TRANSCRIPT_MATCH_FLAG"
        ].eq(0),
        [
            "MUTATION_ROW_ID",
            "COHORT",
            "GENE_SYMBOL",
            "TRANSCRIPT_ID_RAW",
            "HGVSC_RAW",
            "TRANSCRIPT_ID_SELECTED",
            "TRANSCRIPT_MATCH_STATUS",
        ],
    ]
    .head(20)
)


if not unresolved_examples.empty:

    print("\nUnresolved transcript examples:")

    print(
        unresolved_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 14. Hard validations
# ------------------------------------------------------------

if duplicate_mutation_ids.any():
    raise ValueError(
        "Transcript matching duplicated mutation rows."
    )

if matched_gene_mismatches.any():
    raise ValueError(
        "A gene-mismatched transcript entered the matched table."
    )

if matched_missing_sequences.any():
    raise ValueError(
        "A matched transcript does not contain a valid CDS sequence."
    )


print("\n" + "=" * 78)
print("STEP 9C FINISHED")
print("=" * 78)

print("Transcript-matched mutation table:")
print("MUTATION_TRANSCRIPT_MATCH_DF")

print("\nUnmatched rows will remain available for the final exclusion audit.")
```

### Step 9C-1 — Audit ZFHX4 and EPPK1 Transcript Availability

This audit checks whether the transcript identifiers used by unmatched ZFHX4 and EPPK1 mutations:

- exist among the valid Ensembl CDS records
- were excluded as `SEQUENCE UNAVAILABLE`
- have alternative valid transcripts for the same gene

Alternative transcripts are reported for reference only. A transcript-specific HGVSc coordinate will not be transferred to another transcript without an explicit transcript-coordinate mapping.

```python
# ============================================================
# STEP 9C-1 — ZFHX4 AND EPPK1 TRANSCRIPT AVAILABILITY AUDIT
# ============================================================

import pandas as pd


required_objects = [
    "MUTATION_TRANSCRIPT_MATCH_DF",
    "ENSEMBL_CDS_REFERENCE_CLEAN_DF",
    "ENSEMBL_CDS_EXCLUSION_AUDIT_DF",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required objects are missing: "
        + ", ".join(missing_objects)
    )


print("=" * 78)
print("STEP 9C-1 — ZFHX4 AND EPPK1 TRANSCRIPT AUDIT")
print("=" * 78)


audit_genes = ["ZFHX4", "EPPK1"]


# ------------------------------------------------------------
# 1. Valid transcripts for both genes
# ------------------------------------------------------------

valid_gene_transcripts = (
    ENSEMBL_CDS_REFERENCE_CLEAN_DF.loc[
        ENSEMBL_CDS_REFERENCE_CLEAN_DF[
            "GENE_SYMBOL"
        ].isin(audit_genes),
        [
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
            "ENSEMBL_TRANSCRIPT_ID_BASE",
            "CDS_LENGTH",
        ],
    ]
    .copy()
    .sort_values(
        [
            "GENE_SYMBOL",
            "CDS_LENGTH",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
        ],
        ascending=[True, False, True],
    )
)


# ------------------------------------------------------------
# 2. Excluded placeholder transcripts for both genes
# ------------------------------------------------------------

excluded_gene_transcripts = (
    ENSEMBL_CDS_EXCLUSION_AUDIT_DF.loc[
        ENSEMBL_CDS_EXCLUSION_AUDIT_DF[
            "GENE_SYMBOL"
        ].isin(audit_genes),
        [
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
            "ENSEMBL_TRANSCRIPT_ID_BASE",
            "CDS_LENGTH",
            "FASTA_EXCLUSION_REASON",
        ],
    ]
    .copy()
    .sort_values(
        [
            "GENE_SYMBOL",
            "ENSEMBL_TRANSCRIPT_ID_RAW",
        ]
    )
)


# ------------------------------------------------------------
# 3. Unmatched mutation transcript IDs
# ------------------------------------------------------------

unmatched_gene_mutations = (
    MUTATION_TRANSCRIPT_MATCH_DF.loc[
        MUTATION_TRANSCRIPT_MATCH_DF[
            "GENE_SYMBOL"
        ].isin(audit_genes)
        &
        MUTATION_TRANSCRIPT_MATCH_DF[
            "TRANSCRIPT_MATCH_FLAG"
        ].eq(0),
        [
            "MUTATION_ROW_ID",
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "GENE_SYMBOL",
            "TRANSCRIPT_ID_SELECTED",
            "TRANSCRIPT_ID_BASE_SELECTED",
            "HGVSC_RAW",
            "TRANSCRIPT_MATCH_STATUS",
        ],
    ]
    .copy()
)


unmatched_transcript_summary = (
    unmatched_gene_mutations
    .groupby(
        [
            "GENE_SYMBOL",
            "TRANSCRIPT_ID_BASE_SELECTED",
            "TRANSCRIPT_MATCH_STATUS",
        ],
        dropna=False,
        as_index=False,
    )
    .agg(
        UNMATCHED_MUTATION_ROWS=(
            "MUTATION_ROW_ID",
            "size",
        ),
        AFFECTED_PATIENTS=(
            "PATIENT_ID",
            "nunique",
        ),
    )
)


# ------------------------------------------------------------
# 4. Determine where each unmatched transcript exists
# ------------------------------------------------------------

valid_transcript_ids = set(
    valid_gene_transcripts[
        "ENSEMBL_TRANSCRIPT_ID_BASE"
    ]
    .dropna()
    .astype(str)
)

excluded_transcript_ids = set(
    excluded_gene_transcripts[
        "ENSEMBL_TRANSCRIPT_ID_BASE"
    ]
    .dropna()
    .astype(str)
)


unmatched_transcript_summary[
    "FOUND_IN_VALID_CDS"
] = (
    unmatched_transcript_summary[
        "TRANSCRIPT_ID_BASE_SELECTED"
    ]
    .isin(valid_transcript_ids)
)


unmatched_transcript_summary[
    "FOUND_IN_EXCLUDED_PLACEHOLDERS"
] = (
    unmatched_transcript_summary[
        "TRANSCRIPT_ID_BASE_SELECTED"
    ]
    .isin(excluded_transcript_ids)
)


valid_counts_by_gene = (
    valid_gene_transcripts
    .groupby("GENE_SYMBOL")[
        "ENSEMBL_TRANSCRIPT_ID_BASE"
    ]
    .nunique()
    .to_dict()
)


unmatched_transcript_summary[
    "VALID_ALTERNATIVE_TRANSCRIPTS_FOR_GENE"
] = (
    unmatched_transcript_summary[
        "GENE_SYMBOL"
    ]
    .map(valid_counts_by_gene)
    .fillna(0)
    .astype(int)
)


# ------------------------------------------------------------
# 5. Print audit
# ------------------------------------------------------------

for gene in audit_genes:

    gene_valid = valid_gene_transcripts.loc[
        valid_gene_transcripts[
            "GENE_SYMBOL"
        ].eq(gene)
    ]

    gene_excluded = excluded_gene_transcripts.loc[
        excluded_gene_transcripts[
            "GENE_SYMBOL"
        ].eq(gene)
    ]

    gene_unmatched = unmatched_gene_mutations.loc[
        unmatched_gene_mutations[
            "GENE_SYMBOL"
        ].eq(gene)
    ]

    gene_all_mutations = (
        MUTATION_TRANSCRIPT_MATCH_DF.loc[
            MUTATION_TRANSCRIPT_MATCH_DF[
                "GENE_SYMBOL"
            ].eq(gene)
        ]
    )

    print("\n" + "-" * 78)
    print(f"GENE: {gene}")

    print(
        f"Valid CDS transcripts: "
        f"{gene_valid['ENSEMBL_TRANSCRIPT_ID_BASE'].nunique():,}"
    )

    print(
        f"Excluded placeholder transcripts: "
        f"{gene_excluded['ENSEMBL_TRANSCRIPT_ID_BASE'].nunique():,}"
    )

    print(
        f"Total selected-sample mutation rows: "
        f"{len(gene_all_mutations):,}"
    )

    print(
        f"Successfully transcript-matched mutations: "
        f"{gene_all_mutations['TRANSCRIPT_MATCH_FLAG'].sum():,}"
    )

    print(
        f"Unmatched mutation rows: "
        f"{len(gene_unmatched):,}"
    )

    print("\nValid transcript examples:")

    print(
        gene_valid.head(10).to_string(index=False)
        if not gene_valid.empty
        else "None"
    )

    print("\nExcluded placeholder transcripts:")

    print(
        gene_excluded.to_string(index=False)
        if not gene_excluded.empty
        else "None"
    )


print("\n" + "=" * 78)
print("UNMATCHED TRANSCRIPT SUMMARY")
print("=" * 78)

print(
    unmatched_transcript_summary.to_string(
        index=False
    )
)


print("\n" + "=" * 78)
print("STEP 9C-1 FINISHED")
print("=" * 78)

print(
    "Alternative transcripts were audited but were not used "
    "to remap transcript-specific HGVSc coordinates."
)
```

## Step 9D — Parse Coding HGVSc Coordinates and Mutation Operations

This step parses coding-DNA mutation annotations from `HGVSC_RAW`.

It extracts:

- coding expression
- mutation operation
- 1-based CDS start and end coordinates
- HGVSc reference sequence
- HGVSc alternate or inserted sequence

Supported operations are:

- substitution
- deletion
- insertion
- duplication
- deletion-insertion

Intronic, UTR, uncertain and unsupported annotations remain explicitly classified. Parsed coordinates are validated against the matched full Ensembl CDS length.

Mutation alleles are not applied to the reference sequence in this step.

```python
# ============================================================
# STEP 9D — PARSE CODING HGVSc COORDINATES AND OPERATIONS
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_TRANSCRIPT_MATCH_DF" not in globals():
    raise RuntimeError(
        "MUTATION_TRANSCRIPT_MATCH_DF was not found. "
        "Please run Step 9C first."
    )


print("=" * 78)
print("STEP 9D — CODING HGVSc PARSING")
print("=" * 78)


# ------------------------------------------------------------
# 2. Standard empty parser result
# ------------------------------------------------------------

def empty_hgvsc_result(status, expression=pd.NA):
    """
    Return a consistent empty parsing result.
    """

    return {
        "HGVS_C_EXPRESSION": expression,
        "HGVS_OPERATION": pd.NA,
        "CDS_START_1BASED": pd.NA,
        "CDS_END_1BASED": pd.NA,
        "HGVS_REFERENCE_SEQUENCE": pd.NA,
        "HGVS_ALTERNATE_SEQUENCE": pd.NA,
        "HGVS_PARSE_STATUS": status,
    }


# ------------------------------------------------------------
# 3. HGVSc parser
# ------------------------------------------------------------

def parse_hgvsc_annotation(raw_value):
    """
    Parse supported positive coding-DNA HGVS expressions.

    Supported examples:
    c.35G>A
    c.123del
    c.123_125delACT
    c.123_124insA
    c.123dup
    c.123_125dupACT
    c.123_125delinsGG
    """

    if pd.isna(raw_value):
        return empty_hgvsc_result(
            "MISSING_HGVSC"
        )

    raw_text = str(raw_value).strip()

    if raw_text == "":
        return empty_hgvsc_result(
            "MISSING_HGVSC"
        )


    # Extract the coding expression after "c."
    coding_match = re.search(
        r"(?:^|:)c\.(.+)$",
        raw_text,
        flags=re.IGNORECASE,
    )

    if coding_match is None:
        return empty_hgvsc_result(
            "NOT_CODING_HGVSC"
        )


    expression = (
        coding_match
        .group(1)
        .strip()
        .replace(" ", "")
        .upper()
    )


    base_result = {
        "HGVS_C_EXPRESSION": expression,
        "HGVS_OPERATION": pd.NA,
        "CDS_START_1BASED": pd.NA,
        "CDS_END_1BASED": pd.NA,
        "HGVS_REFERENCE_SEQUENCE": pd.NA,
        "HGVS_ALTERNATE_SEQUENCE": pd.NA,
        "HGVS_PARSE_STATUS": pd.NA,
    }


    # --------------------------------------------------------
    # Non-coding or uncertain coordinates
    # --------------------------------------------------------

    if expression.startswith("-"):
        return {
            **base_result,
            "HGVS_PARSE_STATUS": "FIVE_PRIME_UTR_COORDINATE",
        }

    if expression.startswith("*"):
        return {
            **base_result,
            "HGVS_PARSE_STATUS": "THREE_PRIME_UTR_COORDINATE",
        }

    if re.search(r"\d+[+-]\d+", expression):
        return {
            **base_result,
            "HGVS_PARSE_STATUS": "INTRONIC_COORDINATE",
        }

    if "?" in expression:
        return {
            **base_result,
            "HGVS_PARSE_STATUS": "UNCERTAIN_HGVSC",
        }


    # --------------------------------------------------------
    # Substitution
    # Example: c.35G>A
    # --------------------------------------------------------

    substitution_match = re.fullmatch(
        r"(?P<position>\d+)"
        r"(?P<reference>[ACGTN]+)"
        r">"
        r"(?P<alternate>[ACGTN]+)",
        expression,
    )

    if substitution_match is not None:

        position = int(
            substitution_match.group("position")
        )

        return {
            **base_result,
            "HGVS_OPERATION": "SUBSTITUTION",
            "CDS_START_1BASED": position,
            "CDS_END_1BASED": position,
            "HGVS_REFERENCE_SEQUENCE": (
                substitution_match.group("reference")
            ),
            "HGVS_ALTERNATE_SEQUENCE": (
                substitution_match.group("alternate")
            ),
            "HGVS_PARSE_STATUS": "PARSED_SUBSTITUTION",
        }


    # --------------------------------------------------------
    # Deletion-insertion
    # Examples:
    # c.123delinsGG
    # c.123_125delinsGG
    # c.123_125delACTinsGG
    # --------------------------------------------------------

    delins_match = re.fullmatch(
        r"(?P<start>\d+)"
        r"(?:_(?P<end>\d+))?"
        r"DEL"
        r"(?P<reference>[ACGTN]*)"
        r"INS"
        r"(?P<alternate>[ACGTN]+)",
        expression,
    )

    if delins_match is not None:

        start = int(
            delins_match.group("start")
        )

        end = int(
            delins_match.group("end")
            or start
        )

        reference = (
            delins_match.group("reference")
            or pd.NA
        )

        alternate = delins_match.group(
            "alternate"
        )

        return {
            **base_result,
            "HGVS_OPERATION": "DELINS",
            "CDS_START_1BASED": start,
            "CDS_END_1BASED": end,
            "HGVS_REFERENCE_SEQUENCE": reference,
            "HGVS_ALTERNATE_SEQUENCE": alternate,
            "HGVS_PARSE_STATUS": "PARSED_DELINS",
        }


    # --------------------------------------------------------
    # Deletion
    # Examples:
    # c.123del
    # c.123delA
    # c.123_125del
    # c.123_125delACT
    # --------------------------------------------------------

    deletion_match = re.fullmatch(
        r"(?P<start>\d+)"
        r"(?:_(?P<end>\d+))?"
        r"DEL"
        r"(?P<reference>[ACGTN]*)",
        expression,
    )

    if deletion_match is not None:

        start = int(
            deletion_match.group("start")
        )

        end = int(
            deletion_match.group("end")
            or start
        )

        reference = (
            deletion_match.group("reference")
            or pd.NA
        )

        return {
            **base_result,
            "HGVS_OPERATION": "DELETION",
            "CDS_START_1BASED": start,
            "CDS_END_1BASED": end,
            "HGVS_REFERENCE_SEQUENCE": reference,
            "HGVS_ALTERNATE_SEQUENCE": "",
            "HGVS_PARSE_STATUS": "PARSED_DELETION",
        }


    # --------------------------------------------------------
    # Insertion
    # Example: c.123_124insA
    # --------------------------------------------------------

    insertion_match = re.fullmatch(
        r"(?P<start>\d+)"
        r"_"
        r"(?P<end>\d+)"
        r"INS"
        r"(?P<alternate>[ACGTN]+)",
        expression,
    )

    if insertion_match is not None:

        start = int(
            insertion_match.group("start")
        )

        end = int(
            insertion_match.group("end")
        )

        alternate = insertion_match.group(
            "alternate"
        )

        return {
            **base_result,
            "HGVS_OPERATION": "INSERTION",
            "CDS_START_1BASED": start,
            "CDS_END_1BASED": end,
            "HGVS_REFERENCE_SEQUENCE": "",
            "HGVS_ALTERNATE_SEQUENCE": alternate,
            "HGVS_PARSE_STATUS": "PARSED_INSERTION",
        }


    # --------------------------------------------------------
    # Duplication
    # Examples:
    # c.123dup
    # c.123dupA
    # c.123_125dup
    # c.123_125dupACT
    # --------------------------------------------------------

    duplication_match = re.fullmatch(
        r"(?P<start>\d+)"
        r"(?:_(?P<end>\d+))?"
        r"DUP"
        r"(?P<reference>[ACGTN]*)",
        expression,
    )

    if duplication_match is not None:

        start = int(
            duplication_match.group("start")
        )

        end = int(
            duplication_match.group("end")
            or start
        )

        reference = (
            duplication_match.group("reference")
            or pd.NA
        )

        return {
            **base_result,
            "HGVS_OPERATION": "DUPLICATION",
            "CDS_START_1BASED": start,
            "CDS_END_1BASED": end,
            "HGVS_REFERENCE_SEQUENCE": reference,
            "HGVS_ALTERNATE_SEQUENCE": pd.NA,
            "HGVS_PARSE_STATUS": "PARSED_DUPLICATION",
        }


    return {
        **base_result,
        "HGVS_PARSE_STATUS": "UNSUPPORTED_CODING_HGVSC",
    }


# ------------------------------------------------------------
# 4. Prepare source table
# ------------------------------------------------------------

working_df = (
    MUTATION_TRANSCRIPT_MATCH_DF
    .copy()
)


# Use the cleaned HGVSc field created in Step 9C.
# Fall back to HGVSC_RAW if the cleaned field is unavailable.
if "HGVSC_RAW_CLEAN" in working_df.columns:

    hgvsc_source_series = (
        working_df["HGVSC_RAW_CLEAN"]
    )

else:

    hgvsc_source_series = (
        working_df["HGVSC_RAW"]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# 5. Parse every mutation row
# ------------------------------------------------------------

parse_results = pd.DataFrame(
    [
        parse_hgvsc_annotation(value)
        for value in hgvsc_source_series
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        parse_results,
    ],
    axis=1,
)


for column in [
    "CDS_START_1BASED",
    "CDS_END_1BASED",
]:
    working_df[column] = (
        working_df[column]
        .astype("Int64")
    )


# ------------------------------------------------------------
# 6. Identify successfully parsed operations
# ------------------------------------------------------------

parsed_statuses = {
    "PARSED_SUBSTITUTION",
    "PARSED_DELETION",
    "PARSED_INSERTION",
    "PARSED_DUPLICATION",
    "PARSED_DELINS",
}


working_df["HGVSC_PARSED_FLAG"] = (
    working_df["HGVS_PARSE_STATUS"]
    .isin(parsed_statuses)
    .astype("int8")
)


# ------------------------------------------------------------
# 7. Validate coordinate order
# ------------------------------------------------------------

working_df["CDS_POSITION_ORDER_VALID_FLAG"] = (
    working_df["HGVSC_PARSED_FLAG"].eq(1)
    &
    working_df["CDS_START_1BASED"].notna()
    &
    working_df["CDS_END_1BASED"].notna()
    &
    working_df["CDS_START_1BASED"].ge(1)
    &
    working_df["CDS_END_1BASED"].ge(
        working_df["CDS_START_1BASED"]
    )
).astype("int8")


# ------------------------------------------------------------
# 8. Validate insertion coordinates
#
# An insertion such as c.123_124insA must be between
# two adjacent CDS positions.
# ------------------------------------------------------------

insertion_mask = (
    working_df["HGVS_OPERATION"]
    .eq("INSERTION")
)


working_df["INSERTION_ADJACENCY_VALID_FLAG"] = 1


working_df.loc[
    insertion_mask,
    "INSERTION_ADJACENCY_VALID_FLAG",
] = (
    (
        working_df.loc[
            insertion_mask,
            "CDS_END_1BASED",
        ]
        -
        working_df.loc[
            insertion_mask,
            "CDS_START_1BASED",
        ]
    )
    .eq(1)
    .astype("int8")
)


# ------------------------------------------------------------
# 9. Validate coordinates against the matched full CDS
# ------------------------------------------------------------

reference_lengths = pd.to_numeric(
    working_df["REFERENCE_CDS_LENGTH"],
    errors="coerce",
)


working_df["CDS_COORDINATES_WITHIN_REFERENCE_FLAG"] = (
    working_df["HGVSC_PARSED_FLAG"].eq(1)
    &
    working_df["TRANSCRIPT_MATCH_FLAG"].eq(1)
    &
    working_df["CDS_POSITION_ORDER_VALID_FLAG"].eq(1)
    &
    reference_lengths.notna()
    &
    working_df["CDS_END_1BASED"].le(
        reference_lengths
    )
).astype("int8")


# ------------------------------------------------------------
# 10. Final coordinate usability flag
# ------------------------------------------------------------

working_df["HGVSC_COORDINATE_USABLE_FLAG"] = (
    working_df["TRANSCRIPT_MATCH_FLAG"].eq(1)
    &
    working_df["HGVSC_PARSED_FLAG"].eq(1)
    &
    working_df["CDS_POSITION_ORDER_VALID_FLAG"].eq(1)
    &
    working_df[
        "CDS_COORDINATES_WITHIN_REFERENCE_FLAG"
    ].eq(1)
    &
    (
        ~insertion_mask
        |
        working_df[
            "INSERTION_ADJACENCY_VALID_FLAG"
        ].eq(1)
    )
).astype("int8")


# ------------------------------------------------------------
# 11. Mutation span lengths
# ------------------------------------------------------------

working_df["HGVS_REFERENCE_SPAN_LENGTH"] = (
    working_df["CDS_END_1BASED"]
    -
    working_df["CDS_START_1BASED"]
    + 1
).astype("Int64")


# Insertions do not remove either flanking reference base.
working_df.loc[
    insertion_mask,
    "HGVS_REFERENCE_SPAN_LENGTH",
] = 0


working_df["HGVS_ALTERNATE_LENGTH"] = (
    working_df["HGVS_ALTERNATE_SEQUENCE"]
    .fillna("")
    .astype(str)
    .str.len()
    .astype("Int64")
)


# Duplication alternate sequence will later be copied
# directly from the verified CDS reference span.
duplication_mask = (
    working_df["HGVS_OPERATION"]
    .eq("DUPLICATION")
)


working_df.loc[
    duplication_mask,
    "HGVS_ALTERNATE_LENGTH",
] = working_df.loc[
    duplication_mask,
    "HGVS_REFERENCE_SPAN_LENGTH",
]


# ------------------------------------------------------------
# 12. Preserve complete parsed table
# ------------------------------------------------------------

MUTATION_HGVSC_PARSED_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 13. Validation calculations
# ------------------------------------------------------------

duplicate_mutation_ids = (
    MUTATION_HGVSC_PARSED_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


parsed_outside_cds_mask = (
    MUTATION_HGVSC_PARSED_DF[
        "HGVSC_PARSED_FLAG"
    ].eq(1)
    &
    MUTATION_HGVSC_PARSED_DF[
        "TRANSCRIPT_MATCH_FLAG"
    ].eq(1)
    &
    MUTATION_HGVSC_PARSED_DF[
        "CDS_COORDINATES_WITHIN_REFERENCE_FLAG"
    ].eq(0)
)


non_adjacent_insertion_mask = (
    MUTATION_HGVSC_PARSED_DF[
        "HGVS_OPERATION"
    ].eq("INSERTION")
    &
    MUTATION_HGVSC_PARSED_DF[
        "INSERTION_ADJACENCY_VALID_FLAG"
    ].eq(0)
)


usable_missing_reference_mask = (
    MUTATION_HGVSC_PARSED_DF[
        "HGVSC_COORDINATE_USABLE_FLAG"
    ].eq(1)
    &
    MUTATION_HGVSC_PARSED_DF[
        "REFERENCE_CDS_SEQUENCE"
    ].isna()
)


parsed_outside_cds_count = int(
    parsed_outside_cds_mask.sum()
)

non_adjacent_insertion_count = int(
    non_adjacent_insertion_mask.sum()
)

usable_missing_reference_count = int(
    usable_missing_reference_mask.sum()
)


# ------------------------------------------------------------
# 14. Print combined audit
# ------------------------------------------------------------

print(
    f"\nTotal mutation rows: "
    f"{len(MUTATION_HGVSC_PARSED_DF):,}"
)

print(
    f"Transcript-matched rows: "
    f"{int(MUTATION_HGVSC_PARSED_DF['TRANSCRIPT_MATCH_FLAG'].sum()):,}"
)

print(
    f"Successfully parsed HGVSc rows: "
    f"{int(MUTATION_HGVSC_PARSED_DF['HGVSC_PARSED_FLAG'].sum()):,}"
)

print(
    f"Usable matched CDS coordinates: "
    f"{int(MUTATION_HGVSC_PARSED_DF['HGVSC_COORDINATE_USABLE_FLAG'].sum()):,}"
)

print(
    f"Parsed coordinates outside CDS: "
    f"{parsed_outside_cds_count:,}"
)

print(
    f"Non-adjacent insertion coordinates: "
    f"{non_adjacent_insertion_count:,}"
)

print(
    f"Usable rows without CDS sequence: "
    f"{usable_missing_reference_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{int(duplicate_mutation_ids.sum()):,}"
)


print("\nHGVSc parse-status counts:")

print(
    MUTATION_HGVSC_PARSED_DF[
        "HGVS_PARSE_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print("\nParsed operation counts:")

print(
    MUTATION_HGVSC_PARSED_DF[
        "HGVS_OPERATION"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 15. Cohort-level summary
# ------------------------------------------------------------

hgvsc_summary_df = (
    MUTATION_HGVSC_PARSED_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),
        transcript_matched=(
            "TRANSCRIPT_MATCH_FLAG",
            "sum",
        ),
        hgvsc_parsed=(
            "HGVSC_PARSED_FLAG",
            "sum",
        ),
        usable_cds_coordinates=(
            "HGVSC_COORDINATE_USABLE_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print("\nHGVSc parsing by cohort:")

print(
    hgvsc_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 16. Show unresolved matched-transcript examples
# ------------------------------------------------------------

unresolved_examples = (
    MUTATION_HGVSC_PARSED_DF.loc[
        MUTATION_HGVSC_PARSED_DF[
            "TRANSCRIPT_MATCH_FLAG"
        ].eq(1)
        &
        MUTATION_HGVSC_PARSED_DF[
            "HGVSC_COORDINATE_USABLE_FLAG"
        ].eq(0),
        [
            "MUTATION_ROW_ID",
            "COHORT",
            "GENE_SYMBOL",
            "TRANSCRIPT_ID_SELECTED",
            "HGVSC_RAW",
            "HGVS_C_EXPRESSION",
            "HGVS_PARSE_STATUS",
            "CDS_START_1BASED",
            "CDS_END_1BASED",
            "REFERENCE_CDS_LENGTH",
        ],
    ]
    .head(20)
)


if not unresolved_examples.empty:

    print(
        "\nMatched-transcript rows without usable "
        "CDS coordinates:"
    )

    print(
        unresolved_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 17. Hard validations
# ------------------------------------------------------------

if len(MUTATION_HGVSC_PARSED_DF) != len(
    MUTATION_TRANSCRIPT_MATCH_DF
):
    raise ValueError(
        "HGVSc parsing changed the total number of mutation rows."
    )

if duplicate_mutation_ids.any():
    raise ValueError(
        "HGVSc parsing duplicated mutation rows."
    )

if usable_missing_reference_mask.any():
    raise ValueError(
        "A coordinate-usable mutation is missing its CDS sequence."
    )


print("\n" + "=" * 78)
print("STEP 9D FINISHED")
print("=" * 78)

print("Parsed mutation table:")
print("MUTATION_HGVSC_PARSED_DF")

print(
    "\nZFHX4 and EPPK1 transcript-specific unmatched rows "
    "remain preserved for the final exclusion audit."
)
```

## Step 9E — Verify CDS Reference Alleles and Prepare Mutation Payloads

This step validates every coordinate-usable mutation against its matched full Ensembl CDS.

For each supported operation, it derives:

- the actual CDS reference span
- the sequence removed from the reference
- the sequence inserted into the mutated CDS
- mutation application mode
- reference and alternate lengths
- net CDS length change

When an HGVSc reference allele is explicitly provided, it must match the Ensembl CDS sequence at the parsed coding position. For deletions, duplications and deletion-insertions where HGVSc omits the reference bases, the reference span is derived directly from the matched CDS.

Original mutation-file reference and tumour alleles are retained as an orientation audit. They are not used to override transcript-oriented HGVSc alleles because source mutation alleles may be reported on the genomic strand or use a different indel representation.

```python
# ============================================================
# STEP 9E — VERIFY CDS REFERENCE ALLELES AND PREPARE PAYLOADS
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_HGVSC_PARSED_DF" not in globals():
    raise RuntimeError(
        "MUTATION_HGVSC_PARSED_DF was not found. "
        "Run Step 9D first."
    )


print("=" * 78)
print("STEP 9E — CDS REFERENCE-ALLELE VERIFICATION")
print("=" * 78)


# ------------------------------------------------------------
# 2. Constants and helper functions
# ------------------------------------------------------------

DNA_PATTERN = re.compile(r"^[ACGTN]*$")

RC_TABLE = str.maketrans(
    "ACGTN",
    "TGCAN",
)


def clean_hgvs_sequence(value):
    """
    Return None, a valid uppercase DNA sequence,
    or '__INVALID__'.
    """

    if value is None or pd.isna(value):
        return None

    text = str(value).strip().upper()

    if text == "":
        return ""

    if DNA_PATTERN.fullmatch(text):
        return text

    return "__INVALID__"


def normalise_source_allele(value):
    """
    Normalise source mutation-file alleles for audit only.
    """

    if value is None or pd.isna(value):
        return None

    text = (
        str(value)
        .strip()
        .upper()
        .replace(" ", "")
    )

    if text in {
        "",
        "-",
        "--",
        ".",
    }:
        return ""

    if text in {
        "NA",
        "N/A",
        "NAN",
        "NONE",
        "NULL",
        "UNKNOWN",
    }:
        return None

    if DNA_PATTERN.fullmatch(text):
        return text

    return "__INVALID__"


def reverse_complement(sequence):
    """
    Return the reverse complement of an A/C/G/T/N sequence.
    """

    if sequence is None:
        return None

    return sequence.translate(
        RC_TABLE
    )[::-1]


def empty_payload(status):
    """
    Standard result for a mutation that cannot be used.
    """

    return {
        "CDS_REFERENCE_SPAN": pd.NA,
        "HGVS_REFERENCE_PROVIDED_FLAG": 0,
        "HGVS_REFERENCE_MATCH_FLAG": pd.NA,
        "REFERENCE_SPAN_CONTAINS_N_FLAG": pd.NA,
        "REFERENCE_REMOVED_SEQUENCE": pd.NA,
        "ALTERNATE_INSERTED_SEQUENCE": pd.NA,
        "MUTATION_APPLICATION_MODE": pd.NA,
        "REFERENCE_REMOVAL_LENGTH": pd.NA,
        "ALTERNATE_INSERTION_LENGTH": pd.NA,
        "NET_CDS_LENGTH_CHANGE": pd.NA,
        "REFERENCE_VERIFICATION_STATUS": status,
        "REFERENCE_VERIFIED_FLAG": 0,
        "MUTATION_PAYLOAD_USABLE_FLAG": 0,
    }


def mismatch_payload(
    actual_reference,
    reference_provided_flag,
):
    """
    Return an HGVSc/Ensembl reference mismatch result.
    """

    result = empty_payload(
        "HGVSC_REFERENCE_MISMATCH"
    )

    result.update({
        "CDS_REFERENCE_SPAN": actual_reference,
        "HGVS_REFERENCE_PROVIDED_FLAG": (
            reference_provided_flag
        ),
        "HGVS_REFERENCE_MATCH_FLAG": 0,
        "REFERENCE_SPAN_CONTAINS_N_FLAG": int(
            "N" in actual_reference
        ),
    })

    return result


def n_span_payload(
    actual_reference,
    reference_provided_flag,
):
    """
    Return a result for a reference span containing N.
    """

    result = empty_payload(
        "REFERENCE_SPAN_CONTAINS_N"
    )

    result.update({
        "CDS_REFERENCE_SPAN": actual_reference,
        "HGVS_REFERENCE_PROVIDED_FLAG": (
            reference_provided_flag
        ),
        "REFERENCE_SPAN_CONTAINS_N_FLAG": 1,
    })

    return result


# ------------------------------------------------------------
# 3. Verify one mutation and create its payload
# ------------------------------------------------------------

def derive_mutation_payload(row):
    """
    Verify one mutation against its matched full CDS.

    The returned payload describes the exact reference sequence
    removed and tumour sequence inserted.
    """

    transcript_flag = row[
        "TRANSCRIPT_MATCH_FLAG"
    ]

    coordinate_flag = row[
        "HGVSC_COORDINATE_USABLE_FLAG"
    ]


    if (
        pd.isna(transcript_flag)
        or int(transcript_flag) != 1
    ):
        return empty_payload(
            "TRANSCRIPT_NOT_MATCHED"
        )


    if (
        pd.isna(coordinate_flag)
        or int(coordinate_flag) != 1
    ):
        return empty_payload(
            "HGVSC_COORDINATE_NOT_USABLE"
        )


    sequence = row[
        "REFERENCE_CDS_SEQUENCE"
    ]


    if sequence is None or pd.isna(sequence):
        return empty_payload(
            "REFERENCE_CDS_MISSING"
        )


    sequence = (
        str(sequence)
        .strip()
        .upper()
    )


    if (
        not sequence
        or not DNA_PATTERN.fullmatch(sequence)
    ):
        return empty_payload(
            "REFERENCE_CDS_INVALID"
        )


    operation = row[
        "HGVS_OPERATION"
    ]


    if operation is None or pd.isna(operation):
        return empty_payload(
            "MUTATION_OPERATION_MISSING"
        )


    operation = (
        str(operation)
        .strip()
        .upper()
    )


    start = int(
        row["CDS_START_1BASED"]
    )

    end = int(
        row["CDS_END_1BASED"]
    )


    if (
        start < 1
        or end < start
        or end > len(sequence)
    ):
        return empty_payload(
            "CDS_COORDINATE_OUT_OF_RANGE"
        )


    hgvs_reference = clean_hgvs_sequence(
        row["HGVS_REFERENCE_SEQUENCE"]
    )

    hgvs_alternate = clean_hgvs_sequence(
        row["HGVS_ALTERNATE_SEQUENCE"]
    )


    if hgvs_reference == "__INVALID__":
        return empty_payload(
            "INVALID_HGVSC_REFERENCE_SEQUENCE"
        )


    if hgvs_alternate == "__INVALID__":
        return empty_payload(
            "INVALID_HGVSC_ALTERNATE_SEQUENCE"
        )


    reference_provided_flag = int(
        hgvs_reference not in {
            None,
            "",
        }
    )


    # --------------------------------------------------------
    # Substitution
    # --------------------------------------------------------

    if operation == "SUBSTITUTION":

        actual_reference = sequence[
            start - 1:end
        ]


        if (
            hgvs_reference is None
            or hgvs_alternate is None
        ):
            return empty_payload(
                "SUBSTITUTION_ALLELE_MISSING"
            )


        if (
            start != end
            or len(hgvs_reference) != 1
            or len(hgvs_alternate) != 1
        ):
            return empty_payload(
                "SUBSTITUTION_LENGTH_INVALID"
            )


        if actual_reference != hgvs_reference:

            return mismatch_payload(
                actual_reference,
                reference_provided_flag,
            )


        removed_sequence = actual_reference
        inserted_sequence = hgvs_alternate

        application_mode = (
            "REPLACE_REFERENCE_SPAN"
        )

        verification_status = (
            "REFERENCE_MATCHED_HGVSC"
        )

        reference_match_flag = 1


    # --------------------------------------------------------
    # Deletion, delins and duplication
    # --------------------------------------------------------

    elif operation in {
        "DELETION",
        "DELINS",
        "DUPLICATION",
    }:

        actual_reference = sequence[
            start - 1:end
        ]


        if "N" in actual_reference:

            return n_span_payload(
                actual_reference,
                reference_provided_flag,
            )


        if (
            operation == "DELINS"
            and hgvs_alternate in {
                None,
                "",
            }
        ):
            return empty_payload(
                "DELINS_ALTERNATE_MISSING"
            )


        if reference_provided_flag == 1:

            if actual_reference != hgvs_reference:

                return mismatch_payload(
                    actual_reference,
                    reference_provided_flag,
                )


            verification_status = (
                "REFERENCE_MATCHED_HGVSC"
            )

            reference_match_flag = 1


        else:

            verification_status = (
                "REFERENCE_DERIVED_FROM_CDS"
            )

            reference_match_flag = pd.NA


        if operation == "DELETION":

            removed_sequence = actual_reference
            inserted_sequence = ""

            application_mode = (
                "REPLACE_REFERENCE_SPAN"
            )


        elif operation == "DELINS":

            removed_sequence = actual_reference
            inserted_sequence = hgvs_alternate

            application_mode = (
                "REPLACE_REFERENCE_SPAN"
            )


        else:

            # Duplication inserts a copy after the
            # duplicated reference span.
            removed_sequence = ""
            inserted_sequence = actual_reference

            application_mode = (
                "INSERT_AFTER_REFERENCE_SPAN"
            )


    # --------------------------------------------------------
    # Insertion
    # --------------------------------------------------------

    elif operation == "INSERTION":

        if hgvs_alternate in {
            None,
            "",
        }:
            return empty_payload(
                "INSERTION_ALTERNATE_MISSING"
            )


        if end != start + 1:
            return empty_payload(
                "INSERTION_COORDINATES_NOT_ADJACENT"
            )


        actual_reference = ""
        removed_sequence = ""
        inserted_sequence = hgvs_alternate

        application_mode = (
            "INSERT_BETWEEN_CDS_BASES"
        )

        verification_status = (
            "INSERTION_FLANKS_VALID"
        )

        reference_match_flag = pd.NA


    else:

        return empty_payload(
            "UNSUPPORTED_MUTATION_OPERATION"
        )


    # --------------------------------------------------------
    # Final payload validation
    # --------------------------------------------------------

    if not DNA_PATTERN.fullmatch(
        removed_sequence
    ):
        return empty_payload(
            "REMOVED_SEQUENCE_INVALID"
        )


    if not DNA_PATTERN.fullmatch(
        inserted_sequence
    ):
        return empty_payload(
            "INSERTED_SEQUENCE_INVALID"
        )


    removal_length = len(
        removed_sequence
    )

    insertion_length = len(
        inserted_sequence
    )


    return {
        "CDS_REFERENCE_SPAN": actual_reference,

        "HGVS_REFERENCE_PROVIDED_FLAG": (
            reference_provided_flag
        ),

        "HGVS_REFERENCE_MATCH_FLAG": (
            reference_match_flag
        ),

        "REFERENCE_SPAN_CONTAINS_N_FLAG": int(
            "N" in actual_reference
        ),

        "REFERENCE_REMOVED_SEQUENCE": (
            removed_sequence
        ),

        "ALTERNATE_INSERTED_SEQUENCE": (
            inserted_sequence
        ),

        "MUTATION_APPLICATION_MODE": (
            application_mode
        ),

        "REFERENCE_REMOVAL_LENGTH": (
            removal_length
        ),

        "ALTERNATE_INSERTION_LENGTH": (
            insertion_length
        ),

        "NET_CDS_LENGTH_CHANGE": (
            insertion_length
            -
            removal_length
        ),

        "REFERENCE_VERIFICATION_STATUS": (
            verification_status
        ),

        "REFERENCE_VERIFIED_FLAG": 1,

        "MUTATION_PAYLOAD_USABLE_FLAG": 1,
    }


# ------------------------------------------------------------
# 4. Prepare rerunnable working table
# ------------------------------------------------------------

working_df = (
    MUTATION_HGVSC_PARSED_DF
    .copy()
)


derived_columns = [
    "CDS_REFERENCE_SPAN",
    "HGVS_REFERENCE_PROVIDED_FLAG",
    "HGVS_REFERENCE_MATCH_FLAG",
    "REFERENCE_SPAN_CONTAINS_N_FLAG",
    "REFERENCE_REMOVED_SEQUENCE",
    "ALTERNATE_INSERTED_SEQUENCE",
    "MUTATION_APPLICATION_MODE",
    "REFERENCE_REMOVAL_LENGTH",
    "ALTERNATE_INSERTION_LENGTH",
    "NET_CDS_LENGTH_CHANGE",
    "REFERENCE_VERIFICATION_STATUS",
    "REFERENCE_VERIFIED_FLAG",
    "MUTATION_PAYLOAD_USABLE_FLAG",
    "SOURCE_REFERENCE_ALLELE_NORMALISED",
    "SOURCE_TUMOUR_ALLELE_NORMALISED",
    "SOURCE_ALLELE_AUDIT_STATUS",
]


working_df = working_df.drop(
    columns=[
        column
        for column in derived_columns
        if column in working_df.columns
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 5. Derive mutation payloads
# ------------------------------------------------------------

payload_results = pd.DataFrame(
    [
        derive_mutation_payload(row)
        for _, row in working_df.iterrows()
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        payload_results,
    ],
    axis=1,
)


# ------------------------------------------------------------
# 6. Normalise original source alleles
# ------------------------------------------------------------

working_df[
    "SOURCE_REFERENCE_ALLELE_NORMALISED"
] = (
    working_df[
        "REFERENCE_ALLELE_RAW"
    ]
    .apply(normalise_source_allele)
    .astype("string")
)


working_df[
    "SOURCE_TUMOUR_ALLELE_NORMALISED"
] = (
    working_df[
        "TUMOUR_ALLELE_RAW"
    ]
    .apply(normalise_source_allele)
    .astype("string")
)


# ------------------------------------------------------------
# 7. Audit source-file allele orientation
# ------------------------------------------------------------

def audit_source_alleles(row):
    """
    Compare source-file alleles with the verified
    transcript-oriented payload.

    This audit does not override HGVSc/CDS verification.
    """

    if int(
        row["MUTATION_PAYLOAD_USABLE_FLAG"]
    ) != 1:
        return (
            "NOT_EVALUATED_PAYLOAD_UNUSABLE"
        )


    source_reference = row[
        "SOURCE_REFERENCE_ALLELE_NORMALISED"
    ]

    source_tumour = row[
        "SOURCE_TUMOUR_ALLELE_NORMALISED"
    ]


    if (
        pd.isna(source_reference)
        or pd.isna(source_tumour)
    ):
        return "SOURCE_ALLELE_MISSING"


    source_reference = str(
        source_reference
    )

    source_tumour = str(
        source_tumour
    )


    if (
        source_reference == "__INVALID__"
        or source_tumour == "__INVALID__"
    ):
        return "SOURCE_ALLELE_UNSUPPORTED"


    expected_reference = str(
        row["REFERENCE_REMOVED_SEQUENCE"]
    )

    expected_alternate = str(
        row["ALTERNATE_INSERTED_SEQUENCE"]
    )


    if (
        source_reference == expected_reference
        and source_tumour == expected_alternate
    ):
        return (
            "DIRECT_TRANSCRIPT_ORIENTATION_MATCH"
        )


    reverse_reference = reverse_complement(
        source_reference
    )

    reverse_tumour = reverse_complement(
        source_tumour
    )


    if (
        reverse_reference == expected_reference
        and reverse_tumour == expected_alternate
    ):
        return "REVERSE_COMPLEMENT_MATCH"


    return (
        "DIFFERENT_REPRESENTATION_OR_MISMATCH"
    )


working_df[
    "SOURCE_ALLELE_AUDIT_STATUS"
] = (
    working_df.apply(
        audit_source_alleles,
        axis=1,
    )
)


# ------------------------------------------------------------
# 8. Final verified mutation table
# ------------------------------------------------------------

MUTATION_REFERENCE_VERIFIED_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 9. Validation masks
# ------------------------------------------------------------

duplicate_mutation_ids = (
    MUTATION_REFERENCE_VERIFIED_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


coordinate_usable_mask = (
    MUTATION_REFERENCE_VERIFIED_DF[
        "HGVSC_COORDINATE_USABLE_FLAG"
    ]
    .eq(1)
)


usable_mask = (
    MUTATION_REFERENCE_VERIFIED_DF[
        "MUTATION_PAYLOAD_USABLE_FLAG"
    ]
    .eq(1)
)


excluded_during_verification_mask = (
    coordinate_usable_mask
    &
    ~usable_mask
)


usable_missing_payload_mask = (
    usable_mask
    &
    (
        MUTATION_REFERENCE_VERIFIED_DF[
            "REFERENCE_REMOVED_SEQUENCE"
        ]
        .isna()

        |

        MUTATION_REFERENCE_VERIFIED_DF[
            "ALTERNATE_INSERTED_SEQUENCE"
        ]
        .isna()

        |

        MUTATION_REFERENCE_VERIFIED_DF[
            "MUTATION_APPLICATION_MODE"
        ]
        .isna()
    )
)


usable_unverified_mask = (
    usable_mask
    &
    MUTATION_REFERENCE_VERIFIED_DF[
        "REFERENCE_VERIFIED_FLAG"
    ]
    .ne(1)
)


removed_valid_mask = (
    MUTATION_REFERENCE_VERIFIED_DF[
        "REFERENCE_REMOVED_SEQUENCE"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[ACGTN]*")
)


inserted_valid_mask = (
    MUTATION_REFERENCE_VERIFIED_DF[
        "ALTERNATE_INSERTED_SEQUENCE"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[ACGTN]*")
)


payload_character_error_mask = (
    usable_mask
    &
    (
        ~removed_valid_mask
        |
        ~inserted_valid_mask
    )
)


# ------------------------------------------------------------
# 10. Precalculate all printed counts
# ------------------------------------------------------------

total_rows = len(
    MUTATION_REFERENCE_VERIFIED_DF
)


coordinate_usable_count = int(
    coordinate_usable_mask.sum()
)


usable_payload_count = int(
    usable_mask.sum()
)


excluded_verification_count = int(
    excluded_during_verification_mask.sum()
)


duplicate_id_count = int(
    duplicate_mutation_ids.sum()
)


missing_payload_count = int(
    usable_missing_payload_mask.sum()
)


usable_unverified_count = int(
    usable_unverified_mask.sum()
)


invalid_payload_character_count = int(
    payload_character_error_mask.sum()
)


# ------------------------------------------------------------
# 11. Print validation results
# ------------------------------------------------------------

print(
    f"\nTotal mutation rows: "
    f"{total_rows:,}"
)


print(
    "Coordinate-usable rows entering verification: "
    f"{coordinate_usable_count:,}"
)


print(
    "Reference-verified usable payloads: "
    f"{usable_payload_count:,}"
)


print(
    "Rows excluded during reference verification: "
    f"{excluded_verification_count:,}"
)


print(
    f"Duplicate mutation IDs: "
    f"{duplicate_id_count:,}"
)


print(
    "Usable rows with missing payload fields: "
    f"{missing_payload_count:,}"
)


print(
    "Usable rows not reference-verified: "
    f"{usable_unverified_count:,}"
)


print(
    "Usable rows with invalid payload characters: "
    f"{invalid_payload_character_count:,}"
)


print(
    "\nReference-verification status counts:"
)


print(
    MUTATION_REFERENCE_VERIFIED_DF[
        "REFERENCE_VERIFICATION_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nUsable payloads by mutation operation:"
)


print(
    MUTATION_REFERENCE_VERIFIED_DF.loc[
        usable_mask,
        "HGVS_OPERATION",
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nSource-allele audit counts:"
)


print(
    MUTATION_REFERENCE_VERIFIED_DF[
        "SOURCE_ALLELE_AUDIT_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 12. Cohort-level summary
# ------------------------------------------------------------

verification_summary_df = (
    MUTATION_REFERENCE_VERIFIED_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        coordinate_usable=(
            "HGVSC_COORDINATE_USABLE_FLAG",
            "sum",
        ),

        reference_verified=(
            "REFERENCE_VERIFIED_FLAG",
            "sum",
        ),

        payload_usable=(
            "MUTATION_PAYLOAD_USABLE_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print(
    "\nReference verification by cohort:"
)


print(
    verification_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 13. Reference-mismatch examples
# ------------------------------------------------------------

reference_mismatch_examples = (
    MUTATION_REFERENCE_VERIFIED_DF.loc[
        MUTATION_REFERENCE_VERIFIED_DF[
            "REFERENCE_VERIFICATION_STATUS"
        ]
        .eq("HGVSC_REFERENCE_MISMATCH"),

        [
            "MUTATION_ROW_ID",
            "COHORT",
            "GENE_SYMBOL",
            "TRANSCRIPT_ID_SELECTED",
            "HGVSC_RAW",
            "CDS_START_1BASED",
            "CDS_END_1BASED",
            "HGVS_REFERENCE_SEQUENCE",
            "CDS_REFERENCE_SPAN",
            "REFERENCE_ALLELE_RAW",
            "TUMOUR_ALLELE_RAW",
        ],
    ]
    .head(20)
)


if not reference_mismatch_examples.empty:

    print(
        "\nReference-mismatch examples:"
    )

    print(
        reference_mismatch_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 14. Hard validations
# ------------------------------------------------------------

if len(
    MUTATION_REFERENCE_VERIFIED_DF
) != len(
    MUTATION_HGVSC_PARSED_DF
):
    raise ValueError(
        "Reference verification changed the "
        "mutation-row count."
    )


if duplicate_mutation_ids.any():
    raise ValueError(
        "Reference verification duplicated "
        "mutation rows."
    )


if usable_missing_payload_mask.any():
    raise ValueError(
        "A usable mutation is missing "
        "application payload fields."
    )


if usable_unverified_mask.any():
    raise ValueError(
        "An unverified mutation entered the "
        "usable payload set."
    )


if payload_character_error_mask.any():
    raise ValueError(
        "A usable mutation payload contains "
        "invalid characters."
    )


print("\n" + "=" * 78)
print("STEP 9E FINISHED")
print("=" * 78)

print(
    "Reference-verified mutation table:"
)

print(
    "MUTATION_REFERENCE_VERIFIED_DF"
)

print(
    "\nOnly reference-verified payloads will be used "
    "for mutated CDS and 301-bp window construction."
)
```

## Step 9F — Construct Fresh 301-bp Mutation-Centred Reference Windows

For every reference-verified mutation, a new 301-bp reference window is extracted directly from the matched full Ensembl CDS.

Window structure:

- 150 positions before the mutation anchor
- one central anchor position
- 150 positions after the mutation anchor
- total length: 301 bases

Anchor definition:

- substitutions: mutated CDS position
- deletions, delins and duplications: midpoint of the affected CDS span
- insertions: left flanking CDS position

When a window extends beyond the beginning or end of the CDS, only the unavailable boundary positions are padded with `N`.

The step also records the mutation's local window coordinates and checks whether the complete affected span is contained inside the 301-bp window.

```python
# ============================================================
# STEP 9F — CONSTRUCT FRESH 301-BP REFERENCE WINDOWS
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_REFERENCE_VERIFIED_DF" not in globals():
    raise RuntimeError(
        "MUTATION_REFERENCE_VERIFIED_DF was not found. "
        "Run Step 9E first."
    )


print("=" * 78)
print("STEP 9F — FRESH 301-BP REFERENCE WINDOW CONSTRUCTION")
print("=" * 78)


# ------------------------------------------------------------
# 2. Fixed window configuration
# ------------------------------------------------------------

WINDOW_SIZE = 301
HALF_WINDOW = 150

WINDOW_CENTER_INDEX_0BASED = 150
WINDOW_CENTER_INDEX_1BASED = 151

DNA_PATTERN_301 = re.compile(
    r"^[ACGTN]{301}$"
)


# ------------------------------------------------------------
# 3. Standard empty result
# ------------------------------------------------------------

def empty_window_result(status):
    """
    Return a consistent result for a row without a usable
    301-bp reference window.
    """

    return {
        "WINDOW_SIZE": WINDOW_SIZE,
        "WINDOW_HALF_SIZE": HALF_WINDOW,
        "WINDOW_CENTER_INDEX_0BASED": (
            WINDOW_CENTER_INDEX_0BASED
        ),
        "WINDOW_CENTER_INDEX_1BASED": (
            WINDOW_CENTER_INDEX_1BASED
        ),

        "WINDOW_ANCHOR_CDS_POSITION_1BASED": pd.NA,

        "WINDOW_REQUESTED_CDS_START_1BASED": pd.NA,
        "WINDOW_REQUESTED_CDS_END_1BASED": pd.NA,

        "WINDOW_EXTRACTED_CDS_START_1BASED": pd.NA,
        "WINDOW_EXTRACTED_CDS_END_1BASED": pd.NA,

        "WINDOW_LEFT_PADDING_COUNT": pd.NA,
        "WINDOW_RIGHT_PADDING_COUNT": pd.NA,
        "WINDOW_HAS_PADDING_FLAG": pd.NA,

        "REFERENCE_WINDOW_301": pd.NA,
        "REFERENCE_WINDOW_LENGTH": pd.NA,

        "EVENT_WINDOW_START_INDEX_0BASED": pd.NA,
        "EVENT_WINDOW_END_INDEX_0BASED": pd.NA,
        "INSERTION_WINDOW_INDEX_0BASED": pd.NA,

        "EVENT_FULLY_CONTAINED_FLAG": 0,

        "CDS_ANCHOR_BASE": pd.NA,
        "WINDOW_CENTER_BASE": pd.NA,
        "WINDOW_CENTER_BASE_MATCH_FLAG": 0,

        "REFERENCE_WINDOW_BUILT_FLAG": 0,
        "REFERENCE_WINDOW_ELIGIBLE_FLAG": 0,
        "REFERENCE_WINDOW_STATUS": status,
    }


# ------------------------------------------------------------
# 4. Construct one reference window
# ------------------------------------------------------------

def construct_reference_window(row):
    """
    Construct one fixed-length mutation-centred reference window.
    """

    payload_flag = row[
        "MUTATION_PAYLOAD_USABLE_FLAG"
    ]


    if (
        pd.isna(payload_flag)
        or int(payload_flag) != 1
    ):
        return empty_window_result(
            "MUTATION_PAYLOAD_NOT_USABLE"
        )


    sequence = row[
        "REFERENCE_CDS_SEQUENCE"
    ]


    if sequence is None or pd.isna(sequence):
        return empty_window_result(
            "REFERENCE_CDS_MISSING"
        )


    sequence = (
        str(sequence)
        .strip()
        .upper()
    )


    if not sequence:
        return empty_window_result(
            "REFERENCE_CDS_EMPTY"
        )


    if not re.fullmatch(
        r"[ACGTN]+",
        sequence,
    ):
        return empty_window_result(
            "REFERENCE_CDS_INVALID_CHARACTERS"
        )


    start_value = row[
        "CDS_START_1BASED"
    ]

    end_value = row[
        "CDS_END_1BASED"
    ]


    if (
        pd.isna(start_value)
        or pd.isna(end_value)
    ):
        return empty_window_result(
            "CDS_COORDINATE_MISSING"
        )


    start_position = int(
        start_value
    )

    end_position = int(
        end_value
    )

    cds_length = len(sequence)


    if (
        start_position < 1
        or end_position < start_position
        or end_position > cds_length
    ):
        return empty_window_result(
            "CDS_COORDINATE_OUT_OF_RANGE"
        )


    operation = str(
        row["HGVS_OPERATION"]
    ).strip().upper()


    # --------------------------------------------------------
    # Mutation anchor
    #
    # For an even-length affected span, integer division selects
    # the left-middle CDS position.
    #
    # For an insertion, start and end are adjacent positions,
    # so this gives the left flanking position.
    # --------------------------------------------------------

    anchor_position = (
        start_position + end_position
    ) // 2


    requested_start = (
        anchor_position - HALF_WINDOW
    )

    requested_end = (
        anchor_position + HALF_WINDOW
    )


    extracted_start = max(
        1,
        requested_start,
    )

    extracted_end = min(
        cds_length,
        requested_end,
    )


    left_padding = max(
        0,
        1 - requested_start,
    )

    right_padding = max(
        0,
        requested_end - cds_length,
    )


    extracted_sequence = sequence[
        extracted_start - 1:
        extracted_end
    ]


    reference_window = (
        ("N" * left_padding)
        +
        extracted_sequence
        +
        ("N" * right_padding)
    )


    window_length = len(
        reference_window
    )


    if window_length != WINDOW_SIZE:

        result = empty_window_result(
            "REFERENCE_WINDOW_LENGTH_ERROR"
        )

        result.update({
            "WINDOW_ANCHOR_CDS_POSITION_1BASED": (
                anchor_position
            ),
            "WINDOW_REQUESTED_CDS_START_1BASED": (
                requested_start
            ),
            "WINDOW_REQUESTED_CDS_END_1BASED": (
                requested_end
            ),
            "WINDOW_EXTRACTED_CDS_START_1BASED": (
                extracted_start
            ),
            "WINDOW_EXTRACTED_CDS_END_1BASED": (
                extracted_end
            ),
            "WINDOW_LEFT_PADDING_COUNT": (
                left_padding
            ),
            "WINDOW_RIGHT_PADDING_COUNT": (
                right_padding
            ),
            "REFERENCE_WINDOW_LENGTH": (
                window_length
            ),
        })

        return result


    if not DNA_PATTERN_301.fullmatch(
        reference_window
    ):
        return empty_window_result(
            "REFERENCE_WINDOW_INVALID_CHARACTERS"
        )


    # --------------------------------------------------------
    # Convert CDS coordinates into local window coordinates
    # --------------------------------------------------------

    event_start_index = (
        start_position - requested_start
    )

    event_end_index = (
        end_position - requested_start
    )


    insertion_index = pd.NA


    if operation == "INSERTION":

        # String insertion index immediately after the
        # left flanking CDS base.
        insertion_index = (
            start_position
            -
            requested_start
            +
            1
        )


        event_fully_contained = (
            event_start_index >= 0
            and event_start_index < WINDOW_SIZE
            and event_end_index >= 0
            and event_end_index < WINDOW_SIZE
            and insertion_index >= 1
            and insertion_index < WINDOW_SIZE
        )


    else:

        event_fully_contained = (
            event_start_index >= 0
            and event_end_index < WINDOW_SIZE
            and event_start_index <= event_end_index
        )


    # --------------------------------------------------------
    # Confirm that the biological anchor maps to window index 150
    # --------------------------------------------------------

    cds_anchor_base = sequence[
        anchor_position - 1
    ]

    window_center_base = reference_window[
        WINDOW_CENTER_INDEX_0BASED
    ]


    center_base_match = int(
        cds_anchor_base
        ==
        window_center_base
    )


    built_flag = 1

    eligible_flag = int(
        event_fully_contained
        and center_base_match == 1
    )


    if eligible_flag == 1:

        status = (
            "REFERENCE_WINDOW_BUILT_AND_EVENT_CONTAINED"
        )

    elif not event_fully_contained:

        status = (
            "REFERENCE_WINDOW_BUILT_EVENT_NOT_FULLY_CONTAINED"
        )

    else:

        status = (
            "REFERENCE_WINDOW_CENTER_MAPPING_ERROR"
        )


    return {
        "WINDOW_SIZE": WINDOW_SIZE,
        "WINDOW_HALF_SIZE": HALF_WINDOW,

        "WINDOW_CENTER_INDEX_0BASED": (
            WINDOW_CENTER_INDEX_0BASED
        ),

        "WINDOW_CENTER_INDEX_1BASED": (
            WINDOW_CENTER_INDEX_1BASED
        ),

        "WINDOW_ANCHOR_CDS_POSITION_1BASED": (
            anchor_position
        ),

        "WINDOW_REQUESTED_CDS_START_1BASED": (
            requested_start
        ),

        "WINDOW_REQUESTED_CDS_END_1BASED": (
            requested_end
        ),

        "WINDOW_EXTRACTED_CDS_START_1BASED": (
            extracted_start
        ),

        "WINDOW_EXTRACTED_CDS_END_1BASED": (
            extracted_end
        ),

        "WINDOW_LEFT_PADDING_COUNT": (
            left_padding
        ),

        "WINDOW_RIGHT_PADDING_COUNT": (
            right_padding
        ),

        "WINDOW_HAS_PADDING_FLAG": int(
            left_padding > 0
            or right_padding > 0
        ),

        "REFERENCE_WINDOW_301": (
            reference_window
        ),

        "REFERENCE_WINDOW_LENGTH": (
            window_length
        ),

        "EVENT_WINDOW_START_INDEX_0BASED": (
            event_start_index
        ),

        "EVENT_WINDOW_END_INDEX_0BASED": (
            event_end_index
        ),

        "INSERTION_WINDOW_INDEX_0BASED": (
            insertion_index
        ),

        "EVENT_FULLY_CONTAINED_FLAG": int(
            event_fully_contained
        ),

        "CDS_ANCHOR_BASE": (
            cds_anchor_base
        ),

        "WINDOW_CENTER_BASE": (
            window_center_base
        ),

        "WINDOW_CENTER_BASE_MATCH_FLAG": (
            center_base_match
        ),

        "REFERENCE_WINDOW_BUILT_FLAG": (
            built_flag
        ),

        "REFERENCE_WINDOW_ELIGIBLE_FLAG": (
            eligible_flag
        ),

        "REFERENCE_WINDOW_STATUS": (
            status
        ),
    }


# ------------------------------------------------------------
# 5. Prepare rerunnable source table
# ------------------------------------------------------------

working_df = (
    MUTATION_REFERENCE_VERIFIED_DF
    .copy()
)


window_columns = [
    "WINDOW_SIZE",
    "WINDOW_HALF_SIZE",
    "WINDOW_CENTER_INDEX_0BASED",
    "WINDOW_CENTER_INDEX_1BASED",
    "WINDOW_ANCHOR_CDS_POSITION_1BASED",
    "WINDOW_REQUESTED_CDS_START_1BASED",
    "WINDOW_REQUESTED_CDS_END_1BASED",
    "WINDOW_EXTRACTED_CDS_START_1BASED",
    "WINDOW_EXTRACTED_CDS_END_1BASED",
    "WINDOW_LEFT_PADDING_COUNT",
    "WINDOW_RIGHT_PADDING_COUNT",
    "WINDOW_HAS_PADDING_FLAG",
    "REFERENCE_WINDOW_301",
    "REFERENCE_WINDOW_LENGTH",
    "EVENT_WINDOW_START_INDEX_0BASED",
    "EVENT_WINDOW_END_INDEX_0BASED",
    "INSERTION_WINDOW_INDEX_0BASED",
    "EVENT_FULLY_CONTAINED_FLAG",
    "CDS_ANCHOR_BASE",
    "WINDOW_CENTER_BASE",
    "WINDOW_CENTER_BASE_MATCH_FLAG",
    "REFERENCE_WINDOW_BUILT_FLAG",
    "REFERENCE_WINDOW_ELIGIBLE_FLAG",
    "REFERENCE_WINDOW_STATUS",
]


working_df = working_df.drop(
    columns=[
        column
        for column in window_columns
        if column in working_df.columns
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 6. Construct all reference windows
# ------------------------------------------------------------

window_results = pd.DataFrame(
    [
        construct_reference_window(row)
        for _, row in working_df.iterrows()
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        window_results,
    ],
    axis=1,
)


MUTATION_301_REFERENCE_WINDOW_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 7. Convert numerical fields to nullable integers
# ------------------------------------------------------------

nullable_integer_columns = [
    "WINDOW_SIZE",
    "WINDOW_HALF_SIZE",
    "WINDOW_CENTER_INDEX_0BASED",
    "WINDOW_CENTER_INDEX_1BASED",
    "WINDOW_ANCHOR_CDS_POSITION_1BASED",
    "WINDOW_REQUESTED_CDS_START_1BASED",
    "WINDOW_REQUESTED_CDS_END_1BASED",
    "WINDOW_EXTRACTED_CDS_START_1BASED",
    "WINDOW_EXTRACTED_CDS_END_1BASED",
    "WINDOW_LEFT_PADDING_COUNT",
    "WINDOW_RIGHT_PADDING_COUNT",
    "WINDOW_HAS_PADDING_FLAG",
    "REFERENCE_WINDOW_LENGTH",
    "EVENT_WINDOW_START_INDEX_0BASED",
    "EVENT_WINDOW_END_INDEX_0BASED",
    "INSERTION_WINDOW_INDEX_0BASED",
    "EVENT_FULLY_CONTAINED_FLAG",
    "WINDOW_CENTER_BASE_MATCH_FLAG",
    "REFERENCE_WINDOW_BUILT_FLAG",
    "REFERENCE_WINDOW_ELIGIBLE_FLAG",
]


for column in nullable_integer_columns:

    MUTATION_301_REFERENCE_WINDOW_DF[
        column
    ] = pd.to_numeric(
        MUTATION_301_REFERENCE_WINDOW_DF[
            column
        ],
        errors="coerce",
    ).astype("Int64")


# ------------------------------------------------------------
# 8. Validation masks
# ------------------------------------------------------------

payload_usable_mask = (
    MUTATION_301_REFERENCE_WINDOW_DF[
        "MUTATION_PAYLOAD_USABLE_FLAG"
    ]
    .eq(1)
)


window_built_mask = (
    MUTATION_301_REFERENCE_WINDOW_DF[
        "REFERENCE_WINDOW_BUILT_FLAG"
    ]
    .eq(1)
)


window_eligible_mask = (
    MUTATION_301_REFERENCE_WINDOW_DF[
        "REFERENCE_WINDOW_ELIGIBLE_FLAG"
    ]
    .eq(1)
)


window_length_error_mask = (
    window_built_mask
    &
    MUTATION_301_REFERENCE_WINDOW_DF[
        "REFERENCE_WINDOW_LENGTH"
    ]
    .ne(WINDOW_SIZE)
)


window_character_error_mask = (
    window_built_mask
    &
    ~MUTATION_301_REFERENCE_WINDOW_DF[
        "REFERENCE_WINDOW_301"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[ACGTN]{301}")
)


center_mapping_error_mask = (
    window_built_mask
    &
    MUTATION_301_REFERENCE_WINDOW_DF[
        "WINDOW_CENTER_BASE_MATCH_FLAG"
    ]
    .ne(1)
)


duplicate_mutation_id_mask = (
    MUTATION_301_REFERENCE_WINDOW_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


# ------------------------------------------------------------
# 9. Precalculate printed counts
# ------------------------------------------------------------

total_rows = len(
    MUTATION_301_REFERENCE_WINDOW_DF
)

payload_usable_count = int(
    payload_usable_mask.sum()
)

window_built_count = int(
    window_built_mask.sum()
)

window_eligible_count = int(
    window_eligible_mask.sum()
)

event_not_contained_count = int(
    (
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "EVENT_FULLY_CONTAINED_FLAG"
        ]
        .eq(0)
    ).sum()
)

padding_window_count = int(
    (
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "WINDOW_HAS_PADDING_FLAG"
        ]
        .eq(1)
    ).sum()
)

left_padding_window_count = int(
    (
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "WINDOW_LEFT_PADDING_COUNT"
        ]
        .gt(0)
    ).sum()
)

right_padding_window_count = int(
    (
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "WINDOW_RIGHT_PADDING_COUNT"
        ]
        .gt(0)
    ).sum()
)

short_cds_window_count = int(
    (
        window_built_mask
        &
        pd.to_numeric(
            MUTATION_301_REFERENCE_WINDOW_DF[
                "REFERENCE_CDS_LENGTH"
            ],
            errors="coerce",
        )
        .lt(WINDOW_SIZE)
    ).sum()
)

length_error_count = int(
    window_length_error_mask.sum()
)

character_error_count = int(
    window_character_error_mask.sum()
)

center_mapping_error_count = int(
    center_mapping_error_mask.sum()
)

duplicate_mutation_id_count = int(
    duplicate_mutation_id_mask.sum()
)


# ------------------------------------------------------------
# 10. Print validation results
# ------------------------------------------------------------

print(
    f"\nTotal mutation rows: "
    f"{total_rows:,}"
)

print(
    f"Reference-verified mutation payloads: "
    f"{payload_usable_count:,}"
)

print(
    f"301-bp reference windows constructed: "
    f"{window_built_count:,}"
)

print(
    f"Windows eligible for mutation application: "
    f"{window_eligible_count:,}"
)

print(
    f"Events not fully contained in the window: "
    f"{event_not_contained_count:,}"
)

print(
    f"Windows requiring boundary padding: "
    f"{padding_window_count:,}"
)

print(
    f"Windows with left padding: "
    f"{left_padding_window_count:,}"
)

print(
    f"Windows with right padding: "
    f"{right_padding_window_count:,}"
)

print(
    f"Windows from CDS shorter than 301 bases: "
    f"{short_cds_window_count:,}"
)

print(
    f"Window length errors: "
    f"{length_error_count:,}"
)

print(
    f"Window character errors: "
    f"{character_error_count:,}"
)

print(
    f"Window centre-mapping errors: "
    f"{center_mapping_error_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{duplicate_mutation_id_count:,}"
)


print(
    "\nReference-window status counts:"
)

print(
    MUTATION_301_REFERENCE_WINDOW_DF[
        "REFERENCE_WINDOW_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 11. Cohort-level summary
# ------------------------------------------------------------

reference_window_summary_df = (
    MUTATION_301_REFERENCE_WINDOW_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        payload_usable=(
            "MUTATION_PAYLOAD_USABLE_FLAG",
            "sum",
        ),

        reference_windows_built=(
            "REFERENCE_WINDOW_BUILT_FLAG",
            "sum",
        ),

        reference_windows_eligible=(
            "REFERENCE_WINDOW_ELIGIBLE_FLAG",
            "sum",
        ),

        windows_with_padding=(
            "WINDOW_HAS_PADDING_FLAG",
            lambda values: int(
                values.fillna(0).eq(1).sum()
            ),
        ),
    )
    .reset_index()
)


print(
    "\n301-bp reference windows by cohort:"
)

print(
    reference_window_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 12. Show event-not-contained examples
# ------------------------------------------------------------

not_contained_examples = (
    MUTATION_301_REFERENCE_WINDOW_DF.loc[
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "EVENT_FULLY_CONTAINED_FLAG"
        ]
        .eq(0),

        [
            "MUTATION_ROW_ID",
            "COHORT",
            "GENE_SYMBOL",
            "HGVS_OPERATION",
            "HGVSC_RAW",
            "CDS_START_1BASED",
            "CDS_END_1BASED",
            "WINDOW_ANCHOR_CDS_POSITION_1BASED",
            "EVENT_WINDOW_START_INDEX_0BASED",
            "EVENT_WINDOW_END_INDEX_0BASED",
            "REFERENCE_WINDOW_STATUS",
        ],
    ]
    .head(20)
)


if not not_contained_examples.empty:

    print(
        "\nEvents not fully contained in 301 bases:"
    )

    print(
        not_contained_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 13. Show padding examples
# ------------------------------------------------------------

padding_examples = (
    MUTATION_301_REFERENCE_WINDOW_DF.loc[
        window_built_mask
        &
        MUTATION_301_REFERENCE_WINDOW_DF[
            "WINDOW_HAS_PADDING_FLAG"
        ]
        .eq(1),

        [
            "MUTATION_ROW_ID",
            "COHORT",
            "GENE_SYMBOL",
            "HGVSC_RAW",
            "REFERENCE_CDS_LENGTH",
            "WINDOW_ANCHOR_CDS_POSITION_1BASED",
            "WINDOW_LEFT_PADDING_COUNT",
            "WINDOW_RIGHT_PADDING_COUNT",
            "REFERENCE_WINDOW_LENGTH",
        ],
    ]
    .head(10)
)


if not padding_examples.empty:

    print(
        "\nBoundary-padding examples:"
    )

    print(
        padding_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 14. Hard validations
# ------------------------------------------------------------

if len(
    MUTATION_301_REFERENCE_WINDOW_DF
) != len(
    MUTATION_REFERENCE_VERIFIED_DF
):
    raise ValueError(
        "Reference-window construction changed "
        "the mutation-row count."
    )


if duplicate_mutation_id_mask.any():
    raise ValueError(
        "Reference-window construction duplicated "
        "mutation rows."
    )


if window_built_count != payload_usable_count:
    raise ValueError(
        "Not every reference-verified mutation received "
        "a 301-bp reference window."
    )


if window_length_error_mask.any():
    raise ValueError(
        "A constructed reference window does not "
        "contain exactly 301 bases."
    )


if window_character_error_mask.any():
    raise ValueError(
        "A constructed reference window contains "
        "invalid characters."
    )


if center_mapping_error_mask.any():
    raise ValueError(
        "A CDS mutation anchor was not mapped to "
        "the central window position."
    )


# ------------------------------------------------------------
# 15. Keep eligible windows in memory
# ------------------------------------------------------------

REFERENCE_WINDOW_301_ELIGIBLE_DF = (
    MUTATION_301_REFERENCE_WINDOW_DF.loc[
        window_eligible_mask
    ]
    .copy()
    .reset_index(drop=True)
)


print("\n" + "=" * 78)
print("STEP 9F FINISHED")
print("=" * 78)

print(
    "Complete reference-window audit table:"
)

print(
    "MUTATION_301_REFERENCE_WINDOW_DF"
)

print(
    "\nMutation-application eligible table:"
)

print(
    "REFERENCE_WINDOW_301_ELIGIBLE_DF"
)
```

## Step 9G — Apply Mutations and Construct Mutated 301-bp Windows

Each reference-verified mutation is applied to its matched full Ensembl CDS, after which a fixed 301-bp mutated window is extracted using the same left boundary as the corresponding reference window.

Substitutions, deletions, insertions, duplications and deletion-insertions are supported.

Mutated windows that are identical to their reference windows are retained in a separate audit and excluded from the final paired-window table because they provide no observable sequence difference to the model.

```python
# ============================================================
# STEP 9G — APPLY MUTATIONS AND BUILD MUTATED 301-BP WINDOWS
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_301_REFERENCE_WINDOW_DF" not in globals():
    raise RuntimeError(
        "MUTATION_301_REFERENCE_WINDOW_DF was not found. "
        "Run Step 9F first."
    )


print("=" * 78)
print("STEP 9G — MUTATED 301-BP WINDOW CONSTRUCTION")
print("=" * 78)


# ------------------------------------------------------------
# 2. Fixed configuration
# ------------------------------------------------------------

WINDOW_SIZE = 301
WINDOW_CENTER_INDEX_0BASED = 150

DNA_SEQUENCE_PATTERN = re.compile(
    r"^[ACGTN]*$"
)

DNA_WINDOW_PATTERN = re.compile(
    r"^[ACGTN]{301}$"
)


# ------------------------------------------------------------
# 3. Helper functions
# ------------------------------------------------------------

def flag_is_one(value):
    """
    Safely check whether a nullable flag equals 1.
    """

    if value is None or pd.isna(value):
        return False

    return int(value) == 1


def empty_mutated_window_result(status):
    """
    Return a standard result for a mutation row that cannot
    produce a mutated 301-bp window.
    """

    return {
        "MUTATED_CDS_LENGTH": pd.NA,
        "EXPECTED_MUTATED_CDS_LENGTH": pd.NA,
        "MUTATED_CDS_LENGTH_VALID_FLAG": 0,

        "MUTATION_EDIT_CDS_INDEX_0BASED": pd.NA,
        "MUTATION_EDIT_WINDOW_INDEX_0BASED": pd.NA,

        "REFERENCE_LOCAL_EVENT_SEQUENCE": pd.NA,
        "REFERENCE_LOCAL_PAYLOAD_MATCH_FLAG": 0,

        "MUTATED_WINDOW_EXTRACTED_START_1BASED": pd.NA,
        "MUTATED_WINDOW_EXTRACTED_END_1BASED": pd.NA,

        "MUTATED_WINDOW_LEFT_PADDING_COUNT": pd.NA,
        "MUTATED_WINDOW_RIGHT_PADDING_COUNT": pd.NA,
        "MUTATED_WINDOW_HAS_PADDING_FLAG": pd.NA,

        "MUTATED_WINDOW_301": pd.NA,
        "MUTATED_WINDOW_LENGTH": pd.NA,
        "MUTATED_WINDOW_CENTER_BASE": pd.NA,

        "MUTATED_INSERTED_START_INDEX_0BASED": pd.NA,
        "MUTATED_INSERTED_END_INDEX_0BASED": pd.NA,
        "MUTATED_INSERTED_FULLY_CONTAINED_FLAG": 0,

        "MUTATION_EDIT_INDEX_WITHIN_WINDOW_FLAG": 0,
        "MUTATED_WINDOW_DIFFERS_FROM_REFERENCE_FLAG": 0,

        "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG": 0,
        "IDENTICAL_WINDOW_REASON": pd.NA,

        "MUTATED_WINDOW_BUILT_FLAG": 0,
        "MUTATED_WINDOW_ELIGIBLE_FLAG": 0,
        "MUTATED_WINDOW_STATUS": status,
    }


# ------------------------------------------------------------
# 4. Apply one mutation and construct its mutated window
# ------------------------------------------------------------

def apply_mutation_and_build_window(row):
    """
    Apply one verified mutation to the complete CDS and build
    a fixed-length mutated 301-bp window.
    """

    if not flag_is_one(
        row["REFERENCE_WINDOW_ELIGIBLE_FLAG"]
    ):
        return empty_mutated_window_result(
            "REFERENCE_WINDOW_NOT_ELIGIBLE"
        )


    if not flag_is_one(
        row["MUTATION_PAYLOAD_USABLE_FLAG"]
    ):
        return empty_mutated_window_result(
            "MUTATION_PAYLOAD_NOT_USABLE"
        )


    # --------------------------------------------------------
    # Read and validate source sequences
    # --------------------------------------------------------

    reference_cds_value = row[
        "REFERENCE_CDS_SEQUENCE"
    ]

    reference_window_value = row[
        "REFERENCE_WINDOW_301"
    ]


    if (
        reference_cds_value is None
        or pd.isna(reference_cds_value)
    ):
        return empty_mutated_window_result(
            "REFERENCE_CDS_MISSING"
        )


    if (
        reference_window_value is None
        or pd.isna(reference_window_value)
    ):
        return empty_mutated_window_result(
            "REFERENCE_WINDOW_MISSING"
        )


    reference_cds = (
        str(reference_cds_value)
        .strip()
        .upper()
    )

    reference_window = (
        str(reference_window_value)
        .strip()
        .upper()
    )


    if (
        reference_cds == ""
        or not DNA_SEQUENCE_PATTERN.fullmatch(
            reference_cds
        )
    ):
        return empty_mutated_window_result(
            "REFERENCE_CDS_INVALID"
        )


    if not DNA_WINDOW_PATTERN.fullmatch(
        reference_window
    ):
        return empty_mutated_window_result(
            "REFERENCE_WINDOW_INVALID"
        )


    # --------------------------------------------------------
    # Read mutation fields
    # --------------------------------------------------------

    operation_value = row[
        "HGVS_OPERATION"
    ]


    if (
        operation_value is None
        or pd.isna(operation_value)
    ):
        return empty_mutated_window_result(
            "MUTATION_OPERATION_MISSING"
        )


    operation = (
        str(operation_value)
        .strip()
        .upper()
    )


    start_position = int(
        row["CDS_START_1BASED"]
    )

    end_position = int(
        row["CDS_END_1BASED"]
    )


    removed_value = row[
        "REFERENCE_REMOVED_SEQUENCE"
    ]

    inserted_value = row[
        "ALTERNATE_INSERTED_SEQUENCE"
    ]


    if removed_value is None or pd.isna(
        removed_value
    ):
        return empty_mutated_window_result(
            "REMOVED_SEQUENCE_MISSING"
        )


    if inserted_value is None or pd.isna(
        inserted_value
    ):
        return empty_mutated_window_result(
            "INSERTED_SEQUENCE_MISSING"
        )


    removed_sequence = (
        str(removed_value)
        .strip()
        .upper()
    )

    inserted_sequence = (
        str(inserted_value)
        .strip()
        .upper()
    )


    if not DNA_SEQUENCE_PATTERN.fullmatch(
        removed_sequence
    ):
        return empty_mutated_window_result(
            "REMOVED_SEQUENCE_INVALID"
        )


    if not DNA_SEQUENCE_PATTERN.fullmatch(
        inserted_sequence
    ):
        return empty_mutated_window_result(
            "INSERTED_SEQUENCE_INVALID"
        )


    if (
        start_position < 1
        or end_position < start_position
        or end_position > len(reference_cds)
    ):
        return empty_mutated_window_result(
            "CDS_COORDINATE_OUT_OF_RANGE"
        )


    requested_start = int(
        row[
            "WINDOW_REQUESTED_CDS_START_1BASED"
        ]
    )

    event_start_index = int(
        row[
            "EVENT_WINDOW_START_INDEX_0BASED"
        ]
    )

    event_end_index = int(
        row[
            "EVENT_WINDOW_END_INDEX_0BASED"
        ]
    )


    # --------------------------------------------------------
    # Substitution, deletion and delins
    # --------------------------------------------------------

    if operation in {
        "SUBSTITUTION",
        "DELETION",
        "DELINS",
    }:

        actual_cds_reference = reference_cds[
            start_position - 1:
            end_position
        ]


        if actual_cds_reference != removed_sequence:
            return empty_mutated_window_result(
                "FULL_CDS_REFERENCE_PAYLOAD_MISMATCH"
            )


        reference_local_event = reference_window[
            event_start_index:
            event_end_index + 1
        ]


        if reference_local_event != removed_sequence:
            return empty_mutated_window_result(
                "WINDOW_REFERENCE_PAYLOAD_MISMATCH"
            )


        mutated_cds = (
            reference_cds[
                :start_position - 1
            ]
            +
            inserted_sequence
            +
            reference_cds[
                end_position:
            ]
        )


        edit_cds_index = (
            start_position - 1
        )

        edit_window_index = (
            event_start_index
        )

        local_payload_match = 1


    # --------------------------------------------------------
    # Insertion
    # --------------------------------------------------------

    elif operation == "INSERTION":

        if end_position != start_position + 1:
            return empty_mutated_window_result(
                "INSERTION_COORDINATES_NOT_ADJACENT"
            )


        insertion_window_value = row[
            "INSERTION_WINDOW_INDEX_0BASED"
        ]


        if (
            insertion_window_value is None
            or pd.isna(insertion_window_value)
        ):
            return empty_mutated_window_result(
                "INSERTION_WINDOW_INDEX_MISSING"
            )


        insertion_window_index = int(
            insertion_window_value
        )


        if insertion_window_index != (
            event_start_index + 1
        ):
            return empty_mutated_window_result(
                "INSERTION_WINDOW_INDEX_MISMATCH"
            )


        cds_left_flank = reference_cds[
            start_position - 1
        ]

        cds_right_flank = reference_cds[
            end_position - 1
        ]


        window_left_flank = reference_window[
            event_start_index
        ]

        window_right_flank = reference_window[
            event_end_index
        ]


        if (
            cds_left_flank != window_left_flank
            or cds_right_flank
            != window_right_flank
        ):
            return empty_mutated_window_result(
                "INSERTION_FLANK_SEQUENCE_MISMATCH"
            )


        mutated_cds = (
            reference_cds[
                :start_position
            ]
            +
            inserted_sequence
            +
            reference_cds[
                start_position:
            ]
        )


        edit_cds_index = (
            start_position
        )

        edit_window_index = (
            insertion_window_index
        )

        reference_local_event = (
            window_left_flank
            +
            window_right_flank
        )

        local_payload_match = 1


    # --------------------------------------------------------
    # Duplication
    # --------------------------------------------------------

    elif operation == "DUPLICATION":

        duplicated_cds_sequence = reference_cds[
            start_position - 1:
            end_position
        ]


        reference_local_event = reference_window[
            event_start_index:
            event_end_index + 1
        ]


        if (
            duplicated_cds_sequence
            != inserted_sequence
        ):
            return empty_mutated_window_result(
                "DUPLICATION_CDS_SEQUENCE_MISMATCH"
            )


        if (
            reference_local_event
            != inserted_sequence
        ):
            return empty_mutated_window_result(
                "DUPLICATION_WINDOW_SEQUENCE_MISMATCH"
            )


        mutated_cds = (
            reference_cds[
                :end_position
            ]
            +
            inserted_sequence
            +
            reference_cds[
                end_position:
            ]
        )


        edit_cds_index = (
            end_position
        )

        edit_window_index = (
            event_end_index + 1
        )

        local_payload_match = 1


    else:

        return empty_mutated_window_result(
            "UNSUPPORTED_MUTATION_OPERATION"
        )


    # --------------------------------------------------------
    # Validate the complete mutated CDS
    # --------------------------------------------------------

    expected_mutated_cds_length = (
        len(reference_cds)
        -
        len(removed_sequence)
        +
        len(inserted_sequence)
    )


    actual_mutated_cds_length = len(
        mutated_cds
    )


    mutated_cds_length_valid = int(
        actual_mutated_cds_length
        ==
        expected_mutated_cds_length
    )


    if mutated_cds_length_valid != 1:
        return empty_mutated_window_result(
            "MUTATED_CDS_LENGTH_ERROR"
        )


    if not DNA_SEQUENCE_PATTERN.fullmatch(
        mutated_cds
    ):
        return empty_mutated_window_result(
            "MUTATED_CDS_INVALID_CHARACTERS"
        )


    # Confirm the inserted/replacement sequence in mutated CDS.
    inserted_end_cds_index = (
        edit_cds_index
        +
        len(inserted_sequence)
    )


    inserted_cds_check = mutated_cds[
        edit_cds_index:
        inserted_end_cds_index
    ]


    if inserted_cds_check != inserted_sequence:
        return empty_mutated_window_result(
            "MUTATED_CDS_INSERTION_CHECK_FAILED"
        )


    # --------------------------------------------------------
    # Construct mutated window using the same requested
    # left boundary as the reference window
    # --------------------------------------------------------

    extracted_start = max(
        1,
        requested_start,
    )


    left_padding = max(
        0,
        1 - requested_start,
    )


    real_bases_needed = (
        WINDOW_SIZE
        -
        left_padding
    )


    mutated_slice_start_index = (
        extracted_start - 1
    )


    mutated_sequence_slice = mutated_cds[
        mutated_slice_start_index:
        mutated_slice_start_index
        +
        real_bases_needed
    ]


    right_padding = (
        WINDOW_SIZE
        -
        left_padding
        -
        len(mutated_sequence_slice)
    )


    if right_padding < 0:
        return empty_mutated_window_result(
            "NEGATIVE_MUTATED_RIGHT_PADDING"
        )


    mutated_window = (
        ("N" * left_padding)
        +
        mutated_sequence_slice
        +
        ("N" * right_padding)
    )


    mutated_window_length = len(
        mutated_window
    )


    if len(mutated_sequence_slice) > 0:

        extracted_end = (
            extracted_start
            +
            len(mutated_sequence_slice)
            -
            1
        )

    else:

        extracted_end = (
            extracted_start - 1
        )


    if mutated_window_length != WINDOW_SIZE:
        return empty_mutated_window_result(
            "MUTATED_WINDOW_LENGTH_ERROR"
        )


    if not DNA_WINDOW_PATTERN.fullmatch(
        mutated_window
    ):
        return empty_mutated_window_result(
            "MUTATED_WINDOW_INVALID_CHARACTERS"
        )


    # --------------------------------------------------------
    # Locate inserted/replacement sequence inside the window
    # --------------------------------------------------------

    edit_index_within_window = int(
        0 <= edit_window_index < WINDOW_SIZE
    )


    inserted_length = len(
        inserted_sequence
    )


    if inserted_length > 0:

        inserted_start_window_index = (
            edit_window_index
        )

        inserted_end_window_index = (
            edit_window_index
            +
            inserted_length
            -
            1
        )


        inserted_fully_contained = int(
            inserted_start_window_index >= 0
            and inserted_end_window_index
            < WINDOW_SIZE
        )


        if inserted_fully_contained == 1:

            inserted_window_check = (
                mutated_window[
                    inserted_start_window_index:
                    inserted_end_window_index
                    +
                    1
                ]
            )


            if (
                inserted_window_check
                != inserted_sequence
            ):
                return empty_mutated_window_result(
                    "MUTATED_WINDOW_INSERTION_CHECK_FAILED"
                )


    else:

        inserted_start_window_index = pd.NA
        inserted_end_window_index = pd.NA

        inserted_fully_contained = int(
            0 <= edit_window_index
            < WINDOW_SIZE
        )


    # --------------------------------------------------------
    # Compare reference and mutated windows
    # --------------------------------------------------------

    window_differs = int(
        mutated_window != reference_window
    )


    identical_window_flag = int(
        window_differs == 0
    )


    identical_reason = pd.NA


    if identical_window_flag == 1:

        if (
            operation
            in {
                "SUBSTITUTION",
                "DELETION",
                "DELINS",
            }
            and
            removed_sequence
            ==
            inserted_sequence
        ):

            identical_reason = (
                "NO_OP_MUTATION_PAYLOAD"
            )

        else:

            identical_reason = (
                "REPETITIVE_CONTEXT_OR_FIXED_WINDOW_EQUIVALENCE"
            )


    mutated_center_base = mutated_window[
        WINDOW_CENTER_INDEX_0BASED
    ]


    # --------------------------------------------------------
    # Final eligibility
    # --------------------------------------------------------

    final_eligible = int(
        mutated_cds_length_valid == 1
        and local_payload_match == 1
        and edit_index_within_window == 1
        and inserted_fully_contained == 1
        and window_differs == 1
    )


    if final_eligible == 1:

        final_status = (
            "MUTATED_WINDOW_BUILT_AND_EVENT_CONTAINED"
        )


    elif identical_window_flag == 1:

        final_status = (
            "MUTATED_WINDOW_IDENTICAL_TO_REFERENCE_EXCLUDED"
        )


    elif inserted_fully_contained != 1:

        final_status = (
            "MUTATED_WINDOW_BUILT_INSERTED_SEQUENCE_TRUNCATED"
        )


    elif edit_index_within_window != 1:

        final_status = (
            "MUTATED_WINDOW_BUILT_EDIT_OUTSIDE_WINDOW"
        )


    else:

        final_status = (
            "MUTATED_WINDOW_BUILT_NOT_ELIGIBLE"
        )


    return {
        "MUTATED_CDS_LENGTH": (
            actual_mutated_cds_length
        ),

        "EXPECTED_MUTATED_CDS_LENGTH": (
            expected_mutated_cds_length
        ),

        "MUTATED_CDS_LENGTH_VALID_FLAG": (
            mutated_cds_length_valid
        ),

        "MUTATION_EDIT_CDS_INDEX_0BASED": (
            edit_cds_index
        ),

        "MUTATION_EDIT_WINDOW_INDEX_0BASED": (
            edit_window_index
        ),

        "REFERENCE_LOCAL_EVENT_SEQUENCE": (
            reference_local_event
        ),

        "REFERENCE_LOCAL_PAYLOAD_MATCH_FLAG": (
            local_payload_match
        ),

        "MUTATED_WINDOW_EXTRACTED_START_1BASED": (
            extracted_start
        ),

        "MUTATED_WINDOW_EXTRACTED_END_1BASED": (
            extracted_end
        ),

        "MUTATED_WINDOW_LEFT_PADDING_COUNT": (
            left_padding
        ),

        "MUTATED_WINDOW_RIGHT_PADDING_COUNT": (
            right_padding
        ),

        "MUTATED_WINDOW_HAS_PADDING_FLAG": int(
            left_padding > 0
            or right_padding > 0
        ),

        "MUTATED_WINDOW_301": (
            mutated_window
        ),

        "MUTATED_WINDOW_LENGTH": (
            mutated_window_length
        ),

        "MUTATED_WINDOW_CENTER_BASE": (
            mutated_center_base
        ),

        "MUTATED_INSERTED_START_INDEX_0BASED": (
            inserted_start_window_index
        ),

        "MUTATED_INSERTED_END_INDEX_0BASED": (
            inserted_end_window_index
        ),

        "MUTATED_INSERTED_FULLY_CONTAINED_FLAG": (
            inserted_fully_contained
        ),

        "MUTATION_EDIT_INDEX_WITHIN_WINDOW_FLAG": (
            edit_index_within_window
        ),

        "MUTATED_WINDOW_DIFFERS_FROM_REFERENCE_FLAG": (
            window_differs
        ),

        "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG": (
            identical_window_flag
        ),

        "IDENTICAL_WINDOW_REASON": (
            identical_reason
        ),

        "MUTATED_WINDOW_BUILT_FLAG": 1,

        "MUTATED_WINDOW_ELIGIBLE_FLAG": (
            final_eligible
        ),

        "MUTATED_WINDOW_STATUS": (
            final_status
        ),
    }


# ------------------------------------------------------------
# 5. Prepare rerunnable source table
# ------------------------------------------------------------

working_df = (
    MUTATION_301_REFERENCE_WINDOW_DF
    .copy()
)


generated_columns = [
    "MUTATED_CDS_LENGTH",
    "EXPECTED_MUTATED_CDS_LENGTH",
    "MUTATED_CDS_LENGTH_VALID_FLAG",
    "MUTATION_EDIT_CDS_INDEX_0BASED",
    "MUTATION_EDIT_WINDOW_INDEX_0BASED",
    "REFERENCE_LOCAL_EVENT_SEQUENCE",
    "REFERENCE_LOCAL_PAYLOAD_MATCH_FLAG",
    "MUTATED_WINDOW_EXTRACTED_START_1BASED",
    "MUTATED_WINDOW_EXTRACTED_END_1BASED",
    "MUTATED_WINDOW_LEFT_PADDING_COUNT",
    "MUTATED_WINDOW_RIGHT_PADDING_COUNT",
    "MUTATED_WINDOW_HAS_PADDING_FLAG",
    "MUTATED_WINDOW_301",
    "MUTATED_WINDOW_LENGTH",
    "MUTATED_WINDOW_CENTER_BASE",
    "MUTATED_INSERTED_START_INDEX_0BASED",
    "MUTATED_INSERTED_END_INDEX_0BASED",
    "MUTATED_INSERTED_FULLY_CONTAINED_FLAG",
    "MUTATION_EDIT_INDEX_WITHIN_WINDOW_FLAG",
    "MUTATED_WINDOW_DIFFERS_FROM_REFERENCE_FLAG",
    "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG",
    "IDENTICAL_WINDOW_REASON",
    "MUTATED_WINDOW_BUILT_FLAG",
    "MUTATED_WINDOW_ELIGIBLE_FLAG",
    "MUTATED_WINDOW_STATUS",
]


working_df = working_df.drop(
    columns=[
        column
        for column in generated_columns
        if column in working_df.columns
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 6. Apply every mutation
# ------------------------------------------------------------

mutation_results = pd.DataFrame(
    [
        apply_mutation_and_build_window(row)
        for _, row in working_df.iterrows()
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        mutation_results,
    ],
    axis=1,
)


MUTATION_301_MUTATED_WINDOW_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 7. Convert generated numerical fields
# ------------------------------------------------------------

nullable_integer_columns = [
    "MUTATED_CDS_LENGTH",
    "EXPECTED_MUTATED_CDS_LENGTH",
    "MUTATED_CDS_LENGTH_VALID_FLAG",
    "MUTATION_EDIT_CDS_INDEX_0BASED",
    "MUTATION_EDIT_WINDOW_INDEX_0BASED",
    "REFERENCE_LOCAL_PAYLOAD_MATCH_FLAG",
    "MUTATED_WINDOW_EXTRACTED_START_1BASED",
    "MUTATED_WINDOW_EXTRACTED_END_1BASED",
    "MUTATED_WINDOW_LEFT_PADDING_COUNT",
    "MUTATED_WINDOW_RIGHT_PADDING_COUNT",
    "MUTATED_WINDOW_HAS_PADDING_FLAG",
    "MUTATED_WINDOW_LENGTH",
    "MUTATED_INSERTED_START_INDEX_0BASED",
    "MUTATED_INSERTED_END_INDEX_0BASED",
    "MUTATED_INSERTED_FULLY_CONTAINED_FLAG",
    "MUTATION_EDIT_INDEX_WITHIN_WINDOW_FLAG",
    "MUTATED_WINDOW_DIFFERS_FROM_REFERENCE_FLAG",
    "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG",
    "MUTATED_WINDOW_BUILT_FLAG",
    "MUTATED_WINDOW_ELIGIBLE_FLAG",
]


for column in nullable_integer_columns:

    MUTATION_301_MUTATED_WINDOW_DF[
        column
    ] = pd.to_numeric(
        MUTATION_301_MUTATED_WINDOW_DF[
            column
        ],
        errors="coerce",
    ).astype("Int64")


# ------------------------------------------------------------
# 8. Validation masks
# ------------------------------------------------------------

reference_eligible_mask = (
    MUTATION_301_MUTATED_WINDOW_DF[
        "REFERENCE_WINDOW_ELIGIBLE_FLAG"
    ]
    .eq(1)
)


mutated_built_mask = (
    MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_WINDOW_BUILT_FLAG"
    ]
    .eq(1)
)


final_eligible_mask = (
    MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_WINDOW_ELIGIBLE_FLAG"
    ]
    .eq(1)
)


identical_window_mask = (
    MUTATION_301_MUTATED_WINDOW_DF[
        "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG"
    ]
    .eq(1)
)


technical_build_failure_mask = (
    reference_eligible_mask
    &
    ~mutated_built_mask
)


window_length_error_mask = (
    mutated_built_mask
    &
    MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_WINDOW_LENGTH"
    ]
    .ne(WINDOW_SIZE)
)


window_character_error_mask = (
    mutated_built_mask
    &
    ~MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_WINDOW_301"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[ACGTN]{301}")
)


cds_length_error_mask = (
    mutated_built_mask
    &
    MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_CDS_LENGTH_VALID_FLAG"
    ]
    .ne(1)
)


local_payload_error_mask = (
    mutated_built_mask
    &
    MUTATION_301_MUTATED_WINDOW_DF[
        "REFERENCE_LOCAL_PAYLOAD_MATCH_FLAG"
    ]
    .ne(1)
)


duplicate_mutation_id_mask = (
    MUTATION_301_MUTATED_WINDOW_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


other_built_ineligible_mask = (
    mutated_built_mask
    &
    ~final_eligible_mask
    &
    ~identical_window_mask
)


# ------------------------------------------------------------
# 9. Precalculate counts
# ------------------------------------------------------------

total_rows = len(
    MUTATION_301_MUTATED_WINDOW_DF
)

reference_eligible_count = int(
    reference_eligible_mask.sum()
)

mutated_built_count = int(
    mutated_built_mask.sum()
)

final_eligible_count = int(
    final_eligible_mask.sum()
)

identical_window_count = int(
    identical_window_mask.sum()
)

technical_build_failure_count = int(
    technical_build_failure_mask.sum()
)

other_built_ineligible_count = int(
    other_built_ineligible_mask.sum()
)

inserted_truncated_count = int(
    (
        mutated_built_mask
        &
        MUTATION_301_MUTATED_WINDOW_DF[
            "MUTATED_INSERTED_FULLY_CONTAINED_FLAG"
        ]
        .ne(1)
    ).sum()
)

mutated_padding_count = int(
    (
        mutated_built_mask
        &
        MUTATION_301_MUTATED_WINDOW_DF[
            "MUTATED_WINDOW_HAS_PADDING_FLAG"
        ]
        .eq(1)
    ).sum()
)

window_length_error_count = int(
    window_length_error_mask.sum()
)

window_character_error_count = int(
    window_character_error_mask.sum()
)

cds_length_error_count = int(
    cds_length_error_mask.sum()
)

local_payload_error_count = int(
    local_payload_error_mask.sum()
)

duplicate_mutation_id_count = int(
    duplicate_mutation_id_mask.sum()
)


# ------------------------------------------------------------
# 10. Create final in-memory tables
# ------------------------------------------------------------

MUTATION_301_WINDOW_ELIGIBLE_DF = (
    MUTATION_301_MUTATED_WINDOW_DF.loc[
        final_eligible_mask
    ]
    .copy()
    .reset_index(drop=True)
)


IDENTICAL_301_WINDOW_AUDIT_DF = (
    MUTATION_301_MUTATED_WINDOW_DF.loc[
        identical_window_mask
    ]
    .copy()
    .reset_index(drop=True)
)


MUTATED_301_WINDOW_INELIGIBLE_AUDIT_DF = (
    MUTATION_301_MUTATED_WINDOW_DF.loc[
        reference_eligible_mask
        &
        ~final_eligible_mask
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 11. Print validation results
# ------------------------------------------------------------

print(
    f"\nTotal mutation rows: "
    f"{total_rows:,}"
)

print(
    f"Eligible reference windows: "
    f"{reference_eligible_count:,}"
)

print(
    f"Mutated 301-bp windows constructed: "
    f"{mutated_built_count:,}"
)

print(
    f"Final eligible paired windows: "
    f"{final_eligible_count:,}"
)

print(
    f"Identical reference/mutated windows excluded: "
    f"{identical_window_count:,}"
)

print(
    f"Other built but ineligible windows: "
    f"{other_built_ineligible_count:,}"
)

print(
    f"Technical mutation-application failures: "
    f"{technical_build_failure_count:,}"
)

print(
    f"Inserted sequences truncated by window boundary: "
    f"{inserted_truncated_count:,}"
)

print(
    f"Mutated windows requiring boundary padding: "
    f"{mutated_padding_count:,}"
)

print(
    f"Mutated-window length errors: "
    f"{window_length_error_count:,}"
)

print(
    f"Mutated-window character errors: "
    f"{window_character_error_count:,}"
)

print(
    f"Mutated-CDS length errors: "
    f"{cds_length_error_count:,}"
)

print(
    f"Local reference-payload errors: "
    f"{local_payload_error_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{duplicate_mutation_id_count:,}"
)


print(
    "\nMutated-window status counts:"
)

print(
    MUTATION_301_MUTATED_WINDOW_DF[
        "MUTATED_WINDOW_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 12. Summary by mutation operation
# ------------------------------------------------------------

operation_summary_df = (
    MUTATION_301_MUTATED_WINDOW_DF.loc[
        reference_eligible_mask
    ]
    .groupby(
        "HGVS_OPERATION",
        dropna=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        windows_built=(
            "MUTATED_WINDOW_BUILT_FLAG",
            "sum",
        ),

        final_paired_windows=(
            "MUTATED_WINDOW_ELIGIBLE_FLAG",
            "sum",
        ),

        identical_windows=(
            "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG",
            "sum",
        ),

        windows_with_padding=(
            "MUTATED_WINDOW_HAS_PADDING_FLAG",
            lambda values: int(
                values.fillna(0).eq(1).sum()
            ),
        ),
    )
    .reset_index()
)


print(
    "\nMutated windows by operation:"
)

print(
    operation_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 13. Summary by cohort
# ------------------------------------------------------------

cohort_summary_df = (
    MUTATION_301_MUTATED_WINDOW_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        reference_windows_eligible=(
            "REFERENCE_WINDOW_ELIGIBLE_FLAG",
            "sum",
        ),

        mutated_windows_built=(
            "MUTATED_WINDOW_BUILT_FLAG",
            "sum",
        ),

        identical_windows=(
            "IDENTICAL_REFERENCE_MUTATED_WINDOW_FLAG",
            "sum",
        ),

        final_paired_windows=(
            "MUTATED_WINDOW_ELIGIBLE_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print(
    "\nMutated 301-bp windows by cohort:"
)

print(
    cohort_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 14. Identical-window audit
# ------------------------------------------------------------

if identical_window_count > 0:

    identical_summary_df = (
        IDENTICAL_301_WINDOW_AUDIT_DF
        .groupby(
            [
                "HGVS_OPERATION",
                "IDENTICAL_WINDOW_REASON",
            ],
            dropna=False,
        )
        .agg(
            mutation_rows=(
                "MUTATION_ROW_ID",
                "size",
            ),

            affected_patients=(
                "PATIENT_ID",
                "nunique",
            ),
        )
        .reset_index()
    )


    print(
        "\nIdentical-window summary:"
    )

    print(
        identical_summary_df.to_string(
            index=False
        )
    )


    print(
        "\nIdentical-window examples:"
    )

    print(
        IDENTICAL_301_WINDOW_AUDIT_DF[
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "PATIENT_ID",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "REFERENCE_REMOVED_SEQUENCE",
                "ALTERNATE_INSERTED_SEQUENCE",
                "IDENTICAL_WINDOW_REASON",
            ]
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 15. Show technical failure examples
# ------------------------------------------------------------

if technical_build_failure_count > 0:

    print(
        "\nTechnical mutation-application failure examples:"
    )

    print(
        MUTATION_301_MUTATED_WINDOW_DF.loc[
            technical_build_failure_mask,
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "MUTATED_WINDOW_STATUS",
            ],
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 16. Show other ineligible examples
# ------------------------------------------------------------

if other_built_ineligible_count > 0:

    print(
        "\nOther built but ineligible examples:"
    )

    print(
        MUTATION_301_MUTATED_WINDOW_DF.loc[
            other_built_ineligible_mask,
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "MUTATION_EDIT_WINDOW_INDEX_0BASED",
                "MUTATED_INSERTED_START_INDEX_0BASED",
                "MUTATED_INSERTED_END_INDEX_0BASED",
                "MUTATED_WINDOW_STATUS",
            ],
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 17. Hard structural validations
# ------------------------------------------------------------

if len(
    MUTATION_301_MUTATED_WINDOW_DF
) != len(
    MUTATION_301_REFERENCE_WINDOW_DF
):
    raise ValueError(
        "Mutation application changed the mutation-row count."
    )


if duplicate_mutation_id_mask.any():
    raise ValueError(
        "Mutation application duplicated mutation rows."
    )


if window_length_error_mask.any():
    raise ValueError(
        "A built mutated window does not contain "
        "exactly 301 bases."
    )


if window_character_error_mask.any():
    raise ValueError(
        "A built mutated window contains invalid characters."
    )


if cds_length_error_mask.any():
    raise ValueError(
        "A mutated full CDS has an unexpected length."
    )


if local_payload_error_mask.any():
    raise ValueError(
        "A mutation payload does not match its local "
        "reference-window sequence."
    )


if technical_build_failure_count > 0:
    raise ValueError(
        "Some eligible reference windows failed mutation "
        "application. Inspect MUTATED_WINDOW_STATUS."
    )


classified_built_count = (
    final_eligible_count
    +
    identical_window_count
    +
    other_built_ineligible_count
)


if classified_built_count != mutated_built_count:
    raise ValueError(
        "Some built mutated windows were not classified."
    )


print("\n" + "=" * 78)
print("STEP 9G FINISHED")
print("=" * 78)

print(
    "Complete mutated-window audit table:"
)

print(
    "MUTATION_301_MUTATED_WINDOW_DF"
)

print(
    "\nFinal eligible paired-window table:"
)

print(
    "MUTATION_301_WINDOW_ELIGIBLE_DF"
)

print(
    "\nIdentical-window audit table:"
)

print(
    "IDENTICAL_301_WINDOW_AUDIT_DF"
)

print(
    "\nOther ineligible mutated-window audit table:"
)

print(
    "MUTATED_301_WINDOW_INELIGIBLE_AUDIT_DF"
)
```

## Step 9H — Generate 301-bp Mutation-Affected Position Masks

This step generates aligned binary masks for every eligible reference–mutated window pair.

Mask rules:

- substitution: mark the replaced base in both windows
- deletion: mark the deleted reference span and the new junction flanks in the mutated window
- insertion: mark the two reference flanking bases and the inserted bases in the mutated window
- duplication: mark the duplicated reference span and the newly inserted duplicate
- deletion-insertion: mark the removed reference span and the inserted replacement sequence

Each mask contains exactly 301 binary positions:

- `1` indicates a mutation-affected position
- `0` indicates an unaffected position

Masks are validated against event coordinates, sequence boundaries, padding and operation-specific expected lengths.

```python
# ============================================================
# STEP 9H — GENERATE 301-BP MUTATION-AFFECTED MASKS
# ============================================================

import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_301_MUTATED_WINDOW_DF" not in globals():
    raise RuntimeError(
        "MUTATION_301_MUTATED_WINDOW_DF was not found. "
        "Run Step 9G first."
    )


print("=" * 78)
print("STEP 9H — 301-BP MUTATION-AFFECTED POSITION MASKS")
print("=" * 78)


WINDOW_SIZE = 301


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def flag_is_one(value):
    """
    Safely check whether a nullable flag equals 1.
    """

    if value is None or pd.isna(value):
        return False

    return int(value) == 1


def build_binary_mask(indices):
    """
    Construct a 301-character binary mask.
    """

    clean_indices = sorted(
        set(int(index) for index in indices)
    )

    for index in clean_indices:

        if index < 0 or index >= WINDOW_SIZE:
            raise ValueError(
                f"Mask index outside 0–300: {index}"
            )

    mask_values = ["0"] * WINDOW_SIZE

    for index in clean_indices:
        mask_values[index] = "1"

    return "".join(mask_values)


def indices_to_text(indices):
    """
    Convert sorted unique indices to comma-separated text.
    """

    clean_indices = sorted(
        set(int(index) for index in indices)
    )

    return ",".join(
        str(index)
        for index in clean_indices
    )


def empty_mask_result(status):
    """
    Standard result for a row without an eligible mask pair.
    """

    return {
        "REFERENCE_AFFECTED_MASK_301": pd.NA,
        "MUTATED_AFFECTED_MASK_301": pd.NA,

        "REFERENCE_AFFECTED_INDICES_0BASED": pd.NA,
        "MUTATED_AFFECTED_INDICES_0BASED": pd.NA,

        "REFERENCE_AFFECTED_START_INDEX_0BASED": pd.NA,
        "REFERENCE_AFFECTED_END_INDEX_0BASED": pd.NA,

        "MUTATED_AFFECTED_START_INDEX_0BASED": pd.NA,
        "MUTATED_AFFECTED_END_INDEX_0BASED": pd.NA,

        "REFERENCE_AFFECTED_POSITION_COUNT": pd.NA,
        "MUTATED_AFFECTED_POSITION_COUNT": pd.NA,

        "REFERENCE_MASK_EXPECTED_COUNT": pd.NA,
        "MUTATED_MASK_EXPECTED_COUNT": pd.NA,

        "REFERENCE_MASK_COUNT_VALID_FLAG": 0,
        "MUTATED_MASK_COUNT_VALID_FLAG": 0,

        "REFERENCE_MASK_PADDING_OVERLAP_FLAG": pd.NA,
        "MUTATED_MASK_PADDING_OVERLAP_FLAG": pd.NA,

        "MUTATION_MASK_POLICY": pd.NA,
        "MUTATION_MASK_VALID_FLAG": 0,
        "MUTATION_MASK_STATUS": status,
    }


# ------------------------------------------------------------
# 3. Generate masks for one mutation-window pair
# ------------------------------------------------------------

def generate_mutation_masks(row):
    """
    Generate reference and mutated affected-position masks.
    """

    if not flag_is_one(
        row["MUTATED_WINDOW_ELIGIBLE_FLAG"]
    ):
        return empty_mask_result(
            "WINDOW_PAIR_NOT_ELIGIBLE"
        )


    operation = str(
        row["HGVS_OPERATION"]
    ).strip().upper()


    event_start = int(
        row["EVENT_WINDOW_START_INDEX_0BASED"]
    )

    event_end = int(
        row["EVENT_WINDOW_END_INDEX_0BASED"]
    )

    edit_index = int(
        row["MUTATION_EDIT_WINDOW_INDEX_0BASED"]
    )


    removed_length = int(
        row["REFERENCE_REMOVAL_LENGTH"]
    )

    inserted_length = int(
        row["ALTERNATE_INSERTION_LENGTH"]
    )


    reference_left_padding = int(
        row["WINDOW_LEFT_PADDING_COUNT"]
    )

    reference_right_padding = int(
        row["WINDOW_RIGHT_PADDING_COUNT"]
    )

    mutated_left_padding = int(
        row["MUTATED_WINDOW_LEFT_PADDING_COUNT"]
    )

    mutated_right_padding = int(
        row["MUTATED_WINDOW_RIGHT_PADDING_COUNT"]
    )


    reference_real_start = (
        reference_left_padding
    )

    reference_real_end = (
        WINDOW_SIZE
        -
        reference_right_padding
        -
        1
    )

    mutated_real_start = (
        mutated_left_padding
    )

    mutated_real_end = (
        WINDOW_SIZE
        -
        mutated_right_padding
        -
        1
    )


    reference_indices = []
    mutated_indices = []

    policy = pd.NA


    # --------------------------------------------------------
    # Substitution
    # --------------------------------------------------------

    if operation == "SUBSTITUTION":

        reference_indices = [
            event_start
        ]

        mutated_indices = [
            edit_index
        ]

        expected_reference_count = 1
        expected_mutated_count = 1

        policy = (
            "REPLACED_BASE_IN_REFERENCE_AND_MUTATED_WINDOWS"
        )


    # --------------------------------------------------------
    # Deletion
    # --------------------------------------------------------

    elif operation == "DELETION":

        reference_indices = list(
            range(
                event_start,
                event_end + 1,
            )
        )


        # Mark the new deletion junction in the mutated window.
        candidate_junction_indices = [
            edit_index - 1,
            edit_index,
        ]

        mutated_indices = [
            index
            for index in candidate_junction_indices
            if (
                0 <= index < WINDOW_SIZE
                and
                mutated_real_start
                <= index
                <= mutated_real_end
            )
        ]


        if len(mutated_indices) == 0:
            return empty_mask_result(
                "DELETION_JUNCTION_NOT_IN_REAL_WINDOW"
            )


        expected_reference_count = (
            removed_length
        )

        expected_mutated_count = len(
            mutated_indices
        )

        policy = (
            "DELETED_REFERENCE_SPAN_AND_MUTATED_JUNCTION_FLANKS"
        )


    # --------------------------------------------------------
    # Insertion
    # --------------------------------------------------------

    elif operation == "INSERTION":

        reference_indices = [
            event_start,
            event_end,
        ]


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        mutated_indices = list(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        expected_reference_count = 2
        expected_mutated_count = (
            inserted_length
        )

        policy = (
            "REFERENCE_INSERTION_FLANKS_AND_MUTATED_INSERTED_BASES"
        )


    # --------------------------------------------------------
    # Duplication
    # --------------------------------------------------------

    elif operation == "DUPLICATION":

        reference_indices = list(
            range(
                event_start,
                event_end + 1,
            )
        )


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        mutated_indices = list(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        expected_reference_count = (
            inserted_length
        )

        expected_mutated_count = (
            inserted_length
        )

        policy = (
            "DUPLICATED_REFERENCE_SPAN_AND_NEW_DUPLICATE_BASES"
        )


    # --------------------------------------------------------
    # Deletion-insertion
    # --------------------------------------------------------

    elif operation == "DELINS":

        reference_indices = list(
            range(
                event_start,
                event_end + 1,
            )
        )


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        mutated_indices = list(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        expected_reference_count = (
            removed_length
        )

        expected_mutated_count = (
            inserted_length
        )

        policy = (
            "REMOVED_REFERENCE_SPAN_AND_MUTATED_REPLACEMENT_BASES"
        )


    else:

        return empty_mask_result(
            "UNSUPPORTED_MUTATION_OPERATION"
        )


    # --------------------------------------------------------
    # Basic index validation
    # --------------------------------------------------------

    reference_indices = sorted(
        set(reference_indices)
    )

    mutated_indices = sorted(
        set(mutated_indices)
    )


    if len(reference_indices) == 0:
        return empty_mask_result(
            "REFERENCE_MASK_EMPTY"
        )


    if len(mutated_indices) == 0:
        return empty_mask_result(
            "MUTATED_MASK_EMPTY"
        )


    if any(
        index < 0 or index >= WINDOW_SIZE
        for index in reference_indices
    ):
        return empty_mask_result(
            "REFERENCE_MASK_INDEX_OUT_OF_RANGE"
        )


    if any(
        index < 0 or index >= WINDOW_SIZE
        for index in mutated_indices
    ):
        return empty_mask_result(
            "MUTATED_MASK_INDEX_OUT_OF_RANGE"
        )


    # --------------------------------------------------------
    # Check that masks do not mark padding
    # --------------------------------------------------------

    reference_padding_overlap = int(
        any(
            index < reference_real_start
            or index > reference_real_end
            for index in reference_indices
        )
    )


    mutated_padding_overlap = int(
        any(
            index < mutated_real_start
            or index > mutated_real_end
            for index in mutated_indices
        )
    )


    if reference_padding_overlap == 1:
        return empty_mask_result(
            "REFERENCE_MASK_OVERLAPS_PADDING"
        )


    if mutated_padding_overlap == 1:
        return empty_mask_result(
            "MUTATED_MASK_OVERLAPS_PADDING"
        )


    # --------------------------------------------------------
    # Construct binary mask strings
    # --------------------------------------------------------

    reference_mask = build_binary_mask(
        reference_indices
    )

    mutated_mask = build_binary_mask(
        mutated_indices
    )


    reference_count = (
        reference_mask.count("1")
    )

    mutated_count = (
        mutated_mask.count("1")
    )


    reference_count_valid = int(
        reference_count
        ==
        expected_reference_count
    )

    mutated_count_valid = int(
        mutated_count
        ==
        expected_mutated_count
    )


    final_valid_flag = int(
        len(reference_mask) == WINDOW_SIZE
        and len(mutated_mask) == WINDOW_SIZE
        and reference_count > 0
        and mutated_count > 0
        and reference_count_valid == 1
        and mutated_count_valid == 1
        and reference_padding_overlap == 0
        and mutated_padding_overlap == 0
    )


    if final_valid_flag == 1:

        status = (
            "VALID_REFERENCE_AND_MUTATED_MASKS"
        )

    else:

        status = (
            "MASK_VALIDATION_FAILED"
        )


    return {
        "REFERENCE_AFFECTED_MASK_301": (
            reference_mask
        ),

        "MUTATED_AFFECTED_MASK_301": (
            mutated_mask
        ),

        "REFERENCE_AFFECTED_INDICES_0BASED": (
            indices_to_text(
                reference_indices
            )
        ),

        "MUTATED_AFFECTED_INDICES_0BASED": (
            indices_to_text(
                mutated_indices
            )
        ),

        "REFERENCE_AFFECTED_START_INDEX_0BASED": (
            min(reference_indices)
        ),

        "REFERENCE_AFFECTED_END_INDEX_0BASED": (
            max(reference_indices)
        ),

        "MUTATED_AFFECTED_START_INDEX_0BASED": (
            min(mutated_indices)
        ),

        "MUTATED_AFFECTED_END_INDEX_0BASED": (
            max(mutated_indices)
        ),

        "REFERENCE_AFFECTED_POSITION_COUNT": (
            reference_count
        ),

        "MUTATED_AFFECTED_POSITION_COUNT": (
            mutated_count
        ),

        "REFERENCE_MASK_EXPECTED_COUNT": (
            expected_reference_count
        ),

        "MUTATED_MASK_EXPECTED_COUNT": (
            expected_mutated_count
        ),

        "REFERENCE_MASK_COUNT_VALID_FLAG": (
            reference_count_valid
        ),

        "MUTATED_MASK_COUNT_VALID_FLAG": (
            mutated_count_valid
        ),

        "REFERENCE_MASK_PADDING_OVERLAP_FLAG": (
            reference_padding_overlap
        ),

        "MUTATED_MASK_PADDING_OVERLAP_FLAG": (
            mutated_padding_overlap
        ),

        "MUTATION_MASK_POLICY": (
            policy
        ),

        "MUTATION_MASK_VALID_FLAG": (
            final_valid_flag
        ),

        "MUTATION_MASK_STATUS": (
            status
        ),
    }


# ------------------------------------------------------------
# 4. Prepare rerunnable working table
# ------------------------------------------------------------

working_df = (
    MUTATION_301_MUTATED_WINDOW_DF
    .copy()
)


generated_mask_columns = [
    "REFERENCE_AFFECTED_MASK_301",
    "MUTATED_AFFECTED_MASK_301",
    "REFERENCE_AFFECTED_INDICES_0BASED",
    "MUTATED_AFFECTED_INDICES_0BASED",
    "REFERENCE_AFFECTED_START_INDEX_0BASED",
    "REFERENCE_AFFECTED_END_INDEX_0BASED",
    "MUTATED_AFFECTED_START_INDEX_0BASED",
    "MUTATED_AFFECTED_END_INDEX_0BASED",
    "REFERENCE_AFFECTED_POSITION_COUNT",
    "MUTATED_AFFECTED_POSITION_COUNT",
    "REFERENCE_MASK_EXPECTED_COUNT",
    "MUTATED_MASK_EXPECTED_COUNT",
    "REFERENCE_MASK_COUNT_VALID_FLAG",
    "MUTATED_MASK_COUNT_VALID_FLAG",
    "REFERENCE_MASK_PADDING_OVERLAP_FLAG",
    "MUTATED_MASK_PADDING_OVERLAP_FLAG",
    "MUTATION_MASK_POLICY",
    "MUTATION_MASK_VALID_FLAG",
    "MUTATION_MASK_STATUS",
]


working_df = working_df.drop(
    columns=[
        column
        for column in generated_mask_columns
        if column in working_df.columns
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 5. Generate masks for every mutation row
# ------------------------------------------------------------

mask_results = pd.DataFrame(
    [
        generate_mutation_masks(row)
        for _, row in working_df.iterrows()
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        mask_results,
    ],
    axis=1,
)


MUTATION_301_MASK_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 6. Convert generated numeric fields
# ------------------------------------------------------------

nullable_integer_columns = [
    "REFERENCE_AFFECTED_START_INDEX_0BASED",
    "REFERENCE_AFFECTED_END_INDEX_0BASED",
    "MUTATED_AFFECTED_START_INDEX_0BASED",
    "MUTATED_AFFECTED_END_INDEX_0BASED",
    "REFERENCE_AFFECTED_POSITION_COUNT",
    "MUTATED_AFFECTED_POSITION_COUNT",
    "REFERENCE_MASK_EXPECTED_COUNT",
    "MUTATED_MASK_EXPECTED_COUNT",
    "REFERENCE_MASK_COUNT_VALID_FLAG",
    "MUTATED_MASK_COUNT_VALID_FLAG",
    "REFERENCE_MASK_PADDING_OVERLAP_FLAG",
    "MUTATED_MASK_PADDING_OVERLAP_FLAG",
    "MUTATION_MASK_VALID_FLAG",
]


for column in nullable_integer_columns:

    MUTATION_301_MASK_DF[column] = (
        pd.to_numeric(
            MUTATION_301_MASK_DF[column],
            errors="coerce",
        )
        .astype("Int64")
    )


# ------------------------------------------------------------
# 7. Validation masks
# ------------------------------------------------------------

window_pair_eligible_mask = (
    MUTATION_301_MASK_DF[
        "MUTATED_WINDOW_ELIGIBLE_FLAG"
    ]
    .eq(1)
)


mask_valid_mask = (
    MUTATION_301_MASK_DF[
        "MUTATION_MASK_VALID_FLAG"
    ]
    .eq(1)
)


eligible_mask_failure_mask = (
    window_pair_eligible_mask
    &
    ~mask_valid_mask
)


reference_mask_format_error_mask = (
    mask_valid_mask
    &
    ~MUTATION_301_MASK_DF[
        "REFERENCE_AFFECTED_MASK_301"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[01]{301}")
)


mutated_mask_format_error_mask = (
    mask_valid_mask
    &
    ~MUTATION_301_MASK_DF[
        "MUTATED_AFFECTED_MASK_301"
    ]
    .fillna("")
    .astype(str)
    .str.fullmatch(r"[01]{301}")
)


reference_count_error_mask = (
    mask_valid_mask
    &
    MUTATION_301_MASK_DF[
        "REFERENCE_AFFECTED_POSITION_COUNT"
    ]
    .ne(
        MUTATION_301_MASK_DF[
            "REFERENCE_MASK_EXPECTED_COUNT"
        ]
    )
)


mutated_count_error_mask = (
    mask_valid_mask
    &
    MUTATION_301_MASK_DF[
        "MUTATED_AFFECTED_POSITION_COUNT"
    ]
    .ne(
        MUTATION_301_MASK_DF[
            "MUTATED_MASK_EXPECTED_COUNT"
        ]
    )
)


padding_overlap_mask = (
    mask_valid_mask
    &
    (
        MUTATION_301_MASK_DF[
            "REFERENCE_MASK_PADDING_OVERLAP_FLAG"
        ]
        .eq(1)

        |

        MUTATION_301_MASK_DF[
            "MUTATED_MASK_PADDING_OVERLAP_FLAG"
        ]
        .eq(1)
    )
)


duplicate_mutation_id_mask = (
    MUTATION_301_MASK_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


# ------------------------------------------------------------
# 8. Precalculate summary counts
# ------------------------------------------------------------

total_rows = len(
    MUTATION_301_MASK_DF
)

eligible_window_pair_count = int(
    window_pair_eligible_mask.sum()
)

valid_mask_pair_count = int(
    mask_valid_mask.sum()
)

eligible_mask_failure_count = int(
    eligible_mask_failure_mask.sum()
)

reference_format_error_count = int(
    reference_mask_format_error_mask.sum()
)

mutated_format_error_count = int(
    mutated_mask_format_error_mask.sum()
)

reference_count_error_count = int(
    reference_count_error_mask.sum()
)

mutated_count_error_count = int(
    mutated_count_error_mask.sum()
)

padding_overlap_count = int(
    padding_overlap_mask.sum()
)

duplicate_mutation_id_count = int(
    duplicate_mutation_id_mask.sum()
)


# ------------------------------------------------------------
# 9. Create final eligible mask table
# ------------------------------------------------------------

MUTATION_301_MASK_ELIGIBLE_DF = (
    MUTATION_301_MASK_DF.loc[
        mask_valid_mask
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 10. Print validation results
# ------------------------------------------------------------

print(
    f"\nTotal mutation rows: "
    f"{total_rows:,}"
)

print(
    f"Eligible reference–mutated window pairs: "
    f"{eligible_window_pair_count:,}"
)

print(
    f"Valid reference–mutated mask pairs: "
    f"{valid_mask_pair_count:,}"
)

print(
    f"Eligible window pairs without valid masks: "
    f"{eligible_mask_failure_count:,}"
)

print(
    f"Reference-mask format errors: "
    f"{reference_format_error_count:,}"
)

print(
    f"Mutated-mask format errors: "
    f"{mutated_format_error_count:,}"
)

print(
    f"Reference-mask count errors: "
    f"{reference_count_error_count:,}"
)

print(
    f"Mutated-mask count errors: "
    f"{mutated_count_error_count:,}"
)

print(
    f"Masks overlapping padded positions: "
    f"{padding_overlap_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{duplicate_mutation_id_count:,}"
)


print(
    "\nMutation-mask status counts:"
)

print(
    MUTATION_301_MASK_DF[
        "MUTATION_MASK_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 11. Operation-level mask summary
# ------------------------------------------------------------

operation_mask_summary_df = (
    MUTATION_301_MASK_ELIGIBLE_DF
    .groupby(
        "HGVS_OPERATION",
        dropna=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        minimum_reference_mask_count=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "min",
        ),

        median_reference_mask_count=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "median",
        ),

        maximum_reference_mask_count=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "max",
        ),

        minimum_mutated_mask_count=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "min",
        ),

        median_mutated_mask_count=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "median",
        ),

        maximum_mutated_mask_count=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "max",
        ),
    )
    .reset_index()
)


print(
    "\nMask counts by mutation operation:"
)

print(
    operation_mask_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 12. Cohort-level summary
# ------------------------------------------------------------

cohort_mask_summary_df = (
    MUTATION_301_MASK_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        eligible_window_pairs=(
            "MUTATED_WINDOW_ELIGIBLE_FLAG",
            "sum",
        ),

        valid_mask_pairs=(
            "MUTATION_MASK_VALID_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print(
    "\nValid mask pairs by cohort:"
)

print(
    cohort_mask_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 13. Show representative examples
# ------------------------------------------------------------

example_rows = (
    MUTATION_301_MASK_ELIGIBLE_DF
    .sort_values(
        [
            "HGVS_OPERATION",
            "MUTATION_ROW_ID",
        ]
    )
    .groupby(
        "HGVS_OPERATION",
        group_keys=False,
    )
    .head(2)
)


if not example_rows.empty:

    print(
        "\nMask examples:"
    )

    print(
        example_rows[
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "REFERENCE_AFFECTED_INDICES_0BASED",
                "MUTATED_AFFECTED_INDICES_0BASED",
                "REFERENCE_AFFECTED_POSITION_COUNT",
                "MUTATED_AFFECTED_POSITION_COUNT",
                "MUTATION_MASK_POLICY",
            ]
        ]
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 14. Show mask failures when present
# ------------------------------------------------------------

if eligible_mask_failure_count > 0:

    print(
        "\nEligible rows without valid masks:"
    )

    print(
        MUTATION_301_MASK_DF.loc[
            eligible_mask_failure_mask,
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "MUTATION_MASK_STATUS",
            ],
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 15. Hard validations
# ------------------------------------------------------------

if len(
    MUTATION_301_MASK_DF
) != len(
    MUTATION_301_MUTATED_WINDOW_DF
):
    raise ValueError(
        "Mask generation changed the mutation-row count."
    )


if duplicate_mutation_id_mask.any():
    raise ValueError(
        "Mask generation duplicated mutation rows."
    )


if eligible_mask_failure_mask.any():
    raise ValueError(
        "Some eligible reference–mutated window pairs "
        "did not receive valid masks."
    )


if reference_mask_format_error_mask.any():
    raise ValueError(
        "A reference mask is not a 301-position "
        "binary string."
    )


if mutated_mask_format_error_mask.any():
    raise ValueError(
        "A mutated mask is not a 301-position "
        "binary string."
    )


if reference_count_error_mask.any():
    raise ValueError(
        "A reference mask contains an unexpected "
        "number of affected positions."
    )


if mutated_count_error_mask.any():
    raise ValueError(
        "A mutated mask contains an unexpected "
        "number of affected positions."
    )


if padding_overlap_mask.any():
    raise ValueError(
        "A mutation mask overlaps transcript-boundary padding."
    )


if valid_mask_pair_count != eligible_window_pair_count:
    raise ValueError(
        "Valid mask count does not match the eligible "
        "window-pair count."
    )


print("\n" + "=" * 78)
print("STEP 9H FINISHED")
print("=" * 78)

print(
    "Complete mutation-mask audit table:"
)

print(
    "MUTATION_301_MASK_DF"
)

print(
    "\nEligible window and mask table:"
)

print(
    "MUTATION_301_MASK_ELIGIBLE_DF"
)
```

## Step 9I — Final 301-bp Reconstruction and Exclusion Audit

This step performs the final technical validation of every reconstructed mutation.

For each eligible mutation, it verifies:

- reference and mutated windows contain exactly 301 DNA positions
- reference and mutated windows are different
- both masks contain exactly 301 binary positions
- mask counts agree with their stored affected-position counts
- masks do not overlap transcript-boundary padding
- affected reference and inserted sequences agree with the mutation operation
- substitution, deletion, insertion, duplication and delins coordinates are internally consistent

Every original selected-sample mutation row is assigned one deterministic reconstruction outcome:

- included as a valid 301-bp mutation reconstruction, or
- excluded with a specific transcript, HGVSc, reference-verification, window or mask reason

Patient-level coverage is also calculated without using stage or clinical target labels.

```python
# ============================================================
# STEP 9I — FINAL 301-BP RECONSTRUCTION AND EXCLUSION AUDIT
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_301_MASK_DF" not in globals():
    raise RuntimeError(
        "MUTATION_301_MASK_DF was not found. "
        "Run Step 9H first."
    )


print("=" * 78)
print("STEP 9I — FINAL 301-BP RECONSTRUCTION AUDIT")
print("=" * 78)


WINDOW_SIZE = 301

DNA_WINDOW_PATTERN = re.compile(
    r"^[ACGTN]{301}$"
)

BINARY_MASK_PATTERN = re.compile(
    r"^[01]{301}$"
)


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def flag_is_one(value):
    """
    Safely check whether a nullable flag equals 1.
    """

    if value is None or pd.isna(value):
        return False

    return int(value) == 1


def safe_status_text(value):
    """
    Convert a status value to a clean uppercase identifier.
    """

    if value is None or pd.isna(value):
        return "UNKNOWN"

    text = str(value).strip().upper()

    text = re.sub(
        r"[^A-Z0-9]+",
        "_",
        text,
    ).strip("_")

    return text if text else "UNKNOWN"


def mask_one_indices(mask):
    """
    Return all 0-based indices marked as 1.
    """

    return {
        index
        for index, value in enumerate(mask)
        if value == "1"
    }


def check_padding_consistency(
    sequence,
    mask,
    left_padding,
    right_padding,
):
    """
    Confirm padded positions contain N and are not mask-positive.
    """

    left_padding = int(
        left_padding
    )

    right_padding = int(
        right_padding
    )


    if left_padding > 0:

        if sequence[:left_padding] != (
            "N" * left_padding
        ):
            return False

        if "1" in mask[:left_padding]:
            return False


    if right_padding > 0:

        if sequence[-right_padding:] != (
            "N" * right_padding
        ):
            return False

        if "1" in mask[-right_padding:]:
            return False


    return True


# ------------------------------------------------------------
# 3. Validate one eligible reconstruction
# ------------------------------------------------------------

def validate_reconstructed_pair(row):
    """
    Perform operation-specific validation of one eligible
    reference–mutated window and mask pair.
    """

    if not flag_is_one(
        row["MUTATION_MASK_VALID_FLAG"]
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "NOT_EVALUATED_WINDOW_PAIR_INELIGIBLE"
            ),
        }


    reference_window = str(
        row["REFERENCE_WINDOW_301"]
    )

    mutated_window = str(
        row["MUTATED_WINDOW_301"]
    )

    reference_mask = str(
        row["REFERENCE_AFFECTED_MASK_301"]
    )

    mutated_mask = str(
        row["MUTATED_AFFECTED_MASK_301"]
    )


    # --------------------------------------------------------
    # Basic format checks
    # --------------------------------------------------------

    if not DNA_WINDOW_PATTERN.fullmatch(
        reference_window
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "INVALID_REFERENCE_WINDOW_FORMAT"
            ),
        }


    if not DNA_WINDOW_PATTERN.fullmatch(
        mutated_window
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "INVALID_MUTATED_WINDOW_FORMAT"
            ),
        }


    if reference_window == mutated_window:
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "REFERENCE_AND_MUTATED_WINDOWS_IDENTICAL"
            ),
        }


    if not BINARY_MASK_PATTERN.fullmatch(
        reference_mask
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "INVALID_REFERENCE_MASK_FORMAT"
            ),
        }


    if not BINARY_MASK_PATTERN.fullmatch(
        mutated_mask
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "INVALID_MUTATED_MASK_FORMAT"
            ),
        }


    reference_indices = mask_one_indices(
        reference_mask
    )

    mutated_indices = mask_one_indices(
        mutated_mask
    )


    if len(reference_indices) == 0:
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "REFERENCE_MASK_EMPTY"
            ),
        }


    if len(mutated_indices) == 0:
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "MUTATED_MASK_EMPTY"
            ),
        }


    stored_reference_count = int(
        row[
            "REFERENCE_AFFECTED_POSITION_COUNT"
        ]
    )

    stored_mutated_count = int(
        row[
            "MUTATED_AFFECTED_POSITION_COUNT"
        ]
    )


    if len(reference_indices) != (
        stored_reference_count
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "REFERENCE_MASK_COUNT_MISMATCH"
            ),
        }


    if len(mutated_indices) != (
        stored_mutated_count
    ):
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "MUTATED_MASK_COUNT_MISMATCH"
            ),
        }


    # --------------------------------------------------------
    # Padding checks
    # --------------------------------------------------------

    reference_padding_valid = (
        check_padding_consistency(
            sequence=reference_window,
            mask=reference_mask,
            left_padding=row[
                "WINDOW_LEFT_PADDING_COUNT"
            ],
            right_padding=row[
                "WINDOW_RIGHT_PADDING_COUNT"
            ],
        )
    )


    if not reference_padding_valid:
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "REFERENCE_PADDING_OR_MASK_ERROR"
            ),
        }


    mutated_padding_valid = (
        check_padding_consistency(
            sequence=mutated_window,
            mask=mutated_mask,
            left_padding=row[
                "MUTATED_WINDOW_LEFT_PADDING_COUNT"
            ],
            right_padding=row[
                "MUTATED_WINDOW_RIGHT_PADDING_COUNT"
            ],
        )
    )


    if not mutated_padding_valid:
        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "MUTATED_PADDING_OR_MASK_ERROR"
            ),
        }


    # --------------------------------------------------------
    # Read mutation fields
    # --------------------------------------------------------

    operation = str(
        row["HGVS_OPERATION"]
    ).strip().upper()


    event_start = int(
        row[
            "EVENT_WINDOW_START_INDEX_0BASED"
        ]
    )

    event_end = int(
        row[
            "EVENT_WINDOW_END_INDEX_0BASED"
        ]
    )

    edit_index = int(
        row[
            "MUTATION_EDIT_WINDOW_INDEX_0BASED"
        ]
    )


    removed_sequence = str(
        row["REFERENCE_REMOVED_SEQUENCE"]
    )

    inserted_sequence = str(
        row["ALTERNATE_INSERTED_SEQUENCE"]
    )


    expected_reference_event_indices = set(
        range(
            event_start,
            event_end + 1,
        )
    )


    # --------------------------------------------------------
    # Substitution
    # --------------------------------------------------------

    if operation == "SUBSTITUTION":

        if (
            event_start != event_end
            or edit_index != event_start
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "SUBSTITUTION_COORDINATE_ERROR"
                ),
            }


        if reference_indices != {
            event_start
        }:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "SUBSTITUTION_REFERENCE_MASK_ERROR"
                ),
            }


        if mutated_indices != {
            edit_index
        }:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "SUBSTITUTION_MUTATED_MASK_ERROR"
                ),
            }


        if reference_window[
            event_start
        ] != removed_sequence:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "SUBSTITUTION_REFERENCE_BASE_ERROR"
                ),
            }


        if mutated_window[
            edit_index
        ] != inserted_sequence:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "SUBSTITUTION_MUTATED_BASE_ERROR"
                ),
            }


    # --------------------------------------------------------
    # Deletion
    # --------------------------------------------------------

    elif operation == "DELETION":

        reference_event_sequence = (
            reference_window[
                event_start:
                event_end + 1
            ]
        )


        if reference_event_sequence != (
            removed_sequence
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELETION_REFERENCE_SEQUENCE_ERROR"
                ),
            }


        if reference_indices != (
            expected_reference_event_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELETION_REFERENCE_MASK_ERROR"
                ),
            }


        if inserted_sequence != "":
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELETION_INSERTED_SEQUENCE_NOT_EMPTY"
                ),
            }


        allowed_junction_indices = {
            index
            for index in [
                edit_index - 1,
                edit_index,
            ]
            if 0 <= index < WINDOW_SIZE
        }


        if not mutated_indices.issubset(
            allowed_junction_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELETION_JUNCTION_MASK_ERROR"
                ),
            }


    # --------------------------------------------------------
    # Insertion
    # --------------------------------------------------------

    elif operation == "INSERTION":

        expected_reference_indices = {
            event_start,
            event_end,
        }


        if reference_indices != (
            expected_reference_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "INSERTION_REFERENCE_FLANK_MASK_ERROR"
                ),
            }


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        expected_mutated_indices = set(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        if mutated_indices != (
            expected_mutated_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "INSERTION_MUTATED_MASK_ERROR"
                ),
            }


        if mutated_window[
            inserted_start:
            inserted_end + 1
        ] != inserted_sequence:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "INSERTION_SEQUENCE_ERROR"
                ),
            }


    # --------------------------------------------------------
    # Duplication
    # --------------------------------------------------------

    elif operation == "DUPLICATION":

        reference_event_sequence = (
            reference_window[
                event_start:
                event_end + 1
            ]
        )


        if reference_event_sequence != (
            inserted_sequence
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DUPLICATION_REFERENCE_SEQUENCE_ERROR"
                ),
            }


        if reference_indices != (
            expected_reference_event_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DUPLICATION_REFERENCE_MASK_ERROR"
                ),
            }


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        expected_mutated_indices = set(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        if mutated_indices != (
            expected_mutated_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DUPLICATION_MUTATED_MASK_ERROR"
                ),
            }


        if mutated_window[
            inserted_start:
            inserted_end + 1
        ] != inserted_sequence:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DUPLICATION_INSERTED_SEQUENCE_ERROR"
                ),
            }


    # --------------------------------------------------------
    # Deletion-insertion
    # --------------------------------------------------------

    elif operation == "DELINS":

        reference_event_sequence = (
            reference_window[
                event_start:
                event_end + 1
            ]
        )


        if reference_event_sequence != (
            removed_sequence
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELINS_REFERENCE_SEQUENCE_ERROR"
                ),
            }


        if reference_indices != (
            expected_reference_event_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELINS_REFERENCE_MASK_ERROR"
                ),
            }


        inserted_start = int(
            row[
                "MUTATED_INSERTED_START_INDEX_0BASED"
            ]
        )

        inserted_end = int(
            row[
                "MUTATED_INSERTED_END_INDEX_0BASED"
            ]
        )


        expected_mutated_indices = set(
            range(
                inserted_start,
                inserted_end + 1,
            )
        )


        if mutated_indices != (
            expected_mutated_indices
        ):
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELINS_MUTATED_MASK_ERROR"
                ),
            }


        if mutated_window[
            inserted_start:
            inserted_end + 1
        ] != inserted_sequence:
            return {
                "PAIR_VALIDATION_FLAG": 0,
                "PAIR_VALIDATION_STATUS": (
                    "DELINS_INSERTED_SEQUENCE_ERROR"
                ),
            }


    else:

        return {
            "PAIR_VALIDATION_FLAG": 0,
            "PAIR_VALIDATION_STATUS": (
                "UNSUPPORTED_OPERATION"
            ),
        }


    return {
        "PAIR_VALIDATION_FLAG": 1,
        "PAIR_VALIDATION_STATUS": (
            "VALID_301BP_RECONSTRUCTION"
        ),
    }


# ------------------------------------------------------------
# 4. Prepare rerunnable working table
# ------------------------------------------------------------

working_df = (
    MUTATION_301_MASK_DF
    .copy()
)


working_df = working_df.drop(
    columns=[
        "PAIR_VALIDATION_FLAG",
        "PAIR_VALIDATION_STATUS",
        "FINAL_301_RECONSTRUCTION_FLAG",
        "FINAL_301_RECONSTRUCTION_STATUS",
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 5. Validate all mutation rows
# ------------------------------------------------------------

pair_validation_results = pd.DataFrame(
    [
        validate_reconstructed_pair(row)
        for _, row in working_df.iterrows()
    ],
    index=working_df.index,
)


working_df = pd.concat(
    [
        working_df,
        pair_validation_results,
    ],
    axis=1,
)


working_df[
    "PAIR_VALIDATION_FLAG"
] = pd.to_numeric(
    working_df[
        "PAIR_VALIDATION_FLAG"
    ],
    errors="coerce",
).fillna(0).astype("int8")


# ------------------------------------------------------------
# 6. Assign deterministic final reconstruction outcome
# ------------------------------------------------------------

def assign_final_reconstruction_status(row):
    """
    Assign one final inclusion or exclusion reason.
    """

    if flag_is_one(
        row["PAIR_VALIDATION_FLAG"]
    ):
        return (
            "INCLUDED_VALID_301BP_RECONSTRUCTION"
        )


    if not flag_is_one(
        row["TRANSCRIPT_MATCH_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row["TRANSCRIPT_MATCH_STATUS"]
            )
        )


    if not flag_is_one(
        row["HGVSC_PARSED_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row["HGVS_PARSE_STATUS"]
            )
        )


    if not flag_is_one(
        row["HGVSC_COORDINATE_USABLE_FLAG"]
    ):
        return (
            "EXCLUDED_HGVSC_COORDINATE_NOT_USABLE"
        )


    if not flag_is_one(
        row["MUTATION_PAYLOAD_USABLE_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row[
                    "REFERENCE_VERIFICATION_STATUS"
                ]
            )
        )


    if not flag_is_one(
        row["REFERENCE_WINDOW_ELIGIBLE_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row["REFERENCE_WINDOW_STATUS"]
            )
        )


    if not flag_is_one(
        row["MUTATED_WINDOW_ELIGIBLE_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row["MUTATED_WINDOW_STATUS"]
            )
        )


    if not flag_is_one(
        row["MUTATION_MASK_VALID_FLAG"]
    ):
        return (
            "EXCLUDED_"
            +
            safe_status_text(
                row["MUTATION_MASK_STATUS"]
            )
        )


    return (
        "EXCLUDED_"
        +
        safe_status_text(
            row["PAIR_VALIDATION_STATUS"]
        )
    )


working_df[
    "FINAL_301_RECONSTRUCTION_STATUS"
] = (
    working_df.apply(
        assign_final_reconstruction_status,
        axis=1,
    )
    .astype("string")
)


working_df[
    "FINAL_301_RECONSTRUCTION_FLAG"
] = (
    working_df[
        "FINAL_301_RECONSTRUCTION_STATUS"
    ]
    .eq(
        "INCLUDED_VALID_301BP_RECONSTRUCTION"
    )
    .astype("int8")
)


# ------------------------------------------------------------
# 7. Final audit and valid mutation tables
# ------------------------------------------------------------

MUTATION_301_FINAL_AUDIT_DF = (
    working_df.copy()
)


MUTATION_301_VALID_DF = (
    MUTATION_301_FINAL_AUDIT_DF.loc[
        MUTATION_301_FINAL_AUDIT_DF[
            "FINAL_301_RECONSTRUCTION_FLAG"
        ]
        .eq(1)
    ]
    .copy()
    .reset_index(drop=True)
)


MUTATION_301_EXCLUDED_DF = (
    MUTATION_301_FINAL_AUDIT_DF.loc[
        MUTATION_301_FINAL_AUDIT_DF[
            "FINAL_301_RECONSTRUCTION_FLAG"
        ]
        .eq(0)
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. Build exclusion summary
# ------------------------------------------------------------

MUTATION_301_EXCLUSION_SUMMARY_DF = (
    MUTATION_301_FINAL_AUDIT_DF
    .groupby(
        "FINAL_301_RECONSTRUCTION_STATUS",
        dropna=False,
    )
    .agg(
        MUTATION_ROWS=(
            "MUTATION_ROW_ID",
            "size",
        ),

        AFFECTED_PATIENTS=(
            "PATIENT_ID",
            "nunique",
        ),

        AFFECTED_GENES=(
            "GENE_SYMBOL",
            "nunique",
        ),
    )
    .reset_index()
    .sort_values(
        "MUTATION_ROWS",
        ascending=False,
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 9. Patient-level reconstruction coverage
# ------------------------------------------------------------

patient_total_counts = (
    MUTATION_301_FINAL_AUDIT_DF
    .groupby(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ],
        as_index=False,
    )
    .agg(
        TOTAL_SELECTED_MUTATION_ROWS=(
            "MUTATION_ROW_ID",
            "size",
        ),

        TOTAL_SELECTED_MUTATED_GENES=(
            "GENE_SYMBOL",
            "nunique",
        ),
    )
)


patient_valid_counts = (
    MUTATION_301_VALID_DF
    .groupby(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ],
        as_index=False,
    )
    .agg(
        VALID_301_MUTATION_COUNT=(
            "MUTATION_ROW_ID",
            "size",
        ),

        VALID_301_GENE_COUNT=(
            "GENE_SYMBOL",
            "nunique",
        ),
    )
)


PATIENT_301_RECONSTRUCTION_COVERAGE_DF = (
    patient_total_counts.merge(
        patient_valid_counts,
        on=[
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ],
        how="left",
        validate="one_to_one",
    )
)


for column in [
    "VALID_301_MUTATION_COUNT",
    "VALID_301_GENE_COUNT",
]:

    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        column
    ] = (
        PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
            column
        ]
        .fillna(0)
        .astype(int)
    )


PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
    "EXCLUDED_MUTATION_COUNT"
] = (
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        "TOTAL_SELECTED_MUTATION_ROWS"
    ]
    -
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        "VALID_301_MUTATION_COUNT"
    ]
)


PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
    "HAS_VALID_301_RECONSTRUCTION_FLAG"
] = (
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .gt(0)
    .astype("int8")
)


# ------------------------------------------------------------
# 10. Validation masks and counts
# ------------------------------------------------------------

duplicate_mutation_id_mask = (
    MUTATION_301_FINAL_AUDIT_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


eligible_mask_pair_count = int(
    MUTATION_301_FINAL_AUDIT_DF[
        "MUTATION_MASK_VALID_FLAG"
    ]
    .eq(1)
    .sum()
)


pair_validation_failure_mask = (
    MUTATION_301_FINAL_AUDIT_DF[
        "MUTATION_MASK_VALID_FLAG"
    ]
    .eq(1)
    &
    MUTATION_301_FINAL_AUDIT_DF[
        "PAIR_VALIDATION_FLAG"
    ]
    .ne(1)
)


valid_reconstruction_count = int(
    MUTATION_301_FINAL_AUDIT_DF[
        "FINAL_301_RECONSTRUCTION_FLAG"
    ]
    .sum()
)


excluded_reconstruction_count = int(
    MUTATION_301_FINAL_AUDIT_DF[
        "FINAL_301_RECONSTRUCTION_FLAG"
    ]
    .eq(0)
    .sum()
)


unclassified_exclusion_mask = (
    MUTATION_301_FINAL_AUDIT_DF[
        "FINAL_301_RECONSTRUCTION_STATUS"
    ]
    .eq("EXCLUDED_UNKNOWN")
)


total_patient_count = len(
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF
)


patients_with_valid_count = int(
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        "HAS_VALID_301_RECONSTRUCTION_FLAG"
    ]
    .sum()
)


patients_without_valid_count = int(
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        "HAS_VALID_301_RECONSTRUCTION_FLAG"
    ]
    .eq(0)
    .sum()
)


valid_gene_count = int(
    MUTATION_301_VALID_DF[
        "GENE_SYMBOL"
    ]
    .nunique()
)


duplicate_id_count = int(
    duplicate_mutation_id_mask.sum()
)


pair_validation_failure_count = int(
    pair_validation_failure_mask.sum()
)


unclassified_exclusion_count = int(
    unclassified_exclusion_mask.sum()
)


# ------------------------------------------------------------
# 11. Print final reconstruction audit
# ------------------------------------------------------------

print(
    f"\nTotal selected-sample mutation rows: "
    f"{len(MUTATION_301_FINAL_AUDIT_DF):,}"
)

print(
    f"Valid 301-bp reconstructions: "
    f"{valid_reconstruction_count:,}"
)

print(
    f"Excluded mutation rows: "
    f"{excluded_reconstruction_count:,}"
)

print(
    f"Eligible mask pairs entering final validation: "
    f"{eligible_mask_pair_count:,}"
)

print(
    f"Eligible mask pairs failing final validation: "
    f"{pair_validation_failure_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{duplicate_id_count:,}"
)

print(
    f"Unclassified exclusions: "
    f"{unclassified_exclusion_count:,}"
)

print(
    f"Genes represented by valid reconstructions: "
    f"{valid_gene_count:,}"
)

print(
    f"Patients with selected mutations: "
    f"{total_patient_count:,}"
)

print(
    f"Patients with at least one valid 301-bp mutation: "
    f"{patients_with_valid_count:,}"
)

print(
    f"Patients with no valid 301-bp mutation: "
    f"{patients_without_valid_count:,}"
)


print(
    "\nFinal reconstruction status counts:"
)

print(
    MUTATION_301_FINAL_AUDIT_DF[
        "FINAL_301_RECONSTRUCTION_STATUS"
    ]
    .value_counts(dropna=False)
    .to_string()
)


# ------------------------------------------------------------
# 12. Cohort-level reconstruction summary
# ------------------------------------------------------------

cohort_reconstruction_summary_df = (
    MUTATION_301_FINAL_AUDIT_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        valid_301_reconstructions=(
            "FINAL_301_RECONSTRUCTION_FLAG",
            "sum",
        ),

        represented_patients=(
            "PATIENT_ID",
            "nunique",
        ),
    )
    .reset_index()
)


cohort_patient_valid_summary = (
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        patients_with_valid_301=(
            "HAS_VALID_301_RECONSTRUCTION_FLAG",
            "sum",
        ),

        patients_without_valid_301=(
            "HAS_VALID_301_RECONSTRUCTION_FLAG",
            lambda values: int(
                values.eq(0).sum()
            ),
        ),
    )
    .reset_index()
)


cohort_reconstruction_summary_df = (
    cohort_reconstruction_summary_df.merge(
        cohort_patient_valid_summary,
        on="COHORT",
        how="left",
        validate="one_to_one",
    )
)


print(
    "\nFinal reconstruction coverage by cohort:"
)

print(
    cohort_reconstruction_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 13. Show patients without a valid reconstruction
# ------------------------------------------------------------

patients_without_valid_examples = (
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF.loc[
        PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
            "HAS_VALID_301_RECONSTRUCTION_FLAG"
        ]
        .eq(0)
    ]
    .head(20)
)


if not patients_without_valid_examples.empty:

    print(
        "\nPatients without a valid 301-bp reconstruction:"
    )

    print(
        patients_without_valid_examples.to_string(
            index=False
        )
    )


# ------------------------------------------------------------
# 14. Show final validation failures if present
# ------------------------------------------------------------

if pair_validation_failure_count > 0:

    print(
        "\nEligible mask pairs failing final validation:"
    )

    print(
        MUTATION_301_FINAL_AUDIT_DF.loc[
            pair_validation_failure_mask,
            [
                "MUTATION_ROW_ID",
                "COHORT",
                "GENE_SYMBOL",
                "HGVS_OPERATION",
                "HGVSC_RAW",
                "PAIR_VALIDATION_STATUS",
            ],
        ]
        .head(20)
        .to_string(index=False)
    )


# ------------------------------------------------------------
# 15. Hard validations
# ------------------------------------------------------------

if len(
    MUTATION_301_FINAL_AUDIT_DF
) != len(
    MUTATION_301_MASK_DF
):
    raise ValueError(
        "Final reconstruction audit changed "
        "the mutation-row count."
    )


if duplicate_mutation_id_mask.any():
    raise ValueError(
        "Duplicate mutation IDs were detected."
    )


if pair_validation_failure_mask.any():
    raise ValueError(
        "Some mask-eligible reconstruction pairs failed "
        "the final operation-specific validation."
    )


if valid_reconstruction_count != (
    eligible_mask_pair_count
):
    raise ValueError(
        "Final valid reconstruction count does not match "
        "the eligible mask-pair count."
    )


if unclassified_exclusion_mask.any():
    raise ValueError(
        "Some excluded mutations do not have a specific "
        "reconstruction reason."
    )


if (
    valid_reconstruction_count
    +
    excluded_reconstruction_count
) != len(
    MUTATION_301_FINAL_AUDIT_DF
):
    raise ValueError(
        "Included and excluded counts do not equal "
        "the total mutation-row count."
    )


if total_patient_count != 4_998:
    raise ValueError(
        f"Expected 4,998 representative patients, "
        f"found {total_patient_count:,}."
    )


print("\n" + "=" * 78)
print("STEP 9I FINISHED")
print("=" * 78)

print(
    "Valid mutation-level reconstruction table:"
)

print(
    "MUTATION_301_VALID_DF"
)

print(
    "\nComplete mutation inclusion/exclusion audit:"
)

print(
    "MUTATION_301_FINAL_AUDIT_DF"
)

print(
    "\nPatient reconstruction coverage table:"
)

print(
    "PATIENT_301_RECONSTRUCTION_COVERAGE_DF"
)

print(
    "\nReconstruction exclusion summary:"
)

print(
    "MUTATION_301_EXCLUSION_SUMMARY_DF"
)

print(
    "\nStage and clinical target labels were not used."
)
```

## Step 9J — Derive Mutation Consequence and Severity Features

This step creates deterministic mutation-level consequence features for every valid 301-bp reconstruction.

The severity classification uses:

- the original mutation-file variant classification
- mutation operation
- reference and alternate sequence lengths
- inferred reading-frame disruption

Severity categories:

- `HIGH`: frameshift, nonsense, splice-site, nonstop or start-loss consequences
- `MODERATE`: missense, in-frame deletion/insertion, duplication or other coding sequence alteration
- `LOW`: synonymous or silent coding variants
- `UNKNOWN`: insufficient annotation for consequence assignment

This is a computational consequence category, not a clinical prognosis or patient-risk score. Stage and other target labels are not used.

```python
# ============================================================
# STEP 9J — MUTATION CONSEQUENCE AND SEVERITY FEATURES
# ============================================================

import pandas as pd
import re


# ------------------------------------------------------------
# 1. Validate required input
# ------------------------------------------------------------

if "MUTATION_301_VALID_DF" not in globals():
    raise RuntimeError(
        "MUTATION_301_VALID_DF was not found. "
        "Run Step 9I first."
    )

if "selected_genes" not in globals():
    raise RuntimeError(
        "selected_genes was not found."
    )


print("=" * 78)
print("STEP 9J — MUTATION CONSEQUENCE AND SEVERITY FEATURES")
print("=" * 78)


# ------------------------------------------------------------
# 2. Helper functions
# ------------------------------------------------------------

def normalise_variant_classification(value):
    """
    Convert a raw variant classification into a consistent
    uppercase identifier.
    """

    if value is None or pd.isna(value):
        return "UNKNOWN"

    text = str(value).strip().upper()

    if text in {
        "",
        "NA",
        "N/A",
        "NAN",
        "NONE",
        "NULL",
        "UNKNOWN",
        ".",
    }:
        return "UNKNOWN"

    text = re.sub(
        r"[^A-Z0-9]+",
        "_",
        text,
    ).strip("_")

    return text if text else "UNKNOWN"


def classify_source_consequence(classification):
    """
    Classify the original source variant annotation.
    """

    text = str(classification).upper()


    high_terms = [
        "FRAME_SHIFT",
        "NONSENSE",
        "NONSTOP",
        "SPLICE_SITE",
        "SPLICE_ACCEPTOR",
        "SPLICE_DONOR",
        "TRANSLATION_START_SITE",
        "START_CODON",
        "START_LOST",
        "STOP_GAINED",
        "STOP_LOST",
    ]


    moderate_terms = [
        "MISSENSE",
        "IN_FRAME",
        "PROTEIN_ALTERING",
        "CODING_SEQUENCE_VARIANT",
        "DISRUPTIVE_INFRAME",
        "CONSERVATIVE_INFRAME",
    ]


    low_terms = [
        "SILENT",
        "SYNONYMOUS",
        "STOP_RETAINED",
    ]


    if any(
        term in text
        for term in high_terms
    ):
        return "HIGH"


    if any(
        term in text
        for term in moderate_terms
    ):
        return "MODERATE"


    if any(
        term in text
        for term in low_terms
    ):
        return "LOW"


    return "UNKNOWN"


def assign_final_severity(row):
    """
    Assign a deterministic computational consequence severity.

    Priority:
    1. Source high-impact annotation
    2. Inferred frameshift
    3. Source moderate annotation
    4. Other non-frameshift coding indel/duplication
    5. Source low-impact annotation
    6. Unknown
    """

    source_severity = row[
        "SOURCE_CONSEQUENCE_SEVERITY"
    ]

    operation = row[
        "HGVS_OPERATION"
    ]

    frameshift_flag = int(
        row["FRAME_SHIFT_INFERRED_FLAG"]
    )


    if source_severity == "HIGH":

        return pd.Series({
            "MUTATION_CONSEQUENCE_SEVERITY": "HIGH",
            "MUTATION_SEVERITY_SCORE": 3,
            "MUTATION_SEVERITY_RULE": (
                "SOURCE_HIGH_IMPACT_CLASSIFICATION"
            ),
        })


    if frameshift_flag == 1:

        return pd.Series({
            "MUTATION_CONSEQUENCE_SEVERITY": "HIGH",
            "MUTATION_SEVERITY_SCORE": 3,
            "MUTATION_SEVERITY_RULE": (
                "FRAME_SHIFT_INFERRED_FROM_CDS_LENGTH_CHANGE"
            ),
        })


    if source_severity == "MODERATE":

        return pd.Series({
            "MUTATION_CONSEQUENCE_SEVERITY": "MODERATE",
            "MUTATION_SEVERITY_SCORE": 2,
            "MUTATION_SEVERITY_RULE": (
                "SOURCE_MODERATE_IMPACT_CLASSIFICATION"
            ),
        })


    if operation in {
        "DELETION",
        "INSERTION",
        "DUPLICATION",
        "DELINS",
    }:

        return pd.Series({
            "MUTATION_CONSEQUENCE_SEVERITY": "MODERATE",
            "MUTATION_SEVERITY_SCORE": 2,
            "MUTATION_SEVERITY_RULE": (
                "NON_FRAMESHIFT_CODING_SEQUENCE_ALTERATION"
            ),
        })


    if source_severity == "LOW":

        return pd.Series({
            "MUTATION_CONSEQUENCE_SEVERITY": "LOW",
            "MUTATION_SEVERITY_SCORE": 1,
            "MUTATION_SEVERITY_RULE": (
                "SOURCE_LOW_IMPACT_CLASSIFICATION"
            ),
        })


    return pd.Series({
        "MUTATION_CONSEQUENCE_SEVERITY": "UNKNOWN",
        "MUTATION_SEVERITY_SCORE": 0,
        "MUTATION_SEVERITY_RULE": (
            "INSUFFICIENT_CONSEQUENCE_ANNOTATION"
        ),
    })


# ------------------------------------------------------------
# 3. Prepare rerunnable source table
# ------------------------------------------------------------

working_df = (
    MUTATION_301_VALID_DF
    .copy()
)


generated_columns = [
    "VARIANT_CLASSIFICATION_NORMALISED",
    "SOURCE_CONSEQUENCE_SEVERITY",
    "MUTATION_LENGTH_CHANGE_ABS",
    "MUTATION_LENGTH_EFFECT",
    "FRAME_SHIFT_INFERRED_FLAG",
    "IN_FRAME_LENGTH_CHANGE_FLAG",
    "NO_LENGTH_CHANGE_FLAG",
    "MUTATION_OPERATION_GROUP",
    "MUTATION_CONSEQUENCE_SEVERITY",
    "MUTATION_SEVERITY_SCORE",
    "MUTATION_SEVERITY_RULE",
    "REFERENCE_WINDOW_N_COUNT",
    "MUTATED_WINDOW_N_COUNT",
    "WINDOW_PAIR_TOTAL_AFFECTED_POSITIONS",
]


working_df = working_df.drop(
    columns=[
        column
        for column in generated_columns
        if column in working_df.columns
    ],
    errors="ignore",
)


# ------------------------------------------------------------
# 4. Normalise source variant classifications
# ------------------------------------------------------------

working_df[
    "VARIANT_CLASSIFICATION_NORMALISED"
] = (
    working_df[
        "VARIANT_CLASSIFICATION_RAW"
    ]
    .apply(
        normalise_variant_classification
    )
    .astype("string")
)


working_df[
    "SOURCE_CONSEQUENCE_SEVERITY"
] = (
    working_df[
        "VARIANT_CLASSIFICATION_NORMALISED"
    ]
    .apply(
        classify_source_consequence
    )
    .astype("string")
)


# ------------------------------------------------------------
# 5. Validate and derive mutation length effects
# ------------------------------------------------------------

working_df[
    "REFERENCE_REMOVAL_LENGTH"
] = pd.to_numeric(
    working_df[
        "REFERENCE_REMOVAL_LENGTH"
    ],
    errors="coerce",
).astype("Int64")


working_df[
    "ALTERNATE_INSERTION_LENGTH"
] = pd.to_numeric(
    working_df[
        "ALTERNATE_INSERTION_LENGTH"
    ],
    errors="coerce",
).astype("Int64")


working_df[
    "NET_CDS_LENGTH_CHANGE"
] = pd.to_numeric(
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ],
    errors="coerce",
).astype("Int64")


working_df[
    "MUTATION_LENGTH_CHANGE_ABS"
] = (
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .abs()
    .astype("Int64")
)


length_changing_operation_mask = (
    working_df[
        "HGVS_OPERATION"
    ]
    .isin(
        [
            "DELETION",
            "INSERTION",
            "DUPLICATION",
            "DELINS",
        ]
    )
)


working_df[
    "FRAME_SHIFT_INFERRED_FLAG"
] = (
    length_changing_operation_mask
    &
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .ne(0)
    &
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .abs()
    .mod(3)
    .ne(0)
).astype("int8")


working_df[
    "IN_FRAME_LENGTH_CHANGE_FLAG"
] = (
    length_changing_operation_mask
    &
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .ne(0)
    &
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .abs()
    .mod(3)
    .eq(0)
).astype("int8")


working_df[
    "NO_LENGTH_CHANGE_FLAG"
] = (
    working_df[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .eq(0)
    .astype("int8")
)


def assign_length_effect(row):

    if int(
        row["FRAME_SHIFT_INFERRED_FLAG"]
    ) == 1:
        return "FRAME_SHIFT_LENGTH_CHANGE"

    if int(
        row["IN_FRAME_LENGTH_CHANGE_FLAG"]
    ) == 1:
        return "IN_FRAME_LENGTH_CHANGE"

    if int(
        row["NO_LENGTH_CHANGE_FLAG"]
    ) == 1:
        return "NO_CDS_LENGTH_CHANGE"

    return "OTHER_LENGTH_EFFECT"


working_df[
    "MUTATION_LENGTH_EFFECT"
] = (
    working_df.apply(
        assign_length_effect,
        axis=1,
    )
    .astype("string")
)


# ------------------------------------------------------------
# 6. Create broad mutation-operation groups
# ------------------------------------------------------------

operation_group_map = {
    "SUBSTITUTION": "SINGLE_BASE_SUBSTITUTION",
    "DELETION": "CODING_DELETION",
    "INSERTION": "CODING_INSERTION",
    "DUPLICATION": "CODING_DUPLICATION",
    "DELINS": "CODING_DELETION_INSERTION",
}


working_df[
    "MUTATION_OPERATION_GROUP"
] = (
    working_df[
        "HGVS_OPERATION"
    ]
    .map(operation_group_map)
    .fillna("OTHER")
    .astype("string")
)


# ------------------------------------------------------------
# 7. Assign final computational severity
# ------------------------------------------------------------

severity_results = working_df.apply(
    assign_final_severity,
    axis=1,
)


working_df = pd.concat(
    [
        working_df,
        severity_results,
    ],
    axis=1,
)


working_df[
    "MUTATION_SEVERITY_SCORE"
] = pd.to_numeric(
    working_df[
        "MUTATION_SEVERITY_SCORE"
    ],
    errors="coerce",
).astype("Int64")


# ------------------------------------------------------------
# 8. Add compact window summary features
# ------------------------------------------------------------

working_df[
    "REFERENCE_WINDOW_N_COUNT"
] = (
    working_df[
        "REFERENCE_WINDOW_301"
    ]
    .astype(str)
    .str.count("N")
    .astype("Int64")
)


working_df[
    "MUTATED_WINDOW_N_COUNT"
] = (
    working_df[
        "MUTATED_WINDOW_301"
    ]
    .astype(str)
    .str.count("N")
    .astype("Int64")
)


working_df[
    "WINDOW_PAIR_TOTAL_AFFECTED_POSITIONS"
] = (
    working_df[
        "REFERENCE_AFFECTED_POSITION_COUNT"
    ]
    +
    working_df[
        "MUTATED_AFFECTED_POSITION_COUNT"
    ]
).astype("Int64")


# ------------------------------------------------------------
# 9. Final mutation consequence table
# ------------------------------------------------------------

MUTATION_301_CONSEQUENCE_DF = (
    working_df.copy()
)


# ------------------------------------------------------------
# 10. Structural validations
# ------------------------------------------------------------

duplicate_mutation_id_mask = (
    MUTATION_301_CONSEQUENCE_DF
    .duplicated(
        subset=["MUTATION_ROW_ID"],
        keep=False,
    )
)


net_length_mismatch_mask = (
    MUTATION_301_CONSEQUENCE_DF[
        "NET_CDS_LENGTH_CHANGE"
    ]
    .ne(
        MUTATION_301_CONSEQUENCE_DF[
            "ALTERNATE_INSERTION_LENGTH"
        ]
        -
        MUTATION_301_CONSEQUENCE_DF[
            "REFERENCE_REMOVAL_LENGTH"
        ]
    )
)


conflicting_frame_flags_mask = (
    MUTATION_301_CONSEQUENCE_DF[
        "FRAME_SHIFT_INFERRED_FLAG"
    ]
    .eq(1)
    &
    MUTATION_301_CONSEQUENCE_DF[
        "IN_FRAME_LENGTH_CHANGE_FLAG"
    ]
    .eq(1)
)


missing_severity_mask = (
    MUTATION_301_CONSEQUENCE_DF[
        "MUTATION_CONSEQUENCE_SEVERITY"
    ]
    .isna()
    |
    MUTATION_301_CONSEQUENCE_DF[
        "MUTATION_SEVERITY_SCORE"
    ]
    .isna()
)


invalid_severity_score_mask = (
    ~MUTATION_301_CONSEQUENCE_DF[
        "MUTATION_SEVERITY_SCORE"
    ]
    .isin(
        [
            0,
            1,
            2,
            3,
        ]
    )
)


invalid_window_length_mask = (
    MUTATION_301_CONSEQUENCE_DF[
        "REFERENCE_WINDOW_301"
    ]
    .astype(str)
    .str.len()
    .ne(301)

    |

    MUTATION_301_CONSEQUENCE_DF[
        "MUTATED_WINDOW_301"
    ]
    .astype(str)
    .str.len()
    .ne(301)
)


invalid_mask_length_mask = (
    MUTATION_301_CONSEQUENCE_DF[
        "REFERENCE_AFFECTED_MASK_301"
    ]
    .astype(str)
    .str.len()
    .ne(301)

    |

    MUTATION_301_CONSEQUENCE_DF[
        "MUTATED_AFFECTED_MASK_301"
    ]
    .astype(str)
    .str.len()
    .ne(301)
)


# ------------------------------------------------------------
# 11. Precalculate counts
# ------------------------------------------------------------

total_rows = len(
    MUTATION_301_CONSEQUENCE_DF
)

unique_patient_count = (
    MUTATION_301_CONSEQUENCE_DF[
        [
            "COHORT",
            "PATIENT_ID",
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

unique_gene_count = int(
    MUTATION_301_CONSEQUENCE_DF[
        "GENE_SYMBOL"
    ]
    .nunique()
)

duplicate_id_count = int(
    duplicate_mutation_id_mask.sum()
)

net_length_mismatch_count = int(
    net_length_mismatch_mask.sum()
)

conflicting_frame_flag_count = int(
    conflicting_frame_flags_mask.sum()
)

missing_severity_count = int(
    missing_severity_mask.sum()
)

invalid_score_count = int(
    invalid_severity_score_mask.sum()
)

invalid_window_length_count = int(
    invalid_window_length_mask.sum()
)

invalid_mask_length_count = int(
    invalid_mask_length_mask.sum()
)

frameshift_count = int(
    MUTATION_301_CONSEQUENCE_DF[
        "FRAME_SHIFT_INFERRED_FLAG"
    ]
    .sum()
)

in_frame_length_change_count = int(
    MUTATION_301_CONSEQUENCE_DF[
        "IN_FRAME_LENGTH_CHANGE_FLAG"
    ]
    .sum()
)

no_length_change_count = int(
    MUTATION_301_CONSEQUENCE_DF[
        "NO_LENGTH_CHANGE_FLAG"
    ]
    .sum()
)


valid_reconstruction_genes = set(
    MUTATION_301_CONSEQUENCE_DF[
        "GENE_SYMBOL"
    ]
    .dropna()
    .astype(str)
)


PANEL_GENES_WITHOUT_VALID_301_RECONSTRUCTION = sorted(
    selected_genes
    -
    valid_reconstruction_genes
)


# ------------------------------------------------------------
# 12. Print validation results
# ------------------------------------------------------------

print(
    f"\nValid mutation rows: "
    f"{total_rows:,}"
)

print(
    f"Patients represented: "
    f"{unique_patient_count:,}"
)

print(
    f"Genes represented: "
    f"{unique_gene_count:,}"
)

print(
    "Panel genes without a valid reconstruction: "
    f"{PANEL_GENES_WITHOUT_VALID_301_RECONSTRUCTION}"
)

print(
    f"Inferred frameshift mutations: "
    f"{frameshift_count:,}"
)

print(
    f"In-frame length-changing mutations: "
    f"{in_frame_length_change_count:,}"
)

print(
    f"Mutations without CDS length change: "
    f"{no_length_change_count:,}"
)

print(
    f"Duplicate mutation IDs: "
    f"{duplicate_id_count:,}"
)

print(
    f"Net-length calculation mismatches: "
    f"{net_length_mismatch_count:,}"
)

print(
    f"Conflicting frame flags: "
    f"{conflicting_frame_flag_count:,}"
)

print(
    f"Rows with missing severity: "
    f"{missing_severity_count:,}"
)

print(
    f"Invalid severity scores: "
    f"{invalid_score_count:,}"
)

print(
    f"Invalid 301-bp window lengths: "
    f"{invalid_window_length_count:,}"
)

print(
    f"Invalid 301-position mask lengths: "
    f"{invalid_mask_length_count:,}"
)


print(
    "\nMutation consequence severity counts:"
)

print(
    MUTATION_301_CONSEQUENCE_DF[
        "MUTATION_CONSEQUENCE_SEVERITY"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nMutation severity-rule counts:"
)

print(
    MUTATION_301_CONSEQUENCE_DF[
        "MUTATION_SEVERITY_RULE"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nNormalised source-classification counts:"
)

print(
    MUTATION_301_CONSEQUENCE_DF[
        "VARIANT_CLASSIFICATION_NORMALISED"
    ]
    .value_counts(dropna=False)
    .head(20)
    .to_string()
)


# ------------------------------------------------------------
# 13. Severity by mutation operation
# ------------------------------------------------------------

severity_operation_summary_df = (
    MUTATION_301_CONSEQUENCE_DF
    .groupby(
        [
            "HGVS_OPERATION",
            "MUTATION_CONSEQUENCE_SEVERITY",
        ],
        dropna=False,
    )
    .agg(
        MUTATION_ROWS=(
            "MUTATION_ROW_ID",
            "size",
        ),

        AFFECTED_PATIENTS=(
            "PATIENT_ID",
            "nunique",
        ),
    )
    .reset_index()
)


print(
    "\nSeverity by mutation operation:"
)

print(
    severity_operation_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 14. Cohort-level summary
# ------------------------------------------------------------

cohort_consequence_summary_df = (
    MUTATION_301_CONSEQUENCE_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        valid_mutation_rows=(
            "MUTATION_ROW_ID",
            "size",
        ),

        represented_patients=(
            "PATIENT_ID",
            "nunique",
        ),

        high_severity_mutations=(
            "MUTATION_CONSEQUENCE_SEVERITY",
            lambda values: int(
                values.eq("HIGH").sum()
            ),
        ),

        moderate_severity_mutations=(
            "MUTATION_CONSEQUENCE_SEVERITY",
            lambda values: int(
                values.eq("MODERATE").sum()
            ),
        ),

        low_severity_mutations=(
            "MUTATION_CONSEQUENCE_SEVERITY",
            lambda values: int(
                values.eq("LOW").sum()
            ),
        ),

        unknown_severity_mutations=(
            "MUTATION_CONSEQUENCE_SEVERITY",
            lambda values: int(
                values.eq("UNKNOWN").sum()
            ),
        ),
    )
    .reset_index()
)


print(
    "\nMutation consequence summary by cohort:"
)

print(
    cohort_consequence_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 15. Hard validations
# ------------------------------------------------------------

if len(
    MUTATION_301_CONSEQUENCE_DF
) != len(
    MUTATION_301_VALID_DF
):
    raise ValueError(
        "Consequence derivation changed the "
        "valid mutation-row count."
    )


if duplicate_mutation_id_mask.any():
    raise ValueError(
        "Duplicate mutation IDs were detected."
    )


if net_length_mismatch_mask.any():
    raise ValueError(
        "Some mutation length-change calculations "
        "are inconsistent."
    )


if conflicting_frame_flags_mask.any():
    raise ValueError(
        "A mutation was classified as both frameshift "
        "and in-frame."
    )


if missing_severity_mask.any():
    raise ValueError(
        "Some valid mutations do not have a severity category."
    )


if invalid_severity_score_mask.any():
    raise ValueError(
        "An invalid mutation severity score was generated."
    )


if invalid_window_length_mask.any():
    raise ValueError(
        "A valid mutation does not contain two "
        "301-bp windows."
    )


if invalid_mask_length_mask.any():
    raise ValueError(
        "A valid mutation does not contain two "
        "301-position masks."
    )


print("\n" + "=" * 78)
print("STEP 9J FINISHED")
print("=" * 78)

print(
    "Mutation consequence table:"
)

print(
    "MUTATION_301_CONSEQUENCE_DF"
)

print(
    "\nThe severity field is a computational consequence "
    "category, not a clinical-risk label."
)
```

## Step 9K — Save the Essential Mutation-Level Dataset and Reconstruction Audit

This step saves two reproducible outputs:

1. The valid mutation-level 301-bp dataset containing identifiers, mutation annotations, paired windows, masks and consequence features.
2. A compact audit containing every excluded mutation and its exact exclusion reason.

Full CDS sequences and clinical target variables are not included in the mutation-level export. Each CSV is accompanied by a matching TXT documentation file containing counts, definitions, validation results and a SHA-256 checksum.

```python
# ============================================================
# STEP 9K — SAVE ESSENTIAL MUTATION-LEVEL OUTPUTS
# ============================================================

from pathlib import Path
from datetime import datetime
import hashlib
import os
import re
import pandas as pd


# ------------------------------------------------------------
# 1. Validate required in-memory tables
# ------------------------------------------------------------

required_tables = [
    "MUTATION_301_CONSEQUENCE_DF",
    "MUTATION_301_EXCLUDED_DF",
]

missing_tables = [
    name
    for name in required_tables
    if name not in globals()
]

if missing_tables:
    raise RuntimeError(
        "Missing required table(s): "
        + ", ".join(missing_tables)
    )


print("=" * 78)
print("STEP 9K — SAVE ESSENTIAL MUTATION-LEVEL OUTPUTS")
print("=" * 78)


# ------------------------------------------------------------
# 2. Define the approved output location
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Users\LAPTOP\Downloads\luad_reserch_code"
)

FINAL_RESULTS_DIR = (
    PROJECT_ROOT
    / "final_luad_results"
)

DATA_DIR = (
    FINAL_RESULTS_DIR
    / "data"
)

DATA_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


mutation_csv_path = (
    DATA_DIR
    / "luad_301bp_mutation_level_dataset.csv"
)

mutation_txt_path = (
    DATA_DIR
    / "luad_301bp_mutation_level_dataset.txt"
)

exclusion_csv_path = (
    DATA_DIR
    / "luad_301bp_mutation_exclusion_audit.csv"
)

exclusion_txt_path = (
    DATA_DIR
    / "luad_301bp_mutation_exclusion_audit.txt"
)


# ------------------------------------------------------------
# 3. Helper functions
# ------------------------------------------------------------

def select_existing_columns(
    dataframe,
    requested_columns,
):
    """
    Preserve the requested order while selecting only
    columns available in the dataframe.
    """

    return [
        column
        for column in requested_columns
        if column in dataframe.columns
    ]


def atomic_save_csv(
    dataframe,
    output_path,
):
    """
    Save through a temporary file and replace the final file
    only after writing completes.
    """

    temporary_path = output_path.with_suffix(
        output_path.suffix + ".tmp"
    )

    dataframe.to_csv(
        temporary_path,
        index=False,
        encoding="utf-8",
    )

    os.replace(
        temporary_path,
        output_path,
    )


def atomic_save_text(
    text,
    output_path,
):
    """
    Save a UTF-8 text file atomically.
    """

    temporary_path = output_path.with_suffix(
        output_path.suffix + ".tmp"
    )

    temporary_path.write_text(
        text,
        encoding="utf-8",
    )

    os.replace(
        temporary_path,
        output_path,
    )


def sha256_file(file_path):
    """
    Calculate the SHA-256 checksum of a file.
    """

    digest = hashlib.sha256()

    with open(file_path, "rb") as file_handle:

        for block in iter(
            lambda: file_handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()


def file_size_mb(file_path):
    """
    Return the file size in megabytes.
    """

    return (
        file_path.stat().st_size
        / (1024 ** 2)
    )


# ------------------------------------------------------------
# 4. Build the valid mutation-level export
# ------------------------------------------------------------

mutation_export_columns = [
    # Identifiers
    "MUTATION_ROW_ID",
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "GENE_SYMBOL",

    # Transcript and mutation annotation
    "TRANSCRIPT_ID_SELECTED",
    "TRANSCRIPT_ID_MATCHED",
    "HGVSC_RAW",
    "HGVS_C_EXPRESSION",
    "HGVS_OPERATION",
    "VARIANT_CLASSIFICATION_RAW",
    "VARIANT_CLASSIFICATION_NORMALISED",

    # Coding coordinates and mutation payload
    "CDS_START_1BASED",
    "CDS_END_1BASED",
    "REFERENCE_CDS_LENGTH",
    "REFERENCE_REMOVED_SEQUENCE",
    "ALTERNATE_INSERTED_SEQUENCE",
    "REFERENCE_REMOVAL_LENGTH",
    "ALTERNATE_INSERTION_LENGTH",
    "NET_CDS_LENGTH_CHANGE",

    # Window coordinates
    "WINDOW_ANCHOR_CDS_POSITION_1BASED",
    "WINDOW_REQUESTED_CDS_START_1BASED",
    "WINDOW_REQUESTED_CDS_END_1BASED",
    "EVENT_WINDOW_START_INDEX_0BASED",
    "EVENT_WINDOW_END_INDEX_0BASED",
    "MUTATION_EDIT_WINDOW_INDEX_0BASED",

    # Reference and mutated windows
    "REFERENCE_WINDOW_301",
    "MUTATED_WINDOW_301",

    # Window masks
    "REFERENCE_AFFECTED_MASK_301",
    "MUTATED_AFFECTED_MASK_301",
    "REFERENCE_AFFECTED_INDICES_0BASED",
    "MUTATED_AFFECTED_INDICES_0BASED",
    "REFERENCE_AFFECTED_POSITION_COUNT",
    "MUTATED_AFFECTED_POSITION_COUNT",
    "MUTATION_MASK_POLICY",

    # Boundary-padding information
    "WINDOW_LEFT_PADDING_COUNT",
    "WINDOW_RIGHT_PADDING_COUNT",
    "MUTATED_WINDOW_LEFT_PADDING_COUNT",
    "MUTATED_WINDOW_RIGHT_PADDING_COUNT",
    "REFERENCE_WINDOW_N_COUNT",
    "MUTATED_WINDOW_N_COUNT",

    # Consequence features
    "MUTATION_OPERATION_GROUP",
    "MUTATION_LENGTH_CHANGE_ABS",
    "MUTATION_LENGTH_EFFECT",
    "FRAME_SHIFT_INFERRED_FLAG",
    "IN_FRAME_LENGTH_CHANGE_FLAG",
    "NO_LENGTH_CHANGE_FLAG",
    "SOURCE_CONSEQUENCE_SEVERITY",
    "MUTATION_CONSEQUENCE_SEVERITY",
    "MUTATION_SEVERITY_SCORE",
    "MUTATION_SEVERITY_RULE",
    "WINDOW_PAIR_TOTAL_AFFECTED_POSITIONS",

    # Reconstruction status
    "FINAL_301_RECONSTRUCTION_FLAG",
    "FINAL_301_RECONSTRUCTION_STATUS",
]


selected_mutation_columns = (
    select_existing_columns(
        MUTATION_301_CONSEQUENCE_DF,
        mutation_export_columns,
    )
)


critical_mutation_columns = [
    "MUTATION_ROW_ID",
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "GENE_SYMBOL",
    "HGVSC_RAW",
    "HGVS_OPERATION",
    "CDS_START_1BASED",
    "CDS_END_1BASED",
    "REFERENCE_WINDOW_301",
    "MUTATED_WINDOW_301",
    "REFERENCE_AFFECTED_MASK_301",
    "MUTATED_AFFECTED_MASK_301",
    "MUTATION_CONSEQUENCE_SEVERITY",
    "MUTATION_SEVERITY_SCORE",
]


missing_critical_columns = [
    column
    for column in critical_mutation_columns
    if column not in selected_mutation_columns
]

if missing_critical_columns:
    raise ValueError(
        "Critical mutation export columns are missing: "
        + ", ".join(
            missing_critical_columns
        )
    )


MUTATION_LEVEL_EXPORT_DF = (
    MUTATION_301_CONSEQUENCE_DF[
        selected_mutation_columns
    ]
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "MUTATION_ROW_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. Build the compact exclusion audit
# ------------------------------------------------------------

exclusion_export_columns = [
    "MUTATION_ROW_ID",
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "GENE_SYMBOL",
    "TRANSCRIPT_ID_SELECTED",
    "HGVSC_RAW",
    "HGVS_C_EXPRESSION",
    "TRANSCRIPT_MATCH_STATUS",
    "HGVS_PARSE_STATUS",
    "REFERENCE_VERIFICATION_STATUS",
    "REFERENCE_WINDOW_STATUS",
    "MUTATED_WINDOW_STATUS",
    "MUTATION_MASK_STATUS",
    "PAIR_VALIDATION_STATUS",
    "FINAL_301_RECONSTRUCTION_STATUS",
]


selected_exclusion_columns = (
    select_existing_columns(
        MUTATION_301_EXCLUDED_DF,
        exclusion_export_columns,
    )
)


critical_exclusion_columns = [
    "MUTATION_ROW_ID",
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "GENE_SYMBOL",
    "FINAL_301_RECONSTRUCTION_STATUS",
]


missing_exclusion_columns = [
    column
    for column in critical_exclusion_columns
    if column not in selected_exclusion_columns
]

if missing_exclusion_columns:
    raise ValueError(
        "Critical exclusion-audit columns are missing: "
        + ", ".join(
            missing_exclusion_columns
        )
    )


MUTATION_EXCLUSION_EXPORT_DF = (
    MUTATION_301_EXCLUDED_DF[
        selected_exclusion_columns
    ]
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "MUTATION_ROW_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 6. Pre-save validation
# ------------------------------------------------------------

expected_valid_rows = len(
    MUTATION_301_CONSEQUENCE_DF
)

expected_excluded_rows = len(
    MUTATION_301_EXCLUDED_DF
)


if len(MUTATION_LEVEL_EXPORT_DF) != (
    expected_valid_rows
):
    raise ValueError(
        "Mutation export row count changed."
    )


if len(MUTATION_EXCLUSION_EXPORT_DF) != (
    expected_excluded_rows
):
    raise ValueError(
        "Exclusion audit row count changed."
    )


if MUTATION_LEVEL_EXPORT_DF[
    "MUTATION_ROW_ID"
].duplicated().any():
    raise ValueError(
        "Duplicate mutation IDs exist in the valid export."
    )


if MUTATION_EXCLUSION_EXPORT_DF[
    "MUTATION_ROW_ID"
].duplicated().any():
    raise ValueError(
        "Duplicate mutation IDs exist in the exclusion audit."
    )


key_missing_count = int(
    MUTATION_LEVEL_EXPORT_DF[
        [
            "MUTATION_ROW_ID",
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "GENE_SYMBOL",
        ]
    ]
    .isna()
    .any(axis=1)
    .sum()
)


if key_missing_count > 0:
    raise ValueError(
        "Valid mutation export contains missing identifiers."
    )


window_length_error_count = int(
    (
        MUTATION_LEVEL_EXPORT_DF[
            "REFERENCE_WINDOW_301"
        ]
        .astype(str)
        .str.len()
        .ne(301)

        |

        MUTATION_LEVEL_EXPORT_DF[
            "MUTATED_WINDOW_301"
        ]
        .astype(str)
        .str.len()
        .ne(301)
    )
    .sum()
)


mask_length_error_count = int(
    (
        MUTATION_LEVEL_EXPORT_DF[
            "REFERENCE_AFFECTED_MASK_301"
        ]
        .astype(str)
        .str.len()
        .ne(301)

        |

        MUTATION_LEVEL_EXPORT_DF[
            "MUTATED_AFFECTED_MASK_301"
        ]
        .astype(str)
        .str.len()
        .ne(301)
    )
    .sum()
)


identical_window_count = int(
    MUTATION_LEVEL_EXPORT_DF[
        "REFERENCE_WINDOW_301"
    ]
    .eq(
        MUTATION_LEVEL_EXPORT_DF[
            "MUTATED_WINDOW_301"
        ]
    )
    .sum()
)


if window_length_error_count > 0:
    raise ValueError(
        "Invalid 301-bp window length found before saving."
    )


if mask_length_error_count > 0:
    raise ValueError(
        "Invalid 301-position mask length found before saving."
    )


if identical_window_count > 0:
    raise ValueError(
        "Identical reference and mutated windows entered "
        "the valid export."
    )


# ------------------------------------------------------------
# 7. Confirm that target-label columns are absent
# ------------------------------------------------------------

target_column_pattern = re.compile(
    r"(STAGE|TNM|TUMOU?R_SIZE|LYMPH|METASTASIS|"
    r"TARGET|LABEL)",
    flags=re.IGNORECASE,
)


target_like_columns = [
    column
    for column in MUTATION_LEVEL_EXPORT_DF.columns
    if target_column_pattern.search(column)
]


if target_like_columns:
    raise ValueError(
        "Potential target-label columns entered the "
        "mutation export: "
        + ", ".join(target_like_columns)
    )


# ------------------------------------------------------------
# 8. Calculate documentation summaries
# ------------------------------------------------------------

valid_patient_count = (
    MUTATION_LEVEL_EXPORT_DF[
        [
            "COHORT",
            "PATIENT_ID",
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


valid_gene_count = int(
    MUTATION_LEVEL_EXPORT_DF[
        "GENE_SYMBOL"
    ]
    .nunique()
)


operation_counts_text = (
    MUTATION_LEVEL_EXPORT_DF[
        "HGVS_OPERATION"
    ]
    .value_counts()
    .to_string()
)


severity_counts_text = (
    MUTATION_LEVEL_EXPORT_DF[
        "MUTATION_CONSEQUENCE_SEVERITY"
    ]
    .value_counts()
    .to_string()
)


exclusion_counts_text = (
    MUTATION_EXCLUSION_EXPORT_DF[
        "FINAL_301_RECONSTRUCTION_STATUS"
    ]
    .value_counts()
    .to_string()
)


panel_genes_without_valid = globals().get(
    "PANEL_GENES_WITHOUT_VALID_301_RECONSTRUCTION",
    [],
)

panel_genes_without_valid_text = (
    ", ".join(
        panel_genes_without_valid
    )
    if panel_genes_without_valid
    else "None"
)


# ------------------------------------------------------------
# 9. Save both CSV files
# ------------------------------------------------------------

atomic_save_csv(
    MUTATION_LEVEL_EXPORT_DF,
    mutation_csv_path,
)

atomic_save_csv(
    MUTATION_EXCLUSION_EXPORT_DF,
    exclusion_csv_path,
)


mutation_csv_checksum = sha256_file(
    mutation_csv_path
)

exclusion_csv_checksum = sha256_file(
    exclusion_csv_path
)


mutation_csv_size = file_size_mb(
    mutation_csv_path
)

exclusion_csv_size = file_size_mb(
    exclusion_csv_path
)


# ------------------------------------------------------------
# 10. Create matching TXT documentation
# ------------------------------------------------------------

creation_time = datetime.now().strftime(
    "%Y-%m-%d %H:%M:%S"
)


mutation_documentation = f"""
LUAD 301-BP MUTATION-LEVEL DATASET
=================================

File
----
CSV: {mutation_csv_path.name}
Created: {creation_time}
SHA-256: {mutation_csv_checksum}
File size: {mutation_csv_size:.2f} MB

Purpose
-------
This file contains one row per valid reconstructed mutation from
the seven LUAD cohorts. Each mutation has a freshly constructed
301-bp reference window, a corresponding mutated window, aligned
binary affected-position masks, coding coordinates and consequence
features.

Dataset size
------------
Mutation rows: {len(MUTATION_LEVEL_EXPORT_DF):,}
Unique cohort-patient combinations: {valid_patient_count:,}
Represented genes: {valid_gene_count:,}
Number of columns: {len(MUTATION_LEVEL_EXPORT_DF.columns):,}

Genes from the 96-gene panel without a valid 301-bp reconstruction
-----------------------------------------------------------------
{panel_genes_without_valid_text}

Mutation operation counts
-------------------------
{operation_counts_text}

Consequence severity counts
---------------------------
{severity_counts_text}

Important sequence fields
-------------------------
REFERENCE_WINDOW_301:
    Fresh 301-bp window extracted from the matched Ensembl CDS.

MUTATED_WINDOW_301:
    Fixed-length 301-bp window after applying the verified
    transcript-oriented mutation.

REFERENCE_AFFECTED_MASK_301:
    A 301-position binary mask marking the affected reference
    positions.

MUTATED_AFFECTED_MASK_301:
    A 301-position binary mask marking inserted, replaced or
    deletion-junction positions in the mutated window.

Padding
-------
N characters occur only where the requested window extends beyond
a true CDS boundary. Mutation masks do not overlap padded positions.

Consequence severity
--------------------
The severity category is a deterministic computational consequence
feature. It is not a clinical prognosis, stage label or patient-risk
score.

Target leakage
--------------
Stage, TNM, tumour size, lymph-node, metastasis and clinical target
labels are not included in this mutation-level file.

Columns
-------
{chr(10).join(MUTATION_LEVEL_EXPORT_DF.columns)}
""".strip()


exclusion_documentation = f"""
LUAD 301-BP MUTATION EXCLUSION AUDIT
====================================

File
----
CSV: {exclusion_csv_path.name}
Created: {creation_time}
SHA-256: {exclusion_csv_checksum}
File size: {exclusion_csv_size:.2f} MB

Purpose
-------
This file preserves every selected-sample mutation that could not
enter the valid 301-bp reconstruction dataset. Each row contains a
deterministic exclusion reason.

Excluded mutation rows: {len(MUTATION_EXCLUSION_EXPORT_DF):,}
Number of columns: {len(MUTATION_EXCLUSION_EXPORT_DF.columns):,}

Exclusion counts
----------------
{exclusion_counts_text}

Transcript-specific policy
--------------------------
Mutation records were not transferred to alternative transcripts.
This particularly affects ZFHX4 and EPPK1 mutation annotations whose
specified transcripts were absent from the selected Ensembl BioMart
CDS export. HGVSc coding coordinates were treated as transcript-
specific.

Columns
-------
{chr(10).join(MUTATION_EXCLUSION_EXPORT_DF.columns)}
""".strip()


atomic_save_text(
    mutation_documentation,
    mutation_txt_path,
)

atomic_save_text(
    exclusion_documentation,
    exclusion_txt_path,
)


# ------------------------------------------------------------
# 11. Verify the saved CSV files
# ------------------------------------------------------------

saved_mutation_ids = pd.read_csv(
    mutation_csv_path,
    usecols=[
        "MUTATION_ROW_ID",
    ],
    dtype="string",
)


saved_exclusion_ids = pd.read_csv(
    exclusion_csv_path,
    usecols=[
        "MUTATION_ROW_ID",
    ],
    dtype="string",
)


saved_mutation_row_count = len(
    saved_mutation_ids
)

saved_exclusion_row_count = len(
    saved_exclusion_ids
)


if saved_mutation_row_count != len(
    MUTATION_LEVEL_EXPORT_DF
):
    raise ValueError(
        "Saved mutation-level CSV row count is incorrect."
    )


if saved_exclusion_row_count != len(
    MUTATION_EXCLUSION_EXPORT_DF
):
    raise ValueError(
        "Saved exclusion-audit CSV row count is incorrect."
    )


if not mutation_txt_path.exists():
    raise FileNotFoundError(
        "Mutation-level TXT documentation was not saved."
    )


if not exclusion_txt_path.exists():
    raise FileNotFoundError(
        "Exclusion-audit TXT documentation was not saved."
    )


# ------------------------------------------------------------
# 12. Print final save report
# ------------------------------------------------------------

print(
    f"\nValid mutation-level rows saved: "
    f"{saved_mutation_row_count:,}"
)

print(
    f"Excluded mutation rows saved: "
    f"{saved_exclusion_row_count:,}"
)

print(
    f"Valid patients represented: "
    f"{valid_patient_count:,}"
)

print(
    f"Genes represented: "
    f"{valid_gene_count:,}"
)

print(
    f"Window length errors before saving: "
    f"{window_length_error_count:,}"
)

print(
    f"Mask length errors before saving: "
    f"{mask_length_error_count:,}"
)

print(
    f"Target-like columns detected: "
    f"{len(target_like_columns):,}"
)


print("\nSaved files:")

print(
    f"1. {mutation_csv_path}"
)

print(
    f"2. {mutation_txt_path}"
)

print(
    f"3. {exclusion_csv_path}"
)

print(
    f"4. {exclusion_txt_path}"
)


print(
    "\nMutation CSV SHA-256:"
)

print(
    mutation_csv_checksum
)


print(
    "\nExclusion CSV SHA-256:"
)

print(
    exclusion_csv_checksum
)


print("\n" + "=" * 78)
print("STEP 9K FINISHED")
print("=" * 78)

print(
    "Valid mutation-level export table:"
)

print(
    "MUTATION_LEVEL_EXPORT_DF"
)

print(
    "\nCompact exclusion export table:"
)

print(
    "MUTATION_EXCLUSION_EXPORT_DF"
)
```

## Step 10A — Aggregate Mutations into One Row per Patient

This step converts the valid mutation-level reconstructions into a patient-level structure.

All valid mutations for each representative patient are preserved as aligned variable-length lists, including:

- gene symbols
- transcript IDs
- HGVSc annotations
- mutation operations
- reference 301-bp windows
- mutated 301-bp windows
- reference masks
- mutated masks
- mutation severity categories and scores

Patients with no valid reconstruction are retained with empty lists and zero valid-mutation counts. No mutation limit or truncation is applied.

The previously saved separate exclusion-audit CSV and TXT files are removed because only essential dataset files will be retained.

```python
# ============================================================
# STEP 10A — PATIENT-LEVEL VARIABLE-LENGTH LIST AGGREGATION
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. Validate required inputs
# ------------------------------------------------------------

required_tables = [
    "MUTATION_301_CONSEQUENCE_DF",
    "PATIENT_301_RECONSTRUCTION_COVERAGE_DF",
]

missing_tables = [
    table_name
    for table_name in required_tables
    if table_name not in globals()
]

if missing_tables:
    raise RuntimeError(
        "Missing required table(s): "
        + ", ".join(missing_tables)
    )


print("=" * 78)
print("STEP 10A — PATIENT-LEVEL VARIABLE-LENGTH AGGREGATION")
print("=" * 78)


# ------------------------------------------------------------
# 2. Remove the two extra exclusion-audit files
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Users\LAPTOP\Downloads\luad_reserch_code"
)

DATA_DIR = (
    PROJECT_ROOT
    / "final_luad_results"
    / "data"
)


extra_exclusion_files = [
    DATA_DIR
    / "luad_301bp_mutation_exclusion_audit.csv",

    DATA_DIR
    / "luad_301bp_mutation_exclusion_audit.txt",
]


removed_files = []
already_absent_files = []


for file_path in extra_exclusion_files:

    if file_path.exists():

        file_path.unlink()

        removed_files.append(
            str(file_path)
        )

    else:

        already_absent_files.append(
            str(file_path)
        )


print(
    f"\nExtra exclusion files removed: "
    f"{len(removed_files)}"
)


for file_path in removed_files:

    print(
        f"Removed: {file_path}"
    )


if already_absent_files:

    print(
        f"Extra exclusion files already absent: "
        f"{len(already_absent_files)}"
    )


# ------------------------------------------------------------
# 3. Helper for clean Python scalar values
# ------------------------------------------------------------

def to_python_value(value):
    """
    Convert pandas and NumPy scalar values into standard
    Python values suitable for patient-level lists.
    """

    if value is None or pd.isna(value):
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    return value


def aggregate_as_list(series):
    """
    Preserve row order and convert values into a Python list.
    """

    return [
        to_python_value(value)
        for value in series.tolist()
    ]


def unique_preserve_order(values):
    """
    Return unique nonmissing values while preserving their
    first-occurrence order.
    """

    unique_values = []
    observed_values = set()

    for value in values:

        if value is None or pd.isna(value):
            continue

        clean_value = to_python_value(
            value
        )

        if clean_value not in observed_values:

            observed_values.add(
                clean_value
            )

            unique_values.append(
                clean_value
            )

    return unique_values


# ------------------------------------------------------------
# 4. Prepare the representative-patient base
# ------------------------------------------------------------

patient_base_columns = [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "TOTAL_SELECTED_MUTATION_ROWS",
    "TOTAL_SELECTED_MUTATED_GENES",
    "VALID_301_MUTATION_COUNT",
    "VALID_301_GENE_COUNT",
    "EXCLUDED_MUTATION_COUNT",
    "HAS_VALID_301_RECONSTRUCTION_FLAG",
]


missing_patient_base_columns = [
    column
    for column in patient_base_columns
    if column not in
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF.columns
]

if missing_patient_base_columns:
    raise ValueError(
        "Patient reconstruction table is missing: "
        + ", ".join(
            missing_patient_base_columns
        )
    )


patient_base_df = (
    PATIENT_301_RECONSTRUCTION_COVERAGE_DF[
        patient_base_columns
    ]
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. Validate the valid mutation-level source table
# ------------------------------------------------------------

valid_mutation_df = (
    MUTATION_301_CONSEQUENCE_DF
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "MUTATION_ROW_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


critical_mutation_columns = [
    "MUTATION_ROW_ID",
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
    "GENE_SYMBOL",
    "TRANSCRIPT_ID_SELECTED",
    "HGVSC_RAW",
    "HGVS_OPERATION",
    "REFERENCE_WINDOW_301",
    "MUTATED_WINDOW_301",
    "REFERENCE_AFFECTED_MASK_301",
    "MUTATED_AFFECTED_MASK_301",
    "MUTATION_CONSEQUENCE_SEVERITY",
    "MUTATION_SEVERITY_SCORE",
]


missing_mutation_columns = [
    column
    for column in critical_mutation_columns
    if column not in valid_mutation_df.columns
]

if missing_mutation_columns:
    raise ValueError(
        "Valid mutation table is missing: "
        + ", ".join(
            missing_mutation_columns
        )
    )


# ------------------------------------------------------------
# 6. Define aligned variable-length list columns
# ------------------------------------------------------------

list_source_to_output = {
    "MUTATION_ROW_ID":
        "MUTATION_ROW_ID_LIST",

    "GENE_SYMBOL":
        "GENE_SYMBOL_LIST",

    "TRANSCRIPT_ID_SELECTED":
        "TRANSCRIPT_ID_LIST",

    "HGVSC_RAW":
        "HGVSC_LIST",

    "HGVS_OPERATION":
        "HGVS_OPERATION_LIST",

    "VARIANT_CLASSIFICATION_NORMALISED":
        "VARIANT_CLASSIFICATION_LIST",

    "CDS_START_1BASED":
        "CDS_START_1BASED_LIST",

    "CDS_END_1BASED":
        "CDS_END_1BASED_LIST",

    "REFERENCE_REMOVED_SEQUENCE":
        "REFERENCE_REMOVED_SEQUENCE_LIST",

    "ALTERNATE_INSERTED_SEQUENCE":
        "ALTERNATE_INSERTED_SEQUENCE_LIST",

    "REFERENCE_WINDOW_301":
        "REFERENCE_WINDOW_301_LIST",

    "MUTATED_WINDOW_301":
        "MUTATED_WINDOW_301_LIST",

    "REFERENCE_AFFECTED_MASK_301":
        "REFERENCE_AFFECTED_MASK_301_LIST",

    "MUTATED_AFFECTED_MASK_301":
        "MUTATED_AFFECTED_MASK_301_LIST",

    "REFERENCE_AFFECTED_POSITION_COUNT":
        "REFERENCE_AFFECTED_POSITION_COUNT_LIST",

    "MUTATED_AFFECTED_POSITION_COUNT":
        "MUTATED_AFFECTED_POSITION_COUNT_LIST",

    "MUTATION_LENGTH_EFFECT":
        "MUTATION_LENGTH_EFFECT_LIST",

    "FRAME_SHIFT_INFERRED_FLAG":
        "FRAME_SHIFT_FLAG_LIST",

    "IN_FRAME_LENGTH_CHANGE_FLAG":
        "IN_FRAME_LENGTH_CHANGE_FLAG_LIST",

    "MUTATION_CONSEQUENCE_SEVERITY":
        "MUTATION_SEVERITY_LIST",

    "MUTATION_SEVERITY_SCORE":
        "MUTATION_SEVERITY_SCORE_LIST",
}


available_list_mapping = {
    source_column: output_column
    for source_column, output_column
    in list_source_to_output.items()
    if source_column in valid_mutation_df.columns
}


missing_essential_list_sources = [
    column
    for column in critical_mutation_columns
    if column not in available_list_mapping
    and column not in [
        "COHORT",
        "PATIENT_ID",
        "SAMPLE_ID",
    ]
]

if missing_essential_list_sources:
    raise ValueError(
        "Essential list fields are unavailable: "
        + ", ".join(
            missing_essential_list_sources
        )
    )


# ------------------------------------------------------------
# 7. Aggregate aligned mutation lists
# ------------------------------------------------------------

group_columns = [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
]


aggregation_dictionary = {
    source_column: aggregate_as_list
    for source_column
    in available_list_mapping
}


patient_list_aggregation_df = (
    valid_mutation_df
    .groupby(
        group_columns,
        sort=False,
        as_index=False,
    )
    .agg(
        aggregation_dictionary
    )
    .rename(
        columns=available_list_mapping
    )
)


# ------------------------------------------------------------
# 8. Add unique gene lists
# ------------------------------------------------------------

patient_unique_genes_df = (
    valid_mutation_df
    .groupby(
        group_columns,
        sort=False,
        as_index=False,
    )
    .agg(
        UNIQUE_VALID_GENE_SYMBOL_LIST=(
            "GENE_SYMBOL",
            unique_preserve_order,
        )
    )
)


patient_list_aggregation_df = (
    patient_list_aggregation_df.merge(
        patient_unique_genes_df,
        on=group_columns,
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 9. Merge lists onto all 4,998 representative patients
# ------------------------------------------------------------

PATIENT_301_LIST_AGGREGATED_DF = (
    patient_base_df.merge(
        patient_list_aggregation_df,
        on=group_columns,
        how="left",
        validate="one_to_one",
    )
)


all_list_columns = (
    list(
        available_list_mapping.values()
    )
    +
    [
        "UNIQUE_VALID_GENE_SYMBOL_LIST",
    ]
)


# Replace missing aggregation values with independent empty lists.
for column in all_list_columns:

    PATIENT_301_LIST_AGGREGATED_DF[
        column
    ] = [
        value
        if isinstance(value, list)
        else []
        for value in
        PATIENT_301_LIST_AGGREGATED_DF[
            column
        ]
    ]


# ------------------------------------------------------------
# 10. Derive list-based counts
# ------------------------------------------------------------

PATIENT_301_LIST_AGGREGATED_DF[
    "AGGREGATED_MUTATION_LIST_LENGTH"
] = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "MUTATION_ROW_ID_LIST"
    ]
    .apply(len)
    .astype(int)
)


PATIENT_301_LIST_AGGREGATED_DF[
    "AGGREGATED_UNIQUE_GENE_LIST_LENGTH"
] = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "UNIQUE_VALID_GENE_SYMBOL_LIST"
    ]
    .apply(len)
    .astype(int)
)


# ------------------------------------------------------------
# 11. Validate aligned list lengths
# ------------------------------------------------------------

list_length_columns = []

for list_column in all_list_columns:

    length_column = (
        list_column
        +
        "__LENGTH_CHECK"
    )

    PATIENT_301_LIST_AGGREGATED_DF[
        length_column
    ] = (
        PATIENT_301_LIST_AGGREGATED_DF[
            list_column
        ]
        .apply(len)
        .astype(int)
    )

    list_length_columns.append(
        length_column
    )


mutation_aligned_length_columns = [
    column
    for column in list_length_columns
    if not column.startswith(
        "UNIQUE_VALID_GENE_SYMBOL_LIST"
    )
]


list_alignment_error_mask = (
    PATIENT_301_LIST_AGGREGATED_DF[
        mutation_aligned_length_columns
    ]
    .nunique(axis=1)
    .ne(1)
)


mutation_count_mismatch_mask = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "AGGREGATED_MUTATION_LIST_LENGTH"
    ]
    .ne(
        PATIENT_301_LIST_AGGREGATED_DF[
            "VALID_301_MUTATION_COUNT"
        ]
    )
)


gene_count_mismatch_mask = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "AGGREGATED_UNIQUE_GENE_LIST_LENGTH"
    ]
    .ne(
        PATIENT_301_LIST_AGGREGATED_DF[
            "VALID_301_GENE_COUNT"
        ]
    )
)


zero_valid_nonempty_list_mask = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .eq(0)
    &
    PATIENT_301_LIST_AGGREGATED_DF[
        "AGGREGATED_MUTATION_LIST_LENGTH"
    ]
    .ne(0)
)


valid_patient_empty_list_mask = (
    PATIENT_301_LIST_AGGREGATED_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .gt(0)
    &
    PATIENT_301_LIST_AGGREGATED_DF[
        "AGGREGATED_MUTATION_LIST_LENGTH"
    ]
    .eq(0)
)


duplicate_patient_mask = (
    PATIENT_301_LIST_AGGREGATED_DF
    .duplicated(
        subset=[
            "COHORT",
            "PATIENT_ID",
        ],
        keep=False,
    )
)


duplicate_sample_mask = (
    PATIENT_301_LIST_AGGREGATED_DF
    .duplicated(
        subset=[
            "COHORT",
            "SAMPLE_ID",
        ],
        keep=False,
    )
)


# ------------------------------------------------------------
# 12. Drop temporary list-length validation columns
# ------------------------------------------------------------

PATIENT_301_LIST_AGGREGATED_DF = (
    PATIENT_301_LIST_AGGREGATED_DF.drop(
        columns=list_length_columns,
        errors="ignore",
    )
)


# ------------------------------------------------------------
# 13. Precalculate summary counts
# ------------------------------------------------------------

total_patient_count = len(
    PATIENT_301_LIST_AGGREGATED_DF
)


patients_with_valid_mutations = int(
    PATIENT_301_LIST_AGGREGATED_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .gt(0)
    .sum()
)


patients_without_valid_mutations = int(
    PATIENT_301_LIST_AGGREGATED_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .eq(0)
    .sum()
)


total_aggregated_mutations = int(
    PATIENT_301_LIST_AGGREGATED_DF[
        "AGGREGATED_MUTATION_LIST_LENGTH"
    ]
    .sum()
)


minimum_valid_mutations = int(
    PATIENT_301_LIST_AGGREGATED_DF.loc[
        PATIENT_301_LIST_AGGREGATED_DF[
            "VALID_301_MUTATION_COUNT"
        ]
        .gt(0),
        "VALID_301_MUTATION_COUNT",
    ]
    .min()
)


median_valid_mutations = float(
    PATIENT_301_LIST_AGGREGATED_DF.loc[
        PATIENT_301_LIST_AGGREGATED_DF[
            "VALID_301_MUTATION_COUNT"
        ]
        .gt(0),
        "VALID_301_MUTATION_COUNT",
    ]
    .median()
)


maximum_valid_mutations = int(
    PATIENT_301_LIST_AGGREGATED_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .max()
)


list_alignment_error_count = int(
    list_alignment_error_mask.sum()
)


mutation_count_mismatch_count = int(
    mutation_count_mismatch_mask.sum()
)


gene_count_mismatch_count = int(
    gene_count_mismatch_mask.sum()
)


zero_valid_nonempty_count = int(
    zero_valid_nonempty_list_mask.sum()
)


valid_patient_empty_count = int(
    valid_patient_empty_list_mask.sum()
)


duplicate_patient_count = int(
    duplicate_patient_mask.sum()
)


duplicate_sample_count = int(
    duplicate_sample_mask.sum()
)


# ------------------------------------------------------------
# 14. Print validation results
# ------------------------------------------------------------

print(
    f"\nRepresentative patients retained: "
    f"{total_patient_count:,}"
)

print(
    f"Patients with at least one valid mutation: "
    f"{patients_with_valid_mutations:,}"
)

print(
    f"Patients with empty mutation lists: "
    f"{patients_without_valid_mutations:,}"
)

print(
    f"Total mutations preserved in lists: "
    f"{total_aggregated_mutations:,}"
)

print(
    f"Minimum valid mutations per represented patient: "
    f"{minimum_valid_mutations:,}"
)

print(
    f"Median valid mutations per represented patient: "
    f"{median_valid_mutations:.1f}"
)

print(
    f"Maximum valid mutations per patient: "
    f"{maximum_valid_mutations:,}"
)

print(
    f"Patients with misaligned mutation lists: "
    f"{list_alignment_error_count:,}"
)

print(
    f"Mutation-count mismatches: "
    f"{mutation_count_mismatch_count:,}"
)

print(
    f"Unique-gene-count mismatches: "
    f"{gene_count_mismatch_count:,}"
)

print(
    f"Zero-valid patients with nonempty lists: "
    f"{zero_valid_nonempty_count:,}"
)

print(
    f"Valid patients with empty lists: "
    f"{valid_patient_empty_count:,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{duplicate_patient_count:,}"
)

print(
    f"Duplicate cohort-sample rows: "
    f"{duplicate_sample_count:,}"
)


# ------------------------------------------------------------
# 15. Cohort summary
# ------------------------------------------------------------

patient_list_cohort_summary_df = (
    PATIENT_301_LIST_AGGREGATED_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        representative_patients=(
            "PATIENT_ID",
            "size",
        ),

        patients_with_valid_mutations=(
            "HAS_VALID_301_RECONSTRUCTION_FLAG",
            "sum",
        ),

        valid_mutation_rows=(
            "VALID_301_MUTATION_COUNT",
            "sum",
        ),

        median_valid_mutations=(
            "VALID_301_MUTATION_COUNT",
            "median",
        ),

        maximum_valid_mutations=(
            "VALID_301_MUTATION_COUNT",
            "max",
        ),
    )
    .reset_index()
)


print(
    "\nPatient-level aggregation by cohort:"
)

print(
    patient_list_cohort_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 16. Show compact examples
# ------------------------------------------------------------

example_df = (
    PATIENT_301_LIST_AGGREGATED_DF.loc[
        PATIENT_301_LIST_AGGREGATED_DF[
            "VALID_301_MUTATION_COUNT"
        ]
        .gt(0)
    ]
    .head(8)
    .copy()
)


example_df[
    "FIRST_GENE_SYMBOLS"
] = (
    example_df[
        "GENE_SYMBOL_LIST"
    ]
    .apply(
        lambda values: values[:5]
    )
)


example_df[
    "FIRST_OPERATIONS"
] = (
    example_df[
        "HGVS_OPERATION_LIST"
    ]
    .apply(
        lambda values: values[:5]
    )
)


print(
    "\nPatient aggregation examples:"
)

print(
    example_df[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "VALID_301_MUTATION_COUNT",
            "VALID_301_GENE_COUNT",
            "FIRST_GENE_SYMBOLS",
            "FIRST_OPERATIONS",
        ]
    ]
    .to_string(index=False)
)


# ------------------------------------------------------------
# 17. Hard validations
# ------------------------------------------------------------

if total_patient_count != 4_998:
    raise ValueError(
        f"Expected 4,998 representative patients, "
        f"found {total_patient_count:,}."
    )


if patients_with_valid_mutations != 4_881:
    raise ValueError(
        "Expected 4,881 patients with valid "
        "reconstructions."
    )


if patients_without_valid_mutations != 117:
    raise ValueError(
        "Expected 117 patients with no valid "
        "reconstruction."
    )


if total_aggregated_mutations != len(
    MUTATION_301_CONSEQUENCE_DF
):
    raise ValueError(
        "The total number of aggregated mutations does "
        "not match the valid mutation-level table."
    )


if list_alignment_error_mask.any():
    raise ValueError(
        "Some patient mutation lists are not aligned."
    )


if mutation_count_mismatch_mask.any():
    raise ValueError(
        "Some patient mutation-list lengths do not match "
        "their valid mutation counts."
    )


if gene_count_mismatch_mask.any():
    raise ValueError(
        "Some unique gene-list lengths do not match "
        "their valid gene counts."
    )


if zero_valid_nonempty_list_mask.any():
    raise ValueError(
        "A zero-valid patient contains a nonempty "
        "mutation list."
    )


if valid_patient_empty_list_mask.any():
    raise ValueError(
        "A valid patient has an empty mutation list."
    )


if duplicate_patient_mask.any():
    raise ValueError(
        "Duplicate cohort-patient rows were generated."
    )


if duplicate_sample_mask.any():
    raise ValueError(
        "Duplicate cohort-sample rows were generated."
    )


print("\n" + "=" * 78)
print("STEP 10A FINISHED")
print("=" * 78)

print(
    "Patient-level aligned mutation-list table:"
)

print(
    "PATIENT_301_LIST_AGGREGATED_DF"
)

print(
    "\nNo patient train/validation/test split or "
    "model training was performed."
)
```

## Step 10B — Create Patient-Level Mutation Summaries and 96-Gene Indicators

This step derives patient-level features from the valid 301-bp mutation reconstructions.

The features include:

- mutation-operation counts
- severity-category counts
- frameshift and in-frame mutation counts
- severity-score summaries
- affected-position summaries
- padded-window counts
- number of distinct transcripts
- one binary mutation indicator for every gene in the complete 96-gene panel

A gene indicator equals `1` when the patient has at least one valid reconstructed mutation in that gene and `0` otherwise.

Patients without a valid reconstruction remain in the table with empty mutation lists and zero-valued summary features. No mutation lists are truncated, and no clinical stage or target labels are used

```python
# ============================================================
# STEP 10B — PATIENT SUMMARIES AND 96-GENE INDICATORS
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re


# ------------------------------------------------------------
# 1. Validate required inputs
# ------------------------------------------------------------

required_tables = [
    "PATIENT_301_LIST_AGGREGATED_DF",
    "MUTATION_301_CONSEQUENCE_DF",
]

missing_tables = [
    table_name
    for table_name in required_tables
    if table_name not in globals()
]

if missing_tables:
    raise RuntimeError(
        "Missing required table(s): "
        + ", ".join(missing_tables)
    )


if "selected_genes" not in globals():
    raise RuntimeError(
        "selected_genes was not found."
    )


print("=" * 78)
print("STEP 10B — PATIENT SUMMARIES AND 96-GENE INDICATORS")
print("=" * 78)


# ------------------------------------------------------------
# 2. Define patient keys
# ------------------------------------------------------------

patient_keys = [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
]


patient_base_df = (
    PATIENT_301_LIST_AGGREGATED_DF
    .copy()
)


mutation_df = (
    MUTATION_301_CONSEQUENCE_DF
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
            "MUTATION_ROW_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. Obtain the complete 96-gene panel
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Users\LAPTOP\Downloads\luad_reserch_code"
)

panel_file_path = (
    PROJECT_ROOT
    / "final_luad_results"
    / "data"
    / "luad_96_driver_genes.csv"
)


panel_genes = []


if panel_file_path.exists():

    panel_df = pd.read_csv(
        panel_file_path
    )


    candidate_gene_columns = [
        "GENE_SYMBOL",
        "GENE",
        "SYMBOL",
        "DRIVER_GENE",
        "GENE_NAME",
    ]


    panel_gene_column = next(
        (
            column
            for column in candidate_gene_columns
            if column in panel_df.columns
        ),
        None,
    )


    if panel_gene_column is not None:

        panel_genes = (
            panel_df[
                panel_gene_column
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .str.upper()
            .drop_duplicates()
            .tolist()
        )


# Use the already validated in-memory panel when necessary.
if len(panel_genes) != 96:

    panel_genes = sorted(
        {
            str(gene).strip().upper()
            for gene in selected_genes
            if str(gene).strip()
        }
    )


if len(panel_genes) != 96:
    raise ValueError(
        f"Expected 96 panel genes, found "
        f"{len(panel_genes):,}."
    )


if len(set(panel_genes)) != 96:
    raise ValueError(
        "The 96-gene panel contains duplicate genes."
    )


# ------------------------------------------------------------
# 4. Mutation-operation counts
# ------------------------------------------------------------

expected_operations = [
    "SUBSTITUTION",
    "DELETION",
    "INSERTION",
    "DUPLICATION",
    "DELINS",
]


operation_count_df = (
    mutation_df
    .pivot_table(
        index=patient_keys,
        columns="HGVS_OPERATION",
        values="MUTATION_ROW_ID",
        aggfunc="count",
        fill_value=0,
    )
    .reset_index()
)


operation_column_map = {
    operation: (
        "MUTATION_COUNT_"
        + operation
    )
    for operation in expected_operations
}


for operation in expected_operations:

    if operation not in operation_count_df.columns:
        operation_count_df[
            operation
        ] = 0


operation_count_df = (
    operation_count_df.rename(
        columns=operation_column_map
    )
)


operation_feature_columns = list(
    operation_column_map.values()
)


# ------------------------------------------------------------
# 5. Severity-category counts
# ------------------------------------------------------------

expected_severities = [
    "HIGH",
    "MODERATE",
    "LOW",
    "UNKNOWN",
]


severity_count_df = (
    mutation_df
    .pivot_table(
        index=patient_keys,
        columns="MUTATION_CONSEQUENCE_SEVERITY",
        values="MUTATION_ROW_ID",
        aggfunc="count",
        fill_value=0,
    )
    .reset_index()
)


severity_column_map = {
    severity: (
        "SEVERITY_COUNT_"
        + severity
    )
    for severity in expected_severities
}


for severity in expected_severities:

    if severity not in severity_count_df.columns:
        severity_count_df[
            severity
        ] = 0


severity_count_df = (
    severity_count_df.rename(
        columns=severity_column_map
    )
)


severity_feature_columns = list(
    severity_column_map.values()
)


# ------------------------------------------------------------
# 6. General patient-level mutation summaries
# ------------------------------------------------------------

mutation_df[
    "WINDOW_HAS_ANY_PADDING_FLAG"
] = (
    mutation_df[
        "REFERENCE_WINDOW_N_COUNT"
    ]
    .gt(0)
    |
    mutation_df[
        "MUTATED_WINDOW_N_COUNT"
    ]
    .gt(0)
).astype("int8")


general_summary_df = (
    mutation_df
    .groupby(
        patient_keys,
        sort=False,
        as_index=False,
    )
    .agg(
        DISTINCT_VALID_TRANSCRIPT_COUNT=(
            "TRANSCRIPT_ID_SELECTED",
            "nunique",
        ),

        FRAME_SHIFT_MUTATION_COUNT=(
            "FRAME_SHIFT_INFERRED_FLAG",
            "sum",
        ),

        IN_FRAME_LENGTH_CHANGE_COUNT=(
            "IN_FRAME_LENGTH_CHANGE_FLAG",
            "sum",
        ),

        NO_LENGTH_CHANGE_MUTATION_COUNT=(
            "NO_LENGTH_CHANGE_FLAG",
            "sum",
        ),

        HIGH_OR_MODERATE_MUTATION_COUNT=(
            "MUTATION_CONSEQUENCE_SEVERITY",
            lambda values: int(
                values.isin(
                    [
                        "HIGH",
                        "MODERATE",
                    ]
                ).sum()
            ),
        ),

        MUTATION_SEVERITY_SCORE_TOTAL=(
            "MUTATION_SEVERITY_SCORE",
            "sum",
        ),

        MUTATION_SEVERITY_SCORE_MEAN=(
            "MUTATION_SEVERITY_SCORE",
            "mean",
        ),

        MUTATION_SEVERITY_SCORE_MAX=(
            "MUTATION_SEVERITY_SCORE",
            "max",
        ),

        TOTAL_REFERENCE_AFFECTED_POSITIONS=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "sum",
        ),

        TOTAL_MUTATED_AFFECTED_POSITIONS=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "sum",
        ),

        MEAN_REFERENCE_AFFECTED_POSITIONS=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "mean",
        ),

        MEAN_MUTATED_AFFECTED_POSITIONS=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "mean",
        ),

        MAX_REFERENCE_AFFECTED_POSITIONS=(
            "REFERENCE_AFFECTED_POSITION_COUNT",
            "max",
        ),

        MAX_MUTATED_AFFECTED_POSITIONS=(
            "MUTATED_AFFECTED_POSITION_COUNT",
            "max",
        ),

        TOTAL_REFERENCE_WINDOW_N_COUNT=(
            "REFERENCE_WINDOW_N_COUNT",
            "sum",
        ),

        TOTAL_MUTATED_WINDOW_N_COUNT=(
            "MUTATED_WINDOW_N_COUNT",
            "sum",
        ),

        PADDED_WINDOW_PAIR_COUNT=(
            "WINDOW_HAS_ANY_PADDING_FLAG",
            "sum",
        ),
    )
)


# ------------------------------------------------------------
# 7. Build the complete 96-gene indicator matrix
# ------------------------------------------------------------

patient_gene_presence_df = (
    mutation_df[
        patient_keys
        +
        [
            "GENE_SYMBOL",
        ]
    ]
    .drop_duplicates()
    .assign(
        GENE_PRESENT_FLAG=1
    )
)


gene_indicator_df = (
    patient_gene_presence_df
    .pivot_table(
        index=patient_keys,
        columns="GENE_SYMBOL",
        values="GENE_PRESENT_FLAG",
        aggfunc="max",
        fill_value=0,
    )
    .reset_index()
)


def gene_indicator_column_name(gene):
    """
    Create a safe and readable indicator-column name.
    """

    safe_gene = re.sub(
        r"[^A-Z0-9]+",
        "_",
        str(gene).upper(),
    ).strip("_")

    return (
        "GENE_"
        + safe_gene
        + "_MUTATED_FLAG"
    )


gene_indicator_column_map = {
    gene: gene_indicator_column_name(
        gene
    )
    for gene in panel_genes
}


# Add absent genes as all-zero columns.
for gene in panel_genes:

    if gene not in gene_indicator_df.columns:

        gene_indicator_df[
            gene
        ] = 0


# Keep only the patient keys and complete panel.
gene_indicator_df = gene_indicator_df[
    patient_keys
    +
    panel_genes
].copy()


gene_indicator_df = (
    gene_indicator_df.rename(
        columns=gene_indicator_column_map
    )
)


gene_indicator_columns = [
    gene_indicator_column_map[gene]
    for gene in panel_genes
]


# ------------------------------------------------------------
# 8. Merge summaries onto all 4,998 patients
# ------------------------------------------------------------

PATIENT_301_FEATURES_DF = (
    patient_base_df
    .merge(
        operation_count_df[
            patient_keys
            +
            operation_feature_columns
        ],
        on=patient_keys,
        how="left",
        validate="one_to_one",
    )
    .merge(
        severity_count_df[
            patient_keys
            +
            severity_feature_columns
        ],
        on=patient_keys,
        how="left",
        validate="one_to_one",
    )
    .merge(
        general_summary_df,
        on=patient_keys,
        how="left",
        validate="one_to_one",
    )
    .merge(
        gene_indicator_df,
        on=patient_keys,
        how="left",
        validate="one_to_one",
    )
)


# ------------------------------------------------------------
# 9. Fill summary features for zero-valid patients
# ------------------------------------------------------------

integer_summary_columns = (
    operation_feature_columns
    +
    severity_feature_columns
    +
    [
        "DISTINCT_VALID_TRANSCRIPT_COUNT",
        "FRAME_SHIFT_MUTATION_COUNT",
        "IN_FRAME_LENGTH_CHANGE_COUNT",
        "NO_LENGTH_CHANGE_MUTATION_COUNT",
        "HIGH_OR_MODERATE_MUTATION_COUNT",
        "MUTATION_SEVERITY_SCORE_TOTAL",
        "MUTATION_SEVERITY_SCORE_MAX",
        "TOTAL_REFERENCE_AFFECTED_POSITIONS",
        "TOTAL_MUTATED_AFFECTED_POSITIONS",
        "MAX_REFERENCE_AFFECTED_POSITIONS",
        "MAX_MUTATED_AFFECTED_POSITIONS",
        "TOTAL_REFERENCE_WINDOW_N_COUNT",
        "TOTAL_MUTATED_WINDOW_N_COUNT",
        "PADDED_WINDOW_PAIR_COUNT",
    ]
    +
    gene_indicator_columns
)


for column in integer_summary_columns:

    PATIENT_301_FEATURES_DF[
        column
    ] = (
        pd.to_numeric(
            PATIENT_301_FEATURES_DF[
                column
            ],
            errors="coerce",
        )
        .fillna(0)
        .astype(int)
    )


float_summary_columns = [
    "MUTATION_SEVERITY_SCORE_MEAN",
    "MEAN_REFERENCE_AFFECTED_POSITIONS",
    "MEAN_MUTATED_AFFECTED_POSITIONS",
]


for column in float_summary_columns:

    PATIENT_301_FEATURES_DF[
        column
    ] = (
        pd.to_numeric(
            PATIENT_301_FEATURES_DF[
                column
            ],
            errors="coerce",
        )
        .fillna(0.0)
        .astype(float)
    )


# ------------------------------------------------------------
# 10. Add compact derived indicators
# ------------------------------------------------------------

PATIENT_301_FEATURES_DF[
    "HAS_ANY_VALID_MUTATION_FLAG"
] = (
    PATIENT_301_FEATURES_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .gt(0)
    .astype("int8")
)


PATIENT_301_FEATURES_DF[
    "HAS_HIGH_SEVERITY_MUTATION_FLAG"
] = (
    PATIENT_301_FEATURES_DF[
        "SEVERITY_COUNT_HIGH"
    ]
    .gt(0)
    .astype("int8")
)


PATIENT_301_FEATURES_DF[
    "HAS_FRAMESHIFT_MUTATION_FLAG"
] = (
    PATIENT_301_FEATURES_DF[
        "FRAME_SHIFT_MUTATION_COUNT"
    ]
    .gt(0)
    .astype("int8")
)


PATIENT_301_FEATURES_DF[
    "MUTATIONS_PER_VALID_GENE"
] = np.where(
    PATIENT_301_FEATURES_DF[
        "VALID_301_GENE_COUNT"
    ]
    .gt(0),

    PATIENT_301_FEATURES_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    /
    PATIENT_301_FEATURES_DF[
        "VALID_301_GENE_COUNT"
    ],

    0.0,
)


# ------------------------------------------------------------
# 11. Validation calculations
# ------------------------------------------------------------

operation_sum = (
    PATIENT_301_FEATURES_DF[
        operation_feature_columns
    ]
    .sum(axis=1)
)


severity_sum = (
    PATIENT_301_FEATURES_DF[
        severity_feature_columns
    ]
    .sum(axis=1)
)


gene_indicator_sum = (
    PATIENT_301_FEATURES_DF[
        gene_indicator_columns
    ]
    .sum(axis=1)
)


operation_count_mismatch_mask = (
    operation_sum.ne(
        PATIENT_301_FEATURES_DF[
            "VALID_301_MUTATION_COUNT"
        ]
    )
)


severity_count_mismatch_mask = (
    severity_sum.ne(
        PATIENT_301_FEATURES_DF[
            "VALID_301_MUTATION_COUNT"
        ]
    )
)


gene_indicator_mismatch_mask = (
    gene_indicator_sum.ne(
        PATIENT_301_FEATURES_DF[
            "VALID_301_GENE_COUNT"
        ]
    )
)


invalid_gene_indicator_mask = (
    ~PATIENT_301_FEATURES_DF[
        gene_indicator_columns
    ]
    .isin(
        [
            0,
            1,
        ]
    )
    .all(axis=1)
)


duplicate_patient_mask = (
    PATIENT_301_FEATURES_DF
    .duplicated(
        subset=[
            "COHORT",
            "PATIENT_ID",
        ],
        keep=False,
    )
)


duplicate_sample_mask = (
    PATIENT_301_FEATURES_DF
    .duplicated(
        subset=[
            "COHORT",
            "SAMPLE_ID",
        ],
        keep=False,
    )
)


zero_valid_mask = (
    PATIENT_301_FEATURES_DF[
        "VALID_301_MUTATION_COUNT"
    ]
    .eq(0)
)


zero_valid_feature_columns = (
    operation_feature_columns
    +
    severity_feature_columns
    +
    gene_indicator_columns
    +
    [
        "FRAME_SHIFT_MUTATION_COUNT",
        "IN_FRAME_LENGTH_CHANGE_COUNT",
        "NO_LENGTH_CHANGE_MUTATION_COUNT",
        "HIGH_OR_MODERATE_MUTATION_COUNT",
        "MUTATION_SEVERITY_SCORE_TOTAL",
        "TOTAL_REFERENCE_AFFECTED_POSITIONS",
        "TOTAL_MUTATED_AFFECTED_POSITIONS",
        "PADDED_WINDOW_PAIR_COUNT",
    ]
)


zero_valid_nonzero_feature_mask = (
    zero_valid_mask
    &
    PATIENT_301_FEATURES_DF[
        zero_valid_feature_columns
    ]
    .sum(axis=1)
    .ne(0)
)


# ------------------------------------------------------------
# 12. Detect represented and unrepresented panel genes
# ------------------------------------------------------------

gene_indicator_totals = (
    PATIENT_301_FEATURES_DF[
        gene_indicator_columns
    ]
    .sum()
)


PANEL_GENES_WITH_PATIENT_INDICATORS = [
    panel_genes[index]
    for index, column
    in enumerate(
        gene_indicator_columns
    )
    if gene_indicator_totals[
        column
    ] > 0
]


PANEL_GENES_WITH_ZERO_INDICATORS = [
    panel_genes[index]
    for index, column
    in enumerate(
        gene_indicator_columns
    )
    if gene_indicator_totals[
        column
    ] == 0
]


# ------------------------------------------------------------
# 13. Precalculate printed counts
# ------------------------------------------------------------

total_patient_count = len(
    PATIENT_301_FEATURES_DF
)

patient_feature_column_count = len(
    PATIENT_301_FEATURES_DF.columns
)

operation_mismatch_count = int(
    operation_count_mismatch_mask.sum()
)

severity_mismatch_count = int(
    severity_count_mismatch_mask.sum()
)

gene_indicator_mismatch_count = int(
    gene_indicator_mismatch_mask.sum()
)

invalid_gene_indicator_count = int(
    invalid_gene_indicator_mask.sum()
)

duplicate_patient_count = int(
    duplicate_patient_mask.sum()
)

duplicate_sample_count = int(
    duplicate_sample_mask.sum()
)

zero_valid_nonzero_count = int(
    zero_valid_nonzero_feature_mask.sum()
)


# ------------------------------------------------------------
# 14. Print summary
# ------------------------------------------------------------

print(
    f"\nPatient rows: "
    f"{total_patient_count:,}"
)

print(
    f"Total patient-level columns: "
    f"{patient_feature_column_count:,}"
)

print(
    f"Gene-indicator columns created: "
    f"{len(gene_indicator_columns):,}"
)

print(
    f"Genes with at least one positive indicator: "
    f"{len(PANEL_GENES_WITH_PATIENT_INDICATORS):,}"
)

print(
    "Panel genes with all-zero indicators: "
    f"{PANEL_GENES_WITH_ZERO_INDICATORS}"
)

print(
    f"Operation-count mismatches: "
    f"{operation_mismatch_count:,}"
)

print(
    f"Severity-count mismatches: "
    f"{severity_mismatch_count:,}"
)

print(
    f"Gene-indicator/gene-count mismatches: "
    f"{gene_indicator_mismatch_count:,}"
)

print(
    f"Rows with invalid gene-indicator values: "
    f"{invalid_gene_indicator_count:,}"
)

print(
    f"Zero-valid patients with nonzero mutation features: "
    f"{zero_valid_nonzero_count:,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{duplicate_patient_count:,}"
)

print(
    f"Duplicate cohort-sample rows: "
    f"{duplicate_sample_count:,}"
)


# ------------------------------------------------------------
# 15. Compact cohort summary
# ------------------------------------------------------------

cohort_patient_feature_summary_df = (
    PATIENT_301_FEATURES_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        patients=(
            "PATIENT_ID",
            "size",
        ),

        patients_with_valid_mutations=(
            "HAS_ANY_VALID_MUTATION_FLAG",
            "sum",
        ),

        total_valid_mutations=(
            "VALID_301_MUTATION_COUNT",
            "sum",
        ),

        total_valid_genes=(
            "VALID_301_GENE_COUNT",
            "sum",
        ),

        patients_with_high_severity=(
            "HAS_HIGH_SEVERITY_MUTATION_FLAG",
            "sum",
        ),

        patients_with_frameshift=(
            "HAS_FRAMESHIFT_MUTATION_FLAG",
            "sum",
        ),
    )
    .reset_index()
)


print(
    "\nPatient mutation summaries by cohort:"
)

print(
    cohort_patient_feature_summary_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 16. Show compact examples
# ------------------------------------------------------------

example_indicator_columns = [
    column
    for column in gene_indicator_columns
    if PATIENT_301_FEATURES_DF[
        column
    ].sum() > 0
][:8]


example_columns = (
    [
        "COHORT",
        "PATIENT_ID",
        "SAMPLE_ID",
        "VALID_301_MUTATION_COUNT",
        "VALID_301_GENE_COUNT",
        "SEVERITY_COUNT_HIGH",
        "SEVERITY_COUNT_MODERATE",
        "FRAME_SHIFT_MUTATION_COUNT",
        "MUTATIONS_PER_VALID_GENE",
    ]
    +
    example_indicator_columns
)


print(
    "\nPatient-level feature examples:"
)

print(
    PATIENT_301_FEATURES_DF[
        example_columns
    ]
    .head(8)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 17. Hard validations
# ------------------------------------------------------------

if total_patient_count != 4_998:
    raise ValueError(
        f"Expected 4,998 patients, found "
        f"{total_patient_count:,}."
    )


if len(gene_indicator_columns) != 96:
    raise ValueError(
        "Exactly 96 gene indicators were not created."
    )


if operation_count_mismatch_mask.any():
    raise ValueError(
        "Patient operation counts do not equal valid "
        "mutation counts."
    )


if severity_count_mismatch_mask.any():
    raise ValueError(
        "Patient severity counts do not equal valid "
        "mutation counts."
    )


if gene_indicator_mismatch_mask.any():
    raise ValueError(
        "The number of positive gene indicators does not "
        "match the valid unique-gene count."
    )


if invalid_gene_indicator_mask.any():
    raise ValueError(
        "A gene indicator contains a value other than 0 or 1."
    )


if zero_valid_nonzero_feature_mask.any():
    raise ValueError(
        "A patient without valid mutations contains "
        "nonzero mutation features."
    )


if duplicate_patient_mask.any():
    raise ValueError(
        "Duplicate cohort-patient rows were generated."
    )


if duplicate_sample_mask.any():
    raise ValueError(
        "Duplicate cohort-sample rows were generated."
    )


print("\n" + "=" * 78)
print("STEP 10B FINISHED")
print("=" * 78)

print(
    "Patient-level mutation-list and summary-feature table:"
)

print(
    "PATIENT_301_FEATURES_DF"
)

print(
    "\nThe next step will merge the harmonised clinical "
    "information and perform final leakage validation."
)
```

## Step 10C — Merge Clinical Variables and Define Leakage-Safe Model Columns

This step merges the harmonised clinical information with the patient-level mutation dataset.

Clinical variables include:

- age
- sex
- active smoking
- passive smoking
- pack-years

Stage is retained only as the prediction target:

- four-class stage: I, II, III and IV
- binary stage: Early versus Advanced

The cohort, patient and sample identifiers are retained for patient-grouped splitting but are excluded from the candidate model-input columns.

No missing-value imputation, scaling, feature selection, dataset splitting or model training is performed in this notebook.

```python
# ============================================================
# STEP 10C — CLINICAL MERGE AND LEAKAGE-SAFE COLUMN DEFINITION
# ============================================================

import pandas as pd
import numpy as np
import re


# ------------------------------------------------------------
# 1. Validate required tables
# ------------------------------------------------------------

required_tables = [
    "PATIENT_301_FEATURES_DF",
    "SELECTED_SAMPLE_CLINICAL_BASE_DF",
]

missing_tables = [
    table_name
    for table_name in required_tables
    if table_name not in globals()
]

if missing_tables:
    raise RuntimeError(
        "Missing required table(s): "
        + ", ".join(missing_tables)
    )


print("=" * 78)
print("STEP 10C — CLINICAL MERGE AND LEAKAGE VALIDATION")
print("=" * 78)


# Defragment the table after the harmless PerformanceWarning.
patient_feature_df = (
    PATIENT_301_FEATURES_DF
    .copy()
)


clinical_source_df = (
    SELECTED_SAMPLE_CLINICAL_BASE_DF
    .copy()
)


patient_keys = [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
]


# ------------------------------------------------------------
# 2. Validate merge keys
# ------------------------------------------------------------

for table_name, dataframe in [
    (
        "PATIENT_301_FEATURES_DF",
        patient_feature_df,
    ),
    (
        "SELECTED_SAMPLE_CLINICAL_BASE_DF",
        clinical_source_df,
    ),
]:

    missing_key_columns = [
        column
        for column in patient_keys
        if column not in dataframe.columns
    ]

    if missing_key_columns:
        raise ValueError(
            f"{table_name} is missing key column(s): "
            + ", ".join(missing_key_columns)
        )


clinical_duplicate_key_mask = (
    clinical_source_df
    .duplicated(
        subset=patient_keys,
        keep=False,
    )
)


if clinical_duplicate_key_mask.any():
    raise ValueError(
        "The clinical source contains duplicate "
        "cohort-patient-sample rows."
    )


# ------------------------------------------------------------
# 3. Helper for locating harmonised clinical columns
# ------------------------------------------------------------

def find_column(
    dataframe,
    exact_candidates,
    include_patterns=None,
    exclude_patterns=None,
):
    """
    Locate one column using exact names first and then a
    restricted pattern search.
    """

    column_lookup = {
        str(column).upper(): column
        for column in dataframe.columns
    }


    for candidate in exact_candidates:

        candidate_upper = str(
            candidate
        ).upper()

        if candidate_upper in column_lookup:
            return column_lookup[
                candidate_upper
            ]


    include_patterns = (
        include_patterns or []
    )

    exclude_patterns = (
        exclude_patterns or []
    )


    for column in dataframe.columns:

        column_upper = str(
            column
        ).upper()


        include_match = all(
            re.search(
                pattern,
                column_upper,
            )
            for pattern in include_patterns
        )


        exclude_match = any(
            re.search(
                pattern,
                column_upper,
            )
            for pattern in exclude_patterns
        )


        if include_match and not exclude_match:
            return column


    return None


# ------------------------------------------------------------
# 4. Discover source columns
# ------------------------------------------------------------

age_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "AGE_FOR_MODEL",
        "HARMONISED_AGE_FOR_MODEL",
        "AGE_YEARS",
        "AGE",
    ],
    include_patterns=[
        r"AGE",
    ],
    exclude_patterns=[
        r"STAGE",
        r"FLAG",
        r"SOURCE",
        r"MISSING",
        r"RAW",
    ],
)


sex_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "SEX_STANDARDISED",
        "SEX_STANDARDIZED",
        "HARMONISED_SEX",
        "SEX",
    ],
    include_patterns=[
        r"SEX",
    ],
    exclude_patterns=[
        r"ENCODE",
        r"BINARY",
        r"FLAG",
        r"SOURCE",
    ],
)


active_smoking_status_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "ACTIVE_SMOKING_STATUS",
        "ACTIVE_SMOKING_STANDARDISED",
        "ACTIVE_SMOKING_STANDARDIZED",
        "SMOKING_STATUS_STANDARDISED",
        "SMOKING_STATUS_STANDARDIZED",
        "SMOKING_STATUS",
    ],
    include_patterns=[
        r"SMOK",
        r"STATUS",
    ],
    exclude_patterns=[
        r"PASSIVE",
        r"PACK",
        r"BINARY",
        r"ENCODE",
        r"FLAG",
        r"SOURCE",
    ],
)


active_smoking_binary_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "ACTIVE_SMOKING_BINARY",
        "ACTIVE_SMOKING_ENCODED",
        "SMOKING_STATUS_BINARY",
        "SMOKING_BINARY",
    ],
    include_patterns=[
        r"SMOK",
        r"(BINARY|ENCODE)",
    ],
    exclude_patterns=[
        r"PASSIVE",
        r"PACK",
    ],
)


passive_smoking_status_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "PASSIVE_SMOKING_STATUS",
        "PASSIVE_SMOKING_STANDARDISED",
        "PASSIVE_SMOKING_STANDARDIZED",
        "PASSIVE_SMOKING",
    ],
    include_patterns=[
        r"PASSIVE",
        r"SMOK",
    ],
    exclude_patterns=[
        r"BINARY",
        r"ENCODE",
        r"FLAG",
        r"SOURCE",
    ],
)


passive_smoking_binary_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "PASSIVE_SMOKING_BINARY",
        "PASSIVE_SMOKING_ENCODED",
    ],
    include_patterns=[
        r"PASSIVE",
        r"(BINARY|ENCODE)",
    ],
)


pack_years_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "PACK_YEARS_FOR_MODEL",
        "SMOKING_PACK_YEARS_FOR_MODEL",
        "SMOKING_PACK_YEARS",
        "PACK_YEARS",
    ],
    include_patterns=[
        r"PACK",
        r"YEAR",
    ],
    exclude_patterns=[
        r"FLAG",
        r"SOURCE",
        r"RAW",
    ],
)


stage_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "STAGE_STANDARDISED",
        "STAGE_STANDARDIZED",
        "STAGE_NORMALISED",
        "STAGE_NORMALIZED",
        "RESOLVED_STAGE",
        "FINAL_STAGE",
        "STAGE_GROUP",
        "STAGE",
    ],
    include_patterns=[
        r"STAGE",
    ],
    exclude_patterns=[
        r"BINARY",
        r"4CLASS",
        r"LABEL",
        r"FLAG",
        r"CONFLICT",
        r"SOURCE",
        r"RAW",
        r"CLINICAL_STAGE_SOURCE",
    ],
)


stage_four_class_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "LABEL_STAGE_4CLASS",
        "STAGE_LABEL_4CLASS",
        "STAGE_4CLASS_LABEL",
        "STAGE_4CLASS",
    ],
    include_patterns=[
        r"STAGE",
        r"4",
    ],
    exclude_patterns=[
        r"FLAG",
        r"SOURCE",
    ],
)


stage_binary_source_column = find_column(
    clinical_source_df,
    exact_candidates=[
        "LABEL_STAGE_BINARY",
        "STAGE_LABEL_BINARY",
        "STAGE_BINARY_LABEL",
        "STAGE_BINARY",
    ],
    include_patterns=[
        r"STAGE",
        r"BINARY",
    ],
    exclude_patterns=[
        r"FLAG",
        r"SOURCE",
    ],
)


selected_source_columns = {
    "Age": age_source_column,
    "Sex": sex_source_column,
    "Active smoking status":
        active_smoking_status_column,
    "Active smoking binary":
        active_smoking_binary_column,
    "Passive smoking status":
        passive_smoking_status_column,
    "Passive smoking binary":
        passive_smoking_binary_column,
    "Pack-years":
        pack_years_source_column,
    "Stage group":
        stage_source_column,
    "Stage four-class label":
        stage_four_class_source_column,
    "Stage binary label":
        stage_binary_source_column,
}


print("\nSelected clinical source columns:")

for field_name, column_name in (
    selected_source_columns.items()
):

    print(
        f"{field_name}: "
        f"{column_name}"
    )


if (
    stage_source_column is None
    and stage_four_class_source_column is None
):
    raise ValueError(
        "No harmonised stage group or four-class "
        "stage-label column was located."
    )


# ------------------------------------------------------------
# 5. Clinical normalisation helpers
# ------------------------------------------------------------

def normalise_sex(value):

    if value is None or pd.isna(value):
        return pd.NA

    text = str(
        value
    ).strip().upper()

    if text in {
        "F",
        "FEMALE",
        "WOMAN",
    }:
        return "Female"

    if text in {
        "M",
        "MALE",
        "MAN",
    }:
        return "Male"

    return pd.NA


def normalise_active_smoking(value):

    if value is None or pd.isna(value):
        return pd.NA

    text = str(
        value
    ).strip().upper()

    never_terms = [
        "NEVER",
        "NONSMOKER",
        "NON-SMOKER",
        "NO",
    ]

    ever_terms = [
        "EVER",
        "CURRENT",
        "FORMER",
        "YES",
        "SMOKER",
    ]


    if any(
        term in text
        for term in never_terms
    ):
        return "Never"

    if any(
        term in text
        for term in ever_terms
    ):
        return "Ever"

    return pd.NA


def normalise_passive_smoking(value):

    if value is None or pd.isna(value):
        return pd.NA

    text = str(
        value
    ).strip().upper()

    if text in {
        "YES",
        "Y",
        "1",
        "TRUE",
    }:
        return "Yes"

    if text in {
        "NO",
        "N",
        "0",
        "FALSE",
    }:
        return "No"

    return pd.NA


def normalise_stage_group(value):

    if value is None or pd.isna(value):
        return pd.NA

    text = str(
        value
    ).strip().upper()

    if text in {
        "",
        "NA",
        "N/A",
        "NAN",
        "NONE",
        "UNKNOWN",
        "NOT AVAILABLE",
    }:
        return pd.NA


    text = text.replace(
        "PATHOLOGIC",
        ""
    )

    text = text.replace(
        "PATHOLOGICAL",
        ""
    )

    text = text.replace(
        "CLINICAL",
        ""
    )

    text = text.replace(
        "STAGE",
        ""
    )

    text = re.sub(
        r"[^A-Z0-9]+",
        "",
        text,
    )


    if text.startswith("IV") or text == "4":
        return "IV"

    if text.startswith("III") or text == "3":
        return "III"

    if text.startswith("II") or text == "2":
        return "II"

    if text.startswith("I") or text == "1":
        return "I"

    return pd.NA


def normalise_existing_four_class_label(value):

    if value is None or pd.isna(value):
        return pd.NA

    numeric_value = pd.to_numeric(
        value,
        errors="coerce",
    )

    if pd.isna(numeric_value):
        return pd.NA

    integer_value = int(
        numeric_value
    )

    if integer_value in {
        0,
        1,
        2,
        3,
    }:
        return integer_value

    return pd.NA


def normalise_existing_binary_label(value):

    if value is None or pd.isna(value):
        return pd.NA

    numeric_value = pd.to_numeric(
        value,
        errors="coerce",
    )

    if pd.isna(numeric_value):
        return pd.NA

    integer_value = int(
        numeric_value
    )

    if integer_value in {
        0,
        1,
    }:
        return integer_value

    return pd.NA


# ------------------------------------------------------------
# 6. Build canonical clinical table
# ------------------------------------------------------------

clinical_canonical_df = (
    clinical_source_df[
        patient_keys
    ]
    .copy()
)


if age_source_column is not None:

    clinical_canonical_df[
        "CLINICAL_AGE_YEARS"
    ] = pd.to_numeric(
        clinical_source_df[
            age_source_column
        ],
        errors="coerce",
    )

else:

    clinical_canonical_df[
        "CLINICAL_AGE_YEARS"
    ] = np.nan


if sex_source_column is not None:

    clinical_canonical_df[
        "CLINICAL_SEX"
    ] = (
        clinical_source_df[
            sex_source_column
        ]
        .apply(normalise_sex)
        .astype("string")
    )

else:

    clinical_canonical_df[
        "CLINICAL_SEX"
    ] = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="string",
    )


clinical_canonical_df[
    "CLINICAL_SEX_BINARY"
] = (
    clinical_canonical_df[
        "CLINICAL_SEX"
    ]
    .map({
        "Female": 0,
        "Male": 1,
    })
    .astype("Int64")
)


# ------------------------------------------------------------
# 7. Active-smoking variables
# ------------------------------------------------------------

if active_smoking_status_column is not None:

    clinical_canonical_df[
        "CLINICAL_ACTIVE_SMOKING_STATUS"
    ] = (
        clinical_source_df[
            active_smoking_status_column
        ]
        .apply(
            normalise_active_smoking
        )
        .astype("string")
    )

else:

    clinical_canonical_df[
        "CLINICAL_ACTIVE_SMOKING_STATUS"
    ] = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="string",
    )


if active_smoking_binary_column is not None:

    active_binary_source = pd.to_numeric(
        clinical_source_df[
            active_smoking_binary_column
        ],
        errors="coerce",
    )

    active_binary_source = (
        active_binary_source
        .where(
            active_binary_source.isin(
                [
                    0,
                    1,
                ]
            )
        )
    )

else:

    active_binary_source = pd.Series(
        np.nan,
        index=clinical_canonical_df.index,
    )


active_binary_derived = (
    clinical_canonical_df[
        "CLINICAL_ACTIVE_SMOKING_STATUS"
    ]
    .map({
        "Never": 0,
        "Ever": 1,
    })
)


clinical_canonical_df[
    "CLINICAL_ACTIVE_SMOKING_BINARY"
] = (
    active_binary_source
    .combine_first(
        active_binary_derived
    )
    .astype("Int64")
)


# ------------------------------------------------------------
# 8. Passive-smoking variables
# ------------------------------------------------------------

if passive_smoking_status_column is not None:

    clinical_canonical_df[
        "CLINICAL_PASSIVE_SMOKING_STATUS"
    ] = (
        clinical_source_df[
            passive_smoking_status_column
        ]
        .apply(
            normalise_passive_smoking
        )
        .astype("string")
    )

else:

    clinical_canonical_df[
        "CLINICAL_PASSIVE_SMOKING_STATUS"
    ] = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="string",
    )


if passive_smoking_binary_column is not None:

    passive_binary_source = pd.to_numeric(
        clinical_source_df[
            passive_smoking_binary_column
        ],
        errors="coerce",
    )

    passive_binary_source = (
        passive_binary_source
        .where(
            passive_binary_source.isin(
                [
                    0,
                    1,
                ]
            )
        )
    )

else:

    passive_binary_source = pd.Series(
        np.nan,
        index=clinical_canonical_df.index,
    )


passive_binary_derived = (
    clinical_canonical_df[
        "CLINICAL_PASSIVE_SMOKING_STATUS"
    ]
    .map({
        "No": 0,
        "Yes": 1,
    })
)


clinical_canonical_df[
    "CLINICAL_PASSIVE_SMOKING_BINARY"
] = (
    passive_binary_source
    .combine_first(
        passive_binary_derived
    )
    .astype("Int64")
)


# ------------------------------------------------------------
# 9. Pack-years
# ------------------------------------------------------------

if pack_years_source_column is not None:

    clinical_canonical_df[
        "CLINICAL_PACK_YEARS"
    ] = pd.to_numeric(
        clinical_source_df[
            pack_years_source_column
        ],
        errors="coerce",
    )

else:

    clinical_canonical_df[
        "CLINICAL_PACK_YEARS"
    ] = np.nan


# ------------------------------------------------------------
# 10. Build canonical stage targets
# ------------------------------------------------------------

if stage_source_column is not None:

    clinical_canonical_df[
        "TARGET_STAGE_GROUP"
    ] = (
        clinical_source_df[
            stage_source_column
        ]
        .apply(
            normalise_stage_group
        )
        .astype("string")
    )

else:

    clinical_canonical_df[
        "TARGET_STAGE_GROUP"
    ] = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="string",
    )


stage_group_to_four_class = {
    "I": 0,
    "II": 1,
    "III": 2,
    "IV": 3,
}


derived_four_class = (
    clinical_canonical_df[
        "TARGET_STAGE_GROUP"
    ]
    .map(
        stage_group_to_four_class
    )
)


if stage_four_class_source_column is not None:

    existing_four_class = (
        clinical_source_df[
            stage_four_class_source_column
        ]
        .apply(
            normalise_existing_four_class_label
        )
    )

else:

    existing_four_class = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="Int64",
    )


clinical_canonical_df[
    "TARGET_STAGE_4CLASS"
] = (
    derived_four_class
    .combine_first(
        existing_four_class
    )
    .astype("Int64")
)


# Fill stage group from a valid numeric label when needed.
four_class_to_stage_group = {
    0: "I",
    1: "II",
    2: "III",
    3: "IV",
}


missing_stage_group_mask = (
    clinical_canonical_df[
        "TARGET_STAGE_GROUP"
    ]
    .isna()
    &
    clinical_canonical_df[
        "TARGET_STAGE_4CLASS"
    ]
    .notna()
)


clinical_canonical_df.loc[
    missing_stage_group_mask,
    "TARGET_STAGE_GROUP",
] = (
    clinical_canonical_df.loc[
        missing_stage_group_mask,
        "TARGET_STAGE_4CLASS",
    ]
    .map(
        four_class_to_stage_group
    )
)


derived_binary = (
    clinical_canonical_df[
        "TARGET_STAGE_4CLASS"
    ]
    .map({
        0: 0,
        1: 0,
        2: 1,
        3: 1,
    })
)


if stage_binary_source_column is not None:

    existing_binary = (
        clinical_source_df[
            stage_binary_source_column
        ]
        .apply(
            normalise_existing_binary_label
        )
    )

else:

    existing_binary = pd.Series(
        pd.NA,
        index=clinical_canonical_df.index,
        dtype="Int64",
    )


clinical_canonical_df[
    "TARGET_STAGE_BINARY"
] = (
    derived_binary
    .combine_first(
        existing_binary
    )
    .astype("Int64")
)


clinical_canonical_df[
    "TARGET_STAGE_BINARY_NAME"
] = (
    clinical_canonical_df[
        "TARGET_STAGE_BINARY"
    ]
    .map({
        0: "Early",
        1: "Advanced",
    })
    .astype("string")
)


clinical_canonical_df[
    "HAS_STAGE_TARGET_FLAG"
] = (
    clinical_canonical_df[
        "TARGET_STAGE_4CLASS"
    ]
    .notna()
    .astype("int8")
)


# ------------------------------------------------------------
# 11. Merge mutation and clinical tables
# ------------------------------------------------------------

FINAL_LUAD_PATIENT_DATASET_DF = (
    patient_feature_df.merge(
        clinical_canonical_df,
        on=patient_keys,
        how="left",
        validate="one_to_one",
        indicator="_CLINICAL_MERGE_STATUS",
    )
)


# Place identifiers and clinical columns first.
clinical_output_columns = [
    "CLINICAL_AGE_YEARS",
    "CLINICAL_SEX",
    "CLINICAL_SEX_BINARY",
    "CLINICAL_ACTIVE_SMOKING_STATUS",
    "CLINICAL_ACTIVE_SMOKING_BINARY",
    "CLINICAL_PASSIVE_SMOKING_STATUS",
    "CLINICAL_PASSIVE_SMOKING_BINARY",
    "CLINICAL_PACK_YEARS",
]


target_output_columns = [
    "TARGET_STAGE_GROUP",
    "TARGET_STAGE_4CLASS",
    "TARGET_STAGE_BINARY",
    "TARGET_STAGE_BINARY_NAME",
    "HAS_STAGE_TARGET_FLAG",
]


remaining_columns = [
    column
    for column in
    FINAL_LUAD_PATIENT_DATASET_DF.columns
    if column not in (
        patient_keys
        +
        clinical_output_columns
        +
        target_output_columns
        +
        [
            "_CLINICAL_MERGE_STATUS",
        ]
    )
]


FINAL_LUAD_PATIENT_DATASET_DF = (
    FINAL_LUAD_PATIENT_DATASET_DF[
        patient_keys
        +
        clinical_output_columns
        +
        target_output_columns
        +
        remaining_columns
        +
        [
            "_CLINICAL_MERGE_STATUS",
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 12. Define column roles for the next notebook
# ------------------------------------------------------------

FINAL_IDENTIFIER_COLUMNS = [
    "COHORT",
    "PATIENT_ID",
    "SAMPLE_ID",
]


FINAL_TARGET_ONLY_COLUMNS = (
    target_output_columns.copy()
)


FINAL_SAFE_CLINICAL_FEATURE_COLUMNS = [
    "CLINICAL_AGE_YEARS",
    "CLINICAL_SEX",
    "CLINICAL_SEX_BINARY",
    "CLINICAL_ACTIVE_SMOKING_STATUS",
    "CLINICAL_ACTIVE_SMOKING_BINARY",
    "CLINICAL_PASSIVE_SMOKING_STATUS",
    "CLINICAL_PASSIVE_SMOKING_BINARY",
    "CLINICAL_PACK_YEARS",
]


FINAL_SEQUENCE_INPUT_COLUMNS = [
    column
    for column in [
        "GENE_SYMBOL_LIST",
        "REFERENCE_WINDOW_301_LIST",
        "MUTATED_WINDOW_301_LIST",
        "REFERENCE_AFFECTED_MASK_301_LIST",
        "MUTATED_AFFECTED_MASK_301_LIST",
    ]
    if column in
    FINAL_LUAD_PATIENT_DATASET_DF.columns
]


metadata_only_columns = [
    column
    for column in [
        "MUTATION_ROW_ID_LIST",
        "TRANSCRIPT_ID_LIST",
        "HGVSC_LIST",
        "VARIANT_CLASSIFICATION_LIST",
        "CDS_START_1BASED_LIST",
        "CDS_END_1BASED_LIST",
        "_CLINICAL_MERGE_STATUS",
    ]
    if column in
    FINAL_LUAD_PATIENT_DATASET_DF.columns
]


numeric_mutation_feature_columns = [
    column
    for column in
    FINAL_LUAD_PATIENT_DATASET_DF.columns
    if (
        pd.api.types.is_numeric_dtype(
            FINAL_LUAD_PATIENT_DATASET_DF[
                column
            ]
        )
        and column not in (
            FINAL_IDENTIFIER_COLUMNS
            +
            FINAL_TARGET_ONLY_COLUMNS
            +
            FINAL_SAFE_CLINICAL_FEATURE_COLUMNS
        )
    )
]


FINAL_MODEL_CANDIDATE_COLUMNS = list(
    dict.fromkeys(
        FINAL_SEQUENCE_INPUT_COLUMNS
        +
        FINAL_SAFE_CLINICAL_FEATURE_COLUMNS
        +
        numeric_mutation_feature_columns
    )
)


FINAL_METADATA_ONLY_COLUMNS = (
    metadata_only_columns
)


# ------------------------------------------------------------
# 13. Leakage audit
# ------------------------------------------------------------

target_leakage_pattern = re.compile(
    r"(STAGE|TARGET|LABEL|TNM|"
    r"TUMOU?R_SIZE|LYMPH|METASTASIS)",
    flags=re.IGNORECASE,
)


leaking_candidate_columns = [
    column
    for column in
    FINAL_MODEL_CANDIDATE_COLUMNS
    if target_leakage_pattern.search(
        column
    )
]


cohort_in_model_features = (
    "COHORT"
    in FINAL_MODEL_CANDIDATE_COLUMNS
)


identifier_in_model_features = [
    column
    for column in
    FINAL_IDENTIFIER_COLUMNS
    if column in
    FINAL_MODEL_CANDIDATE_COLUMNS
]


# ------------------------------------------------------------
# 14. Validation masks
# ------------------------------------------------------------

clinical_unmatched_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF[
        "_CLINICAL_MERGE_STATUS"
    ]
    .ne("both")
)


duplicate_patient_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF
    .duplicated(
        subset=[
            "COHORT",
            "PATIENT_ID",
        ],
        keep=False,
    )
)


duplicate_sample_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF
    .duplicated(
        subset=[
            "COHORT",
            "SAMPLE_ID",
        ],
        keep=False,
    )
)


four_binary_consistency_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_4CLASS"
    ]
    .notna()
    &
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_BINARY"
    ]
    .notna()
    &
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_BINARY"
    ]
    .ne(
        FINAL_LUAD_PATIENT_DATASET_DF[
            "TARGET_STAGE_4CLASS"
        ]
        .map({
            0: 0,
            1: 0,
            2: 1,
            3: 1,
        })
    )
)


invalid_four_class_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_4CLASS"
    ]
    .notna()
    &
    ~FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_4CLASS"
    ]
    .isin([
        0,
        1,
        2,
        3,
    ])
)


invalid_binary_mask = (
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_BINARY"
    ]
    .notna()
    &
    ~FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_BINARY"
    ]
    .isin([
        0,
        1,
    ])
)


# ------------------------------------------------------------
# 15. Precalculate counts
# ------------------------------------------------------------

total_patient_count = len(
    FINAL_LUAD_PATIENT_DATASET_DF
)

clinical_matched_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "_CLINICAL_MERGE_STATUS"
    ]
    .eq("both")
    .sum()
)

clinical_unmatched_count = int(
    clinical_unmatched_mask.sum()
)

stage_target_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "HAS_STAGE_TARGET_FLAG"
    ]
    .sum()
)

stage_missing_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "HAS_STAGE_TARGET_FLAG"
    ]
    .eq(0)
    .sum()
)

age_available_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "CLINICAL_AGE_YEARS"
    ]
    .notna()
    .sum()
)

sex_available_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "CLINICAL_SEX"
    ]
    .notna()
    .sum()
)

active_smoking_available_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "CLINICAL_ACTIVE_SMOKING_BINARY"
    ]
    .notna()
    .sum()
)

passive_smoking_available_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "CLINICAL_PASSIVE_SMOKING_BINARY"
    ]
    .notna()
    .sum()
)

pack_years_available_count = int(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "CLINICAL_PACK_YEARS"
    ]
    .notna()
    .sum()
)

duplicate_patient_count = int(
    duplicate_patient_mask.sum()
)

duplicate_sample_count = int(
    duplicate_sample_mask.sum()
)

stage_consistency_error_count = int(
    four_binary_consistency_mask.sum()
)

invalid_four_class_count = int(
    invalid_four_class_mask.sum()
)

invalid_binary_count = int(
    invalid_binary_mask.sum()
)


# ------------------------------------------------------------
# 16. Print merge and leakage report
# ------------------------------------------------------------

print(
    f"\nFinal patient rows: "
    f"{total_patient_count:,}"
)

print(
    f"Clinical rows matched: "
    f"{clinical_matched_count:,}"
)

print(
    f"Clinical rows unmatched: "
    f"{clinical_unmatched_count:,}"
)

print(
    f"Patients with stage target: "
    f"{stage_target_count:,}"
)

print(
    f"Patients without stage target: "
    f"{stage_missing_count:,}"
)

print(
    f"Age available: "
    f"{age_available_count:,}"
)

print(
    f"Sex available: "
    f"{sex_available_count:,}"
)

print(
    f"Active-smoking value available: "
    f"{active_smoking_available_count:,}"
)

print(
    f"Passive-smoking value available: "
    f"{passive_smoking_available_count:,}"
)

print(
    f"Pack-years available: "
    f"{pack_years_available_count:,}"
)

print(
    f"Candidate model-input columns: "
    f"{len(FINAL_MODEL_CANDIDATE_COLUMNS):,}"
)

print(
    f"Target-only columns: "
    f"{len(FINAL_TARGET_ONLY_COLUMNS):,}"
)

print(
    f"Target-like candidate input columns: "
    f"{len(leaking_candidate_columns):,}"
)

print(
    f"Identifiers present in model-input columns: "
    f"{len(identifier_in_model_features):,}"
)

print(
    f"Cohort included as a model feature: "
    f"{cohort_in_model_features}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{duplicate_patient_count:,}"
)

print(
    f"Duplicate cohort-sample rows: "
    f"{duplicate_sample_count:,}"
)

print(
    f"Four-class/binary target inconsistencies: "
    f"{stage_consistency_error_count:,}"
)

print(
    f"Invalid four-class targets: "
    f"{invalid_four_class_count:,}"
)

print(
    f"Invalid binary targets: "
    f"{invalid_binary_count:,}"
)


print(
    "\nFour-class stage distribution:"
)

print(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_GROUP"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nBinary stage distribution:"
)

print(
    FINAL_LUAD_PATIENT_DATASET_DF[
        "TARGET_STAGE_BINARY_NAME"
    ]
    .value_counts(dropna=False)
    .to_string()
)


print(
    "\nStage-target coverage by cohort:"
)

print(
    FINAL_LUAD_PATIENT_DATASET_DF
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        patients=(
            "PATIENT_ID",
            "size",
        ),

        patients_with_stage_target=(
            "HAS_STAGE_TARGET_FLAG",
            "sum",
        ),

        patients_with_valid_mutations=(
            "HAS_ANY_VALID_MUTATION_FLAG",
            "sum",
        ),
    )
    .reset_index()
    .to_string(index=False)
)


# ------------------------------------------------------------
# 17. Hard validations
# ------------------------------------------------------------

if total_patient_count != 4_998:
    raise ValueError(
        f"Expected 4,998 patients, found "
        f"{total_patient_count:,}."
    )


if clinical_unmatched_mask.any():
    raise ValueError(
        "Some patient rows did not match the clinical table."
    )


if stage_target_count != 3_558:
    raise ValueError(
        f"Expected 3,558 patients with stage targets, "
        f"found {stage_target_count:,}. "
        "Check the selected stage source column."
    )


if duplicate_patient_mask.any():
    raise ValueError(
        "Duplicate cohort-patient rows were generated."
    )


if duplicate_sample_mask.any():
    raise ValueError(
        "Duplicate cohort-sample rows were generated."
    )


if four_binary_consistency_mask.any():
    raise ValueError(
        "Four-class and binary stage targets are inconsistent."
    )


if invalid_four_class_mask.any():
    raise ValueError(
        "An invalid four-class target was generated."
    )


if invalid_binary_mask.any():
    raise ValueError(
        "An invalid binary target was generated."
    )


if leaking_candidate_columns:
    raise ValueError(
        "Target-like columns entered the candidate "
        "model-input list: "
        + ", ".join(leaking_candidate_columns)
    )


if identifier_in_model_features:
    raise ValueError(
        "Identifier columns entered the candidate "
        "model-input list: "
        + ", ".join(identifier_in_model_features)
    )


if cohort_in_model_features:
    raise ValueError(
        "COHORT must be preserved for grouped evaluation "
        "but must not be used as a model feature."
    )


# Remove merge helper after validation.
FINAL_LUAD_PATIENT_DATASET_DF = (
    FINAL_LUAD_PATIENT_DATASET_DF.drop(
        columns=[
            "_CLINICAL_MERGE_STATUS",
        ],
        errors="ignore",
    )
    .copy()
)


print("\n" + "=" * 78)
print("STEP 10C FINISHED")
print("=" * 78)

print(
    "Final unsplit patient-level dataset:"
)

print(
    "FINAL_LUAD_PATIENT_DATASET_DF"
)

print(
    "\nCandidate model-input columns:"
)

print(
    "FINAL_MODEL_CANDIDATE_COLUMNS"
)

print(
    "\nTarget-only columns:"
)

print(
    "FINAL_TARGET_ONLY_COLUMNS"
)

print(
    "\nIdentifiers for grouped splitting:"
)

print(
    "FINAL_IDENTIFIER_COLUMNS"
)

print(
    "\nNo imputation, scaling, patient splitting or "
    "model training was performed."
)
```

## Step 10D — Save the Final Unsplit Patient-Level Dataset

This step saves the final patient-level LUAD dataset as one CSV file with one matching TXT documentation file.

The dataset retains:

- all 4,998 representative patients
- variable-length mutation lists without truncation
- paired 301-bp reference and mutated windows
- mutation masks
- mutation summary features
- all 96 gene indicators
- harmonised clinical variables
- stage targets stored separately from candidate model inputs

List-valued columns are stored as valid JSON arrays inside the CSV so that they can be restored using `json.loads()` in the model-training notebook.

No train, validation or test split is created in this notebook.

```python
# ============================================================
# STEP 10D — SAVE FINAL PATIENT-LEVEL CSV AND MATCHING TXT
# ============================================================

from pathlib import Path
from datetime import datetime
import pandas as pd
import numpy as np
import hashlib
import json
import os
import re


# ------------------------------------------------------------
# 1. Validate required objects
# ------------------------------------------------------------

required_objects = [
    "FINAL_LUAD_PATIENT_DATASET_DF",
    "FINAL_MODEL_CANDIDATE_COLUMNS",
    "FINAL_TARGET_ONLY_COLUMNS",
    "FINAL_IDENTIFIER_COLUMNS",
    "FINAL_SEQUENCE_INPUT_COLUMNS",
]

missing_objects = [
    object_name
    for object_name in required_objects
    if object_name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Missing required object(s): "
        + ", ".join(missing_objects)
    )


print("=" * 78)
print("STEP 10D — SAVE FINAL UNSPLIT PATIENT-LEVEL DATASET")
print("=" * 78)


# ------------------------------------------------------------
# 2. Define exactly two output files
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Users\LAPTOP\Downloads\luad_reserch_code"
)

DATA_DIR = (
    PROJECT_ROOT
    / "final_luad_results"
    / "data"
)

DATA_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


FINAL_PATIENT_CSV_PATH = (
    DATA_DIR
    / "luad_301bp_patient_level_dataset.csv"
)

FINAL_PATIENT_TXT_PATH = (
    DATA_DIR
    / "luad_301bp_patient_level_dataset.txt"
)


# ------------------------------------------------------------
# 3. Helper functions
# ------------------------------------------------------------

def json_default_converter(value):
    """
    Convert NumPy and pandas values into JSON-compatible
    Python values.
    """

    if value is None or value is pd.NA:
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):

        if np.isnan(value):
            return None

        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    raise TypeError(
        f"Unsupported JSON value type: {type(value)}"
    )


def serialise_list_value(value):
    """
    Convert a Python list to a compact valid JSON array.
    """

    if isinstance(value, list):

        return json.dumps(
            value,
            ensure_ascii=False,
            separators=(",", ":"),
            default=json_default_converter,
        )

    if value is None or pd.isna(value):

        return "[]"

    raise TypeError(
        "A list-valued column contains a non-list value: "
        f"{type(value)}"
    )


def atomic_save_csv(
    dataframe,
    output_path,
):
    """
    Save the CSV through a temporary file and replace the
    final file only when writing has completed.
    """

    temporary_path = Path(
        str(output_path) + ".tmp"
    )

    dataframe.to_csv(
        temporary_path,
        index=False,
        encoding="utf-8",
        lineterminator="\n",
    )

    os.replace(
        temporary_path,
        output_path,
    )


def atomic_save_text(
    text,
    output_path,
):
    """
    Save the TXT documentation atomically.
    """

    temporary_path = Path(
        str(output_path) + ".tmp"
    )

    temporary_path.write_text(
        text,
        encoding="utf-8",
    )

    os.replace(
        temporary_path,
        output_path,
    )


def calculate_sha256(file_path):
    """
    Calculate a SHA-256 checksum.
    """

    digest = hashlib.sha256()

    with open(file_path, "rb") as file_handle:

        for block in iter(
            lambda: file_handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()


def file_size_mb(file_path):
    """
    Return file size in megabytes.
    """

    return (
        file_path.stat().st_size
        /
        (1024 ** 2)
    )


# ------------------------------------------------------------
# 4. Prepare a clean export copy
# ------------------------------------------------------------

final_source_df = (
    FINAL_LUAD_PATIENT_DATASET_DF
    .copy()
    .sort_values(
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ],
        kind="stable",
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. Detect list-valued columns
# ------------------------------------------------------------

list_columns = []


for column in final_source_df.columns:

    contains_list = final_source_df[
        column
    ].apply(
        lambda value: isinstance(
            value,
            list,
        )
    ).any()

    if contains_list:
        list_columns.append(
            column
        )


print(
    f"\nList-valued columns detected: "
    f"{len(list_columns):,}"
)


# ------------------------------------------------------------
# 6. Validate that list columns contain only lists
# ------------------------------------------------------------

invalid_list_value_details = []


for column in list_columns:

    invalid_mask = (
        ~final_source_df[
            column
        ].apply(
            lambda value: (
                isinstance(value, list)
                or value is None
                or pd.isna(value)
            )
        )
    )


    invalid_count = int(
        invalid_mask.sum()
    )


    if invalid_count > 0:

        invalid_list_value_details.append(
            f"{column}: {invalid_count}"
        )


if invalid_list_value_details:
    raise ValueError(
        "Invalid values were found in list columns: "
        + "; ".join(
            invalid_list_value_details
        )
    )


# ------------------------------------------------------------
# 7. Validate aligned mutation-list lengths
# ------------------------------------------------------------

if "MUTATION_ROW_ID_LIST" not in list_columns:
    raise ValueError(
        "MUTATION_ROW_ID_LIST was not found."
    )


mutation_list_length = (
    final_source_df[
        "MUTATION_ROW_ID_LIST"
    ]
    .apply(len)
)


mutation_aligned_list_columns = [
    column
    for column in list_columns
    if column != "UNIQUE_VALID_GENE_SYMBOL_LIST"
]


list_alignment_error_columns = []


for column in mutation_aligned_list_columns:

    column_length = (
        final_source_df[
            column
        ]
        .apply(len)
    )


    mismatch_count = int(
        column_length.ne(
            mutation_list_length
        ).sum()
    )


    if mismatch_count > 0:

        list_alignment_error_columns.append(
            f"{column}: {mismatch_count}"
        )


if list_alignment_error_columns:
    raise ValueError(
        "Mutation lists are not aligned: "
        + "; ".join(
            list_alignment_error_columns
        )
    )


mutation_count_mismatch_count = int(
    mutation_list_length.ne(
        final_source_df[
            "VALID_301_MUTATION_COUNT"
        ]
    ).sum()
)


if mutation_count_mismatch_count > 0:
    raise ValueError(
        "Mutation-list lengths do not match "
        "VALID_301_MUTATION_COUNT."
    )


if (
    "UNIQUE_VALID_GENE_SYMBOL_LIST"
    in final_source_df.columns
):

    unique_gene_list_length = (
        final_source_df[
            "UNIQUE_VALID_GENE_SYMBOL_LIST"
        ]
        .apply(len)
    )


    unique_gene_count_mismatch = int(
        unique_gene_list_length.ne(
            final_source_df[
                "VALID_301_GENE_COUNT"
            ]
        ).sum()
    )


    if unique_gene_count_mismatch > 0:
        raise ValueError(
            "Unique gene-list lengths do not match "
            "VALID_301_GENE_COUNT."
        )


# ------------------------------------------------------------
# 8. Structural dataset validation
# ------------------------------------------------------------

total_patient_count = len(
    final_source_df
)


duplicate_patient_count = int(
    final_source_df.duplicated(
        subset=[
            "COHORT",
            "PATIENT_ID",
        ],
        keep=False,
    ).sum()
)


duplicate_sample_count = int(
    final_source_df.duplicated(
        subset=[
            "COHORT",
            "SAMPLE_ID",
        ],
        keep=False,
    ).sum()
)


missing_identifier_count = int(
    final_source_df[
        [
            "COHORT",
            "PATIENT_ID",
            "SAMPLE_ID",
        ]
    ]
    .isna()
    .any(axis=1)
    .sum()
)


patients_with_valid_mutations = int(
    final_source_df[
        "VALID_301_MUTATION_COUNT"
    ]
    .gt(0)
    .sum()
)


patients_without_valid_mutations = int(
    final_source_df[
        "VALID_301_MUTATION_COUNT"
    ]
    .eq(0)
    .sum()
)


total_valid_mutations = int(
    final_source_df[
        "VALID_301_MUTATION_COUNT"
    ]
    .sum()
)


patients_with_stage_target = int(
    final_source_df[
        "HAS_STAGE_TARGET_FLAG"
    ]
    .eq(1)
    .sum()
)


patients_without_stage_target = int(
    final_source_df[
        "HAS_STAGE_TARGET_FLAG"
    ]
    .eq(0)
    .sum()
)


gene_indicator_columns = [
    column
    for column in final_source_df.columns
    if re.fullmatch(
        r"GENE_[A-Z0-9_]+_MUTATED_FLAG",
        str(column),
    )
]


positive_gene_indicator_count = int(
    (
        final_source_df[
            gene_indicator_columns
        ]
        .sum(axis=0)
        .gt(0)
    )
    .sum()
)


all_zero_gene_indicator_columns = [
    column
    for column in gene_indicator_columns
    if final_source_df[
        column
    ].sum() == 0
]


all_zero_gene_symbols = [
    column
    .removeprefix("GENE_")
    .removesuffix("_MUTATED_FLAG")

    for column in
    all_zero_gene_indicator_columns
]


# ------------------------------------------------------------
# 9. Validate saved column-role definitions
# ------------------------------------------------------------

missing_model_candidate_columns = [
    column
    for column in FINAL_MODEL_CANDIDATE_COLUMNS
    if column not in final_source_df.columns
]


missing_target_columns = [
    column
    for column in FINAL_TARGET_ONLY_COLUMNS
    if column not in final_source_df.columns
]


missing_identifier_columns = [
    column
    for column in FINAL_IDENTIFIER_COLUMNS
    if column not in final_source_df.columns
]


if missing_model_candidate_columns:
    raise ValueError(
        "Candidate model columns are missing: "
        + ", ".join(
            missing_model_candidate_columns
        )
    )


if missing_target_columns:
    raise ValueError(
        "Target columns are missing: "
        + ", ".join(
            missing_target_columns
        )
    )


if missing_identifier_columns:
    raise ValueError(
        "Identifier columns are missing: "
        + ", ".join(
            missing_identifier_columns
        )
    )


target_leakage_pattern = re.compile(
    r"(STAGE|TARGET|LABEL|TNM|"
    r"TUMOU?R_SIZE|LYMPH|METASTASIS)",
    flags=re.IGNORECASE,
)


target_like_model_columns = [
    column
    for column in FINAL_MODEL_CANDIDATE_COLUMNS
    if target_leakage_pattern.search(
        str(column)
    )
]


identifier_model_overlap = sorted(
    set(
        FINAL_IDENTIFIER_COLUMNS
    )
    &
    set(
        FINAL_MODEL_CANDIDATE_COLUMNS
    )
)


# ------------------------------------------------------------
# 10. Hard pre-save validations
# ------------------------------------------------------------

if total_patient_count != 4_998:
    raise ValueError(
        f"Expected 4,998 patients, found "
        f"{total_patient_count:,}."
    )


if patients_with_valid_mutations != 4_881:
    raise ValueError(
        "Expected 4,881 patients with valid mutations."
    )


if patients_without_valid_mutations != 117:
    raise ValueError(
        "Expected 117 patients with empty mutation lists."
    )


if total_valid_mutations != 17_987:
    raise ValueError(
        f"Expected 17,987 valid mutations, found "
        f"{total_valid_mutations:,}."
    )


if patients_with_stage_target != 3_558:
    raise ValueError(
        f"Expected 3,558 patients with stage targets, "
        f"found {patients_with_stage_target:,}."
    )


if duplicate_patient_count > 0:
    raise ValueError(
        "Duplicate cohort-patient rows were detected."
    )


if duplicate_sample_count > 0:
    raise ValueError(
        "Duplicate cohort-sample rows were detected."
    )


if missing_identifier_count > 0:
    raise ValueError(
        "Missing patient identifiers were detected."
    )


if len(gene_indicator_columns) != 96:
    raise ValueError(
        f"Expected 96 gene indicators, found "
        f"{len(gene_indicator_columns):,}."
    )


if target_like_model_columns:
    raise ValueError(
        "Target-like columns entered the candidate "
        "model-input list: "
        + ", ".join(
            target_like_model_columns
        )
    )


if identifier_model_overlap:
    raise ValueError(
        "Identifier columns entered the candidate "
        "model-input list: "
        + ", ".join(
            identifier_model_overlap
        )
    )


# ------------------------------------------------------------
# 11. Convert list columns to JSON arrays
# ------------------------------------------------------------

FINAL_LUAD_PATIENT_EXPORT_DF = (
    final_source_df.copy()
)


for column in list_columns:

    FINAL_LUAD_PATIENT_EXPORT_DF[
        column
    ] = (
        FINAL_LUAD_PATIENT_EXPORT_DF[
            column
        ]
        .apply(
            serialise_list_value
        )
        .astype("string")
    )


# ------------------------------------------------------------
# 12. Build documentation summaries
# ------------------------------------------------------------

stage_four_class_counts_text = (
    final_source_df[
        "TARGET_STAGE_GROUP"
    ]
    .fillna("Missing")
    .value_counts(dropna=False)
    .to_string()
)


stage_binary_counts_text = (
    final_source_df[
        "TARGET_STAGE_BINARY_NAME"
    ]
    .fillna("Missing")
    .value_counts(dropna=False)
    .to_string()
)


cohort_summary_text = (
    final_source_df
    .groupby(
        "COHORT",
        sort=False,
    )
    .agg(
        PATIENTS=(
            "PATIENT_ID",
            "size",
        ),

        PATIENTS_WITH_STAGE=(
            "HAS_STAGE_TARGET_FLAG",
            "sum",
        ),

        PATIENTS_WITH_VALID_MUTATIONS=(
            "HAS_ANY_VALID_MUTATION_FLAG",
            "sum",
        ),

        VALID_MUTATIONS=(
            "VALID_301_MUTATION_COUNT",
            "sum",
        ),
    )
    .reset_index()
    .to_string(index=False)
)


identifier_columns_text = "\n".join(
    FINAL_IDENTIFIER_COLUMNS
)


target_columns_text = "\n".join(
    FINAL_TARGET_ONLY_COLUMNS
)


sequence_columns_text = "\n".join(
    FINAL_SEQUENCE_INPUT_COLUMNS
)


model_candidate_columns_text = "\n".join(
    FINAL_MODEL_CANDIDATE_COLUMNS
)


list_columns_text = "\n".join(
    list_columns
)


all_columns_text = "\n".join(
    FINAL_LUAD_PATIENT_EXPORT_DF.columns
)


zero_gene_text = (
    ", ".join(
        all_zero_gene_symbols
    )
    if all_zero_gene_symbols
    else "None"
)


# ------------------------------------------------------------
# 13. Save only the final patient CSV
# ------------------------------------------------------------

atomic_save_csv(
    FINAL_LUAD_PATIENT_EXPORT_DF,
    FINAL_PATIENT_CSV_PATH,
)


patient_csv_checksum = calculate_sha256(
    FINAL_PATIENT_CSV_PATH
)


patient_csv_size = file_size_mb(
    FINAL_PATIENT_CSV_PATH
)


# ------------------------------------------------------------
# 14. Create the matching TXT documentation
# ------------------------------------------------------------

creation_time = datetime.now().strftime(
    "%Y-%m-%d %H:%M:%S"
)


documentation_text = f"""
LUAD 301-BP PATIENT-LEVEL DATASET
================================

FILE INFORMATION
----------------
CSV file: {FINAL_PATIENT_CSV_PATH.name}
Created: {creation_time}
SHA-256: {patient_csv_checksum}
File size: {patient_csv_size:.2f} MB

PURPOSE
-------
This is the final unsplit patient-level dataset produced by
04_301bp_luad_dataset_from_scratch.ipynb.

Each row represents one representative patient and tumour sample.
All valid mutation reconstructions are retained as aligned,
variable-length JSON lists. No patient mutation list was truncated.

DATASET SIZE
------------
Patient rows: {total_patient_count:,}
Columns: {len(FINAL_LUAD_PATIENT_EXPORT_DF.columns):,}
Patients with at least one valid mutation: {patients_with_valid_mutations:,}
Patients with empty valid-mutation lists: {patients_without_valid_mutations:,}
Total valid 301-bp mutations: {total_valid_mutations:,}
Patients with stage targets: {patients_with_stage_target:,}
Patients without stage targets: {patients_without_stage_target:,}

GENE PANEL
----------
Gene-indicator columns: {len(gene_indicator_columns):,}
Genes with at least one positive indicator: {positive_gene_indicator_count:,}
All-zero gene indicators: {zero_gene_text}

FOUR-CLASS STAGE DISTRIBUTION
-----------------------------
{stage_four_class_counts_text}

BINARY STAGE DISTRIBUTION
-------------------------
{stage_binary_counts_text}

COHORT SUMMARY
--------------
{cohort_summary_text}

LIST SERIALISATION
------------------
List-valued columns are stored as valid JSON arrays inside CSV cells.

Example restoration in the model notebook:

    import json
    df["MUTATED_WINDOW_301_LIST"] = (
        df["MUTATED_WINDOW_301_LIST"].apply(json.loads)
    )

An empty mutation list is stored as:

    []

LIST-VALUED COLUMNS
-------------------
{list_columns_text}

IDENTIFIER COLUMNS
------------------
These columns must be retained for patient-grouped splitting and
evaluation. They must not be used as model features.

{identifier_columns_text}

TARGET-ONLY COLUMNS
-------------------
These columns are labels or target metadata. They must never enter
the input feature matrix.

{target_columns_text}

SEQUENCE INPUT COLUMNS
----------------------
{sequence_columns_text}

CANDIDATE MODEL-INPUT COLUMNS
-----------------------------
The following columns passed the leakage-name audit. The final model
notebook may choose a subset depending on the selected architecture.

{model_candidate_columns_text}

MODEL-TRAINING NOTEBOOK RULES
-----------------------------
1. Use only patients with HAS_STAGE_TARGET_FLAG equal to 1 for
   supervised stage prediction.

2. Create train, validation and test sets in the new notebook.

3. Split using the unique COHORT + PATIENT_ID grouping key.

4. Keep all mutation windows belonging to one patient in the same
   split.

5. Preserve COHORT for grouped evaluation, but do not use it as an
   input feature.

6. Fit imputation, encoding, scaling and feature-selection procedures
   using training data only.

7. Calculate class weights using training labels only.

8. Use validation data for threshold selection and hyperparameter
   tuning.

9. Do not use the test set for model selection or preprocessing
   decisions.

10. Do not oversample before patient-level splitting.

11. The 117 patients with empty valid-mutation lists are retained.
    Their handling must be defined after splitting in the model
    notebook.

12. Passive-smoking and pack-years variables have substantial
    missingness. Missing-value handling must be learned from the
    training split only.

LEAKAGE VALIDATION
------------------
Target-like candidate input columns: {len(target_like_model_columns):,}
Identifier/model-input overlaps: {len(identifier_model_overlap):,}
COHORT used as model feature: No
Dataset split performed: No
Imputation performed: No
Scaling performed: No
Model training performed: No

ALL DATASET COLUMNS
-------------------
{all_columns_text}
""".strip()


atomic_save_text(
    documentation_text,
    FINAL_PATIENT_TXT_PATH,
)


# ------------------------------------------------------------
# 15. Verify the saved CSV
# ------------------------------------------------------------

saved_key_df = pd.read_csv(
    FINAL_PATIENT_CSV_PATH,
    usecols=[
        "COHORT",
        "PATIENT_ID",
        "SAMPLE_ID",
        "VALID_301_MUTATION_COUNT",
        "HAS_STAGE_TARGET_FLAG",
        "MUTATION_ROW_ID_LIST",
    ],
    dtype={
        "COHORT": "string",
        "PATIENT_ID": "string",
        "SAMPLE_ID": "string",
        "MUTATION_ROW_ID_LIST": "string",
    },
)


saved_patient_count = len(
    saved_key_df
)


saved_valid_mutation_total = int(
    pd.to_numeric(
        saved_key_df[
            "VALID_301_MUTATION_COUNT"
        ],
        errors="raise",
    ).sum()
)


saved_stage_target_count = int(
    pd.to_numeric(
        saved_key_df[
            "HAS_STAGE_TARGET_FLAG"
        ],
        errors="raise",
    ).sum()
)


saved_json_parse_error_count = 0
saved_json_mutation_count = 0


for value in saved_key_df[
    "MUTATION_ROW_ID_LIST"
]:

    try:

        parsed_value = json.loads(
            value
        )

        if not isinstance(
            parsed_value,
            list,
        ):
            saved_json_parse_error_count += 1

        else:
            saved_json_mutation_count += len(
                parsed_value
            )

    except Exception:
        saved_json_parse_error_count += 1


if saved_patient_count != 4_998:
    raise ValueError(
        "Saved CSV patient count is incorrect."
    )


if saved_valid_mutation_total != 17_987:
    raise ValueError(
        "Saved CSV valid-mutation total is incorrect."
    )


if saved_json_mutation_count != 17_987:
    raise ValueError(
        "Saved JSON mutation-list total is incorrect."
    )


if saved_stage_target_count != 3_558:
    raise ValueError(
        "Saved CSV stage-target count is incorrect."
    )


if saved_json_parse_error_count > 0:
    raise ValueError(
        "Some saved mutation-list cells are not valid JSON."
    )


if not FINAL_PATIENT_TXT_PATH.exists():
    raise FileNotFoundError(
        "Matching TXT documentation was not saved."
    )


# ------------------------------------------------------------
# 16. Final report
# ------------------------------------------------------------

print(
    f"\nPatient rows saved: "
    f"{saved_patient_count:,}"
)

print(
    f"Columns saved: "
    f"{len(FINAL_LUAD_PATIENT_EXPORT_DF.columns):,}"
)

print(
    f"Valid mutations preserved in JSON lists: "
    f"{saved_json_mutation_count:,}"
)

print(
    f"Patients with stage targets: "
    f"{saved_stage_target_count:,}"
)

print(
    f"List-valued columns serialised as JSON: "
    f"{len(list_columns):,}"
)

print(
    f"JSON parsing errors after saving: "
    f"{saved_json_parse_error_count:,}"
)

print(
    f"Duplicate cohort-patient rows: "
    f"{duplicate_patient_count:,}"
)

print(
    f"Target-like model-input columns: "
    f"{len(target_like_model_columns):,}"
)

print(
    f"Identifier/model-input overlaps: "
    f"{len(identifier_model_overlap):,}"
)


print(
    "\nSaved files — only two files created:"
)

print(
    f"1. {FINAL_PATIENT_CSV_PATH}"
)

print(
    f"2. {FINAL_PATIENT_TXT_PATH}"
)


print(
    "\nPatient CSV SHA-256:"
)

print(
    patient_csv_checksum
)


print("\n" + "=" * 78)
print("STEP 10D FINISHED")
print("=" * 78)

print(
    "The 301-bp dataset-construction notebook is complete."
)

print(
    "\nModel splitting, preprocessing and training will be "
    "performed in a new Jupyter Notebook."
)
```
